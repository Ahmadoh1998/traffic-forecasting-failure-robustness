```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 4: Initialize output structure and inspect modeling data
# ============================================================

import os
import json
import pandas as pd
import numpy as np

BASE_PATH = str(OUTPUT_ROOT)

NB3_PATH = os.path.join(BASE_PATH, "Notebook_3_Output")
NB4_PATH = os.path.join(BASE_PATH, "Notebook_4_Output")
NB5_PATH = os.path.join(BASE_PATH, "Notebook_5_Output")

# ------------------------------------------------------------
# 1. Create Notebook 5 output structure
# ------------------------------------------------------------

subfolders = [
    "Data",
    "Models",
    "Results",
    "Tables",
    "Figures"
]

for folder in subfolders:
    os.makedirs(os.path.join(NB5_PATH, folder), exist_ok=True)

print("=" * 110)
print("NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING")
print("=" * 110)

print("\n✅ Notebook 5 output folders created:")
for folder in subfolders:
    print(f"   {os.path.join(NB5_PATH, folder)}")

# ------------------------------------------------------------
# 2. Inspect monthly forecasting datasets
# ------------------------------------------------------------

FORECASTING_PATH = os.path.join(
    NB3_PATH,
    "Data",
    "Forecasting_Monthly"
)

monthly_files = sorted([
    f for f in os.listdir(FORECASTING_PATH)
    if f.endswith(".parquet")
])

print("\n" + "=" * 110)
print("FORECASTING DATA FILES")
print("=" * 110)

for f in monthly_files:
    full_path = os.path.join(FORECASTING_PATH, f)
    size_mb = os.path.getsize(full_path) / (1024 ** 2)
    print(f"{f:<40} {size_mb:>10.2f} MB")

# ------------------------------------------------------------
# 3. Inspect representative TRAIN and TEST months
# ------------------------------------------------------------

train_file = os.path.join(
    FORECASTING_PATH,
    "FORECASTING_2026-06.parquet"
)

test_file = os.path.join(
    FORECASTING_PATH,
    "FORECASTING_2026-08.parquet"
)

train_sample = pd.read_parquet(train_file)
test_sample = pd.read_parquet(test_file)

print("\n" + "=" * 110)
print("DATASET STRUCTURE")
print("=" * 110)

print(f"\nJune shape   : {train_sample.shape}")
print(f"August shape : {test_sample.shape}")

print("\nColumns:")
for i, col in enumerate(train_sample.columns, start=1):
    print(f"{i:>3}. {col}")

# ------------------------------------------------------------
# 4. Missing-value structure
# ------------------------------------------------------------

lag_columns = [
    c for c in train_sample.columns
    if c.lower().startswith("lag_")
]

print("\n" + "=" * 110)
print("LAG FEATURE INSPECTION")
print("=" * 110)

print(f"\nDetected lag columns ({len(lag_columns)}):")
print(lag_columns)

for label, df in [
    ("June", train_sample),
    ("August", test_sample)
]:
    print(f"\n--- {label} ---")

    if lag_columns:
        missing_by_lag = df[lag_columns].isna().sum()

        print("Missing values by lag:")
        print(missing_by_lag.to_string())

        rows_with_missing = df[lag_columns].isna().any(axis=1).sum()

        print(
            f"\nRows with >=1 missing lag: "
            f"{rows_with_missing:,} / {len(df):,} "
            f"({100 * rows_with_missing / len(df):.3f}%)"
        )

# ------------------------------------------------------------
# 5. Inspect model-ready indicator / target candidates
# ------------------------------------------------------------

candidate_columns = [
    c for c in train_sample.columns
    if any(
        keyword in c.lower()
        for keyword in [
            "target",
            "actual",
            "volume",
            "model_ready",
            "valid"
        ]
    )
]

print("\n" + "=" * 110)
print("POTENTIAL TARGET / MODEL-READY COLUMNS")
print("=" * 110)

for c in candidate_columns:
    print(c)

# ------------------------------------------------------------
# 6. Load Notebook 4 controlled failure design metadata
# ------------------------------------------------------------

design_path = os.path.join(
    NB4_PATH,
    "Results",
    "Controlled_Failure_Injection_Design.json"
)

print("\n" + "=" * 110)
print("NOTEBOOK 4 FAILURE-INJECTION DESIGN")
print("=" * 110)

if os.path.exists(design_path):
    with open(design_path, "r", encoding="utf-8") as f:
        design = json.load(f)

    print(json.dumps(
        design,
        indent=2,
        ensure_ascii=False,
        default=str
    ))
else:
    print("❌ Controlled failure design file not found.")

print("\n" + "=" * 110)
print("STEP 4 COMPLETED")
print("=" * 110)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 5: Recover exact model specification and failure-event structure
# ============================================================

import os
import json
import pandas as pd
import lightgbm as lgb

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(BASE_PATH, "Notebook_3_Output")
NB4 = os.path.join(BASE_PATH, "Notebook_4_Output")

print("=" * 110)
print("NOTEBOOK 5 — EXACT EXPERIMENT RECOVERY")
print("=" * 110)

# ------------------------------------------------------------
# 1. Notebook 3 forecasting feature configuration
# ------------------------------------------------------------

feature_config_path = os.path.join(
    NB3,
    "Results",
    "Forecasting_Feature_Configuration.json"
)

print("\n" + "=" * 110)
print("1. NOTEBOOK 3 FEATURE CONFIGURATION")
print("=" * 110)

if os.path.exists(feature_config_path):
    with open(feature_config_path, "r", encoding="utf-8") as f:
        feature_config = json.load(f)

    print(json.dumps(
        feature_config,
        indent=2,
        ensure_ascii=False,
        default=str
    ))
else:
    print("❌ Feature configuration file not found.")

# ------------------------------------------------------------
# 2. Load the FINAL LightGBM model
# ------------------------------------------------------------

model_path = os.path.join(
    NB3,
    "Models",
    "LightGBM_FINAL_JanJul_1200.txt"
)

print("\n" + "=" * 110)
print("2. FINAL LIGHTGBM MODEL")
print("=" * 110)

if os.path.exists(model_path):

    booster = lgb.Booster(model_file=model_path)

    print("✅ Final LightGBM model loaded")
    print(f"Number of trees : {booster.num_trees()}")
    print(f"Features        : {booster.num_feature()}")

    print("\nModel feature names:")
    for i, f in enumerate(booster.feature_name(), start=1):
        print(f"{i:>2}. {f}")

    print("\nSelected model parameters:")
    model_params = booster.params

    important_params = [
        "objective",
        "metric",
        "learning_rate",
        "num_leaves",
        "max_depth",
        "min_data_in_leaf",
        "feature_fraction",
        "bagging_fraction",
        "bagging_freq",
        "lambda_l1",
        "lambda_l2",
        "seed"
    ]

    for p in important_params:
        if p in model_params:
            print(f"{p:<25}: {model_params[p]}")

else:
    print("❌ Final LightGBM model not found.")

# ------------------------------------------------------------
# 3. Notebook 4 controlled injection events
# ------------------------------------------------------------

events_path = os.path.join(
    NB4,
    "Data",
    "Controlled_August_Failure_Injection_Events.csv"
)

print("\n" + "=" * 110)
print("3. CONTROLLED AUGUST FAILURE EVENTS")
print("=" * 110)

if os.path.exists(events_path):

    events = pd.read_csv(events_path)

    print(f"Shape: {events.shape}")

    print("\nColumns:")
    for i, c in enumerate(events.columns, start=1):
        print(f"{i:>2}. {c}")

    print("\nFirst 10 events:")
    print(events.head(10).to_string(index=False))

    print("\nUnique detectors:",
          events["Detector_Code"].nunique()
          if "Detector_Code" in events.columns else "N/A")

    print("\nRun-length distribution:")
    run_candidates = [
        c for c in events.columns
        if "run" in c.lower() or "length" in c.lower()
    ]

    if run_candidates:
        for c in run_candidates:
            print(f"\n{c}:")
            print(events[c].describe().to_string())

else:
    print("❌ Controlled event file not found.")

# ------------------------------------------------------------
# 4. Event-level robustness output structure
# ------------------------------------------------------------

event_result_path = os.path.join(
    NB4,
    "Data",
    "Controlled_Failure_Event_Level_Robustness.parquet"
)

print("\n" + "=" * 110)
print("4. EVENT-LEVEL ROBUSTNESS DATA")
print("=" * 110)

if os.path.exists(event_result_path):

    event_results = pd.read_parquet(event_result_path)

    print(f"Shape: {event_results.shape}")

    print("\nColumns:")
    for i, c in enumerate(event_results.columns, start=1):
        print(f"{i:>2}. {c}")

    print("\nFirst 5 rows:")
    print(event_results.head().to_string(index=False))

else:
    print("❌ Event-level robustness file not found.")

# ------------------------------------------------------------
# 5. Affected injected prediction dataset
# ------------------------------------------------------------

affected_path = os.path.join(
    NB4,
    "Data",
    "Controlled_Failure_Injected_Affected_Predictions.parquet"
)

print("\n" + "=" * 110)
print("5. INJECTED AFFECTED PREDICTIONS")
print("=" * 110)

if os.path.exists(affected_path):

    affected = pd.read_parquet(affected_path)

    print(f"Shape: {affected.shape}")

    print("\nColumns:")
    for i, c in enumerate(affected.columns, start=1):
        print(f"{i:>2}. {c}")

    print("\nFirst 5 rows:")
    print(affected.head().to_string(index=False))

else:
    print("❌ Injected affected-prediction file not found.")

print("\n" + "=" * 110)
print("STEP 5 COMPLETED")
print("=" * 110)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 6: Build leakage-free training failure-pattern inventory
# ============================================================

import os
import pandas as pd
import numpy as np

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(BASE_PATH, "Notebook_3_Output")
NB4 = os.path.join(BASE_PATH, "Notebook_4_Output")
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

pattern_path = os.path.join(
    NB4,
    "Data",
    "PreTest_Real_Failure_Pattern_Library_JanJul.csv"
)

print("=" * 110)
print("NOTEBOOK 5 — LEAKAGE-FREE TRAINING FAILURE PATTERN INVENTORY")
print("=" * 110)

# ------------------------------------------------------------
# 1. Load empirical failure library
# ------------------------------------------------------------

patterns = pd.read_csv(pattern_path)

print("\n" + "=" * 110)
print("1. FULL JAN–JUL FAILURE LIBRARY")
print("=" * 110)

print(f"Shape: {patterns.shape}")

print("\nColumns:")
for i, c in enumerate(patterns.columns, start=1):
    print(f"{i:>2}. {c}")

print("\nFirst 10 rows:")
print(patterns.head(10).to_string(index=False))

# ------------------------------------------------------------
# 2. Detect date column
# ------------------------------------------------------------

date_candidates = [
    c for c in patterns.columns
    if "date" in c.lower()
]

print("\nPossible date columns:")
print(date_candidates)

if len(date_candidates) == 0:
    raise ValueError("No date column detected in failure-pattern library.")

# Prefer Source_Failure_Date if available
if "Source_Failure_Date" in patterns.columns:
    date_col = "Source_Failure_Date"
elif "Failure_Date" in patterns.columns:
    date_col = "Failure_Date"
else:
    date_col = date_candidates[0]

patterns[date_col] = pd.to_datetime(patterns[date_col])

print(f"\nUsing date column: {date_col}")

# ------------------------------------------------------------
# 3. Restrict patterns to TRAINING period only
#    January 1 → June 30, 2026
# ------------------------------------------------------------

train_patterns = patterns[
    (patterns[date_col] >= "2026-01-01") &
    (patterns[date_col] <= "2026-06-30")
].copy()

july_patterns = patterns[
    (patterns[date_col] >= "2026-07-01") &
    (patterns[date_col] <= "2026-07-31")
].copy()

print("\n" + "=" * 110)
print("2. TEMPORAL LEAKAGE CONTROL")
print("=" * 110)

print(f"Jan–Jun training patterns : {len(train_patterns):,}")
print(f"July excluded patterns    : {len(july_patterns):,}")
print(f"Total Jan–Jul patterns    : {len(patterns):,}")

if "Detector_Code" in train_patterns.columns:
    print(
        f"Unique training detectors : "
        f"{train_patterns['Detector_Code'].nunique():,}"
    )

# ------------------------------------------------------------
# 4. Monthly distribution
# ------------------------------------------------------------

train_patterns["Source_Month"] = (
    train_patterns[date_col]
    .dt.to_period("M")
    .astype(str)
)

monthly_counts = (
    train_patterns
    .groupby("Source_Month")
    .size()
    .reset_index(name="Failure_Runs")
)

print("\n" + "=" * 110)
print("3. TRAINING FAILURE PATTERNS BY MONTH")
print("=" * 110)

print(monthly_counts.to_string(index=False))

# ------------------------------------------------------------
# 5. Run-length distribution
# ------------------------------------------------------------

run_candidates = [
    c for c in train_patterns.columns
    if "run" in c.lower() or "length" in c.lower()
]

print("\n" + "=" * 110)
print("4. FAILURE RUN-LENGTH DISTRIBUTION")
print("=" * 110)

print("Possible run-length columns:")
print(run_candidates)

run_col = None

preferred_run_names = [
    "Run_Length",
    "Failure_Run_Length",
    "RunLength"
]

for c in preferred_run_names:
    if c in train_patterns.columns:
        run_col = c
        break

if run_col is None and run_candidates:
    run_col = run_candidates[0]

if run_col is not None:

    print(f"\nUsing run-length column: {run_col}")

    print("\nDescriptive statistics:")
    print(
        train_patterns[run_col]
        .describe(
            percentiles=[
                0.25,
                0.50,
                0.75,
                0.90,
                0.95,
                0.99
            ]
        )
        .to_string()
    )

    print("\nMost common run lengths:")
    print(
        train_patterns[run_col]
        .value_counts()
        .sort_index()
        .head(30)
        .to_string()
    )

# ------------------------------------------------------------
# 6. Start-interval distribution
# ------------------------------------------------------------

start_candidates = [
    c for c in train_patterns.columns
    if "start" in c.lower() and "interval" in c.lower()
]

print("\n" + "=" * 110)
print("5. START-INTERVAL INFORMATION")
print("=" * 110)

print("Possible start-interval columns:")
print(start_candidates)

if start_candidates:
    start_col = start_candidates[0]

    print(f"\nUsing start column: {start_col}")

    print("\nStart interval descriptive statistics:")
    print(train_patterns[start_col].describe().to_string())

# ------------------------------------------------------------
# 7. Day-of-week distribution
# ------------------------------------------------------------

dow_candidates = [
    c for c in train_patterns.columns
    if "dayofweek" in c.lower()
    or "day_of_week" in c.lower()
]

print("\n" + "=" * 110)
print("6. DAY-OF-WEEK INFORMATION")
print("=" * 110)

print("Possible day-of-week columns:")
print(dow_candidates)

if dow_candidates:

    dow_col = dow_candidates[0]

    print(f"\nUsing day-of-week column: {dow_col}")

    print(
        train_patterns[dow_col]
        .value_counts()
        .sort_index()
        .to_string()
    )

# ------------------------------------------------------------
# 8. Failure-pattern frequency per detector
# ------------------------------------------------------------

if "Detector_Code" in train_patterns.columns:

    detector_pattern_counts = (
        train_patterns
        .groupby("Detector_Code")
        .size()
        .rename("Failure_Runs")
        .reset_index()
    )

    print("\n" + "=" * 110)
    print("7. FAILURE PATTERNS PER DETECTOR")
    print("=" * 110)

    print(
        detector_pattern_counts["Failure_Runs"]
        .describe(
            percentiles=[
                0.25,
                0.50,
                0.75,
                0.90,
                0.95,
                0.99
            ]
        )
        .to_string()
    )

    print("\nDetectors with most observed training-period failures:")
    print(
        detector_pattern_counts
        .sort_values(
            "Failure_Runs",
            ascending=False
        )
        .head(20)
        .to_string(index=False)
    )

# ------------------------------------------------------------
# 9. Save leakage-free pattern library
# ------------------------------------------------------------

output_path = os.path.join(
    NB5,
    "Data",
    "Training_Failure_Pattern_Library_JanJun.csv"
)

train_patterns.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 110)
print("8. SAVED TRAINING FAILURE LIBRARY")
print("=" * 110)

print(f"✅ Saved to:")
print(output_path)

print("\n" + "=" * 110)
print("STEP 6 COMPLETED")
print("=" * 110)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 7: Audit naturally occurring missing-lag training samples
# ============================================================

import os
import pandas as pd
import numpy as np

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(BASE_PATH, "Notebook_3_Output")
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

lag_cols = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

base_feature_cols = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend"
]

needed_cols = [
    "Timestamp",
    "Detector_Code",
    "Interval_Index",
    "Target_Volume",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96",
    "All_Lags_Available",
    "Split"
]

training_months = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06"
]

print("=" * 115)
print("NOTEBOOK 5 — NATURAL MISSING-LAG TRAINING CANDIDATE AUDIT")
print("=" * 115)

monthly_summary = []
candidate_parts = []

# ============================================================
# 1. Process Jan-Jun separately to limit RAM usage
# ============================================================

for month in training_months:

    path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    print("\n" + "-" * 115)
    print(f"PROCESSING {month}")
    print("-" * 115)

    df = pd.read_parquet(
        path,
        columns=needed_cols
    )

    df["Timestamp"] = pd.to_datetime(df["Timestamp"])

    # --------------------------------------------------------
    # Identify missing historical inputs
    # --------------------------------------------------------

    lag_missing_mask = df[lag_cols].isna().any(axis=1)

    # Valid forecasting target
    valid_target_mask = (
        df["Target_Volume"].notna()
        & np.isfinite(df["Target_Volume"])
        & (df["Target_Volume"] >= 0)
    )

    # Check non-lag predictor integrity
    valid_base_features = (
        df[base_feature_cols]
        .notna()
        .all(axis=1)
    )

    # Natural failure-aware candidate:
    # valid target + valid identifiers/time features +
    # at least one missing historical lag
    candidate_mask = (
        lag_missing_mask
        & valid_target_mask
        & valid_base_features
    )

    total_rows = len(df)
    missing_lag_rows = int(lag_missing_mask.sum())
    valid_target_missing_lag_rows = int(
        candidate_mask.sum()
    )

    invalid_target_missing_lag_rows = int(
        (
            lag_missing_mask
            & ~valid_target_mask
        ).sum()
    )

    clean_model_ready_rows = int(
        (
            ~lag_missing_mask
            & valid_target_mask
            & valid_base_features
        ).sum()
    )

    print(f"Total rows                         : {total_rows:,}")
    print(f"Rows with >=1 missing lag         : {missing_lag_rows:,}")
    print(
        f"Missing-lag rows with VALID target: "
        f"{valid_target_missing_lag_rows:,}"
    )
    print(
        f"Missing-lag rows INVALID target   : "
        f"{invalid_target_missing_lag_rows:,}"
    )
    print(
        f"Clean model-ready rows            : "
        f"{clean_model_ready_rows:,}"
    )

    # --------------------------------------------------------
    # Extract valid natural failure-aware candidates
    # --------------------------------------------------------

    cand = df.loc[candidate_mask].copy()

    if len(cand) > 0:

        cand["Missing_Lag_Count"] = (
            cand[lag_cols]
            .isna()
            .sum(axis=1)
            .astype("int8")
        )

        # Create interpretable missing-lag pattern
        def build_pattern(row):
            missing = [
                col
                for col in lag_cols
                if pd.isna(row[col])
            ]
            return "+".join(missing)

        cand["Missing_Lag_Pattern"] = cand.apply(
            build_pattern,
            axis=1
        )

        cand["Source_Month"] = month

        candidate_parts.append(cand)

    monthly_summary.append({
        "Month": month,
        "Total_Rows": total_rows,
        "Missing_Lag_Rows": missing_lag_rows,
        "Missing_Lag_Percent":
            100 * missing_lag_rows / total_rows,
        "Valid_Target_Missing_Lag_Rows":
            valid_target_missing_lag_rows,
        "Invalid_Target_Missing_Lag_Rows":
            invalid_target_missing_lag_rows,
        "Clean_Model_Ready_Rows":
            clean_model_ready_rows
    })

    del df

# ============================================================
# 2. Combine all Jan-Jun candidates
# ============================================================

summary_df = pd.DataFrame(monthly_summary)

if candidate_parts:
    candidates = pd.concat(
        candidate_parts,
        ignore_index=True
    )
else:
    candidates = pd.DataFrame()

print("\n" + "=" * 115)
print("1. MONTHLY SUMMARY")
print("=" * 115)

print(summary_df.to_string(index=False))

print("\n" + "=" * 115)
print("2. JAN–JUN OVERALL SUMMARY")
print("=" * 115)

total_rows = summary_df["Total_Rows"].sum()
total_missing = summary_df["Missing_Lag_Rows"].sum()
total_valid_missing = summary_df[
    "Valid_Target_Missing_Lag_Rows"
].sum()

total_clean = summary_df[
    "Clean_Model_Ready_Rows"
].sum()

print(f"Total Jan–Jun rows                  : {total_rows:,}")
print(f"Clean model-ready rows              : {total_clean:,}")
print(f"Rows with >=1 missing lag           : {total_missing:,}")
print(
    f"Valid-target missing-lag candidates : "
    f"{total_valid_missing:,}"
)

print(
    f"\nObserved missing-lag prevalence     : "
    f"{100 * total_missing / total_rows:.4f}%"
)

if total_missing > 0:
    print(
        f"Missing-lag rows retaining valid target: "
        f"{100 * total_valid_missing / total_missing:.4f}%"
    )

# ============================================================
# 3. Candidate quality checks
# ============================================================

if len(candidates) > 0:

    print("\n" + "=" * 115)
    print("3. FAILURE-AWARE CANDIDATE QUALITY")
    print("=" * 115)

    print(f"Candidate rows          : {len(candidates):,}")
    print(
        f"Unique detectors       : "
        f"{candidates['Detector_Code'].nunique():,}"
    )
    print(
        f"Unique dates           : "
        f"{candidates['Timestamp'].dt.date.nunique():,}"
    )

    print(
        f"Target minimum         : "
        f"{candidates['Target_Volume'].min():.3f}"
    )
    print(
        f"Target maximum         : "
        f"{candidates['Target_Volume'].max():.3f}"
    )

    # --------------------------------------------------------
    # Missing-lag count distribution
    # --------------------------------------------------------

    print("\n" + "=" * 115)
    print("4. MISSING-LAG COUNT DISTRIBUTION")
    print("=" * 115)

    count_distribution = (
        candidates["Missing_Lag_Count"]
        .value_counts()
        .sort_index()
        .rename_axis("Missing_Lag_Count")
        .reset_index(name="Rows")
    )

    count_distribution["Percent"] = (
        100
        * count_distribution["Rows"]
        / len(candidates)
    )

    print(
        count_distribution.to_string(
            index=False
        )
    )

    # --------------------------------------------------------
    # Missing-lag pattern distribution
    # --------------------------------------------------------

    print("\n" + "=" * 115)
    print("5. MOST COMMON NATURAL MISSING-LAG PATTERNS")
    print("=" * 115)

    pattern_distribution = (
        candidates["Missing_Lag_Pattern"]
        .value_counts()
        .rename_axis("Missing_Lag_Pattern")
        .reset_index(name="Rows")
    )

    pattern_distribution["Percent"] = (
        100
        * pattern_distribution["Rows"]
        / len(candidates)
    )

    print(
        pattern_distribution
        .head(25)
        .to_string(index=False)
    )

    # --------------------------------------------------------
    # Individual lag missingness
    # --------------------------------------------------------

    print("\n" + "=" * 115)
    print("6. INDIVIDUAL LAG MISSINGNESS")
    print("=" * 115)

    individual_missing = []

    for col in lag_cols:

        n = int(candidates[col].isna().sum())

        individual_missing.append({
            "Lag": col,
            "Missing_Rows": n,
            "Percent_of_Candidates":
                100 * n / len(candidates)
        })

    individual_missing_df = pd.DataFrame(
        individual_missing
    )

    print(
        individual_missing_df.to_string(
            index=False
        )
    )

    # --------------------------------------------------------
    # Monthly candidate distribution
    # --------------------------------------------------------

    print("\n" + "=" * 115)
    print("7. FAILURE-AWARE CANDIDATES BY MONTH")
    print("=" * 115)

    candidate_months = (
        candidates
        .groupby("Source_Month")
        .size()
        .reset_index(name="Candidate_Rows")
    )

    candidate_months["Percent"] = (
        100
        * candidate_months["Candidate_Rows"]
        / len(candidates)
    )

    print(
        candidate_months.to_string(
            index=False
        )
    )

# ============================================================
# 4. Save outputs
# ============================================================

summary_path = os.path.join(
    NB5,
    "Tables",
    "Table_Natural_Missing_Lag_Training_Audit.csv"
)

summary_df.to_csv(
    summary_path,
    index=False
)

print("\n" + "=" * 115)
print("8. SAVING OUTPUTS")
print("=" * 115)

print(f"✅ Saved audit table:\n{summary_path}")

if len(candidates) > 0:

    candidate_path = os.path.join(
        NB5,
        "Data",
        "Natural_Failure_Aware_Training_Candidates.parquet"
    )

    candidates.to_parquet(
        candidate_path,
        index=False
    )

    print(
        f"\n✅ Saved natural failure-aware candidates:\n"
        f"{candidate_path}"
    )

    pattern_path = os.path.join(
        NB5,
        "Tables",
        "Table_Natural_Missing_Lag_Patterns.csv"
    )

    pattern_distribution.to_csv(
        pattern_path,
        index=False
    )

    print(
        f"\n✅ Saved missing-pattern table:\n"
        f"{pattern_path}"
    )

print("\n" + "=" * 115)
print("STEP 7 COMPLETED")
print("=" * 115)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 8: Audit provenance and representativeness of natural
#         missing-lag training samples
# ============================================================

import os
import pandas as pd
import numpy as np

BASE_PATH = str(OUTPUT_ROOT)

NB4 = os.path.join(BASE_PATH, "Notebook_4_Output")
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

lag_offsets = {
    "Lag_1": 15,
    "Lag_2": 30,
    "Lag_4": 60,
    "Lag_8": 120,
    "Lag_96": 1440
}

# ============================================================
# 1. Load data
# ============================================================

candidate_path = os.path.join(
    NB5,
    "Data",
    "Natural_Failure_Aware_Training_Candidates.parquet"
)

failure_library_path = os.path.join(
    NB5,
    "Data",
    "Training_Failure_Pattern_Library_JanJun.csv"
)

controlled_path = os.path.join(
    NB4,
    "Data",
    "Controlled_Failure_Injected_Affected_Predictions.parquet"
)

candidates = pd.read_parquet(candidate_path)
failures = pd.read_csv(failure_library_path)
controlled = pd.read_parquet(
    controlled_path,
    columns=[
        "Missing_Lag_Count",
        "Missing_Lag_Pattern"
    ]
)

candidates["Timestamp"] = pd.to_datetime(
    candidates["Timestamp"]
)

failures["Date"] = pd.to_datetime(
    failures["Date"]
).dt.normalize()

print("=" * 120)
print("NOTEBOOK 5 — NATURAL MISSINGNESS PROVENANCE & REPRESENTATIVENESS AUDIT")
print("=" * 120)

print(f"\nNatural candidate rows : {len(candidates):,}")
print(f"Training failure runs  : {len(failures):,}")
print(f"Controlled test rows   : {len(controlled):,}")


# ============================================================
# 2. Expand documented Jan-Jun real failure runs
#    into exact failed 15-minute timestamps
# ============================================================

expanded_parts = []

for row in failures.itertuples(index=False):

    intervals = np.arange(
        int(row.Start_Interval_Index),
        int(row.End_Interval_Index) + 1
    )

    timestamps = (
        row.Date
        + pd.to_timedelta(intervals * 15, unit="m")
    )

    expanded_parts.append(
        pd.DataFrame({
            "Detector_Code": row.Detector_Code,
            "Failure_Timestamp": timestamps
        })
    )

failure_intervals = (
    pd.concat(
        expanded_parts,
        ignore_index=True
    )
    .drop_duplicates()
)

print("\n" + "=" * 120)
print("1. EXPANDED EMPIRICAL FAILURE INTERVALS")
print("=" * 120)

print(
    f"Unique documented failed 15-min intervals: "
    f"{len(failure_intervals):,}"
)

print(
    f"Unique detectors with documented failures: "
    f"{failure_intervals['Detector_Code'].nunique():,}"
)


# ============================================================
# 3. Build indices for exact failure interval and
#    detector-day failure occurrence
# ============================================================

failure_interval_index = pd.MultiIndex.from_frame(
    failure_intervals[
        ["Detector_Code", "Failure_Timestamp"]
    ]
)

failure_days = (
    failures[
        ["Detector_Code", "Date"]
    ]
    .drop_duplicates()
)

failure_day_index = pd.MultiIndex.from_frame(
    failure_days[
        ["Detector_Code", "Date"]
    ]
)


# ============================================================
# 4. For every missing lag, determine whether its source
#    timestamp corresponds to:
#       A) exact documented failed interval
#       B) same detector-day containing any documented failure
# ============================================================

source_match_results = []

print("\n" + "=" * 120)
print("2. MISSING-LAG SOURCE PROVENANCE")
print("=" * 120)

for lag, offset_minutes in lag_offsets.items():

    sub = candidates.loc[
        candidates[lag].isna(),
        ["Detector_Code", "Timestamp"]
    ].copy()

    sub["Source_Timestamp"] = (
        sub["Timestamp"]
        - pd.Timedelta(minutes=offset_minutes)
    )

    sub["Source_Date"] = (
        sub["Source_Timestamp"]
        .dt.normalize()
    )

    source_interval_index = pd.MultiIndex.from_frame(
        sub[
            ["Detector_Code", "Source_Timestamp"]
        ]
    )

    source_day_index = pd.MultiIndex.from_frame(
        sub[
            ["Detector_Code", "Source_Date"]
        ].rename(
            columns={"Source_Date": "Date"}
        )
    )

    exact_matches = source_interval_index.isin(
        failure_interval_index
    )

    failure_day_matches = source_day_index.isin(
        failure_day_index
    )

    n = len(sub)
    exact_n = int(exact_matches.sum())
    day_n = int(failure_day_matches.sum())

    source_match_results.append({
        "Lag": lag,
        "Missing_Occurrences": n,
        "Exact_Failure_Interval_Matches": exact_n,
        "Exact_Match_Percent":
            100 * exact_n / n if n else np.nan,
        "Source_Day_Had_Failure": day_n,
        "Failure_Day_Match_Percent":
            100 * day_n / n if n else np.nan
    })

source_match_df = pd.DataFrame(
    source_match_results
)

print(
    source_match_df.to_string(
        index=False
    )
)


# ============================================================
# 5. Determine whether natural candidates appear in
#    complete detector-day blocks
# ============================================================

candidates["Candidate_Date"] = (
    candidates["Timestamp"]
    .dt.normalize()
)

detector_day_counts = (
    candidates
    .groupby(
        ["Detector_Code", "Candidate_Date"]
    )
    .size()
    .reset_index(name="Candidate_Rows")
)

print("\n" + "=" * 120)
print("3. NATURAL MISSINGNESS BY DETECTOR-DAY")
print("=" * 120)

print(
    detector_day_counts["Candidate_Rows"]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .to_string()
)

full_96_blocks = int(
    (detector_day_counts["Candidate_Rows"] == 96).sum()
)

print(
    f"\nDetector-days containing exactly 96 candidate rows: "
    f"{full_96_blocks:,} / {len(detector_day_counts):,} "
    f"({100 * full_96_blocks / len(detector_day_counts):.2f}%)"
)

print("\nMost common candidate-row counts per detector-day:")

print(
    detector_day_counts["Candidate_Rows"]
    .value_counts()
    .head(20)
    .rename_axis("Candidate_Rows")
    .reset_index(name="Detector_Days")
    .to_string(index=False)
)


# ============================================================
# 6. Compare natural training missingness patterns with
#    Notebook 4 controlled test missingness patterns
# ============================================================

natural_pattern_dist = (
    candidates["Missing_Lag_Pattern"]
    .value_counts()
    .rename_axis("Missing_Lag_Pattern")
    .reset_index(name="Natural_Training_Rows")
)

controlled_pattern_dist = (
    controlled["Missing_Lag_Pattern"]
    .value_counts()
    .rename_axis("Missing_Lag_Pattern")
    .reset_index(name="Controlled_Test_Rows")
)

pattern_comparison = pd.merge(
    natural_pattern_dist,
    controlled_pattern_dist,
    on="Missing_Lag_Pattern",
    how="outer"
).fillna(0)

pattern_comparison[
    "Natural_Training_Rows"
] = pattern_comparison[
    "Natural_Training_Rows"
].astype(int)

pattern_comparison[
    "Controlled_Test_Rows"
] = pattern_comparison[
    "Controlled_Test_Rows"
].astype(int)

pattern_comparison["Seen_in_Natural_Training"] = (
    pattern_comparison["Natural_Training_Rows"] > 0
)

pattern_comparison = pattern_comparison.sort_values(
    "Controlled_Test_Rows",
    ascending=False
)

print("\n" + "=" * 120)
print("4. NATURAL TRAINING vs CONTROLLED TEST — MISSING-LAG PATTERNS")
print("=" * 120)

print(
    pattern_comparison
    .to_string(index=False)
)


# ============================================================
# 7. Controlled test coverage by natural-training support
# ============================================================

natural_pattern_set = set(
    candidates["Missing_Lag_Pattern"].unique()
)

controlled["Pattern_Seen_In_Natural_Training"] = (
    controlled["Missing_Lag_Pattern"]
    .isin(natural_pattern_set)
)

covered_rows = int(
    controlled[
        "Pattern_Seen_In_Natural_Training"
    ].sum()
)

coverage_percent = (
    100 * covered_rows / len(controlled)
)

print("\n" + "=" * 120)
print("5. CONTROLLED TEST PATTERN COVERAGE")
print("=" * 120)

print(
    f"Controlled affected rows whose missing-lag pattern "
    f"was seen naturally in training:\n"
    f"{covered_rows:,} / {len(controlled):,} "
    f"({coverage_percent:.2f}%)"
)


# ============================================================
# 8. Compare Missing_Lag_Count distribution
# ============================================================

natural_count = (
    candidates["Missing_Lag_Count"]
    .value_counts()
    .sort_index()
    .rename("Natural_Training_Rows")
)

controlled_count = (
    controlled["Missing_Lag_Count"]
    .value_counts()
    .sort_index()
    .rename("Controlled_Test_Rows")
)

count_comparison = pd.concat(
    [natural_count, controlled_count],
    axis=1
).fillna(0).astype(int)

count_comparison[
    "Natural_Percent"
] = (
    100
    * count_comparison["Natural_Training_Rows"]
    / len(candidates)
)

count_comparison[
    "Controlled_Percent"
] = (
    100
    * count_comparison["Controlled_Test_Rows"]
    / len(controlled)
)

print("\n" + "=" * 120)
print("6. MISSING-LAG COUNT DISTRIBUTION COMPARISON")
print("=" * 120)

print(
    count_comparison
    .reset_index()
    .rename(
        columns={"index": "Missing_Lag_Count"}
    )
    .to_string(index=False)
)


# ============================================================
# 9. Save audit outputs
# ============================================================

source_match_path = os.path.join(
    NB5,
    "Tables",
    "Table_Natural_Missing_Lag_Source_Provenance.csv"
)

pattern_comparison_path = os.path.join(
    NB5,
    "Tables",
    "Table_Natural_vs_Controlled_Missing_Patterns.csv"
)

day_block_path = os.path.join(
    NB5,
    "Tables",
    "Table_Natural_Missingness_Detector_Day_Blocks.csv"
)

source_match_df.to_csv(
    source_match_path,
    index=False
)

pattern_comparison.to_csv(
    pattern_comparison_path,
    index=False
)

detector_day_counts.to_csv(
    day_block_path,
    index=False
)

print("\n" + "=" * 120)
print("7. SAVED AUDIT OUTPUTS")
print("=" * 120)

print(f"✅ {source_match_path}")
print(f"✅ {pattern_comparison_path}")
print(f"✅ {day_block_path}")

print("\n" + "=" * 120)
print("STEP 8 COMPLETED")
print("=" * 120)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 9: Audit feasibility of leakage-free empirical
#         failure-pattern augmentation in Jan–Jun training data
# ============================================================

import os
import pandas as pd
import numpy as np

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(BASE_PATH, "Notebook_3_Output")
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

failure_path = os.path.join(
    NB5,
    "Data",
    "Training_Failure_Pattern_Library_JanJun.csv"
)

failures = pd.read_csv(failure_path)

failures["Date"] = pd.to_datetime(
    failures["Date"]
)

failures["Source_Month"] = (
    failures["Date"]
    .dt.to_period("M")
    .astype(str)
)

failures["Source_DayOfWeek"] = (
    failures["Date"]
    .dt.dayofweek
)

print("=" * 120)
print("NOTEBOOK 5 — TRAINING AUGMENTATION FEASIBILITY AUDIT")
print("=" * 120)

print(f"\nTraining failure patterns: {len(failures):,}")
print(
    f"Unique affected detectors: "
    f"{failures['Detector_Code'].nunique():,}"
)


# ============================================================
# 1. Build clean detector-day inventory from Jan–Jun
# ============================================================

training_months = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06"
]

clean_day_parts = []

needed_cols = [
    "Timestamp",
    "Detector_Code",
    "Target_Volume",
    "DayOfWeek",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

lag_cols = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

print("\n" + "=" * 120)
print("1. BUILDING CLEAN DETECTOR-DAY INVENTORY")
print("=" * 120)

for month in training_months:

    path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    df = pd.read_parquet(
        path,
        columns=needed_cols
    )

    df["Timestamp"] = pd.to_datetime(
        df["Timestamp"]
    )

    df["Date"] = (
        df["Timestamp"]
        .dt.normalize()
    )

    df["Valid_Target"] = (
        df["Target_Volume"].notna()
        & np.isfinite(df["Target_Volume"])
        & (df["Target_Volume"] >= 0)
    )

    df["All_Lags_Valid"] = (
        df[lag_cols]
        .notna()
        .all(axis=1)
    )

    daily = (
        df
        .groupby(
            ["Detector_Code", "Date"],
            as_index=False
        )
        .agg(
            Rows=("Timestamp", "size"),
            Valid_Target_Rows=("Valid_Target", "sum"),
            All_Lags_Valid_Rows=("All_Lags_Valid", "sum"),
            DayOfWeek=("DayOfWeek", "first")
        )
    )

    daily["Month"] = month

    # Strict clean detector-day:
    # all 96 intervals present
    # all targets valid
    # all five lag features valid
    daily["Strict_Clean_Day"] = (
        (daily["Rows"] == 96)
        & (daily["Valid_Target_Rows"] == 96)
        & (daily["All_Lags_Valid_Rows"] == 96)
    )

    clean_day_parts.append(
        daily[
            daily["Strict_Clean_Day"]
        ].copy()
    )

    print(
        f"{month}: "
        f"{daily['Strict_Clean_Day'].sum():,} "
        f"strict clean detector-days"
    )

    del df, daily


clean_days = pd.concat(
    clean_day_parts,
    ignore_index=True
)

print(
    f"\nTotal strict clean detector-days: "
    f"{len(clean_days):,}"
)

print(
    f"Unique detectors with >=1 clean day: "
    f"{clean_days['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Count candidate injection days for every empirical
#    failure pattern
#
# Priority:
#   A. Same detector + same month + same day-of-week
#   B. Same detector + same day-of-week anywhere Jan-Jun
#
# Source failure date itself is excluded.
# ============================================================

print("\n" + "=" * 120)
print("2. MATCHING EMPIRICAL FAILURE PATTERNS TO CLEAN DAYS")
print("=" * 120)

same_month_counts = []
same_dow_counts = []

for row in failures.itertuples(index=False):

    detector_days = clean_days[
        clean_days["Detector_Code"]
        == row.Detector_Code
    ]

    # Exclude original source failure date
    detector_days = detector_days[
        detector_days["Date"]
        != row.Date.normalize()
    ]

    same_month_dow = detector_days[
        (detector_days["Month"] == row.Source_Month)
        & (
            detector_days["DayOfWeek"]
            == row.Source_DayOfWeek
        )
    ]

    same_dow_any_month = detector_days[
        detector_days["DayOfWeek"]
        == row.Source_DayOfWeek
    ]

    same_month_counts.append(
        len(same_month_dow)
    )

    same_dow_counts.append(
        len(same_dow_any_month)
    )


failures[
    "Clean_Candidate_Days_SameMonth_SameDOW"
] = same_month_counts

failures[
    "Clean_Candidate_Days_SameDOW_AnyMonth"
] = same_dow_counts


# ============================================================
# 3. Feasibility summary
# ============================================================

same_month_available = (
    failures[
        "Clean_Candidate_Days_SameMonth_SameDOW"
    ] > 0
)

same_dow_available = (
    failures[
        "Clean_Candidate_Days_SameDOW_AnyMonth"
    ] > 0
)

no_match = ~same_dow_available

print(
    f"\nPatterns with SAME MONTH + SAME DOW clean match:"
    f"\n{same_month_available.sum():,} / {len(failures):,} "
    f"({100 * same_month_available.mean():.2f}%)"
)

print(
    f"\nPatterns with SAME DOW match anywhere Jan–Jun:"
    f"\n{same_dow_available.sum():,} / {len(failures):,} "
    f"({100 * same_dow_available.mean():.2f}%)"
)

print(
    f"\nPatterns with NO suitable clean day:"
    f"\n{no_match.sum():,} / {len(failures):,} "
    f"({100 * no_match.mean():.2f}%)"
)


# ============================================================
# 4. Candidate-count distributions
# ============================================================

print("\n" + "=" * 120)
print("3. CLEAN-DAY MATCH AVAILABILITY DISTRIBUTION")
print("=" * 120)

print("\nSame month + same day-of-week:")
print(
    failures[
        "Clean_Candidate_Days_SameMonth_SameDOW"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .to_string()
)

print("\nSame day-of-week anywhere Jan–Jun:")
print(
    failures[
        "Clean_Candidate_Days_SameDOW_AnyMonth"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .to_string()
)


# ============================================================
# 5. Feasibility by failure duration
# ============================================================

duration_summary = (
    failures
    .groupby(
        "Failure_Duration_Group",
        observed=True
    )
    .agg(
        Failure_Patterns=("Run_Length", "size"),
        SameMonth_SameDOW_Available=(
            "Clean_Candidate_Days_SameMonth_SameDOW",
            lambda x: int((x > 0).sum())
        ),
        SameDOW_AnyMonth_Available=(
            "Clean_Candidate_Days_SameDOW_AnyMonth",
            lambda x: int((x > 0).sum())
        )
    )
    .reset_index()
)

duration_summary[
    "SameMonth_SameDOW_Percent"
] = (
    100
    * duration_summary[
        "SameMonth_SameDOW_Available"
    ]
    / duration_summary[
        "Failure_Patterns"
    ]
)

duration_summary[
    "SameDOW_AnyMonth_Percent"
] = (
    100
    * duration_summary[
        "SameDOW_AnyMonth_Available"
    ]
    / duration_summary[
        "Failure_Patterns"
    ]
)

print("\n" + "=" * 120)
print("4. MATCH FEASIBILITY BY FAILURE DURATION")
print("=" * 120)

print(
    duration_summary
    .to_string(index=False)
)


# ============================================================
# 6. Feasibility by month
# ============================================================

month_summary = (
    failures
    .groupby("Source_Month")
    .agg(
        Failure_Patterns=("Run_Length", "size"),
        SameMonth_SameDOW_Available=(
            "Clean_Candidate_Days_SameMonth_SameDOW",
            lambda x: int((x > 0).sum())
        ),
        SameDOW_AnyMonth_Available=(
            "Clean_Candidate_Days_SameDOW_AnyMonth",
            lambda x: int((x > 0).sum())
        )
    )
    .reset_index()
)

month_summary[
    "SameMonth_SameDOW_Percent"
] = (
    100
    * month_summary[
        "SameMonth_SameDOW_Available"
    ]
    / month_summary[
        "Failure_Patterns"
    ]
)

month_summary[
    "SameDOW_AnyMonth_Percent"
] = (
    100
    * month_summary[
        "SameDOW_AnyMonth_Available"
    ]
    / month_summary[
        "Failure_Patterns"
    ]
)

print("\n" + "=" * 120)
print("5. MATCH FEASIBILITY BY SOURCE MONTH")
print("=" * 120)

print(
    month_summary
    .to_string(index=False)
)


# ============================================================
# 7. Detectors with no feasible mapping
# ============================================================

unmatched = failures[
    no_match
].copy()

print("\n" + "=" * 120)
print("6. UNMATCHED FAILURE PATTERNS")
print("=" * 120)

print(
    f"Unmatched patterns : {len(unmatched):,}"
)

if len(unmatched) > 0:

    print(
        f"Unique unmatched detectors: "
        f"{unmatched['Detector_Code'].nunique():,}"
    )

    print("\nSample unmatched patterns:")

    cols_to_show = [
        "Date",
        "Detector_Code",
        "Start_Interval_Index",
        "Run_Length",
        "Failure_Duration_Group",
        "Source_Month",
        "Source_DayOfWeek"
    ]

    print(
        unmatched[
            cols_to_show
        ]
        .head(25)
        .to_string(index=False)
    )


# ============================================================
# 8. Save outputs
# ============================================================

event_audit_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Match_Feasibility.csv"
)

duration_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Feasibility_by_Duration.csv"
)

month_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Feasibility_by_Month.csv"
)

clean_day_path = os.path.join(
    NB5,
    "Data",
    "Strict_Clean_Training_Detector_Days.parquet"
)

failures.to_csv(
    event_audit_path,
    index=False
)

duration_summary.to_csv(
    duration_path,
    index=False
)

month_summary.to_csv(
    month_path,
    index=False
)

clean_days.to_parquet(
    clean_day_path,
    index=False
)

print("\n" + "=" * 120)
print("7. SAVED OUTPUTS")
print("=" * 120)

print(f"✅ {event_audit_path}")
print(f"✅ {duration_path}")
print(f"✅ {month_path}")
print(f"✅ {clean_day_path}")

print("\n" + "=" * 120)
print("STEP 9 COMPLETED")
print("=" * 120)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 10: Build deterministic empirical augmentation event map
#          using clean 48-hour baseline windows
# ============================================================

import os
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

failure_path = os.path.join(
    NB5,
    "Data",
    "Training_Failure_Pattern_Library_JanJun.csv"
)

clean_day_path = os.path.join(
    NB5,
    "Data",
    "Strict_Clean_Training_Detector_Days.parquet"
)

# ------------------------------------------------------------
# Reproducibility
# ------------------------------------------------------------

RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

print("=" * 125)
print("NOTEBOOK 5 — DETERMINISTIC TRAINING AUGMENTATION EVENT MAPPING")
print("=" * 125)

print(f"\nRandom seed: {RANDOM_SEED}")


# ============================================================
# 1. Load source failure patterns
# ============================================================

failures = pd.read_csv(failure_path)

failures["Date"] = pd.to_datetime(
    failures["Date"]
).dt.normalize()

failures["Source_Month"] = (
    failures["Date"]
    .dt.to_period("M")
    .astype(str)
)

failures["Source_DayOfWeek"] = (
    failures["Date"]
    .dt.dayofweek
)

failures = failures.reset_index(drop=True)

failures["Training_Aug_Event_ID"] = (
    np.arange(1, len(failures) + 1)
)

print("\n" + "=" * 125)
print("1. SOURCE EMPIRICAL FAILURE PATTERNS")
print("=" * 125)

print(f"Failure patterns : {len(failures):,}")

print(
    f"Unique detectors: "
    f"{failures['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Load strict clean detector-day inventory
# ============================================================

clean_days = pd.read_parquet(
    clean_day_path
)

clean_days["Date"] = pd.to_datetime(
    clean_days["Date"]
).dt.normalize()

clean_days["Month"] = (
    clean_days["Date"]
    .dt.to_period("M")
    .astype(str)
)

print("\n" + "=" * 125)
print("2. STRICT CLEAN DETECTOR-DAYS")
print("=" * 125)

print(
    f"Strict clean detector-days: "
    f"{len(clean_days):,}"
)


# ============================================================
# 3. Identify clean 48-hour windows
#
# Requirement:
#   target injection day = strict clean
#   following detector-day = strict clean
#
# This guarantees that:
# - same-day short lag effects begin from clean data
# - next-day Lag_96 effects begin from clean data
# - synthetic missingness is attributable only to injection
# ============================================================

clean_key_set = set(
    zip(
        clean_days["Detector_Code"].astype(int),
        clean_days["Date"]
    )
)

clean_days["Next_Date"] = (
    clean_days["Date"]
    + pd.Timedelta(days=1)
)

clean_days["Next_Day_Strict_Clean"] = [
    (
        int(detector),
        next_date
    ) in clean_key_set
    for detector, next_date in zip(
        clean_days["Detector_Code"],
        clean_days["Next_Date"]
    )
]

clean_48h = clean_days[
    clean_days["Next_Day_Strict_Clean"]
].copy()

# Never allow evaluation window to enter July
clean_48h = clean_48h[
    clean_48h["Next_Date"]
    <= pd.Timestamp("2026-06-30")
].copy()

print("\n" + "=" * 125)
print("3. CLEAN 48-HOUR BASELINE WINDOWS")
print("=" * 125)

print(
    f"Eligible clean 48-hour starting days: "
    f"{len(clean_48h):,}"
)

print(
    f"Unique detectors represented: "
    f"{clean_48h['Detector_Code'].nunique():,}"
)

print("\nEligible windows by month:")

print(
    clean_48h["Month"]
    .value_counts()
    .sort_index()
    .rename_axis("Month")
    .reset_index(name="Clean_48h_Windows")
    .to_string(index=False)
)


# ============================================================
# 4. Deterministic matching
#
# Priority:
#   1) Same detector + same month + same DOW
#   2) Same detector + same DOW anywhere Jan-Jun
#
# Prefer previously unused target detector-days.
# ============================================================

assignments = []

used_target_days = set()

for row in failures.itertuples(index=False):

    detector = int(row.Detector_Code)

    detector_candidates = clean_48h[
        clean_48h["Detector_Code"]
        == detector
    ].copy()

    # Do not map event back to original source date
    detector_candidates = detector_candidates[
        detector_candidates["Date"]
        != row.Date
    ]

    # --------------------------------------------------------
    # Priority 1
    # Same detector + same month + same day-of-week
    # --------------------------------------------------------

    priority_1 = detector_candidates[
        (
            detector_candidates["Month"]
            == row.Source_Month
        )
        &
        (
            detector_candidates["DayOfWeek"]
            == row.Source_DayOfWeek
        )
    ].copy()

    # Prefer unused target detector-day
    priority_1_unused = priority_1[
        ~priority_1.apply(
            lambda x:
            (
                detector,
                x["Date"]
            ) in used_target_days,
            axis=1
        )
    ]

    match_type = None
    reused_target = False

    if len(priority_1_unused) > 0:

        candidate_pool = priority_1_unused
        match_type = "SameMonth_SameDOW"

    else:

        # ----------------------------------------------------
        # Priority 2
        # Same detector + same DOW anywhere Jan-Jun
        # ----------------------------------------------------

        priority_2 = detector_candidates[
            detector_candidates["DayOfWeek"]
            == row.Source_DayOfWeek
        ].copy()

        priority_2_unused = priority_2[
            ~priority_2.apply(
                lambda x:
                (
                    detector,
                    x["Date"]
                ) in used_target_days,
                axis=1
            )
        ]

        if len(priority_2_unused) > 0:

            candidate_pool = priority_2_unused
            match_type = "SameDOW_Fallback"

        elif len(priority_2) > 0:

            # Only if all suitable detector-days have
            # already been used
            candidate_pool = priority_2
            match_type = "SameDOW_Reused"
            reused_target = True

        else:

            candidate_pool = pd.DataFrame()

    # --------------------------------------------------------
    # Select one deterministic random candidate
    # --------------------------------------------------------

    if len(candidate_pool) == 0:

        assignments.append({
            "Training_Aug_Event_ID":
                row.Training_Aug_Event_ID,

            "Detector_Code":
                detector,

            "Source_Failure_Date":
                row.Date,

            "Target_Injection_Date":
                pd.NaT,

            "Match_Type":
                "NO_MATCH",

            "Target_Day_Reused":
                False,

            "Source_DayOfWeek":
                row.Source_DayOfWeek,

            "Target_DayOfWeek":
                np.nan,

            "Source_Month":
                row.Source_Month,

            "Target_Month":
                None,

            "Start_Interval_Index":
                row.Start_Interval_Index,

            "End_Interval_Index":
                row.End_Interval_Index,

            "Run_Length":
                row.Run_Length,

            "Failure_Duration_Minutes":
                row.Failure_Duration_Minutes,

            "Failure_Duration_Hours":
                row.Failure_Duration_Hours,

            "Failure_Duration_Group":
                row.Failure_Duration_Group
        })

        continue

    selected_position = rng.integers(
        0,
        len(candidate_pool)
    )

    selected = candidate_pool.iloc[
        selected_position
    ]

    target_date = pd.Timestamp(
        selected["Date"]
    )

    used_target_days.add(
        (
            detector,
            target_date
        )
    )

    # --------------------------------------------------------
    # Reconstruct synthetic failure timestamps
    # --------------------------------------------------------

    failure_start = (
        target_date
        + pd.to_timedelta(
            int(row.Start_Interval_Index) * 15,
            unit="m"
        )
    )

    failure_end = (
        target_date
        + pd.to_timedelta(
            int(row.End_Interval_Index) * 15,
            unit="m"
        )
    )

    # Maximum dependency = Lag_96 = 24 hours
    evaluation_end = (
        failure_end
        + pd.Timedelta(hours=24)
    )

    assignments.append({
        "Training_Aug_Event_ID":
            row.Training_Aug_Event_ID,

        "Detector_Code":
            detector,

        "Source_Failure_Date":
            row.Date,

        "Target_Injection_Date":
            target_date,

        "Match_Type":
            match_type,

        "Target_Day_Reused":
            reused_target,

        "Source_DayOfWeek":
            row.Source_DayOfWeek,

        "Target_DayOfWeek":
            int(selected["DayOfWeek"]),

        "Source_Month":
            row.Source_Month,

        "Target_Month":
            selected["Month"],

        "Start_Interval_Index":
            int(row.Start_Interval_Index),

        "End_Interval_Index":
            int(row.End_Interval_Index),

        "Run_Length":
            int(row.Run_Length),

        "Failure_Duration_Minutes":
            row.Failure_Duration_Minutes,

        "Failure_Duration_Hours":
            row.Failure_Duration_Hours,

        "Failure_Duration_Group":
            row.Failure_Duration_Group,

        "Failure_Start":
            failure_start,

        "Failure_End":
            failure_end,

        "Evaluation_End":
            evaluation_end
    })


# ============================================================
# 5. Build assignment dataframe
# ============================================================

assignment_df = pd.DataFrame(
    assignments
)

print("\n" + "=" * 125)
print("4. AUGMENTATION EVENT ASSIGNMENT RESULTS")
print("=" * 125)

print(
    f"Events assigned: "
    f"{assignment_df['Target_Injection_Date'].notna().sum():,}"
    f" / {len(assignment_df):,}"
)

print(
    f"Events unmatched: "
    f"{assignment_df['Target_Injection_Date'].isna().sum():,}"
)

print("\nMatch types:")

print(
    assignment_df["Match_Type"]
    .value_counts(dropna=False)
    .rename_axis("Match_Type")
    .reset_index(name="Events")
    .to_string(index=False)
)

print(
    f"\nReused target detector-days: "
    f"{assignment_df['Target_Day_Reused'].sum():,}"
)


# ============================================================
# 6. Validation checks
# ============================================================

assigned = assignment_df[
    assignment_df[
        "Target_Injection_Date"
    ].notna()
].copy()

print("\n" + "=" * 125)
print("5. VALIDATION CHECKS")
print("=" * 125)

same_detector_preserved = (
    len(assigned)
    == len(assignment_df)
)

same_dow_preserved = (
    assigned["Source_DayOfWeek"]
    == assigned["Target_DayOfWeek"]
).all()

evaluation_within_training = (
    assigned["Evaluation_End"]
    <= pd.Timestamp(
        "2026-06-30 23:45:00"
    )
).all()

unique_event_ids = (
    assignment_df[
        "Training_Aug_Event_ID"
    ].is_unique
)

print(
    f"All empirical events mapped        : "
    f"{same_detector_preserved}"
)

print(
    f"Day-of-week preserved              : "
    f"{same_dow_preserved}"
)

print(
    f"All evaluation windows stay Jan-Jun: "
    f"{evaluation_within_training}"
)

print(
    f"Unique augmentation event IDs      : "
    f"{unique_event_ids}"
)


# ============================================================
# 7. Distribution comparison
# ============================================================

print("\n" + "=" * 125)
print("6. SOURCE vs AUGMENTATION DURATION DISTRIBUTION")
print("=" * 125)

duration_check = (
    assignment_df[
        "Failure_Duration_Group"
    ]
    .value_counts()
    .rename_axis(
        "Failure_Duration_Group"
    )
    .reset_index(
        name="Augmentation_Events"
    )
)

print(
    duration_check.to_string(
        index=False
    )
)


# ============================================================
# 8. Preview
# ============================================================

print("\n" + "=" * 125)
print("7. SAMPLE AUGMENTATION EVENTS")
print("=" * 125)

preview_cols = [
    "Training_Aug_Event_ID",
    "Detector_Code",
    "Source_Failure_Date",
    "Target_Injection_Date",
    "Match_Type",
    "Start_Interval_Index",
    "Run_Length",
    "Failure_Start",
    "Failure_End",
    "Evaluation_End"
]

print(
    assignment_df[
        preview_cols
    ]
    .head(20)
    .to_string(index=False)
)


# ============================================================
# 9. Save outputs
# ============================================================

event_map_path = os.path.join(
    NB5,
    "Data",
    "Training_Empirical_Augmentation_Event_Map.csv"
)

summary_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Event_Mapping_Summary.csv"
)

assignment_df.to_csv(
    event_map_path,
    index=False
)

mapping_summary = (
    assignment_df["Match_Type"]
    .value_counts(dropna=False)
    .rename_axis("Match_Type")
    .reset_index(name="Events")
)

mapping_summary["Percent"] = (
    100
    * mapping_summary["Events"]
    / len(assignment_df)
)

mapping_summary.to_csv(
    summary_path,
    index=False
)

print("\n" + "=" * 125)
print("8. SAVED OUTPUTS")
print("=" * 125)

print(f"✅ Event map:\n{event_map_path}")

print(
    f"\n✅ Mapping summary:\n"
    f"{summary_path}"
)

print("\n" + "=" * 125)
print("STEP 10 COMPLETED")
print("=" * 125)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 11: Exact augmentation collision audit
# ============================================================

import os
import pandas as pd
import numpy as np

BASE_PATH = str(OUTPUT_ROOT)
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

event_map_path = os.path.join(
    NB5,
    "Data",
    "Training_Empirical_Augmentation_Event_Map.csv"
)

events = pd.read_csv(event_map_path)

# ------------------------------------------------------------
# Parse datetime columns
# ------------------------------------------------------------

datetime_cols = [
    "Source_Failure_Date",
    "Target_Injection_Date",
    "Failure_Start",
    "Failure_End",
    "Evaluation_End"
]

for c in datetime_cols:
    events[c] = pd.to_datetime(events[c])

lag_offsets = {
    "Lag_1": 15,
    "Lag_2": 30,
    "Lag_4": 60,
    "Lag_8": 120,
    "Lag_96": 1440
}

print("=" * 125)
print("NOTEBOOK 5 — EXACT AUGMENTATION COLLISION AUDIT")
print("=" * 125)

print(f"\nMapped augmentation events: {len(events):,}")


# ============================================================
# 1. Expand every injected failure into exact failed
#    historical timestamps
# ============================================================

affected_records = []

for row in events.itertuples(index=False):

    # Exact failed source timestamps
    failure_times = pd.date_range(
        start=row.Failure_Start,
        periods=int(row.Run_Length),
        freq="15min"
    )

    # --------------------------------------------------------
    # Every failed historical timestamp can affect forecasts
    # through Lag_1, Lag_2, Lag_4, Lag_8 and Lag_96
    # --------------------------------------------------------

    for lag_name, offset_minutes in lag_offsets.items():

        target_times = (
            failure_times
            + pd.Timedelta(
                minutes=offset_minutes
            )
        )

        for source_ts, target_ts in zip(
            failure_times,
            target_times
        ):

            affected_records.append({
                "Training_Aug_Event_ID":
                    int(row.Training_Aug_Event_ID),

                "Detector_Code":
                    int(row.Detector_Code),

                "Source_Failure_Timestamp":
                    source_ts,

                "Affected_Target_Timestamp":
                    target_ts,

                "Affected_Lag":
                    lag_name,

                "Run_Length":
                    int(row.Run_Length),

                "Failure_Duration_Group":
                    row.Failure_Duration_Group,

                "Match_Type":
                    row.Match_Type
            })


expanded = pd.DataFrame(
    affected_records
)

print("\n" + "=" * 125)
print("1. EXPANDED FAILURE-DEPENDENCY RECORDS")
print("=" * 125)

print(
    f"Raw lag-impact records: "
    f"{len(expanded):,}"
)

print(
    f"Unique event IDs: "
    f"{expanded['Training_Aug_Event_ID'].nunique():,}"
)

print(
    f"Unique detectors: "
    f"{expanded['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Collapse multiple missing lags caused by SAME event
#    at the SAME forecasting target
# ============================================================

event_target = (
    expanded
    .groupby(
        [
            "Training_Aug_Event_ID",
            "Detector_Code",
            "Affected_Target_Timestamp"
        ],
        as_index=False
    )
    .agg(
        Missing_Lag_Count=(
            "Affected_Lag",
            "nunique"
        ),

        Missing_Lags=(
            "Affected_Lag",
            lambda x:
            "+".join(
                sorted(
                    set(x),
                    key=lambda z:
                    list(lag_offsets.keys()).index(z)
                )
            )
        ),

        Run_Length=(
            "Run_Length",
            "first"
        ),

        Failure_Duration_Group=(
            "Failure_Duration_Group",
            "first"
        ),

        Match_Type=(
            "Match_Type",
            "first"
        )
    )
)

print("\n" + "=" * 125)
print("2. UNIQUE EVENT-TARGET PAIRS")
print("=" * 125)

print(
    f"Unique event-target rows: "
    f"{len(event_target):,}"
)

print(
    f"Mean affected targets/event: "
    f"{len(event_target) / len(events):.2f}"
)


# ============================================================
# 3. EXACT collision check
#
# Collision =
# same Detector_Code + Affected_Target_Timestamp
# belongs to more than one augmentation event
# ============================================================

target_collision_summary = (
    event_target
    .groupby(
        [
            "Detector_Code",
            "Affected_Target_Timestamp"
        ],
        as_index=False
    )
    .agg(
        Number_of_Events=(
            "Training_Aug_Event_ID",
            "nunique"
        ),

        Event_IDs=(
            "Training_Aug_Event_ID",
            lambda x:
            ",".join(
                map(
                    str,
                    sorted(set(x))
                )
            )
        )
    )
)

collisions = target_collision_summary[
    target_collision_summary[
        "Number_of_Events"
    ] > 1
].copy()

print("\n" + "=" * 125)
print("3. EXACT TARGET COLLISION CHECK")
print("=" * 125)

print(
    f"Unique detector-target timestamps: "
    f"{len(target_collision_summary):,}"
)

print(
    f"Colliding detector-target timestamps: "
    f"{len(collisions):,}"
)

if len(collisions) > 0:

    collision_rows = event_target.merge(
        collisions[
            [
                "Detector_Code",
                "Affected_Target_Timestamp"
            ]
        ],
        on=[
            "Detector_Code",
            "Affected_Target_Timestamp"
        ],
        how="inner"
    )

    affected_collision_events = (
        collision_rows[
            "Training_Aug_Event_ID"
        ]
        .nunique()
    )

    print(
        f"Events involved in collisions: "
        f"{affected_collision_events:,}"
    )

    print("\nSample exact collisions:")

    print(
        collisions
        .head(30)
        .to_string(index=False)
    )

else:

    collision_rows = pd.DataFrame()

    print(
        "✅ No exact detector-target timestamp "
        "collisions detected."
    )


# ============================================================
# 4. Broad evaluation-window overlap audit
#
# This is secondary.
# Exact target collisions above are the critical criterion.
# ============================================================

window_overlaps = []

for detector, group in events.groupby(
    "Detector_Code"
):

    g = group.sort_values(
        "Failure_Start"
    ).reset_index(drop=True)

    for i in range(len(g)):

        for j in range(i + 1, len(g)):

            # Once the next failure begins after the
            # current evaluation window, later events
            # also cannot overlap.
            if (
                g.loc[j, "Failure_Start"]
                > g.loc[i, "Evaluation_End"]
            ):
                break

            overlap_start = max(
                g.loc[i, "Failure_Start"],
                g.loc[j, "Failure_Start"]
            )

            overlap_end = min(
                g.loc[i, "Evaluation_End"],
                g.loc[j, "Evaluation_End"]
            )

            if overlap_start <= overlap_end:

                window_overlaps.append({
                    "Detector_Code":
                        int(detector),

                    "Event_ID_1":
                        int(
                            g.loc[
                                i,
                                "Training_Aug_Event_ID"
                            ]
                        ),

                    "Event_ID_2":
                        int(
                            g.loc[
                                j,
                                "Training_Aug_Event_ID"
                            ]
                        ),

                    "Overlap_Start":
                        overlap_start,

                    "Overlap_End":
                        overlap_end,

                    "Overlap_Hours":
                        (
                            overlap_end
                            - overlap_start
                        ).total_seconds() / 3600
                })


window_overlap_df = pd.DataFrame(
    window_overlaps
)

print("\n" + "=" * 125)
print("4. BROAD EVALUATION-WINDOW OVERLAP")
print("=" * 125)

print(
    f"Overlapping event-window pairs: "
    f"{len(window_overlap_df):,}"
)

if len(window_overlap_df) > 0:

    print(
        f"Events involved: "
        f"{len(set(window_overlap_df['Event_ID_1']) | set(window_overlap_df['Event_ID_2'])):,}"
    )

    print("\nSample broad overlaps:")

    print(
        window_overlap_df
        .head(20)
        .to_string(index=False)
    )

else:

    print("✅ No broad event-window overlap detected.")


# ============================================================
# 5. Missing-lag pattern distribution produced by
#    the augmentation map
# ============================================================

pattern_distribution = (
    event_target[
        "Missing_Lags"
    ]
    .value_counts()
    .rename_axis(
        "Missing_Lag_Pattern"
    )
    .reset_index(
        name="Rows"
    )
)

pattern_distribution[
    "Percent"
] = (
    100
    * pattern_distribution["Rows"]
    / len(event_target)
)

print("\n" + "=" * 125)
print("5. AUGMENTED MISSING-LAG PATTERN DISTRIBUTION")
print("=" * 125)

print(
    pattern_distribution
    .to_string(index=False)
)


# ============================================================
# 6. Missing-lag count distribution
# ============================================================

count_distribution = (
    event_target[
        "Missing_Lag_Count"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "Missing_Lag_Count"
    )
    .reset_index(
        name="Rows"
    )
)

count_distribution[
    "Percent"
] = (
    100
    * count_distribution["Rows"]
    / len(event_target)
)

print("\n" + "=" * 125)
print("6. AUGMENTED MISSING-LAG COUNT DISTRIBUTION")
print("=" * 125)

print(
    count_distribution
    .to_string(index=False)
)


# ============================================================
# 7. Boundary validation
# ============================================================

min_target = (
    event_target[
        "Affected_Target_Timestamp"
    ].min()
)

max_target = (
    event_target[
        "Affected_Target_Timestamp"
    ].max()
)

all_inside_training = (
    (min_target >= pd.Timestamp("2026-01-01"))
    and
    (
        max_target
        <= pd.Timestamp(
            "2026-06-30 23:45:00"
        )
    )
)

print("\n" + "=" * 125)
print("7. TEMPORAL BOUNDARY VALIDATION")
print("=" * 125)

print(
    f"Earliest affected target: "
    f"{min_target}"
)

print(
    f"Latest affected target  : "
    f"{max_target}"
)

print(
    f"All affected targets remain Jan–Jun: "
    f"{all_inside_training}"
)


# ============================================================
# 8. Save outputs
# ============================================================

affected_map_path = os.path.join(
    NB5,
    "Data",
    "Training_Augmentation_Affected_Target_Map.parquet"
)

collision_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Exact_Collisions.csv"
)

window_overlap_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Window_Overlaps.csv"
)

pattern_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Missing_Patterns.csv"
)

count_path = os.path.join(
    NB5,
    "Tables",
    "Table_Training_Augmentation_Missing_Counts.csv"
)

event_target.to_parquet(
    affected_map_path,
    index=False
)

collisions.to_csv(
    collision_path,
    index=False
)

window_overlap_df.to_csv(
    window_overlap_path,
    index=False
)

pattern_distribution.to_csv(
    pattern_path,
    index=False
)

count_distribution.to_csv(
    count_path,
    index=False
)

print("\n" + "=" * 125)
print("8. SAVED OUTPUTS")
print("=" * 125)

print(f"✅ {affected_map_path}")
print(f"✅ {collision_path}")
print(f"✅ {window_overlap_path}")
print(f"✅ {pattern_path}")
print(f"✅ {count_path}")

print("\n" + "=" * 125)
print("STEP 11 COMPLETED")
print("=" * 125)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 12: Build conflict-free empirical augmentation event map
# ============================================================

import os
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

failure_path = os.path.join(
    NB5,
    "Data",
    "Training_Failure_Pattern_Library_JanJun.csv"
)

clean_day_path = os.path.join(
    NB5,
    "Data",
    "Strict_Clean_Training_Detector_Days.parquet"
)

RANDOM_SEED = 42

lag_offsets_minutes = [
    15,    # Lag_1
    30,    # Lag_2
    60,    # Lag_4
    120,   # Lag_8
    1440   # Lag_96
]

print("=" * 130)
print("NOTEBOOK 5 — CONFLICT-FREE TRAINING AUGMENTATION MAPPING")
print("=" * 130)

print(f"\nRandom seed: {RANDOM_SEED}")


# ============================================================
# 1. Load empirical failure patterns
# ============================================================

failures = pd.read_csv(failure_path)

failures["Date"] = pd.to_datetime(
    failures["Date"]
).dt.normalize()

failures["Source_Month"] = (
    failures["Date"]
    .dt.to_period("M")
    .astype(str)
)

failures["Source_DayOfWeek"] = (
    failures["Date"]
    .dt.dayofweek
)

failures = failures.reset_index(drop=True)

failures["Training_Aug_Event_ID"] = np.arange(
    1,
    len(failures) + 1
)

print("\n" + "=" * 130)
print("1. SOURCE FAILURE PATTERNS")
print("=" * 130)

print(f"Failure patterns : {len(failures):,}")
print(
    f"Unique detectors: "
    f"{failures['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Rebuild clean 48-hour candidate windows
# ============================================================

clean_days = pd.read_parquet(
    clean_day_path
)

clean_days["Date"] = pd.to_datetime(
    clean_days["Date"]
).dt.normalize()

clean_days["Month"] = (
    clean_days["Date"]
    .dt.to_period("M")
    .astype(str)
)

clean_key_set = set(
    zip(
        clean_days["Detector_Code"].astype(int),
        clean_days["Date"]
    )
)

clean_days["Next_Date"] = (
    clean_days["Date"]
    + pd.Timedelta(days=1)
)

clean_days["Next_Day_Strict_Clean"] = [
    (
        int(detector),
        next_date
    ) in clean_key_set

    for detector, next_date in zip(
        clean_days["Detector_Code"],
        clean_days["Next_Date"]
    )
]

clean_48h = clean_days[
    clean_days["Next_Day_Strict_Clean"]
].copy()

# Keep all dependency effects inside training period
clean_48h = clean_48h[
    clean_48h["Next_Date"]
    <= pd.Timestamp("2026-06-30")
].copy()

print("\n" + "=" * 130)
print("2. CLEAN 48-HOUR WINDOWS")
print("=" * 130)

print(
    f"Eligible windows: "
    f"{len(clean_48h):,}"
)

print(
    f"Unique detectors: "
    f"{clean_48h['Detector_Code'].nunique():,}"
)


# ============================================================
# 3. Helper: calculate exact forecast-target footprint
# ============================================================

def get_affected_targets(
    target_date,
    start_interval,
    run_length
):
    """
    Returns exact forecasting timestamps affected when the
    historical observations in this empirical failure run
    become unavailable.
    """

    failure_times = (
        target_date
        + pd.to_timedelta(
            (
                int(start_interval)
                + np.arange(int(run_length))
            ) * 15,
            unit="m"
        )
    )

    affected = set()

    for offset in lag_offsets_minutes:

        shifted = (
            failure_times
            + pd.Timedelta(minutes=offset)
        )

        affected.update(
            pd.Timestamp(x)
            for x in shifted
        )

    return affected


# ============================================================
# 4. Precompute candidate clean dates for every event
# ============================================================

event_candidates = {}

for row in failures.itertuples(index=False):

    detector = int(row.Detector_Code)

    pool = clean_48h[
        clean_48h["Detector_Code"]
        == detector
    ].copy()

    # Exclude original real-failure date
    pool = pool[
        pool["Date"] != row.Date
    ]

    # Same DOW is mandatory
    pool = pool[
        pool["DayOfWeek"]
        == row.Source_DayOfWeek
    ].copy()

    same_month = (
        pool[
            pool["Month"]
            == row.Source_Month
        ]["Date"]
        .tolist()
    )

    fallback = (
        pool[
            pool["Month"]
            != row.Source_Month
        ]["Date"]
        .tolist()
    )

    event_candidates[
        int(row.Training_Aug_Event_ID)
    ] = {
        "priority_1": same_month,
        "priority_2": fallback,
        "total":
            len(same_month)
            + len(fallback)
    }


# ============================================================
# 5. Conflict-aware assignment function
# ============================================================

def assign_detector_events(
    detector_events,
    detector_code,
    attempt_seed
):

    rng = np.random.default_rng(
        attempt_seed
    )

    # Hardest events first:
    # fewer candidate days first,
    # then longer failure runs
    working = detector_events.copy()

    working["Total_Candidates"] = (
        working["Training_Aug_Event_ID"]
        .map(
            lambda x:
            event_candidates[int(x)]["total"]
        )
    )

    # Random tie-breaker, deterministic by attempt seed
    working["_Tie"] = rng.random(
        len(working)
    )

    working = working.sort_values(
        by=[
            "Total_Candidates",
            "Run_Length",
            "_Tie"
        ],
        ascending=[
            True,
            False,
            True
        ]
    )

    occupied_days = set()
    occupied_targets = set()

    assignments = []

    for row in working.itertuples(index=False):

        event_id = int(
            row.Training_Aug_Event_ID
        )

        candidate_info = (
            event_candidates[event_id]
        )

        p1 = list(
            candidate_info["priority_1"]
        )

        p2 = list(
            candidate_info["priority_2"]
        )

        # Deterministic random order inside each priority
        if len(p1) > 1:
            p1 = list(
                rng.permutation(p1)
            )

        if len(p2) > 1:
            p2 = list(
                rng.permutation(p2)
            )

        selected_date = None
        selected_type = None
        selected_footprint = None

        # --------------------------------------------
        # Priority 1:
        # same month + same day-of-week
        # --------------------------------------------

        for candidate_date in p1:

            candidate_date = pd.Timestamp(
                candidate_date
            )

            if candidate_date in occupied_days:
                continue

            footprint = get_affected_targets(
                candidate_date,
                row.Start_Interval_Index,
                row.Run_Length
            )

            if not footprint.isdisjoint(
                occupied_targets
            ):
                continue

            selected_date = candidate_date
            selected_type = (
                "SameMonth_SameDOW"
            )
            selected_footprint = footprint
            break

        # --------------------------------------------
        # Priority 2:
        # same DOW, different month
        # --------------------------------------------

        if selected_date is None:

            for candidate_date in p2:

                candidate_date = pd.Timestamp(
                    candidate_date
                )

                if candidate_date in occupied_days:
                    continue

                footprint = get_affected_targets(
                    candidate_date,
                    row.Start_Interval_Index,
                    row.Run_Length
                )

                if not footprint.isdisjoint(
                    occupied_targets
                ):
                    continue

                selected_date = candidate_date
                selected_type = (
                    "SameDOW_Fallback"
                )
                selected_footprint = footprint
                break

        # --------------------------------------------
        # Failed assignment
        # --------------------------------------------

        if selected_date is None:

            return None

        occupied_days.add(
            selected_date
        )

        occupied_targets.update(
            selected_footprint
        )

        failure_start = (
            selected_date
            + pd.to_timedelta(
                int(
                    row.Start_Interval_Index
                ) * 15,
                unit="m"
            )
        )

        failure_end = (
            failure_start
            + pd.to_timedelta(
                (
                    int(row.Run_Length) - 1
                ) * 15,
                unit="m"
            )
        )

        assignments.append({
            "Training_Aug_Event_ID":
                event_id,

            "Detector_Code":
                int(detector_code),

            "Source_Failure_Date":
                row.Date,

            "Target_Injection_Date":
                selected_date,

            "Match_Type":
                selected_type,

            "Source_DayOfWeek":
                int(
                    row.Source_DayOfWeek
                ),

            "Target_DayOfWeek":
                int(
                    selected_date.dayofweek
                ),

            "Source_Month":
                row.Source_Month,

            "Target_Month":
                selected_date
                .to_period("M")
                .strftime("%Y-%m"),

            "Start_Interval_Index":
                int(
                    row.Start_Interval_Index
                ),

            "End_Interval_Index":
                int(
                    row.End_Interval_Index
                ),

            "Run_Length":
                int(row.Run_Length),

            "Failure_Duration_Minutes":
                row.Failure_Duration_Minutes,

            "Failure_Duration_Hours":
                row.Failure_Duration_Hours,

            "Failure_Duration_Group":
                row.Failure_Duration_Group,

            "Failure_Start":
                failure_start,

            "Failure_End":
                failure_end,

            "Evaluation_End":
                failure_end
                + pd.Timedelta(hours=24),

            "Affected_Target_Count":
                len(
                    selected_footprint
                )
        })

    return assignments


# ============================================================
# 6. Assign independently within every detector
#
# Multiple deterministic attempts are allowed ONLY to find
# a collision-free mapping. Model results are not involved.
# ============================================================

all_assignments = []

detector_failures = []

max_attempts_per_detector = 250

for detector, detector_events in failures.groupby(
    "Detector_Code"
):

    detector = int(detector)

    successful = None
    attempts_used = None

    for attempt in range(
        max_attempts_per_detector
    ):

        attempt_seed = (
            RANDOM_SEED
            + detector * 1000
            + attempt
        )

        result = assign_detector_events(
            detector_events,
            detector,
            attempt_seed
        )

        if result is not None:

            successful = result
            attempts_used = attempt + 1
            break

    if successful is None:

        detector_failures.append({
            "Detector_Code":
                detector,

            "Events":
                len(detector_events)
        })

    else:

        for item in successful:
            item["Assignment_Attempts"] = (
                attempts_used
            )

        all_assignments.extend(
            successful
        )


assignment_df = pd.DataFrame(
    all_assignments
)

assignment_df = assignment_df.sort_values(
    "Training_Aug_Event_ID"
).reset_index(drop=True)


# ============================================================
# 7. Overall assignment results
# ============================================================

print("\n" + "=" * 130)
print("3. CONFLICT-AWARE ASSIGNMENT RESULTS")
print("=" * 130)

print(
    f"Events successfully assigned: "
    f"{len(assignment_df):,} / {len(failures):,}"
)

print(
    f"Unresolved detectors: "
    f"{len(detector_failures):,}"
)

if len(detector_failures) > 0:

    detector_failures_df = pd.DataFrame(
        detector_failures
    )

    print("\nUnresolved detectors:")
    print(
        detector_failures_df
        .to_string(index=False)
    )

else:

    detector_failures_df = pd.DataFrame(
        columns=[
            "Detector_Code",
            "Events"
        ]
    )

    print(
        "✅ All detectors successfully mapped."
    )


# ============================================================
# 8. Exact independent collision verification
# ============================================================

verification_records = []

for row in assignment_df.itertuples(
    index=False
):

    footprint = get_affected_targets(
        pd.Timestamp(
            row.Target_Injection_Date
        ),
        row.Start_Interval_Index,
        row.Run_Length
    )

    for ts in footprint:

        verification_records.append({
            "Training_Aug_Event_ID":
                row.Training_Aug_Event_ID,

            "Detector_Code":
                row.Detector_Code,

            "Affected_Target_Timestamp":
                ts
        })


verification_df = pd.DataFrame(
    verification_records
)

collision_check = (
    verification_df
    .groupby(
        [
            "Detector_Code",
            "Affected_Target_Timestamp"
        ]
    )[
        "Training_Aug_Event_ID"
    ]
    .nunique()
)

exact_collisions = (
    collision_check > 1
).sum()

print("\n" + "=" * 130)
print("4. EXACT COLLISION VERIFICATION")
print("=" * 130)

print(
    f"Unique affected detector-target timestamps: "
    f"{len(collision_check):,}"
)

print(
    f"Exact collisions: "
    f"{exact_collisions:,}"
)

if exact_collisions == 0:
    print(
        "✅ Conflict-free augmentation map confirmed."
    )
else:
    print(
        "❌ Unexpected collisions remain."
    )


# ============================================================
# 9. Design validation
# ============================================================

print("\n" + "=" * 130)
print("5. DESIGN VALIDATION")
print("=" * 130)

all_events_mapped = (
    len(assignment_df)
    == len(failures)
)

same_dow = (
    assignment_df[
        "Source_DayOfWeek"
    ]
    ==
    assignment_df[
        "Target_DayOfWeek"
    ]
).all()

inside_training = (
    assignment_df[
        "Evaluation_End"
    ].max()
    <= pd.Timestamp(
        "2026-06-30 23:45:00"
    )
)

duplicate_target_days = (
    assignment_df
    .duplicated(
        subset=[
            "Detector_Code",
            "Target_Injection_Date"
        ]
    )
    .sum()
)

print(
    f"All 1,898 events mapped       : "
    f"{all_events_mapped}"
)

print(
    f"Day-of-week preserved         : "
    f"{same_dow}"
)

print(
    f"All dependency effects Jan-Jun: "
    f"{inside_training}"
)

print(
    f"Reused detector-target days   : "
    f"{duplicate_target_days:,}"
)

print(
    f"Exact target collisions       : "
    f"{exact_collisions:,}"
)


# ============================================================
# 10. Match-type summary
# ============================================================

match_summary = (
    assignment_df[
        "Match_Type"
    ]
    .value_counts()
    .rename_axis(
        "Match_Type"
    )
    .reset_index(
        name="Events"
    )
)

match_summary[
    "Percent"
] = (
    100
    * match_summary["Events"]
    / len(assignment_df)
)

print("\n" + "=" * 130)
print("6. FINAL MATCH TYPES")
print("=" * 130)

print(
    match_summary
    .to_string(index=False)
)


# ============================================================
# 11. Attempts required
# ============================================================

attempt_summary = (
    assignment_df[
        [
            "Detector_Code",
            "Assignment_Attempts"
        ]
    ]
    .drop_duplicates()
)

print("\n" + "=" * 130)
print("7. ASSIGNMENT SEARCH EFFORT")
print("=" * 130)

print(
    attempt_summary[
        "Assignment_Attempts"
    ]
    .describe()
    .to_string()
)


# ============================================================
# 12. Sample final mapping
# ============================================================

print("\n" + "=" * 130)
print("8. SAMPLE CONFLICT-FREE EVENTS")
print("=" * 130)

preview_cols = [
    "Training_Aug_Event_ID",
    "Detector_Code",
    "Source_Failure_Date",
    "Target_Injection_Date",
    "Match_Type",
    "Start_Interval_Index",
    "Run_Length",
    "Failure_Start",
    "Failure_End",
    "Evaluation_End",
    "Affected_Target_Count"
]

print(
    assignment_df[
        preview_cols
    ]
    .head(20)
    .to_string(index=False)
)


# ============================================================
# 13. Save FINAL conflict-free map
# ============================================================

final_map_path = os.path.join(
    NB5,
    "Data",
    "Training_Empirical_Augmentation_Event_Map_ConflictFree.csv"
)

summary_path = os.path.join(
    NB5,
    "Tables",
    "Table_ConflictFree_Augmentation_Mapping_Summary.csv"
)

verification_path = os.path.join(
    NB5,
    "Data",
    "Training_ConflictFree_Affected_Target_Verification.parquet"
)

assignment_df.to_csv(
    final_map_path,
    index=False
)

match_summary.to_csv(
    summary_path,
    index=False
)

verification_df.to_parquet(
    verification_path,
    index=False
)

print("\n" + "=" * 130)
print("9. SAVED FINAL OUTPUTS")
print("=" * 130)

print(
    f"✅ Conflict-free event map:\n"
    f"{final_map_path}"
)

print(
    f"\n✅ Mapping summary:\n"
    f"{summary_path}"
)

print(
    f"\n✅ Independent target verification:\n"
    f"{verification_path}"
)

print("\n" + "=" * 130)
print("STEP 12 COMPLETED")
print("=" * 130)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 13: Materialize empirical failure-augmented training rows
# ============================================================

import os
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB4 = os.path.join(
    BASE_PATH,
    "Notebook_4_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

event_map_path = os.path.join(
    NB5,
    "Data",
    "Training_Empirical_Augmentation_Event_Map_ConflictFree.csv"
)

controlled_test_path = os.path.join(
    NB4,
    "Data",
    "Controlled_Failure_Injected_Affected_Predictions.parquet"
)

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

lag_offsets = {
    "Lag_1": 15,
    "Lag_2": 30,
    "Lag_4": 60,
    "Lag_8": 120,
    "Lag_96": 1440
}

lag_cols = list(lag_offsets.keys())

model_features = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

training_months = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06"
]

print("=" * 130)
print("NOTEBOOK 5 — MATERIALIZING EMPIRICAL FAILURE-AUGMENTED TRAINING ROWS")
print("=" * 130)


# ============================================================
# 1. Load conflict-free augmentation event map
# ============================================================

events = pd.read_csv(event_map_path)

datetime_cols = [
    "Source_Failure_Date",
    "Target_Injection_Date",
    "Failure_Start",
    "Failure_End",
    "Evaluation_End"
]

for c in datetime_cols:
    events[c] = pd.to_datetime(events[c])

print("\n" + "=" * 130)
print("1. CONFLICT-FREE AUGMENTATION EVENTS")
print("=" * 130)

print(f"Events: {len(events):,}")
print(
    f"Unique detectors: "
    f"{events['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Reconstruct exact affected forecasting targets
#    and lag features to remove
# ============================================================

impact_records = []

for row in events.itertuples(index=False):

    failure_times = pd.date_range(
        start=row.Failure_Start,
        periods=int(row.Run_Length),
        freq="15min"
    )

    for lag_name, offset_minutes in lag_offsets.items():

        target_times = (
            failure_times
            + pd.Timedelta(
                minutes=offset_minutes
            )
        )

        for target_ts in target_times:

            impact_records.append({
                "Training_Aug_Event_ID":
                    int(row.Training_Aug_Event_ID),

                "Detector_Code":
                    int(row.Detector_Code),

                "Timestamp":
                    pd.Timestamp(target_ts),

                "Missing_Lag":
                    lag_name,

                "Failure_Duration_Group":
                    row.Failure_Duration_Group,

                "Failure_Run_Length":
                    int(row.Run_Length),

                "Match_Type":
                    row.Match_Type
            })


impact_long = pd.DataFrame(
    impact_records
).drop_duplicates(
    subset=[
        "Training_Aug_Event_ID",
        "Detector_Code",
        "Timestamp",
        "Missing_Lag"
    ]
)

# ------------------------------------------------------------
# Collapse to one row per affected forecast target
# ------------------------------------------------------------

base_info = (
    impact_long
    .groupby(
        [
            "Training_Aug_Event_ID",
            "Detector_Code",
            "Timestamp"
        ],
        as_index=False
    )
    .agg(
        Failure_Run_Length=(
            "Failure_Run_Length",
            "first"
        ),

        Failure_Duration_Group=(
            "Failure_Duration_Group",
            "first"
        ),

        Match_Type=(
            "Match_Type",
            "first"
        )
    )
)

for lag in lag_cols:

    flag = (
        impact_long[
            impact_long["Missing_Lag"] == lag
        ][
            [
                "Training_Aug_Event_ID",
                "Detector_Code",
                "Timestamp"
            ]
        ]
        .drop_duplicates()
        .copy()
    )

    flag[f"Missing_{lag}"] = 1

    base_info = base_info.merge(
        flag,
        on=[
            "Training_Aug_Event_ID",
            "Detector_Code",
            "Timestamp"
        ],
        how="left"
    )

for lag in lag_cols:
    base_info[f"Missing_{lag}"] = (
        base_info[f"Missing_{lag}"]
        .fillna(0)
        .astype("int8")
    )

base_info["Missing_Lag_Count"] = (
    base_info[
        [f"Missing_{x}" for x in lag_cols]
    ]
    .sum(axis=1)
    .astype("int8")
)


def build_pattern(row):

    missing = [
        lag
        for lag in lag_cols
        if row[f"Missing_{lag}"] == 1
    ]

    return "+".join(missing)


base_info["Missing_Lag_Pattern"] = (
    base_info.apply(
        build_pattern,
        axis=1
    )
)

base_info["Target_Month"] = (
    base_info["Timestamp"]
    .dt.to_period("M")
    .astype(str)
)

print("\n" + "=" * 130)
print("2. EXACT AUGMENTATION IMPACT MAP")
print("=" * 130)

print(
    f"Unique affected training targets: "
    f"{len(base_info):,}"
)

print(
    f"Unique event IDs represented: "
    f"{base_info['Training_Aug_Event_ID'].nunique():,}"
)

print(
    f"Duplicate detector-target timestamps: "
    f"{base_info.duplicated(['Detector_Code', 'Timestamp']).sum():,}"
)


# ============================================================
# 3. Extract clean baseline forecasting rows month-by-month
# ============================================================

columns_to_read = [
    "Timestamp",
    "Detector_Code",
    "Interval_Index",
    "Target_Volume",
    "Hour",
    "Minute",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96",
    "All_Lags_Available",
    "Split"
]

augmented_parts = []

month_audit = []

print("\n" + "=" * 130)
print("3. EXTRACTING CLEAN BASELINE ROWS")
print("=" * 130)

for month in training_months:

    impact_month = base_info[
        base_info["Target_Month"]
        == month
    ].copy()

    expected_rows = len(
        impact_month
    )

    if expected_rows == 0:
        continue

    file_path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    df = pd.read_parquet(
        file_path,
        columns=columns_to_read
    )

    df["Timestamp"] = pd.to_datetime(
        df["Timestamp"]
    )

    extracted = df.merge(
        impact_month.drop(
            columns=["Target_Month"]
        ),
        on=[
            "Detector_Code",
            "Timestamp"
        ],
        how="inner",
        validate="one_to_one"
    )

    found_rows = len(
        extracted
    )

    original_missing_lags = int(
        extracted[
            lag_cols
        ]
        .isna()
        .any(axis=1)
        .sum()
    )

    invalid_targets = int(
        (
            extracted[
                "Target_Volume"
            ].isna()
            |
            ~np.isfinite(
                extracted[
                    "Target_Volume"
                ]
            )
            |
            (
                extracted[
                    "Target_Volume"
                ] < 0
            )
        ).sum()
    )

    print(
        f"{month}: expected={expected_rows:,} | "
        f"found={found_rows:,} | "
        f"baseline_missing_lags={original_missing_lags:,} | "
        f"invalid_targets={invalid_targets:,}"
    )

    month_audit.append({
        "Month": month,
        "Expected_Affected_Rows":
            expected_rows,

        "Extracted_Rows":
            found_rows,

        "Baseline_Rows_With_Missing_Lag":
            original_missing_lags,

        "Invalid_Target_Rows":
            invalid_targets
    })

    augmented_parts.append(
        extracted
    )

    del df


augmented = pd.concat(
    augmented_parts,
    ignore_index=True
)

month_audit_df = pd.DataFrame(
    month_audit
)


# ============================================================
# 4. Critical baseline validation BEFORE injection
# ============================================================

print("\n" + "=" * 130)
print("4. PRE-INJECTION BASELINE VALIDATION")
print("=" * 130)

expected_total = len(base_info)
actual_total = len(augmented)

baseline_missing = int(
    augmented[
        lag_cols
    ]
    .isna()
    .any(axis=1)
    .sum()
)

baseline_all_lags_available = int(
    augmented[
        "All_Lags_Available"
    ].fillna(False).sum()
)

invalid_targets = int(
    (
        augmented[
            "Target_Volume"
        ].isna()
        |
        ~np.isfinite(
            augmented[
                "Target_Volume"
            ]
        )
        |
        (
            augmented[
                "Target_Volume"
            ] < 0
        )
    ).sum()
)

duplicate_rows = int(
    augmented
    .duplicated(
        subset=[
            "Detector_Code",
            "Timestamp"
        ]
    )
    .sum()
)

print(
    f"Expected affected targets        : "
    f"{expected_total:,}"
)

print(
    f"Extracted baseline targets       : "
    f"{actual_total:,}"
)

print(
    f"Baseline rows with missing lag   : "
    f"{baseline_missing:,}"
)

print(
    f"Rows marked All_Lags_Available   : "
    f"{baseline_all_lags_available:,}"
)

print(
    f"Invalid targets                  : "
    f"{invalid_targets:,}"
)

print(
    f"Duplicate detector-target rows   : "
    f"{duplicate_rows:,}"
)


# ============================================================
# 5. Preserve CLEAN lag values for audit
# ============================================================

for lag in lag_cols:

    augmented[
        f"{lag}_Clean"
    ] = augmented[
        lag
    ].copy()


# ============================================================
# 6. Inject missingness into exact historical lag features
# ============================================================

for lag in lag_cols:

    mask = (
        augmented[
            f"Missing_{lag}"
        ] == 1
    )

    augmented.loc[
        mask,
        lag
    ] = np.nan


augmented["All_Lags_Available"] = (
    ~augmented[
        lag_cols
    ]
    .isna()
    .any(axis=1)
)

augmented[
    "Training_Source"
] = "Empirical_Failure_Augmentation"


# ============================================================
# 7. Validate injection correctness
# ============================================================

print("\n" + "=" * 130)
print("5. POST-INJECTION VALIDATION")
print("=" * 130)

validation_rows = []

for lag in lag_cols:

    intended = int(
        augmented[
            f"Missing_{lag}"
        ].sum()
    )

    actual_missing = int(
        augmented[
            lag
        ].isna().sum()
    )

    unexpected_missing = int(
        (
            augmented[
                lag
            ].isna()
            &
            (
                augmented[
                    f"Missing_{lag}"
                ] == 0
            )
        ).sum()
    )

    validation_rows.append({
        "Lag":
            lag,

        "Intended_Missing":
            intended,

        "Actual_Missing":
            actual_missing,

        "Unexpected_Missing":
            unexpected_missing,

        "Exact_Match":
            intended == actual_missing
            and unexpected_missing == 0
    })


injection_validation = pd.DataFrame(
    validation_rows
)

print(
    injection_validation
    .to_string(index=False)
)

all_injections_exact = bool(
    injection_validation[
        "Exact_Match"
    ].all()
)

print(
    f"\nAll lag injections exact: "
    f"{all_injections_exact}"
)


# ============================================================
# 8. Confirm targets were NEVER modified
# ============================================================

target_min = (
    augmented[
        "Target_Volume"
    ].min()
)

target_max = (
    augmented[
        "Target_Volume"
    ].max()
)

target_missing = int(
    augmented[
        "Target_Volume"
    ].isna().sum()
)

print("\n" + "=" * 130)
print("6. TARGET INTEGRITY")
print("=" * 130)

print(
    f"Target minimum       : "
    f"{target_min:.3f}"
)

print(
    f"Target maximum       : "
    f"{target_max:.3f}"
)

print(
    f"Missing targets      : "
    f"{target_missing:,}"
)

print(
    "✅ Target_Volume was preserved; "
    "only historical lag inputs were altered."
)


# ============================================================
# 9. Final augmented missing-pattern distribution
# ============================================================

aug_pattern = (
    augmented[
        "Missing_Lag_Pattern"
    ]
    .value_counts()
    .rename_axis(
        "Missing_Lag_Pattern"
    )
    .reset_index(
        name="Augmented_Training_Rows"
    )
)

aug_pattern[
    "Augmented_Training_Percent"
] = (
    100
    * aug_pattern[
        "Augmented_Training_Rows"
    ]
    / len(augmented)
)

print("\n" + "=" * 130)
print("7. FINAL AUGMENTED TRAINING MISSING-LAG PATTERNS")
print("=" * 130)

print(
    aug_pattern
    .to_string(index=False)
)


# ============================================================
# 10. Compare with controlled August test missingness
# ============================================================

controlled = pd.read_parquet(
    controlled_test_path,
    columns=[
        "Missing_Lag_Count",
        "Missing_Lag_Pattern"
    ]
)

test_pattern = (
    controlled[
        "Missing_Lag_Pattern"
    ]
    .value_counts()
    .rename_axis(
        "Missing_Lag_Pattern"
    )
    .reset_index(
        name="Controlled_August_Rows"
    )
)

test_pattern[
    "Controlled_August_Percent"
] = (
    100
    * test_pattern[
        "Controlled_August_Rows"
    ]
    / len(controlled)
)

pattern_comparison = pd.merge(
    aug_pattern,
    test_pattern,
    on="Missing_Lag_Pattern",
    how="outer"
).fillna(0)

pattern_comparison[
    "Seen_In_Augmented_Training"
] = (
    pattern_comparison[
        "Augmented_Training_Rows"
    ] > 0
)

pattern_comparison = (
    pattern_comparison
    .sort_values(
        "Controlled_August_Rows",
        ascending=False
    )
)

covered_test_rows = int(
    pattern_comparison.loc[
        pattern_comparison[
            "Seen_In_Augmented_Training"
        ],
        "Controlled_August_Rows"
    ].sum()
)

coverage_percent = (
    100
    * covered_test_rows
    / len(controlled)
)

print("\n" + "=" * 130)
print("8. AUGMENTED TRAINING vs CONTROLLED AUGUST TEST")
print("=" * 130)

print(
    pattern_comparison
    .to_string(index=False)
)

print(
    f"\nControlled August rows whose missing pattern "
    f"is represented in empirical augmentation:"
)

print(
    f"{covered_test_rows:,} / {len(controlled):,} "
    f"({coverage_percent:.2f}%)"
)


# ============================================================
# 11. Missing-lag COUNT comparison
# ============================================================

training_count_dist = (
    augmented[
        "Missing_Lag_Count"
    ]
    .value_counts()
    .sort_index()
    .rename(
        "Augmented_Training_Rows"
    )
)

test_count_dist = (
    controlled[
        "Missing_Lag_Count"
    ]
    .value_counts()
    .sort_index()
    .rename(
        "Controlled_August_Rows"
    )
)

count_comparison = pd.concat(
    [
        training_count_dist,
        test_count_dist
    ],
    axis=1
).fillna(0).astype(int)

count_comparison[
    "Augmented_Training_Percent"
] = (
    100
    * count_comparison[
        "Augmented_Training_Rows"
    ]
    / len(augmented)
)

count_comparison[
    "Controlled_August_Percent"
] = (
    100
    * count_comparison[
        "Controlled_August_Rows"
    ]
    / len(controlled)
)

print("\n" + "=" * 130)
print("9. MISSING-LAG COUNT COMPARISON")
print("=" * 130)

print(
    count_comparison
    .reset_index()
    .rename(
        columns={
            "index":
            "Missing_Lag_Count"
        }
    )
    .to_string(index=False)
)


# ============================================================
# 12. Overall integrity decision
# ============================================================

integrity_pass = (
    expected_total == actual_total
    and baseline_missing == 0
    and invalid_targets == 0
    and duplicate_rows == 0
    and all_injections_exact
    and coverage_percent == 100.0
)

print("\n" + "=" * 130)
print("10. STEP 13 INTEGRITY DECISION")
print("=" * 130)

print(
    f"All expected baseline rows recovered : "
    f"{expected_total == actual_total}"
)

print(
    f"All baselines originally clean       : "
    f"{baseline_missing == 0}"
)

print(
    f"All targets valid                    : "
    f"{invalid_targets == 0}"
)

print(
    f"No duplicate target rows             : "
    f"{duplicate_rows == 0}"
)

print(
    f"All injections exact                 : "
    f"{all_injections_exact}"
)

print(
    f"Controlled-test pattern coverage 100%: "
    f"{coverage_percent == 100.0}"
)

print(
    f"\nFINAL STEP 13 INTEGRITY PASS: "
    f"{integrity_pass}"
)


# ============================================================
# 13. Save outputs
# ============================================================

augmented_path = os.path.join(
    NB5,
    "Data",
    "Empirical_Failure_Augmented_Training_Rows.parquet"
)

pattern_path = os.path.join(
    NB5,
    "Tables",
    "Table_Empirical_Augmented_vs_August_Missing_Patterns.csv"
)

count_path = os.path.join(
    NB5,
    "Tables",
    "Table_Empirical_Augmented_vs_August_Missing_Counts.csv"
)

validation_path = os.path.join(
    NB5,
    "Tables",
    "Table_Empirical_Augmentation_Injection_Validation.csv"
)

monthly_path = os.path.join(
    NB5,
    "Tables",
    "Table_Empirical_Augmentation_Monthly_Extraction_Audit.csv"
)

augmented.to_parquet(
    augmented_path,
    index=False
)

pattern_comparison.to_csv(
    pattern_path,
    index=False
)

count_comparison.reset_index().to_csv(
    count_path,
    index=False
)

injection_validation.to_csv(
    validation_path,
    index=False
)

month_audit_df.to_csv(
    monthly_path,
    index=False
)

print("\n" + "=" * 130)
print("11. SAVED OUTPUTS")
print("=" * 130)

print(f"✅ {augmented_path}")
print(f"✅ {pattern_path}")
print(f"✅ {count_path}")
print(f"✅ {validation_path}")
print(f"✅ {monthly_path}")

print("\n" + "=" * 130)
print("STEP 13 COMPLETED")
print("=" * 130)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 14: Audit feasibility of an independent controlled
#          failure-validation experiment in July 2026
# ============================================================

import os
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

july_path = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly",
    "FORECASTING_2026-07.parquet"
)

failure_path = os.path.join(
    NB5,
    "Data",
    "Training_Failure_Pattern_Library_JanJun.csv"
)

print("=" * 130)
print("NOTEBOOK 5 — CONTROLLED JULY VALIDATION FEASIBILITY AUDIT")
print("=" * 130)


# ============================================================
# 1. Load Jan-Jun empirical failure-pattern library
# ============================================================

patterns = pd.read_csv(
    failure_path
)

patterns["Date"] = pd.to_datetime(
    patterns["Date"]
).dt.normalize()

patterns["Source_DayOfWeek"] = (
    patterns["Date"]
    .dt.dayofweek
)

print("\n" + "=" * 130)
print("1. SOURCE FAILURE PATTERNS — JAN–JUN ONLY")
print("=" * 130)

print(
    f"Failure patterns : "
    f"{len(patterns):,}"
)

print(
    f"Unique detectors: "
    f"{patterns['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Load July forecasting data
# ============================================================

columns = [
    "Timestamp",
    "Detector_Code",
    "Target_Volume",
    "DayOfWeek",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

lag_cols = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

july = pd.read_parquet(
    july_path,
    columns=columns
)

july["Timestamp"] = pd.to_datetime(
    july["Timestamp"]
)

july["Date"] = (
    july["Timestamp"]
    .dt.normalize()
)

july["Valid_Target"] = (
    july["Target_Volume"].notna()
    &
    np.isfinite(
        july["Target_Volume"]
    )
    &
    (
        july["Target_Volume"] >= 0
    )
)

july["All_Lags_Valid"] = (
    july[
        lag_cols
    ]
    .notna()
    .all(axis=1)
)

print("\n" + "=" * 130)
print("2. JULY DATA")
print("=" * 130)

print(
    f"Rows             : "
    f"{len(july):,}"
)

print(
    f"Unique detectors : "
    f"{july['Detector_Code'].nunique():,}"
)

print(
    f"Dates            : "
    f"{july['Date'].nunique():,}"
)

print(
    f"Coverage         : "
    f"{july['Date'].min().date()} → "
    f"{july['Date'].max().date()}"
)


# ============================================================
# 3. Build strict clean July detector-day inventory
# ============================================================

daily = (
    july
    .groupby(
        [
            "Detector_Code",
            "Date"
        ],
        as_index=False
    )
    .agg(
        Rows=(
            "Timestamp",
            "size"
        ),

        Valid_Target_Rows=(
            "Valid_Target",
            "sum"
        ),

        All_Lags_Valid_Rows=(
            "All_Lags_Valid",
            "sum"
        ),

        DayOfWeek=(
            "DayOfWeek",
            "first"
        )
    )
)

daily["Strict_Clean_Day"] = (
    (daily["Rows"] == 96)
    &
    (
        daily["Valid_Target_Rows"]
        == 96
    )
    &
    (
        daily["All_Lags_Valid_Rows"]
        == 96
    )
)

clean_days = daily[
    daily["Strict_Clean_Day"]
].copy()

print("\n" + "=" * 130)
print("3. STRICT CLEAN JULY DETECTOR-DAYS")
print("=" * 130)

print(
    f"Strict clean detector-days: "
    f"{len(clean_days):,}"
)

print(
    f"Detectors with >=1 clean day: "
    f"{clean_days['Detector_Code'].nunique():,}"
)


# ============================================================
# 4. Require clean following day as well
#
# This creates clean 48-hour baseline windows so Lag_96
# propagation is fully controlled.
# ============================================================

clean_key_set = set(
    zip(
        clean_days[
            "Detector_Code"
        ].astype(int),

        clean_days[
            "Date"
        ]
    )
)

clean_days["Next_Date"] = (
    clean_days["Date"]
    + pd.Timedelta(days=1)
)

clean_days[
    "Next_Day_Strict_Clean"
] = [
    (
        int(detector),
        next_date
    ) in clean_key_set

    for detector, next_date in zip(
        clean_days["Detector_Code"],
        clean_days["Next_Date"]
    )
]

july_48h = clean_days[
    clean_days[
        "Next_Day_Strict_Clean"
    ]
].copy()

# Ensure the full dependency window remains in July
july_48h = july_48h[
    july_48h[
        "Next_Date"
    ]
    <= pd.Timestamp(
        "2026-07-31"
    )
].copy()

print("\n" + "=" * 130)
print("4. CLEAN 48-HOUR JULY WINDOWS")
print("=" * 130)

print(
    f"Eligible 48-hour windows: "
    f"{len(july_48h):,}"
)

print(
    f"Unique detectors represented: "
    f"{july_48h['Detector_Code'].nunique():,}"
)

print("\nWindows by starting day-of-week:")

print(
    july_48h[
        "DayOfWeek"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "DayOfWeek"
    )
    .reset_index(
        name="Windows"
    )
    .to_string(index=False)
)


# ============================================================
# 5. Match each Jan-Jun failure pattern to potential
#    July clean 48h windows:
#
# Primary requirement:
#   SAME DETECTOR + SAME DAY-OF-WEEK
# ============================================================

candidate_counts = []

for row in patterns.itertuples(
    index=False
):

    pool = july_48h[
        (
            july_48h[
                "Detector_Code"
            ]
            ==
            int(row.Detector_Code)
        )
        &
        (
            july_48h[
                "DayOfWeek"
            ]
            ==
            int(
                row.Source_DayOfWeek
            )
        )
    ]

    # Ensure evaluation endpoint remains in July even
    # for long runs ending late in the day
    valid_dates = []

    for target_date in pool[
        "Date"
    ]:

        failure_start = (
            target_date
            +
            pd.to_timedelta(
                int(
                    row.Start_Interval_Index
                ) * 15,
                unit="m"
            )
        )

        failure_end = (
            failure_start
            +
            pd.to_timedelta(
                (
                    int(row.Run_Length)
                    - 1
                ) * 15,
                unit="m"
            )
        )

        evaluation_end = (
            failure_end
            +
            pd.Timedelta(
                hours=24
            )
        )

        if (
            evaluation_end
            <=
            pd.Timestamp(
                "2026-07-31 23:45:00"
            )
        ):
            valid_dates.append(
                target_date
            )

    candidate_counts.append(
        len(valid_dates)
    )


patterns[
    "July_SameDOW_Clean48h_Candidates"
] = candidate_counts

patterns[
    "July_Validation_Feasible"
] = (
    patterns[
        "July_SameDOW_Clean48h_Candidates"
    ] > 0
)


# ============================================================
# 6. Pattern-level feasibility
# ============================================================

pattern_feasible = int(
    patterns[
        "July_Validation_Feasible"
    ].sum()
)

print("\n" + "=" * 130)
print("5. PATTERN-LEVEL JULY VALIDATION FEASIBILITY")
print("=" * 130)

print(
    f"Patterns with >=1 valid July match:"
)

print(
    f"{pattern_feasible:,} / "
    f"{len(patterns):,} "
    f"("
    f"{100 * pattern_feasible / len(patterns):.2f}%"
    f")"
)

print(
    "\nCandidate July windows per source pattern:"
)

print(
    patterns[
        "July_SameDOW_Clean48h_Candidates"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .to_string()
)


# ============================================================
# 7. Detector-level feasibility
#
# For validation we need at most one controlled event per
# detector, as in the held-out August design.
# ============================================================

detector_feasibility = (
    patterns
    .groupby(
        "Detector_Code"
    )
    .agg(
        Source_Patterns=(
            "Run_Length",
            "size"
        ),

        Feasible_Source_Patterns=(
            "July_Validation_Feasible",
            "sum"
        ),

        Maximum_July_Candidates=(
            "July_SameDOW_Clean48h_Candidates",
            "max"
        )
    )
    .reset_index()
)

detector_feasibility[
    "Detector_Validation_Feasible"
] = (
    detector_feasibility[
        "Feasible_Source_Patterns"
    ] > 0
)

detector_feasible_n = int(
    detector_feasibility[
        "Detector_Validation_Feasible"
    ].sum()
)

print("\n" + "=" * 130)
print("6. DETECTOR-LEVEL JULY VALIDATION FEASIBILITY")
print("=" * 130)

print(
    f"Source detectors with >=1 feasible controlled July event:"
)

print(
    f"{detector_feasible_n:,} / "
    f"{detector_feasibility.shape[0]:,} "
    f"("
    f"{100 * detector_feasible_n / detector_feasibility.shape[0]:.2f}%"
    f")"
)


# ============================================================
# 8. Feasibility by duration group
# ============================================================

duration_summary = (
    patterns
    .groupby(
        "Failure_Duration_Group",
        observed=True
    )
    .agg(
        Source_Patterns=(
            "Run_Length",
            "size"
        ),

        Feasible_Patterns=(
            "July_Validation_Feasible",
            "sum"
        )
    )
    .reset_index()
)

duration_summary[
    "Feasible_Percent"
] = (
    100
    *
    duration_summary[
        "Feasible_Patterns"
    ]
    /
    duration_summary[
        "Source_Patterns"
    ]
)

print("\n" + "=" * 130)
print("7. FEASIBILITY BY FAILURE DURATION")
print("=" * 130)

print(
    duration_summary
    .to_string(index=False)
)


# ============================================================
# 9. Source pattern representation among feasible detectors
# ============================================================

feasible_patterns = patterns[
    patterns[
        "July_Validation_Feasible"
    ]
].copy()

print("\n" + "=" * 130)
print("8. FEASIBLE PATTERN RUN-LENGTH DISTRIBUTION")
print("=" * 130)

print(
    feasible_patterns[
        "Run_Length"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .to_string()
)


# ============================================================
# 10. Save feasibility outputs
# ============================================================

pattern_path = os.path.join(
    NB5,
    "Tables",
    "Table_July_Controlled_Validation_Pattern_Feasibility.csv"
)

detector_path = os.path.join(
    NB5,
    "Tables",
    "Table_July_Controlled_Validation_Detector_Feasibility.csv"
)

duration_path = os.path.join(
    NB5,
    "Tables",
    "Table_July_Controlled_Validation_Feasibility_by_Duration.csv"
)

clean48_path = os.path.join(
    NB5,
    "Data",
    "July_Strict_Clean_48h_Validation_Windows.parquet"
)

patterns.to_csv(
    pattern_path,
    index=False
)

detector_feasibility.to_csv(
    detector_path,
    index=False
)

duration_summary.to_csv(
    duration_path,
    index=False
)

july_48h.to_parquet(
    clean48_path,
    index=False
)

print("\n" + "=" * 130)
print("9. SAVED OUTPUTS")
print("=" * 130)

print(f"✅ {pattern_path}")
print(f"✅ {detector_path}")
print(f"✅ {duration_path}")
print(f"✅ {clean48_path}")

print("\n" + "=" * 130)
print("STEP 14 COMPLETED")
print("=" * 130)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 15: Build controlled July failure-validation events
# ============================================================

import os
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

pattern_path = os.path.join(
    NB5,
    "Tables",
    "Table_July_Controlled_Validation_Pattern_Feasibility.csv"
)

clean48_path = os.path.join(
    NB5,
    "Data",
    "July_Strict_Clean_48h_Validation_Windows.parquet"
)

RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

print("=" * 130)
print("NOTEBOOK 5 — CONTROLLED JULY FAILURE-VALIDATION EVENT SELECTION")
print("=" * 130)

print(f"\nRandom seed: {RANDOM_SEED}")


# ============================================================
# 1. Load feasible Jan-Jun source patterns
# ============================================================

patterns = pd.read_csv(
    pattern_path
)

patterns["Date"] = pd.to_datetime(
    patterns["Date"]
).dt.normalize()

patterns = patterns[
    patterns[
        "July_Validation_Feasible"
    ] == True
].copy()

patterns["Source_DayOfWeek"] = (
    patterns["Date"]
    .dt.dayofweek
)

print("\n" + "=" * 130)
print("1. FEASIBLE SOURCE PATTERNS")
print("=" * 130)

print(
    f"Feasible patterns : "
    f"{len(patterns):,}"
)

print(
    f"Unique detectors  : "
    f"{patterns['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Load clean July 48-hour windows
# ============================================================

clean48 = pd.read_parquet(
    clean48_path
)

clean48["Date"] = pd.to_datetime(
    clean48["Date"]
).dt.normalize()

print("\n" + "=" * 130)
print("2. CLEAN JULY 48-HOUR WINDOWS")
print("=" * 130)

print(
    f"Eligible windows : "
    f"{len(clean48):,}"
)

print(
    f"Unique detectors : "
    f"{clean48['Detector_Code'].nunique():,}"
)


# ============================================================
# 3. Select ONE empirical source pattern per detector
#
# IMPORTANT:
# Selection is random with fixed seed and does NOT use
# forecasting performance.
# ============================================================

selected_sources = []

for detector, group in patterns.groupby(
    "Detector_Code"
):

    group = group.reset_index(
        drop=True
    )

    selected_idx = rng.integers(
        0,
        len(group)
    )

    selected_sources.append(
        group.iloc[
            selected_idx
        ].copy()
    )


selected_sources = pd.DataFrame(
    selected_sources
).reset_index(drop=True)

selected_sources[
    "July_Validation_Event_ID"
] = np.arange(
    1,
    len(selected_sources) + 1
)

print("\n" + "=" * 130)
print("3. SELECTED EMPIRICAL SOURCE EVENTS")
print("=" * 130)

print(
    f"Selected events   : "
    f"{len(selected_sources):,}"
)

print(
    f"Unique detectors  : "
    f"{selected_sources['Detector_Code'].nunique():,}"
)


# ============================================================
# 4. Assign each source event to a clean July day
#    with the SAME detector + SAME day-of-week
# ============================================================

assignments = []

for row in selected_sources.itertuples(
    index=False
):

    detector = int(
        row.Detector_Code
    )

    dow = int(
        row.Source_DayOfWeek
    )

    pool = clean48[
        (
            clean48[
                "Detector_Code"
            ] == detector
        )
        &
        (
            clean48[
                "DayOfWeek"
            ] == dow
        )
    ].copy()

    # --------------------------------------------------------
    # Keep only candidate dates whose entire evaluation
    # window remains within July.
    # --------------------------------------------------------

    valid_rows = []

    for candidate in pool.itertuples(
        index=False
    ):

        target_date = pd.Timestamp(
            candidate.Date
        )

        failure_start = (
            target_date
            +
            pd.to_timedelta(
                int(
                    row.Start_Interval_Index
                ) * 15,
                unit="m"
            )
        )

        failure_end = (
            failure_start
            +
            pd.to_timedelta(
                (
                    int(row.Run_Length)
                    - 1
                ) * 15,
                unit="m"
            )
        )

        evaluation_end = (
            failure_end
            +
            pd.Timedelta(
                hours=24
            )
        )

        if (
            evaluation_end
            <=
            pd.Timestamp(
                "2026-07-31 23:45:00"
            )
        ):
            valid_rows.append({
                "Target_Injection_Date":
                    target_date,

                "Failure_Start":
                    failure_start,

                "Failure_End":
                    failure_end,

                "Evaluation_End":
                    evaluation_end
            })

    if len(valid_rows) == 0:

        assignments.append({
            "July_Validation_Event_ID":
                row.July_Validation_Event_ID,

            "Detector_Code":
                detector,

            "Assignment_Status":
                "NO_MATCH"
        })

        continue

    selected_candidate = valid_rows[
        rng.integers(
            0,
            len(valid_rows)
        )
    ]

    assignments.append({
        "July_Validation_Event_ID":
            int(
                row.July_Validation_Event_ID
            ),

        "Detector_Code":
            detector,

        "Source_Failure_Date":
            row.Date,

        "Source_DayOfWeek":
            dow,

        "Target_Injection_Date":
            selected_candidate[
                "Target_Injection_Date"
            ],

        "Target_DayOfWeek":
            selected_candidate[
                "Target_Injection_Date"
            ].dayofweek,

        "Start_Interval_Index":
            int(
                row.Start_Interval_Index
            ),

        "End_Interval_Index":
            int(
                row.End_Interval_Index
            ),

        "Run_Length":
            int(
                row.Run_Length
            ),

        "Failure_Duration_Minutes":
            row.Failure_Duration_Minutes,

        "Failure_Duration_Hours":
            row.Failure_Duration_Hours,

        "Failure_Duration_Group":
            row.Failure_Duration_Group,

        "Failure_Start":
            selected_candidate[
                "Failure_Start"
            ],

        "Failure_End":
            selected_candidate[
                "Failure_End"
            ],

        "Evaluation_End":
            selected_candidate[
                "Evaluation_End"
            ],

        "Assignment_Status":
            "ASSIGNED"
    })


events = pd.DataFrame(
    assignments
)

assigned = events[
    events[
        "Assignment_Status"
    ] == "ASSIGNED"
].copy()


# ============================================================
# 5. Design validation
# ============================================================

print("\n" + "=" * 130)
print("4. JULY EVENT ASSIGNMENT RESULTS")
print("=" * 130)

print(
    f"Events assigned : "
    f"{len(assigned):,} / "
    f"{len(selected_sources):,}"
)

print(
    f"Unmatched       : "
    f"{(events['Assignment_Status'] != 'ASSIGNED').sum():,}"
)

print(
    f"Unique detectors: "
    f"{assigned['Detector_Code'].nunique():,}"
)


same_detector_one_event = (
    assigned[
        "Detector_Code"
    ].is_unique
)

same_dow = (
    assigned[
        "Source_DayOfWeek"
    ]
    ==
    assigned[
        "Target_DayOfWeek"
    ]
).all()

inside_july = (
    assigned[
        "Evaluation_End"
    ].max()
    <=
    pd.Timestamp(
        "2026-07-31 23:45:00"
    )
)

print("\n" + "=" * 130)
print("5. VALIDATION CHECKS")
print("=" * 130)

print(
    f"Maximum one event per detector : "
    f"{same_detector_one_event}"
)

print(
    f"Day-of-week preserved          : "
    f"{same_dow}"
)

print(
    f"All evaluation windows in July : "
    f"{inside_july}"
)


# ============================================================
# 6. Duration distribution
# ============================================================

duration_summary = (
    assigned[
        "Failure_Duration_Group"
    ]
    .value_counts()
    .rename_axis(
        "Failure_Duration_Group"
    )
    .reset_index(
        name="Events"
    )
)

duration_summary[
    "Percent"
] = (
    100
    * duration_summary[
        "Events"
    ]
    / len(assigned)
)

print("\n" + "=" * 130)
print("6. SELECTED JULY FAILURE DURATION DISTRIBUTION")
print("=" * 130)

print(
    duration_summary
    .to_string(index=False)
)


# ============================================================
# 7. Run-length statistics
# ============================================================

print("\n" + "=" * 130)
print("7. SELECTED JULY RUN-LENGTH STATISTICS")
print("=" * 130)

print(
    assigned[
        "Run_Length"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .to_string()
)


# ============================================================
# 8. Distribution across July injection dates
# ============================================================

date_distribution = (
    assigned[
        "Target_Injection_Date"
    ]
    .dt.date
    .value_counts()
    .sort_index()
    .rename_axis(
        "Target_Injection_Date"
    )
    .reset_index(
        name="Events"
    )
)

print("\n" + "=" * 130)
print("8. JULY INJECTION DATE DISTRIBUTION")
print("=" * 130)

print(
    date_distribution
    .to_string(index=False)
)


# ============================================================
# 9. Preview selected events
# ============================================================

print("\n" + "=" * 130)
print("9. SAMPLE CONTROLLED JULY EVENTS")
print("=" * 130)

preview_cols = [
    "July_Validation_Event_ID",
    "Detector_Code",
    "Source_Failure_Date",
    "Target_Injection_Date",
    "Start_Interval_Index",
    "Run_Length",
    "Failure_Duration_Group",
    "Failure_Start",
    "Failure_End",
    "Evaluation_End"
]

print(
    assigned[
        preview_cols
    ]
    .head(20)
    .to_string(index=False)
)


# ============================================================
# 10. Save controlled July design
# ============================================================

event_path = os.path.join(
    NB5,
    "Data",
    "Controlled_July_Failure_Validation_Events.csv"
)

duration_path = os.path.join(
    NB5,
    "Tables",
    "Table_Controlled_July_Validation_Duration_Distribution.csv"
)

date_path = os.path.join(
    NB5,
    "Tables",
    "Table_Controlled_July_Validation_Date_Distribution.csv"
)

design_path = os.path.join(
    NB5,
    "Results",
    "Controlled_July_Validation_Design.json"
)

assigned.to_csv(
    event_path,
    index=False
)

duration_summary.to_csv(
    duration_path,
    index=False
)

date_distribution.to_csv(
    date_path,
    index=False
)

import json

design_metadata = {
    "purpose":
        "Independent mitigation model selection",

    "source_failure_period":
        "2026-01-01 to 2026-06-30",

    "validation_period":
        "2026-07-01 to 2026-07-31",

    "final_test_period_reserved":
        "2026-08-01 to 2026-08-30",

    "event_selection":
        "Maximum one empirical failure event per detector",

    "selected_events":
        int(len(assigned)),

    "unique_detectors":
        int(
            assigned[
                "Detector_Code"
            ].nunique()
        ),

    "same_detector_preserved":
        True,

    "day_of_week_preserved":
        bool(same_dow),

    "ground_truth_modified":
        False,

    "random_seed":
        RANDOM_SEED
}

with open(
    design_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        design_metadata,
        f,
        indent=2
    )

print("\n" + "=" * 130)
print("10. SAVED OUTPUTS")
print("=" * 130)

print(f"✅ {event_path}")
print(f"✅ {duration_path}")
print(f"✅ {date_path}")
print(f"✅ {design_path}")

print("\n" + "=" * 130)
print("STEP 15 COMPLETED")
print("=" * 130)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 16: Materialize controlled July clean/injected
#          validation rows
# ============================================================

import os
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

JULY_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly",
    "FORECASTING_2026-07.parquet"
)

EVENT_PATH = os.path.join(
    NB5,
    "Data",
    "Controlled_July_Failure_Validation_Events.csv"
)

lag_offsets = {
    "Lag_1": 15,
    "Lag_2": 30,
    "Lag_4": 60,
    "Lag_8": 120,
    "Lag_96": 1440
}

lag_cols = list(lag_offsets.keys())

print("=" * 130)
print("NOTEBOOK 5 — MATERIALIZING CONTROLLED JULY VALIDATION ROWS")
print("=" * 130)


# ============================================================
# 1. Load July controlled events
# ============================================================

events = pd.read_csv(
    EVENT_PATH
)

datetime_cols = [
    "Source_Failure_Date",
    "Target_Injection_Date",
    "Failure_Start",
    "Failure_End",
    "Evaluation_End"
]

for c in datetime_cols:
    events[c] = pd.to_datetime(
        events[c]
    )

print("\n" + "=" * 130)
print("1. CONTROLLED JULY EVENTS")
print("=" * 130)

print(
    f"Events           : {len(events):,}"
)

print(
    f"Unique detectors : "
    f"{events['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Build exact lag-impact map
# ============================================================

impact_records = []

for row in events.itertuples(
    index=False
):

    failure_times = pd.date_range(
        start=row.Failure_Start,
        periods=int(row.Run_Length),
        freq="15min"
    )

    for lag_name, offset_minutes in lag_offsets.items():

        target_times = (
            failure_times
            + pd.Timedelta(
                minutes=offset_minutes
            )
        )

        for target_ts in target_times:

            impact_records.append({
                "July_Validation_Event_ID":
                    int(
                        row.July_Validation_Event_ID
                    ),

                "Detector_Code":
                    int(
                        row.Detector_Code
                    ),

                "Timestamp":
                    pd.Timestamp(
                        target_ts
                    ),

                "Missing_Lag":
                    lag_name,

                "Failure_Run_Length":
                    int(
                        row.Run_Length
                    ),

                "Failure_Duration_Hours":
                    row.Failure_Duration_Hours,

                "Failure_Duration_Group":
                    row.Failure_Duration_Group
            })


impact_long = pd.DataFrame(
    impact_records
).drop_duplicates(
    subset=[
        "July_Validation_Event_ID",
        "Detector_Code",
        "Timestamp",
        "Missing_Lag"
    ]
)

print("\n" + "=" * 130)
print("2. RAW FAILURE-DEPENDENCY IMPACTS")
print("=" * 130)

print(
    f"Raw lag-impact records : "
    f"{len(impact_long):,}"
)


# ============================================================
# 3. Collapse to one row per affected forecasting target
# ============================================================

target_map = (
    impact_long
    .groupby(
        [
            "July_Validation_Event_ID",
            "Detector_Code",
            "Timestamp"
        ],
        as_index=False
    )
    .agg(
        Failure_Run_Length=(
            "Failure_Run_Length",
            "first"
        ),

        Failure_Duration_Hours=(
            "Failure_Duration_Hours",
            "first"
        ),

        Failure_Duration_Group=(
            "Failure_Duration_Group",
            "first"
        )
    )
)

for lag in lag_cols:

    flag = (
        impact_long[
            impact_long["Missing_Lag"] == lag
        ][
            [
                "July_Validation_Event_ID",
                "Detector_Code",
                "Timestamp"
            ]
        ]
        .drop_duplicates()
        .copy()
    )

    flag[f"Missing_{lag}"] = 1

    target_map = target_map.merge(
        flag,
        on=[
            "July_Validation_Event_ID",
            "Detector_Code",
            "Timestamp"
        ],
        how="left"
    )


for lag in lag_cols:

    target_map[
        f"Missing_{lag}"
    ] = (
        target_map[
            f"Missing_{lag}"
        ]
        .fillna(0)
        .astype("int8")
    )


target_map[
    "Missing_Lag_Count"
] = (
    target_map[
        [
            f"Missing_{lag}"
            for lag in lag_cols
        ]
    ]
    .sum(axis=1)
    .astype("int8")
)


def make_pattern(row):

    missing = [
        lag
        for lag in lag_cols
        if row[
            f"Missing_{lag}"
        ] == 1
    ]

    return "+".join(
        missing
    )


target_map[
    "Missing_Lag_Pattern"
] = target_map.apply(
    make_pattern,
    axis=1
)

print("\n" + "=" * 130)
print("3. UNIQUE AFFECTED JULY TARGETS")
print("=" * 130)

print(
    f"Affected targets : "
    f"{len(target_map):,}"
)

print(
    f"Events represented: "
    f"{target_map['July_Validation_Event_ID'].nunique():,}"
)

print(
    f"Duplicate detector-target timestamps: "
    f"{target_map.duplicated(['Detector_Code','Timestamp']).sum():,}"
)


# ============================================================
# 4. Load July forecasting baseline data
# ============================================================

columns = [
    "Timestamp",
    "Detector_Code",
    "Interval_Index",
    "Target_Volume",
    "Hour",
    "Minute",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96",
    "All_Lags_Available",
    "Split"
]

july = pd.read_parquet(
    JULY_PATH,
    columns=columns
)

july["Timestamp"] = pd.to_datetime(
    july["Timestamp"]
)

validation = july.merge(
    target_map,
    on=[
        "Detector_Code",
        "Timestamp"
    ],
    how="inner",
    validate="one_to_one"
)

print("\n" + "=" * 130)
print("4. BASELINE ROW EXTRACTION")
print("=" * 130)

print(
    f"Expected affected targets : "
    f"{len(target_map):,}"
)

print(
    f"Extracted July rows       : "
    f"{len(validation):,}"
)


# ============================================================
# 5. Pre-injection integrity checks
# ============================================================

baseline_missing = int(
    validation[
        lag_cols
    ]
    .isna()
    .any(axis=1)
    .sum()
)

invalid_targets = int(
    (
        validation[
            "Target_Volume"
        ].isna()
        |
        ~np.isfinite(
            validation[
                "Target_Volume"
            ]
        )
        |
        (
            validation[
                "Target_Volume"
            ] < 0
        )
    ).sum()
)

duplicates = int(
    validation
    .duplicated(
        subset=[
            "Detector_Code",
            "Timestamp"
        ]
    )
    .sum()
)

print("\n" + "=" * 130)
print("5. PRE-INJECTION INTEGRITY")
print("=" * 130)

print(
    f"Rows with baseline missing lag : "
    f"{baseline_missing:,}"
)

print(
    f"Invalid targets                : "
    f"{invalid_targets:,}"
)

print(
    f"Duplicate rows                 : "
    f"{duplicates:,}"
)


# ============================================================
# 6. Preserve CLEAN lag features
# ============================================================

for lag in lag_cols:

    validation[
        f"{lag}_Clean"
    ] = validation[
        lag
    ].copy()


# ============================================================
# 7. Create injected feature columns
# ============================================================

for lag in lag_cols:

    validation[
        f"{lag}_Injected"
    ] = validation[
        lag
    ].copy()

    injection_mask = (
        validation[
            f"Missing_{lag}"
        ] == 1
    )

    validation.loc[
        injection_mask,
        f"{lag}_Injected"
    ] = np.nan


# ============================================================
# 8. Validate exact lag injection
# ============================================================

validation_checks = []

for lag in lag_cols:

    intended_missing = int(
        validation[
            f"Missing_{lag}"
        ].sum()
    )

    actual_missing = int(
        validation[
            f"{lag}_Injected"
        ].isna().sum()
    )

    unexpected_missing = int(
        (
            validation[
                f"{lag}_Injected"
            ].isna()
            &
            (
                validation[
                    f"Missing_{lag}"
                ] == 0
            )
        ).sum()
    )

    validation_checks.append({
        "Lag":
            lag,

        "Intended_Missing":
            intended_missing,

        "Actual_Injected_Missing":
            actual_missing,

        "Unexpected_Missing":
            unexpected_missing,

        "Exact_Match":
            (
                intended_missing
                ==
                actual_missing
                and
                unexpected_missing == 0
            )
    })


validation_check_df = pd.DataFrame(
    validation_checks
)

print("\n" + "=" * 130)
print("6. INJECTION VALIDATION")
print("=" * 130)

print(
    validation_check_df
    .to_string(index=False)
)

all_exact = bool(
    validation_check_df[
        "Exact_Match"
    ].all()
)

print(
    f"\nAll injections exact: "
    f"{all_exact}"
)


# ============================================================
# 9. Target integrity
# ============================================================

print("\n" + "=" * 130)
print("7. TARGET INTEGRITY")
print("=" * 130)

print(
    f"Target minimum : "
    f"{validation['Target_Volume'].min():.3f}"
)

print(
    f"Target maximum : "
    f"{validation['Target_Volume'].max():.3f}"
)

print(
    f"Missing targets: "
    f"{validation['Target_Volume'].isna().sum():,}"
)


# ============================================================
# 10. Missing-lag pattern distribution
# ============================================================

pattern_dist = (
    validation[
        "Missing_Lag_Pattern"
    ]
    .value_counts()
    .rename_axis(
        "Missing_Lag_Pattern"
    )
    .reset_index(
        name="Rows"
    )
)

pattern_dist[
    "Percent"
] = (
    100
    * pattern_dist[
        "Rows"
    ]
    / len(validation)
)

print("\n" + "=" * 130)
print("8. JULY CONTROLLED MISSING-LAG PATTERNS")
print("=" * 130)

print(
    pattern_dist
    .to_string(index=False)
)


# ============================================================
# 11. Missing-lag count distribution
# ============================================================

count_dist = (
    validation[
        "Missing_Lag_Count"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "Missing_Lag_Count"
    )
    .reset_index(
        name="Rows"
    )
)

count_dist[
    "Percent"
] = (
    100
    * count_dist[
        "Rows"
    ]
    / len(validation)
)

print("\n" + "=" * 130)
print("9. JULY MISSING-LAG COUNT DISTRIBUTION")
print("=" * 130)

print(
    count_dist
    .to_string(index=False)
)


# ============================================================
# 12. Compare July validation patterns with empirical
#     training augmentation patterns
# ============================================================

training_pattern_path = os.path.join(
    NB5,
    "Tables",
    "Table_Empirical_Augmented_vs_August_Missing_Patterns.csv"
)

training_patterns = pd.read_csv(
    training_pattern_path
)

training_pattern_set = set(
    training_patterns.loc[
        training_patterns[
            "Augmented_Training_Rows"
        ] > 0,
        "Missing_Lag_Pattern"
    ]
)

validation[
    "Pattern_Seen_In_Augmented_Training"
] = (
    validation[
        "Missing_Lag_Pattern"
    ]
    .isin(
        training_pattern_set
    )
)

covered_rows = int(
    validation[
        "Pattern_Seen_In_Augmented_Training"
    ].sum()
)

coverage_percent = (
    100
    * covered_rows
    / len(validation)
)

print("\n" + "=" * 130)
print("10. TRAINING PATTERN COVERAGE OF JULY VALIDATION")
print("=" * 130)

print(
    f"Covered July affected rows:"
)

print(
    f"{covered_rows:,} / "
    f"{len(validation):,} "
    f"({coverage_percent:.2f}%)"
)


# ============================================================
# 13. Event-level affected row counts
# ============================================================

event_summary = (
    validation
    .groupby(
        "July_Validation_Event_ID",
        as_index=False
    )
    .agg(
        Detector_Code=(
            "Detector_Code",
            "first"
        ),

        Failure_Run_Length=(
            "Failure_Run_Length",
            "first"
        ),

        Failure_Duration_Hours=(
            "Failure_Duration_Hours",
            "first"
        ),

        Failure_Duration_Group=(
            "Failure_Duration_Group",
            "first"
        ),

        Affected_Rows=(
            "Timestamp",
            "size"
        ),

        Maximum_Missing_Lag_Count=(
            "Missing_Lag_Count",
            "max"
        )
    )
)

print("\n" + "=" * 130)
print("11. EVENT-LEVEL VALIDATION SUMMARY")
print("=" * 130)

print(
    event_summary[
        "Affected_Rows"
    ]
    .describe()
    .to_string()
)


# ============================================================
# 14. Final integrity decision
# ============================================================

integrity_pass = (
    len(validation)
    ==
    len(target_map)

    and baseline_missing == 0

    and invalid_targets == 0

    and duplicates == 0

    and all_exact

    and coverage_percent == 100.0

    and validation[
        "July_Validation_Event_ID"
    ].nunique() == 581
)

print("\n" + "=" * 130)
print("12. STEP 16 INTEGRITY DECISION")
print("=" * 130)

print(
    f"All expected rows recovered        : "
    f"{len(validation) == len(target_map)}"
)

print(
    f"All clean baselines originally valid: "
    f"{baseline_missing == 0}"
)

print(
    f"All targets valid                  : "
    f"{invalid_targets == 0}"
)

print(
    f"No duplicate targets               : "
    f"{duplicates == 0}"
)

print(
    f"All lag injections exact           : "
    f"{all_exact}"
)

print(
    f"581 events represented             : "
    f"{validation['July_Validation_Event_ID'].nunique() == 581}"
)

print(
    f"Training-pattern coverage = 100%   : "
    f"{coverage_percent == 100.0}"
)

print(
    f"\nFINAL STEP 16 INTEGRITY PASS: "
    f"{integrity_pass}"
)


# ============================================================
# 15. Save outputs
# ============================================================

validation_path = os.path.join(
    NB5,
    "Data",
    "Controlled_July_Clean_Injected_Validation_Rows.parquet"
)

event_summary_path = os.path.join(
    NB5,
    "Data",
    "Controlled_July_Event_Level_Validation_Summary.parquet"
)

pattern_path = os.path.join(
    NB5,
    "Tables",
    "Table_Controlled_July_Missing_Patterns.csv"
)

count_path = os.path.join(
    NB5,
    "Tables",
    "Table_Controlled_July_Missing_Counts.csv"
)

injection_validation_path = os.path.join(
    NB5,
    "Tables",
    "Table_Controlled_July_Injection_Validation.csv"
)

validation.to_parquet(
    validation_path,
    index=False
)

event_summary.to_parquet(
    event_summary_path,
    index=False
)

pattern_dist.to_csv(
    pattern_path,
    index=False
)

count_dist.to_csv(
    count_path,
    index=False
)

validation_check_df.to_csv(
    injection_validation_path,
    index=False
)

print("\n" + "=" * 130)
print("13. SAVED OUTPUTS")
print("=" * 130)

print(f"✅ {validation_path}")
print(f"✅ {event_summary_path}")
print(f"✅ {pattern_path}")
print(f"✅ {count_path}")
print(f"✅ {injection_validation_path}")

print("\n" + "=" * 130)
print("STEP 16 COMPLETED")
print("=" * 130)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 17: Recover and reproduce the ORIGINAL clean-only
#          Jan-Jun LightGBM baseline (R0)
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import lightgbm as lgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

MODEL_PATH = os.path.join(
    NB3,
    "Models",
    "LightGBM_Clean_15min.txt"
)

JULY_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly",
    "FORECASTING_2026-07.parquet"
)

NB3_PRED_PATH = os.path.join(
    NB3,
    "Data",
    "Validation_LightGBM_Predictions.parquet"
)

CONTROLLED_JULY_PATH = os.path.join(
    NB5,
    "Data",
    "Controlled_July_Clean_Injected_Validation_Rows.parquet"
)

print("=" * 130)
print("NOTEBOOK 5 — R0 ORIGINAL CLEAN-ONLY LIGHTGBM RECOVERY")
print("=" * 130)


# ============================================================
# 1. Load original Notebook 3 clean-only LightGBM
# ============================================================

r0 = lgb.Booster(
    model_file=MODEL_PATH
)

feature_names = (
    r0.feature_name()
)

print("\n" + "=" * 130)
print("1. ORIGINAL R0 MODEL")
print("=" * 130)

print("✅ Model loaded")
print(
    f"Trees    : "
    f"{r0.num_trees():,}"
)

print(
    f"Features : "
    f"{r0.num_feature():,}"
)

print("\nFeature names:")

for i, feature in enumerate(
    feature_names,
    start=1
):
    print(
        f"{i:>2}. {feature}"
    )


# ============================================================
# 2. Inspect exact model parameters
# ============================================================

params_to_show = [
    "objective",
    "metric",
    "learning_rate",
    "num_leaves",
    "max_depth",
    "min_data_in_leaf",
    "feature_fraction",
    "bagging_fraction",
    "bagging_freq",
    "lambda_l1",
    "lambda_l2",
    "seed"
]

print("\n" + "=" * 130)
print("2. R0 MODEL PARAMETERS")
print("=" * 130)

for p in params_to_show:

    if p in r0.params:
        print(
            f"{p:<25}: "
            f"{r0.params[p]}"
        )


# ============================================================
# 3. Load July forecasting data
# ============================================================

needed_columns = list(
    dict.fromkeys(
        [
            "Timestamp",
            "Detector_Code",
            "Target_Volume",
            "All_Lags_Available",
            "Split"
        ]
        +
        feature_names
    )
)

july = pd.read_parquet(
    JULY_PATH,
    columns=needed_columns
)

july["Timestamp"] = pd.to_datetime(
    july["Timestamp"]
)

valid_target = (
    july[
        "Target_Volume"
    ].notna()
    &
    np.isfinite(
        july[
            "Target_Volume"
        ]
    )
    &
    (
        july[
            "Target_Volume"
        ] >= 0
    )
)

all_features_available = (
    july[
        feature_names
    ]
    .notna()
    .all(axis=1)
)

model_ready_mask = (
    valid_target
    &
    all_features_available
)

july_ready = july.loc[
    model_ready_mask
].copy()

print("\n" + "=" * 130)
print("3. JULY CLEAN MODEL-READY DATA")
print("=" * 130)

print(
    f"Full July rows       : "
    f"{len(july):,}"
)

print(
    f"Model-ready rows     : "
    f"{len(july_ready):,}"
)

print(
    f"Excluded rows        : "
    f"{len(july)-len(july_ready):,}"
)

print(
    f"Model-ready percent  : "
    f"{100*len(july_ready)/len(july):.4f}%"
)


# ============================================================
# 4. Predict using ORIGINAL R0
# ============================================================

X_july = july_ready[
    feature_names
]

y_july = (
    july_ready[
        "Target_Volume"
    ]
    .to_numpy(
        dtype=np.float64
    )
)

r0_pred_raw = r0.predict(
    X_july
)

# Notebook 3/4 post-processing:
# traffic-volume predictions cannot be negative
r0_pred = np.maximum(
    r0_pred_raw,
    0
)


# ============================================================
# 5. Metric function
# ============================================================

def calculate_metrics(
    y_true,
    y_pred
):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    denominator = np.sum(
        np.abs(
            y_true
        )
    )

    if denominator > 0:

        wape = (
            100
            * np.sum(
                np.abs(
                    y_true
                    - y_pred
                )
            )
            / denominator
        )

    else:
        wape = np.nan

    bias = np.mean(
        y_pred
        - y_true
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "WAPE": wape,
        "Bias": bias
    }


full_july_metrics = (
    calculate_metrics(
        y_july,
        r0_pred
    )
)

print("\n" + "=" * 130)
print("4. R0 FULL JULY CLEAN PERFORMANCE")
print("=" * 130)

for k, v in full_july_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )


# ============================================================
# 6. Compare with Notebook 3 saved validation predictions
# ============================================================

print("\n" + "=" * 130)
print("5. NOTEBOOK 3 PREDICTION REPRODUCIBILITY")
print("=" * 130)

if os.path.exists(
    NB3_PRED_PATH
):

    nb3_pred = pd.read_parquet(
        NB3_PRED_PATH
    )

    print(
        f"Notebook 3 saved prediction rows: "
        f"{len(nb3_pred):,}"
    )

    print("\nSaved prediction columns:")

    for i, c in enumerate(
        nb3_pred.columns,
        start=1
    ):
        print(
            f"{i:>2}. {c}"
        )

    # --------------------------------------------------------
    # Find prediction column automatically
    # --------------------------------------------------------

    pred_candidates = [
        c
        for c in nb3_pred.columns
        if (
            "pred" in c.lower()
            and
            "xgb" not in c.lower()
        )
    ]

    print(
        "\nPotential prediction columns:"
    )
    print(
        pred_candidates
    )

    # --------------------------------------------------------
    # Build current prediction table
    # --------------------------------------------------------

    current_pred_df = pd.DataFrame({
        "Timestamp":
            july_ready[
                "Timestamp"
            ].values,

        "Detector_Code":
            july_ready[
                "Detector_Code"
            ].values,

        "Target_Volume_Current":
            y_july,

        "R0_Reproduced_Prediction":
            r0_pred
    })

    reproducibility_result = {
        "Comparison_Performed":
            False
    }

    # Need shared keys
    if (
        "Timestamp" in nb3_pred.columns
        and
        "Detector_Code" in nb3_pred.columns
        and
        len(pred_candidates) > 0
    ):

        nb3_pred[
            "Timestamp"
        ] = pd.to_datetime(
            nb3_pred[
                "Timestamp"
            ]
        )

        # Prefer explicitly LightGBM-named column
        lgb_candidates = [
            c
            for c in pred_candidates
            if "lightgbm" in c.lower()
            or "lgb" in c.lower()
        ]

        if len(
            lgb_candidates
        ) > 0:

            stored_pred_col = (
                lgb_candidates[0]
            )

        else:

            stored_pred_col = (
                pred_candidates[0]
            )

        compare = current_pred_df.merge(
            nb3_pred[
                [
                    "Timestamp",
                    "Detector_Code",
                    stored_pred_col
                ]
            ],
            on=[
                "Timestamp",
                "Detector_Code"
            ],
            how="inner"
        )

        stored_values = (
            compare[
                stored_pred_col
            ]
            .to_numpy(
                dtype=np.float64
            )
        )

        # Same zero-clipping policy
        stored_values = np.maximum(
            stored_values,
            0
        )

        reproduced_values = (
            compare[
                "R0_Reproduced_Prediction"
            ]
            .to_numpy(
                dtype=np.float64
            )
        )

        absolute_difference = (
            np.abs(
                reproduced_values
                - stored_values
            )
        )

        reproducibility_result = {
            "Comparison_Performed":
                True,

            "Stored_Prediction_Column":
                stored_pred_col,

            "Matched_Rows":
                int(
                    len(compare)
                ),

            "Mean_Absolute_Difference":
                float(
                    absolute_difference.mean()
                ),

            "Median_Absolute_Difference":
                float(
                    np.median(
                        absolute_difference
                    )
                ),

            "Maximum_Absolute_Difference":
                float(
                    absolute_difference.max()
                ),

            "Rows_Difference_GT_1e-6":
                int(
                    (
                        absolute_difference
                        > 1e-6
                    ).sum()
                ),

            "Rows_Difference_GT_1e-4":
                int(
                    (
                        absolute_difference
                        > 1e-4
                    ).sum()
                )
        }

        print(
            f"\nStored prediction column : "
            f"{stored_pred_col}"
        )

        print(
            f"Matched rows             : "
            f"{len(compare):,}"
        )

        print(
            f"Mean absolute difference : "
            f"{absolute_difference.mean():.10f}"
        )

        print(
            f"Median absolute difference: "
            f"{np.median(absolute_difference):.10f}"
        )

        print(
            f"Maximum absolute difference: "
            f"{absolute_difference.max():.10f}"
        )

        print(
            f"Difference > 1e-6        : "
            f"{(absolute_difference > 1e-6).sum():,}"
        )

        print(
            f"Difference > 1e-4        : "
            f"{(absolute_difference > 1e-4).sum():,}"
        )

    else:

        print(
            "\n⚠️ Automatic row-level comparison "
            "could not be completed from available columns."
        )

else:

    print(
        "❌ Notebook 3 validation "
        "prediction file not found."
    )

    reproducibility_result = {
        "Comparison_Performed":
            False
    }


# ============================================================
# 7. Evaluate R0 on EXACT controlled July rows
#    CLEAN condition
# ============================================================

controlled = pd.read_parquet(
    CONTROLLED_JULY_PATH
)

controlled[
    "Timestamp"
] = pd.to_datetime(
    controlled[
        "Timestamp"
    ]
)

X_controlled_clean = (
    controlled[
        feature_names
    ]
)

y_controlled = (
    controlled[
        "Target_Volume"
    ]
    .to_numpy(
        dtype=np.float64
    )
)

r0_controlled_clean_pred = (
    np.maximum(
        r0.predict(
            X_controlled_clean
        ),
        0
    )
)

controlled_clean_metrics = (
    calculate_metrics(
        y_controlled,
        r0_controlled_clean_pred
    )
)

print("\n" + "=" * 130)
print("6. R0 CONTROLLED JULY — CLEAN CONDITION")
print("=" * 130)

for k, v in controlled_clean_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )


# ============================================================
# 8. Evaluate R0 under CONTROLLED FAILURE injection
# ============================================================

controlled_injected = (
    controlled.copy()
)

for lag in [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]:

    controlled_injected[
        lag
    ] = controlled_injected[
        f"{lag}_Injected"
    ]


X_controlled_injected = (
    controlled_injected[
        feature_names
    ]
)

r0_controlled_injected_pred = (
    np.maximum(
        r0.predict(
            X_controlled_injected
        ),
        0
    )
)

controlled_injected_metrics = (
    calculate_metrics(
        y_controlled,
        r0_controlled_injected_pred
    )
)

print("\n" + "=" * 130)
print("7. R0 CONTROLLED JULY — FAILURE-INJECTED CONDITION")
print("=" * 130)

for k, v in controlled_injected_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )


# ============================================================
# 9. R0 degradation summary
# ============================================================

r0_degradation = {
    "MAE_Delta":
        controlled_injected_metrics[
            "MAE"
        ]
        -
        controlled_clean_metrics[
            "MAE"
        ],

    "RMSE_Delta":
        controlled_injected_metrics[
            "RMSE"
        ]
        -
        controlled_clean_metrics[
            "RMSE"
        ],

    "WAPE_Delta_pp":
        controlled_injected_metrics[
            "WAPE"
        ]
        -
        controlled_clean_metrics[
            "WAPE"
        ],

    "WAPE_Degradation_Percent":
        100
        *
        (
            controlled_injected_metrics[
                "WAPE"
            ]
            -
            controlled_clean_metrics[
                "WAPE"
            ]
        )
        /
        controlled_clean_metrics[
            "WAPE"
        ],

    "Bias_Shift":
        controlled_injected_metrics[
            "Bias"
        ]
        -
        controlled_clean_metrics[
            "Bias"
        ]
}

print("\n" + "=" * 130)
print("8. R0 JULY ROBUSTNESS DEGRADATION")
print("=" * 130)

for k, v in r0_degradation.items():

    print(
        f"{k:<28}: "
        f"{v:.6f}"
    )


# ============================================================
# 10. Save R0 controlled July predictions
# ============================================================

r0_prediction_table = pd.DataFrame({
    "July_Validation_Event_ID":
        controlled[
            "July_Validation_Event_ID"
        ].values,

    "Detector_Code":
        controlled[
            "Detector_Code"
        ].values,

    "Timestamp":
        controlled[
            "Timestamp"
        ].values,

    "Actual":
        y_controlled,

    "Missing_Lag_Count":
        controlled[
            "Missing_Lag_Count"
        ].values,

    "Missing_Lag_Pattern":
        controlled[
            "Missing_Lag_Pattern"
        ].values,

    "R0_Clean_Prediction":
        r0_controlled_clean_pred,

    "R0_Injected_Prediction":
        r0_controlled_injected_pred
})


# ============================================================
# 11. Save results
# ============================================================

results_table = pd.DataFrame([
    {
        "Model":
            "R0_Original_CleanOnly",

        "Condition":
            "Full_July_Clean",

        **full_july_metrics
    },
    {
        "Model":
            "R0_Original_CleanOnly",

        "Condition":
            "Controlled_July_Clean",

        **controlled_clean_metrics
    },
    {
        "Model":
            "R0_Original_CleanOnly",

        "Condition":
            "Controlled_July_Injected",

        **controlled_injected_metrics
    }
])

result_path = os.path.join(
    NB5,
    "Tables",
    "Table_R0_July_Baseline_Performance.csv"
)

prediction_path = os.path.join(
    NB5,
    "Data",
    "R0_Controlled_July_Predictions.parquet"
)

metadata_path = os.path.join(
    NB5,
    "Results",
    "R0_Original_Model_Recovery.json"
)

results_table.to_csv(
    result_path,
    index=False
)

r0_prediction_table.to_parquet(
    prediction_path,
    index=False
)

metadata = {
    "regime":
        "R0_Original_CleanOnly",

    "source_model":
        MODEL_PATH,

    "number_of_trees":
        int(
            r0.num_trees()
        ),

    "feature_names":
        feature_names,

    "model_parameters":
        {
            k: r0.params.get(k)
            for k in params_to_show
            if k in r0.params
        },

    "full_july_model_ready_rows":
        int(
            len(july_ready)
        ),

    "full_july_metrics":
        {
            k: float(v)
            for k, v
            in full_july_metrics.items()
        },

    "controlled_july_clean_metrics":
        {
            k: float(v)
            for k, v
            in controlled_clean_metrics.items()
        },

    "controlled_july_injected_metrics":
        {
            k: float(v)
            for k, v
            in controlled_injected_metrics.items()
        },

    "controlled_july_degradation":
        {
            k: float(v)
            for k, v
            in r0_degradation.items()
        },

    "notebook3_reproducibility":
        reproducibility_result,

    "zero_clipping_applied":
        True
}

with open(
    metadata_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )

print("\n" + "=" * 130)
print("9. SAVED OUTPUTS")
print("=" * 130)

print(f"✅ {result_path}")
print(f"✅ {prediction_path}")
print(f"✅ {metadata_path}")

print("\n" + "=" * 130)
print("STEP 17 COMPLETED")
print("=" * 130)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 18: Train R1 — Failure-Aware LightGBM
# ============================================================

import os
import gc
import json
import time
import numpy as np
import pandas as pd
import lightgbm as lgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

NATURAL_PATH = os.path.join(
    NB5,
    "Data",
    "Natural_Failure_Aware_Training_Candidates.parquet"
)

AUGMENTED_PATH = os.path.join(
    NB5,
    "Data",
    "Empirical_Failure_Augmented_Training_Rows.parquet"
)

CONTROLLED_JULY_PATH = os.path.join(
    NB5,
    "Data",
    "Controlled_July_Clean_Injected_Validation_Rows.parquet"
)

R0_RESULT_PATH = os.path.join(
    NB5,
    "Tables",
    "Table_R0_July_Baseline_Performance.csv"
)

# ------------------------------------------------------------
# Exact feature set used by Notebook 3
# ------------------------------------------------------------

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

LAG_COLS = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TRAIN_MONTHS = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06"
]

print("=" * 135)
print("NOTEBOOK 5 — R1 FAILURE-AWARE LIGHTGBM TRAINING")
print("=" * 135)


# ============================================================
# 1. Helper functions
# ============================================================

def downcast_training_frame(df):

    # Integer / categorical-like predictors
    for col in [
        "Detector_Code",
        "Interval_Index",
        "DayOfWeek",
        "IsWeekend"
    ]:
        df[col] = pd.to_numeric(
            df[col],
            downcast="integer"
        )

    # Continuous traffic-volume predictors
    for col in LAG_COLS:
        df[col] = df[col].astype(
            "float32"
        )

    df[TARGET] = df[TARGET].astype(
        "float32"
    )

    return df


def calculate_metrics(
    y_true,
    y_pred
):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    denominator = np.sum(
        np.abs(y_true)
    )

    wape = (
        100
        * np.sum(
            np.abs(
                y_true - y_pred
            )
        )
        / denominator
    )

    bias = np.mean(
        y_pred - y_true
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "WAPE": float(wape),
        "Bias": float(bias)
    }


# ============================================================
# 2. Load CLEAN Jan-Jun model-ready training rows
# ============================================================

print("\n" + "=" * 135)
print("1. LOADING CLEAN JAN–JUN TRAINING DATA")
print("=" * 135)

clean_parts = []
clean_month_audit = []

for month in TRAIN_MONTHS:

    path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    df = pd.read_parquet(
        path,
        columns=FEATURES + [TARGET]
    )

    valid_target = (
        df[TARGET].notna()
        &
        np.isfinite(
            df[TARGET]
        )
        &
        (
            df[TARGET] >= 0
        )
    )

    all_features_valid = (
        df[FEATURES]
        .notna()
        .all(axis=1)
    )

    clean = df.loc[
        valid_target
        &
        all_features_valid,
        FEATURES + [TARGET]
    ].copy()

    clean = downcast_training_frame(
        clean
    )

    print(
        f"{month}: "
        f"{len(clean):,} clean model-ready rows"
    )

    clean_month_audit.append({
        "Month": month,
        "Clean_Model_Ready_Rows":
            int(len(clean))
    })

    clean_parts.append(
        clean
    )

    del df, clean
    gc.collect()


clean_training = pd.concat(
    clean_parts,
    ignore_index=True
)

del clean_parts
gc.collect()

print(
    f"\nTotal clean Jan–Jun rows: "
    f"{len(clean_training):,}"
)


# ============================================================
# 3. Load NATURAL missing-history training rows
# ============================================================

print("\n" + "=" * 135)
print("2. LOADING NATURAL MISSING-HISTORY ROWS")
print("=" * 135)

natural_training = pd.read_parquet(
    NATURAL_PATH,
    columns=FEATURES + [TARGET]
)

natural_training = (
    downcast_training_frame(
        natural_training
    )
)

print(
    f"Natural missing-history rows: "
    f"{len(natural_training):,}"
)

print(
    f"Rows with >=1 missing feature: "
    f"{natural_training[FEATURES].isna().any(axis=1).sum():,}"
)

print(
    f"Missing targets: "
    f"{natural_training[TARGET].isna().sum():,}"
)


# ============================================================
# 4. Load EMPIRICAL FAILURE-AUGMENTED rows
# ============================================================

print("\n" + "=" * 135)
print("3. LOADING EMPIRICAL FAILURE-AUGMENTED ROWS")
print("=" * 135)

augmented_training = pd.read_parquet(
    AUGMENTED_PATH,
    columns=FEATURES + [TARGET]
)

augmented_training = (
    downcast_training_frame(
        augmented_training
    )
)

print(
    f"Empirical augmented rows: "
    f"{len(augmented_training):,}"
)

print(
    f"Rows with >=1 missing feature: "
    f"{augmented_training[FEATURES].isna().any(axis=1).sum():,}"
)

print(
    f"Missing targets: "
    f"{augmented_training[TARGET].isna().sum():,}"
)


# ============================================================
# 5. Assemble R1 training dataset
# ============================================================

print("\n" + "=" * 135)
print("4. ASSEMBLING R1 TRAINING DATA")
print("=" * 135)

n_clean = len(
    clean_training
)

n_natural = len(
    natural_training
)

n_augmented = len(
    augmented_training
)

r1_training = pd.concat(
    [
        clean_training,
        natural_training,
        augmented_training
    ],
    ignore_index=True
)

expected_total = (
    n_clean
    + n_natural
    + n_augmented
)

print(
    f"Clean rows                : "
    f"{n_clean:,}"
)

print(
    f"Natural missing rows      : "
    f"{n_natural:,}"
)

print(
    f"Empirical augmented rows  : "
    f"{n_augmented:,}"
)

print(
    f"--------------------------------"
)

print(
    f"Expected total            : "
    f"{expected_total:,}"
)

print(
    f"Actual combined total     : "
    f"{len(r1_training):,}"
)

missing_input_rows = int(
    r1_training[
        FEATURES
    ]
    .isna()
    .any(axis=1)
    .sum()
)

print(
    f"\nRows with >=1 missing predictor: "
    f"{missing_input_rows:,}"
)

print(
    f"Missing-input prevalence       : "
    f"{100*missing_input_rows/len(r1_training):.4f}%"
)

print(
    f"Invalid targets                : "
    f"{r1_training[TARGET].isna().sum():,}"
)


# ============================================================
# 6. Free source DataFrames before LightGBM conversion
# ============================================================

del clean_training
del natural_training
del augmented_training

gc.collect()


# ============================================================
# 7. Construct LightGBM training matrices
# ============================================================

print("\n" + "=" * 135)
print("5. PREPARING LIGHTGBM DATASET")
print("=" * 135)

X_train = r1_training[
    FEATURES
]

y_train = r1_training[
    TARGET
]

print(
    f"X_train shape: "
    f"{X_train.shape}"
)

print(
    f"y_train shape: "
    f"{y_train.shape}"
)

train_dataset = lgb.Dataset(
    X_train,
    label=y_train,
    feature_name=FEATURES,
    free_raw_data=False
)


# ============================================================
# 8. EXACT fixed R0 architecture / hyperparameters
# ============================================================

R1_PARAMS = {
    "objective":
        "regression",

    "metric":
        [
            "l1",
            "rmse",
            "l2"
        ],

    "learning_rate":
        0.05,

    "num_leaves":
        31,

    "max_depth":
        -1,

    "min_data_in_leaf":
        100,

    "feature_fraction":
        0.9,

    "bagging_fraction":
        0.8,

    "bagging_freq":
        1,

    "lambda_l1":
        0,

    "lambda_l2":
        1,

    "seed":
        42,

    "feature_fraction_seed":
        42,

    "bagging_seed":
        42,

    "data_random_seed":
        42,

    "verbosity":
        -1,

    "num_threads":
        -1
}

NUM_BOOST_ROUND = 1200

print("\n" + "=" * 135)
print("6. R1 FIXED TRAINING CONFIGURATION")
print("=" * 135)

print(
    f"Boosting rounds : "
    f"{NUM_BOOST_ROUND}"
)

for k, v in R1_PARAMS.items():

    print(
        f"{k:<25}: {v}"
    )


# ============================================================
# 9. Train R1
# ============================================================

print("\n" + "=" * 135)
print("7. TRAINING R1 FAILURE-AWARE LIGHTGBM")
print("=" * 135)

training_start = time.time()

r1 = lgb.train(
    params=R1_PARAMS,
    train_set=train_dataset,
    num_boost_round=NUM_BOOST_ROUND
)

training_seconds = (
    time.time()
    - training_start
)

print(
    f"\n✅ R1 training completed"
)

print(
    f"Training time: "
    f"{training_seconds:.2f} seconds"
)

print(
    f"Trees: "
    f"{r1.num_trees():,}"
)


# ============================================================
# 10. Save model immediately
# ============================================================

MODEL_OUTPUT_PATH = os.path.join(
    NB5,
    "Models",
    "R1_FailureAware_LightGBM_1200.txt"
)

r1.save_model(
    MODEL_OUTPUT_PATH
)

print(
    f"\n✅ R1 model saved:\n"
    f"{MODEL_OUTPUT_PATH}"
)


# ============================================================
# 11. Release large training matrix before validation
# ============================================================

del train_dataset
del X_train
del y_train
del r1_training

gc.collect()


# ============================================================
# 12. FULL JULY clean evaluation
# ============================================================

print("\n" + "=" * 135)
print("8. R1 FULL JULY CLEAN EVALUATION")
print("=" * 135)

JULY_PATH = os.path.join(
    FORECASTING_PATH,
    "FORECASTING_2026-07.parquet"
)

july = pd.read_parquet(
    JULY_PATH,
    columns=FEATURES + [TARGET]
)

valid_target = (
    july[TARGET].notna()
    &
    np.isfinite(
        july[TARGET]
    )
    &
    (
        july[TARGET] >= 0
    )
)

clean_features = (
    july[FEATURES]
    .notna()
    .all(axis=1)
)

july_clean = july.loc[
    valid_target
    &
    clean_features
].copy()

X_july_clean = july_clean[
    FEATURES
]

y_july_clean = july_clean[
    TARGET
].to_numpy(
    dtype=np.float64
)

r1_full_july_pred = np.maximum(
    r1.predict(
        X_july_clean
    ),
    0
)

r1_full_july_metrics = (
    calculate_metrics(
        y_july_clean,
        r1_full_july_pred
    )
)

for k, v in r1_full_july_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )

del july
del july_clean
del X_july_clean

gc.collect()


# ============================================================
# 13. Controlled July clean + injected evaluation
# ============================================================

print("\n" + "=" * 135)
print("9. R1 CONTROLLED JULY ROBUSTNESS EVALUATION")
print("=" * 135)

controlled = pd.read_parquet(
    CONTROLLED_JULY_PATH
)

y_controlled = (
    controlled[
        TARGET
    ]
    .to_numpy(
        dtype=np.float64
    )
)

# CLEAN condition
X_controlled_clean = (
    controlled[
        FEATURES
    ]
)

r1_clean_pred = np.maximum(
    r1.predict(
        X_controlled_clean
    ),
    0
)

r1_clean_metrics = (
    calculate_metrics(
        y_controlled,
        r1_clean_pred
    )
)


# FAILURE-INJECTED condition
controlled_injected = (
    controlled[
        FEATURES
    ].copy()
)

for lag in LAG_COLS:

    controlled_injected[
        lag
    ] = controlled[
        f"{lag}_Injected"
    ]

r1_injected_pred = np.maximum(
    r1.predict(
        controlled_injected
    ),
    0
)

r1_injected_metrics = (
    calculate_metrics(
        y_controlled,
        r1_injected_pred
    )
)


print("\nCLEAN CONDITION")
print("-" * 60)

for k, v in r1_clean_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )


print("\nFAILURE-INJECTED CONDITION")
print("-" * 60)

for k, v in r1_injected_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )


# ============================================================
# 14. R1 degradation
# ============================================================

r1_degradation = {
    "MAE_Delta":
        (
            r1_injected_metrics["MAE"]
            -
            r1_clean_metrics["MAE"]
        ),

    "RMSE_Delta":
        (
            r1_injected_metrics["RMSE"]
            -
            r1_clean_metrics["RMSE"]
        ),

    "WAPE_Delta_pp":
        (
            r1_injected_metrics["WAPE"]
            -
            r1_clean_metrics["WAPE"]
        ),

    "WAPE_Degradation_Percent":
        (
            100
            *
            (
                r1_injected_metrics["WAPE"]
                -
                r1_clean_metrics["WAPE"]
            )
            /
            r1_clean_metrics["WAPE"]
        ),

    "Bias_Shift":
        (
            r1_injected_metrics["Bias"]
            -
            r1_clean_metrics["Bias"]
        )
}


print("\n" + "=" * 135)
print("10. R1 ROBUSTNESS DEGRADATION")
print("=" * 135)

for k, v in r1_degradation.items():

    print(
        f"{k:<28}: "
        f"{v:.6f}"
    )


# ============================================================
# 15. Load R0 metrics for direct comparison
# ============================================================

r0_table = pd.read_csv(
    R0_RESULT_PATH
)

r0_full = r0_table[
    r0_table["Condition"]
    ==
    "Full_July_Clean"
].iloc[0]

r0_clean = r0_table[
    r0_table["Condition"]
    ==
    "Controlled_July_Clean"
].iloc[0]

r0_injected = r0_table[
    r0_table["Condition"]
    ==
    "Controlled_July_Injected"
].iloc[0]


comparison = pd.DataFrame([
    {
        "Metric":
            "Full_July_Clean_WAPE",

        "R0":
            r0_full["WAPE"],

        "R1":
            r1_full_july_metrics["WAPE"]
    },

    {
        "Metric":
            "Controlled_Clean_WAPE",

        "R0":
            r0_clean["WAPE"],

        "R1":
            r1_clean_metrics["WAPE"]
    },

    {
        "Metric":
            "Controlled_Injected_WAPE",

        "R0":
            r0_injected["WAPE"],

        "R1":
            r1_injected_metrics["WAPE"]
    },

    {
        "Metric":
            "Controlled_Injected_MAE",

        "R0":
            r0_injected["MAE"],

        "R1":
            r1_injected_metrics["MAE"]
    },

    {
        "Metric":
            "Controlled_Injected_RMSE",

        "R0":
            r0_injected["RMSE"],

        "R1":
            r1_injected_metrics["RMSE"]
    },

    {
        "Metric":
            "Controlled_Injected_Bias",

        "R0":
            r0_injected["Bias"],

        "R1":
            r1_injected_metrics["Bias"]
    }
])


comparison[
    "R1_minus_R0"
] = (
    comparison["R1"]
    -
    comparison["R0"]
)

comparison[
    "Percent_Change_vs_R0"
] = (
    100
    *
    (
        comparison["R1"]
        -
        comparison["R0"]
    )
    /
    np.abs(
        comparison["R0"]
    )
)


print("\n" + "=" * 135)
print("11. R0 vs R1 — JULY COMPARISON")
print("=" * 135)

print(
    comparison
    .to_string(
        index=False
    )
)


# ============================================================
# 16. Save controlled July predictions
# ============================================================

prediction_table = pd.DataFrame({
    "July_Validation_Event_ID":
        controlled[
            "July_Validation_Event_ID"
        ].values,

    "Detector_Code":
        controlled[
            "Detector_Code"
        ].values,

    "Timestamp":
        controlled[
            "Timestamp"
        ].values,

    "Actual":
        y_controlled,

    "Missing_Lag_Count":
        controlled[
            "Missing_Lag_Count"
        ].values,

    "Missing_Lag_Pattern":
        controlled[
            "Missing_Lag_Pattern"
        ].values,

    "R1_Clean_Prediction":
        r1_clean_pred,

    "R1_Injected_Prediction":
        r1_injected_pred
})


# ============================================================
# 17. Save results
# ============================================================

performance_table = pd.DataFrame([
    {
        "Model":
            "R1_FailureAware",

        "Condition":
            "Full_July_Clean",

        **r1_full_july_metrics
    },

    {
        "Model":
            "R1_FailureAware",

        "Condition":
            "Controlled_July_Clean",

        **r1_clean_metrics
    },

    {
        "Model":
            "R1_FailureAware",

        "Condition":
            "Controlled_July_Injected",

        **r1_injected_metrics
    }
])


performance_path = os.path.join(
    NB5,
    "Tables",
    "Table_R1_July_Performance.csv"
)

comparison_path = os.path.join(
    NB5,
    "Tables",
    "Table_R0_vs_R1_July_Comparison.csv"
)

prediction_path = os.path.join(
    NB5,
    "Data",
    "R1_Controlled_July_Predictions.parquet"
)

metadata_path = os.path.join(
    NB5,
    "Results",
    "R1_FailureAware_Training_Metadata.json"
)

training_audit_path = os.path.join(
    NB5,
    "Tables",
    "Table_R1_Training_Composition.csv"
)


performance_table.to_csv(
    performance_path,
    index=False
)

comparison.to_csv(
    comparison_path,
    index=False
)

prediction_table.to_parquet(
    prediction_path,
    index=False
)


training_composition = pd.DataFrame([
    {
        "Training_Source":
            "Clean_Model_Ready",

        "Rows":
            n_clean
    },

    {
        "Training_Source":
            "Natural_Missing_History",

        "Rows":
            n_natural
    },

    {
        "Training_Source":
            "Empirical_Failure_Augmentation",

        "Rows":
            n_augmented
    }
])

training_composition[
    "Percent"
] = (
    100
    *
    training_composition["Rows"]
    /
    training_composition["Rows"].sum()
)

training_composition.to_csv(
    training_audit_path,
    index=False
)


metadata = {
    "regime":
        "R1_FailureAware",

    "training_period":
        "2026-01-01 to 2026-06-30",

    "validation_period":
        "2026-07-01 to 2026-07-31",

    "august_used_for_model_selection":
        False,

    "features":
        FEATURES,

    "training_rows": {
        "clean":
            int(n_clean),

        "natural_missing":
            int(n_natural),

        "empirical_augmented":
            int(n_augmented),

        "total":
            int(expected_total),

        "rows_with_missing_predictors":
            int(missing_input_rows)
    },

    "parameters":
        R1_PARAMS,

    "num_boost_round":
        NUM_BOOST_ROUND,

    "training_seconds":
        float(training_seconds),

    "full_july_clean_metrics":
        r1_full_july_metrics,

    "controlled_july_clean_metrics":
        r1_clean_metrics,

    "controlled_july_injected_metrics":
        r1_injected_metrics,

    "controlled_july_degradation":
        {
            k: float(v)
            for k, v
            in r1_degradation.items()
        }
}


with open(
    metadata_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )


print("\n" + "=" * 135)
print("12. SAVED OUTPUTS")
print("=" * 135)

print(f"✅ {MODEL_OUTPUT_PATH}")
print(f"✅ {performance_path}")
print(f"✅ {comparison_path}")
print(f"✅ {prediction_path}")
print(f"✅ {training_audit_path}")
print(f"✅ {metadata_path}")

print("\n" + "=" * 135)
print("STEP 18 COMPLETED")
print("=" * 135)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 19: Train R2 — Failure-Aware LightGBM
#          + Explicit Missingness Indicators
# ============================================================

import os
import gc
import json
import time
import numpy as np
import pandas as pd
import lightgbm as lgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

NATURAL_PATH = os.path.join(
    NB5,
    "Data",
    "Natural_Failure_Aware_Training_Candidates.parquet"
)

AUGMENTED_PATH = os.path.join(
    NB5,
    "Data",
    "Empirical_Failure_Augmented_Training_Rows.parquet"
)

CONTROLLED_JULY_PATH = os.path.join(
    NB5,
    "Data",
    "Controlled_July_Clean_Injected_Validation_Rows.parquet"
)

R0_PATH = os.path.join(
    NB5,
    "Tables",
    "Table_R0_July_Baseline_Performance.csv"
)

R1_PATH = os.path.join(
    NB5,
    "Tables",
    "Table_R1_July_Performance.csv"
)

# ------------------------------------------------------------
# Base features
# ------------------------------------------------------------

BASE_FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

LAG_COLS = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

INDICATOR_COLS = [
    f"Missing_{lag}"
    for lag in LAG_COLS
]

FEATURES_R2 = (
    BASE_FEATURES
    + INDICATOR_COLS
)

TARGET = "Target_Volume"

TRAIN_MONTHS = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06"
]

print("=" * 135)
print("NOTEBOOK 5 — R2 FAILURE-AWARE + MISSINGNESS INDICATORS")
print("=" * 135)

print(
    f"\nTotal R2 features: "
    f"{len(FEATURES_R2)}"
)

for i, feature in enumerate(
    FEATURES_R2,
    start=1
):
    print(
        f"{i:>2}. {feature}"
    )


# ============================================================
# 1. Helper functions
# ============================================================

def add_missing_indicators(df):

    for lag in LAG_COLS:

        df[
            f"Missing_{lag}"
        ] = (
            df[lag]
            .isna()
            .astype("int8")
        )

    return df


def downcast_frame(df):

    for col in [
        "Detector_Code",
        "Interval_Index",
        "DayOfWeek",
        "IsWeekend"
    ]:

        df[col] = pd.to_numeric(
            df[col],
            downcast="integer"
        )

    for lag in LAG_COLS:

        df[lag] = df[lag].astype(
            "float32"
        )

    for col in INDICATOR_COLS:

        df[col] = df[col].astype(
            "int8"
        )

    df[TARGET] = df[TARGET].astype(
        "float32"
    )

    return df


def calculate_metrics(
    y_true,
    y_pred
):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    denominator = np.sum(
        np.abs(y_true)
    )

    wape = (
        100
        * np.sum(
            np.abs(
                y_true - y_pred
            )
        )
        / denominator
    )

    bias = np.mean(
        y_pred - y_true
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "WAPE": float(wape),
        "Bias": float(bias)
    }


# ============================================================
# 2. Load CLEAN Jan-Jun rows
#    All missingness indicators = 0
# ============================================================

print("\n" + "=" * 135)
print("1. LOADING CLEAN JAN–JUN TRAINING ROWS")
print("=" * 135)

clean_parts = []

for month in TRAIN_MONTHS:

    path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    df = pd.read_parquet(
        path,
        columns=BASE_FEATURES + [TARGET]
    )

    valid_target = (
        df[TARGET].notna()
        &
        np.isfinite(
            df[TARGET]
        )
        &
        (
            df[TARGET] >= 0
        )
    )

    clean_features = (
        df[BASE_FEATURES]
        .notna()
        .all(axis=1)
    )

    clean = df.loc[
        valid_target
        &
        clean_features,
        BASE_FEATURES + [TARGET]
    ].copy()

    clean = add_missing_indicators(
        clean
    )

    clean = downcast_frame(
        clean
    )

    print(
        f"{month}: "
        f"{len(clean):,} rows"
    )

    clean_parts.append(
        clean
    )

    del df, clean
    gc.collect()


clean_training = pd.concat(
    clean_parts,
    ignore_index=True
)

del clean_parts
gc.collect()

n_clean = len(
    clean_training
)

print(
    f"\nTotal clean rows: "
    f"{n_clean:,}"
)


# ============================================================
# 3. Load NATURAL missing-history rows
# ============================================================

print("\n" + "=" * 135)
print("2. LOADING NATURAL MISSING-HISTORY ROWS")
print("=" * 135)

natural_training = pd.read_parquet(
    NATURAL_PATH,
    columns=BASE_FEATURES + [TARGET]
)

natural_training = add_missing_indicators(
    natural_training
)

natural_training = downcast_frame(
    natural_training
)

n_natural = len(
    natural_training
)

print(
    f"Natural missing rows: "
    f"{n_natural:,}"
)

print("\nIndicator counts:")

for indicator in INDICATOR_COLS:

    print(
        f"{indicator:<18}: "
        f"{natural_training[indicator].sum():,}"
    )


# ============================================================
# 4. Load empirical failure-augmented rows
# ============================================================

print("\n" + "=" * 135)
print("3. LOADING EMPIRICAL FAILURE-AUGMENTED ROWS")
print("=" * 135)

augmented_training = pd.read_parquet(
    AUGMENTED_PATH,
    columns=BASE_FEATURES + [TARGET]
)

augmented_training = add_missing_indicators(
    augmented_training
)

augmented_training = downcast_frame(
    augmented_training
)

n_augmented = len(
    augmented_training
)

print(
    f"Empirical augmented rows: "
    f"{n_augmented:,}"
)

print("\nIndicator counts:")

for indicator in INDICATOR_COLS:

    print(
        f"{indicator:<18}: "
        f"{augmented_training[indicator].sum():,}"
    )


# ============================================================
# 5. Assemble R2 training dataset
# ============================================================

print("\n" + "=" * 135)
print("4. ASSEMBLING R2 TRAINING DATA")
print("=" * 135)

r2_training = pd.concat(
    [
        clean_training,
        natural_training,
        augmented_training
    ],
    ignore_index=True
)

expected_total = (
    n_clean
    + n_natural
    + n_augmented
)

print(
    f"Clean rows               : "
    f"{n_clean:,}"
)

print(
    f"Natural missing rows     : "
    f"{n_natural:,}"
)

print(
    f"Empirical augmented rows : "
    f"{n_augmented:,}"
)

print(
    f"Expected total           : "
    f"{expected_total:,}"
)

print(
    f"Actual total             : "
    f"{len(r2_training):,}"
)

print(
    f"Rows with missing lag    : "
    f"{r2_training[LAG_COLS].isna().any(axis=1).sum():,}"
)

print(
    f"Invalid targets          : "
    f"{r2_training[TARGET].isna().sum():,}"
)


# ============================================================
# 6. Validate missingness indicators
# ============================================================

print("\n" + "=" * 135)
print("5. MISSINGNESS INDICATOR VALIDATION")
print("=" * 135)

indicator_validation = []

for lag in LAG_COLS:

    indicator = f"Missing_{lag}"

    actual_missing = (
        r2_training[
            lag
        ].isna()
    )

    indicator_missing = (
        r2_training[
            indicator
        ] == 1
    )

    exact = bool(
        (
            actual_missing
            ==
            indicator_missing
        ).all()
    )

    indicator_validation.append({
        "Lag":
            lag,

        "Actual_Missing":
            int(
                actual_missing.sum()
            ),

        "Indicator_One":
            int(
                indicator_missing.sum()
            ),

        "Exact_Match":
            exact
    })


indicator_validation_df = (
    pd.DataFrame(
        indicator_validation
    )
)

print(
    indicator_validation_df
    .to_string(index=False)
)

indicators_exact = bool(
    indicator_validation_df[
        "Exact_Match"
    ].all()
)

print(
    f"\nAll indicators exact: "
    f"{indicators_exact}"
)


# ============================================================
# 7. Free source frames
# ============================================================

del clean_training
del natural_training
del augmented_training

gc.collect()


# ============================================================
# 8. Prepare LightGBM dataset
# ============================================================

print("\n" + "=" * 135)
print("6. PREPARING LIGHTGBM DATASET")
print("=" * 135)

X_train = r2_training[
    FEATURES_R2
]

y_train = r2_training[
    TARGET
]

print(
    f"X_train shape: "
    f"{X_train.shape}"
)

print(
    f"y_train shape: "
    f"{y_train.shape}"
)

train_dataset = lgb.Dataset(
    X_train,
    label=y_train,
    feature_name=FEATURES_R2,
    free_raw_data=False
)


# ============================================================
# 9. Fixed hyperparameters
#    Same architecture as R0/R1
# ============================================================

R2_PARAMS = {
    "objective":
        "regression",

    "metric":
        [
            "l1",
            "rmse",
            "l2"
        ],

    "learning_rate":
        0.05,

    "num_leaves":
        31,

    "max_depth":
        -1,

    "min_data_in_leaf":
        100,

    "feature_fraction":
        0.9,

    "bagging_fraction":
        0.8,

    "bagging_freq":
        1,

    "lambda_l1":
        0,

    "lambda_l2":
        1,

    "seed":
        42,

    "feature_fraction_seed":
        42,

    "bagging_seed":
        42,

    "data_random_seed":
        42,

    "verbosity":
        -1,

    "num_threads":
        -1
}

NUM_BOOST_ROUND = 1200


# ============================================================
# 10. Visible training heartbeat callback
# ============================================================

training_start = time.time()


def training_progress(period=50):

    def _callback(env):

        iteration = (
            env.iteration + 1
        )

        if (
            iteration == 1
            or iteration % period == 0
            or iteration == env.end_iteration
        ):

            elapsed = (
                time.time()
                - training_start
            )

            print(
                f"[R2 TRAINING] "
                f"Iteration "
                f"{iteration:,} / "
                f"{env.end_iteration:,} "
                f"| Elapsed: "
                f"{elapsed:.1f} sec",
                flush=True
            )

    _callback.order = 10

    return _callback


print("\n" + "=" * 135)
print("7. TRAINING R2 FAILURE-AWARE + INDICATORS")
print("=" * 135)

print(
    "Progress will be displayed every "
    "50 iterations."
)

print(
    f"Boosting rounds: "
    f"{NUM_BOOST_ROUND:,}"
)


# ============================================================
# 11. Train R2
# ============================================================

r2 = lgb.train(
    params=R2_PARAMS,
    train_set=train_dataset,
    num_boost_round=NUM_BOOST_ROUND,
    callbacks=[
        training_progress(
            period=50
        )
    ]
)

training_seconds = (
    time.time()
    - training_start
)

print(
    "\n✅ R2 training completed"
)

print(
    f"Training time: "
    f"{training_seconds:.2f} seconds"
)

print(
    f"Trees: "
    f"{r2.num_trees():,}"
)


# ============================================================
# 12. Save model immediately
# ============================================================

MODEL_OUTPUT = os.path.join(
    NB5,
    "Models",
    "R2_FailureAware_MissingIndicators_LightGBM_1200.txt"
)

r2.save_model(
    MODEL_OUTPUT
)

print(
    f"\n✅ R2 model saved:\n"
    f"{MODEL_OUTPUT}"
)


# ============================================================
# 13. Release training matrix
# ============================================================

del train_dataset
del X_train
del y_train
del r2_training

gc.collect()


# ============================================================
# 14. FULL JULY CLEAN evaluation
#
# Clean July rows:
# all missingness indicators = 0
# ============================================================

print("\n" + "=" * 135)
print("8. R2 FULL JULY CLEAN EVALUATION")
print("=" * 135)

JULY_PATH = os.path.join(
    FORECASTING_PATH,
    "FORECASTING_2026-07.parquet"
)

july = pd.read_parquet(
    JULY_PATH,
    columns=BASE_FEATURES + [TARGET]
)

valid_target = (
    july[TARGET].notna()
    &
    np.isfinite(
        july[TARGET]
    )
    &
    (
        july[TARGET] >= 0
    )
)

clean_mask = (
    july[
        BASE_FEATURES
    ]
    .notna()
    .all(axis=1)
)

july_clean = july.loc[
    valid_target
    &
    clean_mask,
    BASE_FEATURES + [TARGET]
].copy()

for indicator in INDICATOR_COLS:

    july_clean[
        indicator
    ] = np.int8(0)


y_july = july_clean[
    TARGET
].to_numpy(
    dtype=np.float64
)

r2_full_july_pred = np.maximum(
    r2.predict(
        july_clean[
            FEATURES_R2
        ]
    ),
    0
)

r2_full_july_metrics = (
    calculate_metrics(
        y_july,
        r2_full_july_pred
    )
)

for k, v in r2_full_july_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )

del july
del july_clean

gc.collect()


# ============================================================
# 15. Controlled July evaluation
# ============================================================

print("\n" + "=" * 135)
print("9. R2 CONTROLLED JULY ROBUSTNESS EVALUATION")
print("=" * 135)

controlled = pd.read_parquet(
    CONTROLLED_JULY_PATH
)

y_controlled = controlled[
    TARGET
].to_numpy(
    dtype=np.float64
)


# ------------------------------------------------------------
# CLEAN condition
# ------------------------------------------------------------

controlled_clean = controlled[
    BASE_FEATURES
].copy()

for indicator in INDICATOR_COLS:

    controlled_clean[
        indicator
    ] = np.int8(0)


r2_clean_pred = np.maximum(
    r2.predict(
        controlled_clean[
            FEATURES_R2
        ]
    ),
    0
)

r2_clean_metrics = (
    calculate_metrics(
        y_controlled,
        r2_clean_pred
    )
)


# ------------------------------------------------------------
# FAILURE-INJECTED condition
# ------------------------------------------------------------

controlled_injected = (
    controlled[
        BASE_FEATURES
    ].copy()
)

for lag in LAG_COLS:

    controlled_injected[
        lag
    ] = controlled[
        f"{lag}_Injected"
    ]

    controlled_injected[
        f"Missing_{lag}"
    ] = (
        controlled_injected[
            lag
        ]
        .isna()
        .astype("int8")
    )


r2_injected_pred = np.maximum(
    r2.predict(
        controlled_injected[
            FEATURES_R2
        ]
    ),
    0
)

r2_injected_metrics = (
    calculate_metrics(
        y_controlled,
        r2_injected_pred
    )
)


print("\nCLEAN CONDITION")
print("-" * 60)

for k, v in r2_clean_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )


print("\nFAILURE-INJECTED CONDITION")
print("-" * 60)

for k, v in r2_injected_metrics.items():

    print(
        f"{k:<10}: "
        f"{v:.6f}"
    )


# ============================================================
# 16. Robustness degradation
# ============================================================

r2_degradation = {
    "MAE_Delta":
        (
            r2_injected_metrics["MAE"]
            -
            r2_clean_metrics["MAE"]
        ),

    "RMSE_Delta":
        (
            r2_injected_metrics["RMSE"]
            -
            r2_clean_metrics["RMSE"]
        ),

    "WAPE_Delta_pp":
        (
            r2_injected_metrics["WAPE"]
            -
            r2_clean_metrics["WAPE"]
        ),

    "WAPE_Degradation_Percent":
        (
            100
            *
            (
                r2_injected_metrics["WAPE"]
                -
                r2_clean_metrics["WAPE"]
            )
            /
            r2_clean_metrics["WAPE"]
        ),

    "Bias_Shift":
        (
            r2_injected_metrics["Bias"]
            -
            r2_clean_metrics["Bias"]
        )
}


print("\n" + "=" * 135)
print("10. R2 ROBUSTNESS DEGRADATION")
print("=" * 135)

for k, v in r2_degradation.items():

    print(
        f"{k:<28}: "
        f"{v:.6f}"
    )


# ============================================================
# 17. R0 / R1 / R2 comparison
# ============================================================

r0 = pd.read_csv(
    R0_PATH
)

r1 = pd.read_csv(
    R1_PATH
)


def get_condition(
    df,
    condition
):

    return df[
        df["Condition"]
        ==
        condition
    ].iloc[0]


r0_full = get_condition(
    r0,
    "Full_July_Clean"
)

r0_clean = get_condition(
    r0,
    "Controlled_July_Clean"
)

r0_injected = get_condition(
    r0,
    "Controlled_July_Injected"
)

r1_full = get_condition(
    r1,
    "Full_July_Clean"
)

r1_clean = get_condition(
    r1,
    "Controlled_July_Clean"
)

r1_injected = get_condition(
    r1,
    "Controlled_July_Injected"
)


comparison = pd.DataFrame([
    {
        "Model":
            "R0_CleanOnly",

        "Full_July_Clean_WAPE":
            r0_full["WAPE"],

        "Controlled_Clean_WAPE":
            r0_clean["WAPE"],

        "Controlled_Injected_WAPE":
            r0_injected["WAPE"],

        "Injected_MAE":
            r0_injected["MAE"],

        "Injected_RMSE":
            r0_injected["RMSE"],

        "Injected_Bias":
            r0_injected["Bias"]
    },

    {
        "Model":
            "R1_FailureAware",

        "Full_July_Clean_WAPE":
            r1_full["WAPE"],

        "Controlled_Clean_WAPE":
            r1_clean["WAPE"],

        "Controlled_Injected_WAPE":
            r1_injected["WAPE"],

        "Injected_MAE":
            r1_injected["MAE"],

        "Injected_RMSE":
            r1_injected["RMSE"],

        "Injected_Bias":
            r1_injected["Bias"]
    },

    {
        "Model":
            "R2_FailureAware_Indicators",

        "Full_July_Clean_WAPE":
            r2_full_july_metrics["WAPE"],

        "Controlled_Clean_WAPE":
            r2_clean_metrics["WAPE"],

        "Controlled_Injected_WAPE":
            r2_injected_metrics["WAPE"],

        "Injected_MAE":
            r2_injected_metrics["MAE"],

        "Injected_RMSE":
            r2_injected_metrics["RMSE"],

        "Injected_Bias":
            r2_injected_metrics["Bias"]
    }
])


comparison[
    "Clean_WAPE_Change_vs_R0_pp"
] = (
    comparison[
        "Full_July_Clean_WAPE"
    ]
    -
    float(
        r0_full["WAPE"]
    )
)

comparison[
    "Injected_WAPE_Improvement_vs_R0_Percent"
] = (
    100
    *
    (
        float(
            r0_injected["WAPE"]
        )
        -
        comparison[
            "Controlled_Injected_WAPE"
        ]
    )
    /
    float(
        r0_injected["WAPE"]
    )
)


print("\n" + "=" * 135)
print("11. R0 vs R1 vs R2 — JULY MODEL-SELECTION TABLE")
print("=" * 135)

print(
    comparison
    .to_string(
        index=False
    )
)


# ============================================================
# 18. Save controlled predictions
# ============================================================

prediction_table = pd.DataFrame({
    "July_Validation_Event_ID":
        controlled[
            "July_Validation_Event_ID"
        ].values,

    "Detector_Code":
        controlled[
            "Detector_Code"
        ].values,

    "Timestamp":
        controlled[
            "Timestamp"
        ].values,

    "Actual":
        y_controlled,

    "Missing_Lag_Count":
        controlled[
            "Missing_Lag_Count"
        ].values,

    "Missing_Lag_Pattern":
        controlled[
            "Missing_Lag_Pattern"
        ].values,

    "R2_Clean_Prediction":
        r2_clean_pred,

    "R2_Injected_Prediction":
        r2_injected_pred
})


# ============================================================
# 19. Save results
# ============================================================

performance_table = pd.DataFrame([
    {
        "Model":
            "R2_FailureAware_Indicators",

        "Condition":
            "Full_July_Clean",

        **r2_full_july_metrics
    },

    {
        "Model":
            "R2_FailureAware_Indicators",

        "Condition":
            "Controlled_July_Clean",

        **r2_clean_metrics
    },

    {
        "Model":
            "R2_FailureAware_Indicators",

        "Condition":
            "Controlled_July_Injected",

        **r2_injected_metrics
    }
])


performance_path = os.path.join(
    NB5,
    "Tables",
    "Table_R2_July_Performance.csv"
)

comparison_path = os.path.join(
    NB5,
    "Tables",
    "Table_R0_R1_R2_July_Model_Selection.csv"
)

prediction_path = os.path.join(
    NB5,
    "Data",
    "R2_Controlled_July_Predictions.parquet"
)

indicator_audit_path = os.path.join(
    NB5,
    "Tables",
    "Table_R2_Missingness_Indicator_Validation.csv"
)

metadata_path = os.path.join(
    NB5,
    "Results",
    "R2_FailureAware_Indicators_Training_Metadata.json"
)


performance_table.to_csv(
    performance_path,
    index=False
)

comparison.to_csv(
    comparison_path,
    index=False
)

prediction_table.to_parquet(
    prediction_path,
    index=False
)

indicator_validation_df.to_csv(
    indicator_audit_path,
    index=False
)


metadata = {
    "regime":
        "R2_FailureAware_MissingIndicators",

    "training_period":
        "2026-01-01 to 2026-06-30",

    "validation_period":
        "2026-07-01 to 2026-07-31",

    "august_used_for_model_selection":
        False,

    "base_features":
        BASE_FEATURES,

    "missingness_indicators":
        INDICATOR_COLS,

    "total_features":
        FEATURES_R2,

    "training_rows": {
        "clean":
            int(n_clean),

        "natural_missing":
            int(n_natural),

        "empirical_augmented":
            int(n_augmented),

        "total":
            int(expected_total)
    },

    "parameters":
        R2_PARAMS,

    "num_boost_round":
        NUM_BOOST_ROUND,

    "training_seconds":
        float(
            training_seconds
        ),

    "full_july_clean_metrics":
        r2_full_july_metrics,

    "controlled_july_clean_metrics":
        r2_clean_metrics,

    "controlled_july_injected_metrics":
        r2_injected_metrics,

    "controlled_july_degradation":
        {
            k: float(v)
            for k, v
            in r2_degradation.items()
        },

    "indicator_validation_pass":
        bool(
            indicators_exact
        )
}


with open(
    metadata_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )


print("\n" + "=" * 135)
print("12. SAVED OUTPUTS")
print("=" * 135)

print(f"✅ {MODEL_OUTPUT}")
print(f"✅ {performance_path}")
print(f"✅ {comparison_path}")
print(f"✅ {prediction_path}")
print(f"✅ {indicator_audit_path}")
print(f"✅ {metadata_path}")

print("\n" + "=" * 135)
print("STEP 19 COMPLETED")
print("=" * 135)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 20: Final July model selection and statistical check
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from scipy.stats import wilcoxon

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

SELECTION_TABLE_PATH = os.path.join(
    NB5,
    "Tables",
    "Table_R0_R1_R2_July_Model_Selection.csv"
)

R1_PRED_PATH = os.path.join(
    NB5,
    "Data",
    "R1_Controlled_July_Predictions.parquet"
)

R2_PRED_PATH = os.path.join(
    NB5,
    "Data",
    "R2_Controlled_July_Predictions.parquet"
)

print("=" * 135)
print("NOTEBOOK 5 — FINAL JULY MODEL SELECTION")
print("=" * 135)


# ============================================================
# 1. Load overall July model-selection results
# ============================================================

selection = pd.read_csv(
    SELECTION_TABLE_PATH
)

print("\n" + "=" * 135)
print("1. OVERALL JULY RESULTS")
print("=" * 135)

print(
    selection.to_string(index=False)
)


# ============================================================
# 2. Identify primary-metric winner
#
# Primary model-selection metric:
# Controlled failure-injected WAPE
#
# Clean-data performance must remain comparable to R0.
# ============================================================

winner_row = selection.loc[
    selection[
        "Controlled_Injected_WAPE"
    ].idxmin()
]

primary_winner = (
    winner_row["Model"]
)

print("\n" + "=" * 135)
print("2. PRIMARY MODEL-SELECTION RESULT")
print("=" * 135)

print(
    f"Lowest controlled injected WAPE: "
    f"{primary_winner}"
)

print(
    f"WAPE = "
    f"{winner_row['Controlled_Injected_WAPE']:.6f}%"
)


# ============================================================
# 3. Load R1 and R2 controlled July predictions
# ============================================================

r1 = pd.read_parquet(
    R1_PRED_PATH
)

r2 = pd.read_parquet(
    R2_PRED_PATH
)

keys = [
    "July_Validation_Event_ID",
    "Detector_Code",
    "Timestamp"
]

r1["Timestamp"] = pd.to_datetime(
    r1["Timestamp"]
)

r2["Timestamp"] = pd.to_datetime(
    r2["Timestamp"]
)

paired = r1[
    keys
    +
    [
        "Actual",
        "R1_Clean_Prediction",
        "R1_Injected_Prediction"
    ]
].merge(
    r2[
        keys
        +
        [
            "R2_Clean_Prediction",
            "R2_Injected_Prediction"
        ]
    ],
    on=keys,
    how="inner",
    validate="one_to_one"
)

print("\n" + "=" * 135)
print("3. PAIRED JULY CONTROLLED PREDICTIONS")
print("=" * 135)

print(
    f"R1 rows       : {len(r1):,}"
)

print(
    f"R2 rows       : {len(r2):,}"
)

print(
    f"Matched rows  : {len(paired):,}"
)

print(
    f"Events        : "
    f"{paired['July_Validation_Event_ID'].nunique():,}"
)


# ============================================================
# 4. Event-level injected MAE
# ============================================================

paired[
    "R1_Injected_AE"
] = np.abs(
    paired[
        "Actual"
    ]
    -
    paired[
        "R1_Injected_Prediction"
    ]
)

paired[
    "R2_Injected_AE"
] = np.abs(
    paired[
        "Actual"
    ]
    -
    paired[
        "R2_Injected_Prediction"
    ]
)

event_level = (
    paired
    .groupby(
        "July_Validation_Event_ID",
        as_index=False
    )
    .agg(
        Detector_Code=(
            "Detector_Code",
            "first"
        ),

        Rows=(
            "Timestamp",
            "size"
        ),

        R1_Injected_MAE=(
            "R1_Injected_AE",
            "mean"
        ),

        R2_Injected_MAE=(
            "R2_Injected_AE",
            "mean"
        )
    )
)

event_level[
    "R2_minus_R1_MAE"
] = (
    event_level[
        "R2_Injected_MAE"
    ]
    -
    event_level[
        "R1_Injected_MAE"
    ]
)

r1_better_events = int(
    (
        event_level[
            "R2_minus_R1_MAE"
        ] > 0
    ).sum()
)

r2_better_events = int(
    (
        event_level[
            "R2_minus_R1_MAE"
        ] < 0
    ).sum()
)

tie_events = int(
    (
        np.isclose(
            event_level[
                "R2_minus_R1_MAE"
            ],
            0,
            atol=1e-12
        )
    ).sum()
)

mean_difference = float(
    event_level[
        "R2_minus_R1_MAE"
    ].mean()
)

median_difference = float(
    event_level[
        "R2_minus_R1_MAE"
    ].median()
)

print("\n" + "=" * 135)
print("4. EVENT-LEVEL R1 vs R2")
print("=" * 135)

print(
    f"R1 lower MAE events : "
    f"{r1_better_events:,} / {len(event_level):,}"
)

print(
    f"R2 lower MAE events : "
    f"{r2_better_events:,} / {len(event_level):,}"
)

print(
    f"Ties                : "
    f"{tie_events:,}"
)

print(
    f"\nMean (R2 - R1) event MAE : "
    f"{mean_difference:.6f}"
)

print(
    f"Median (R2 - R1) event MAE: "
    f"{median_difference:.6f}"
)


# ============================================================
# 5. Bootstrap confidence interval
#    for mean event-level R2 - R1 MAE
# ============================================================

RANDOM_SEED = 42
N_BOOT = 10000

rng = np.random.default_rng(
    RANDOM_SEED
)

diffs = event_level[
    "R2_minus_R1_MAE"
].to_numpy()

bootstrap_means = np.empty(
    N_BOOT,
    dtype=np.float64
)

for i in range(
    N_BOOT
):

    sample = rng.choice(
        diffs,
        size=len(diffs),
        replace=True
    )

    bootstrap_means[i] = (
        sample.mean()
    )


ci_low = float(
    np.percentile(
        bootstrap_means,
        2.5
    )
)

ci_high = float(
    np.percentile(
        bootstrap_means,
        97.5
    )
)

print("\n" + "=" * 135)
print("5. BOOTSTRAP EVENT-LEVEL DIFFERENCE")
print("=" * 135)

print(
    f"Mean R2 - R1 MAE : "
    f"{mean_difference:.6f}"
)

print(
    f"95% Bootstrap CI : "
    f"[{ci_low:.6f}, {ci_high:.6f}]"
)


# ============================================================
# 6. Paired Wilcoxon test
# ============================================================

nonzero_diffs = diffs[
    ~np.isclose(
        diffs,
        0,
        atol=1e-12
    )
]

if len(
    nonzero_diffs
) > 0:

    wilcox = wilcoxon(
        nonzero_diffs,
        alternative="two-sided",
        zero_method="wilcox"
    )

    wilcoxon_stat = float(
        wilcox.statistic
    )

    wilcoxon_p = float(
        wilcox.pvalue
    )

else:

    wilcoxon_stat = np.nan
    wilcoxon_p = np.nan


print("\n" + "=" * 135)
print("6. PAIRED WILCOXON TEST")
print("=" * 135)

print(
    f"Statistic : "
    f"{wilcoxon_stat}"
)

print(
    f"p-value   : "
    f"{wilcoxon_p:.10g}"
)


# ============================================================
# 7. Direct R1 vs R2 overall differences
# ============================================================

r1_row = selection[
    selection[
        "Model"
    ] == "R1_FailureAware"
].iloc[0]

r2_row = selection[
    selection[
        "Model"
    ] == "R2_FailureAware_Indicators"
].iloc[0]

direct_comparison = pd.DataFrame([
    {
        "Metric":
            "Full_July_Clean_WAPE",

        "R1":
            r1_row[
                "Full_July_Clean_WAPE"
            ],

        "R2":
            r2_row[
                "Full_July_Clean_WAPE"
            ]
    },

    {
        "Metric":
            "Controlled_Injected_WAPE",

        "R1":
            r1_row[
                "Controlled_Injected_WAPE"
            ],

        "R2":
            r2_row[
                "Controlled_Injected_WAPE"
            ]
    },

    {
        "Metric":
            "Injected_MAE",

        "R1":
            r1_row[
                "Injected_MAE"
            ],

        "R2":
            r2_row[
                "Injected_MAE"
            ]
    },

    {
        "Metric":
            "Injected_RMSE",

        "R1":
            r1_row[
                "Injected_RMSE"
            ],

        "R2":
            r2_row[
                "Injected_RMSE"
            ]
    },

    {
        "Metric":
            "Absolute_Injected_Bias",

        "R1":
            abs(
                r1_row[
                    "Injected_Bias"
                ]
            ),

        "R2":
            abs(
                r2_row[
                    "Injected_Bias"
                ]
            )
    }
])

direct_comparison[
    "R2_minus_R1"
] = (
    direct_comparison[
        "R2"
    ]
    -
    direct_comparison[
        "R1"
    ]
)

print("\n" + "=" * 135)
print("7. DIRECT R1 vs R2 COMPARISON")
print("=" * 135)

print(
    direct_comparison
    .to_string(
        index=False
    )
)


# ============================================================
# 8. FINAL MODEL-SELECTION DECISION
# ============================================================

# R1 wins the primary metric and is also the simpler regime.
# R2 adds five indicator features but does not improve the
# primary failure-injected WAPE.

SELECTED_REGIME = (
    "R1_FailureAware"
)

SELECTED_FEATURE_COUNT = 9

SELECTION_REASON = (
    "R1 achieved the lowest controlled failure-injected "
    "WAPE while preserving clean-data accuracy. "
    "R2 added five explicit missingness indicators but "
    "did not improve the primary robustness metric. "
    "R1 was therefore selected based on predictive "
    "performance and parsimony."
)

print("\n" + "=" * 135)
print("8. FINAL JULY MODEL-SELECTION DECISION")
print("=" * 135)

print(
    f"SELECTED REGIME: "
    f"{SELECTED_REGIME}"
)

print(
    f"Feature count  : "
    f"{SELECTED_FEATURE_COUNT}"
)

print(
    "\nReason:"
)

print(
    SELECTION_REASON
)

print(
    "\nIMPORTANT:"
)

print(
    "August has NOT been used for this model-selection decision."
)


# ============================================================
# 9. Save outputs
# ============================================================

event_path = os.path.join(
    NB5,
    "Tables",
    "Table_R1_vs_R2_July_Event_Level_Comparison.csv"
)

direct_path = os.path.join(
    NB5,
    "Tables",
    "Table_R1_vs_R2_Final_July_Comparison.csv"
)

decision_path = os.path.join(
    NB5,
    "Results",
    "Final_Mitigation_Model_Selection_Decision.json"
)

event_level.to_csv(
    event_path,
    index=False
)

direct_comparison.to_csv(
    direct_path,
    index=False
)


decision = {
    "selection_period":
        "July 2026",

    "final_test_period_reserved":
        "August 2026",

    "august_used_for_selection":
        False,

    "selected_regime":
        SELECTED_REGIME,

    "selected_feature_count":
        SELECTED_FEATURE_COUNT,

    "primary_selection_metric":
        "Controlled failure-injected WAPE",

    "R1_injected_WAPE":
        float(
            r1_row[
                "Controlled_Injected_WAPE"
            ]
        ),

    "R2_injected_WAPE":
        float(
            r2_row[
                "Controlled_Injected_WAPE"
            ]
        ),

    "R1_full_clean_WAPE":
        float(
            r1_row[
                "Full_July_Clean_WAPE"
            ]
        ),

    "R2_full_clean_WAPE":
        float(
            r2_row[
                "Full_July_Clean_WAPE"
            ]
        ),

    "event_level_mean_R2_minus_R1_MAE":
        mean_difference,

    "event_level_bootstrap_95_CI":
        [
            ci_low,
            ci_high
        ],

    "paired_wilcoxon_p_value":
        wilcoxon_p,

    "selection_reason":
        SELECTION_REASON
}


with open(
    decision_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        decision,
        f,
        indent=2
    )


print("\n" + "=" * 135)
print("9. SAVED OUTPUTS")
print("=" * 135)

print(f"✅ {event_path}")
print(f"✅ {direct_path}")
print(f"✅ {decision_path}")

print("\n" + "=" * 135)
print("STEP 20 COMPLETED")
print("=" * 135)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 21: Build FINAL conflict-free Jan-Jul
#          empirical augmentation map
# ============================================================

import os
import gc
import json
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB4 = os.path.join(
    BASE_PATH,
    "Notebook_4_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

PATTERN_PATH = os.path.join(
    NB4,
    "Data",
    "PreTest_Real_Failure_Pattern_Library_JanJul.csv"
)

MONTHS = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06",
    "2026-07"
]

LAG_OFFSETS_MINUTES = {
    "Lag_1": 15,
    "Lag_2": 30,
    "Lag_4": 60,
    "Lag_8": 120,
    "Lag_96": 1440
}

RANDOM_SEED = 42

print("=" * 140)
print("NOTEBOOK 5 — FINAL JAN–JUL CONFLICT-FREE AUGMENTATION MAP")
print("=" * 140)

print(
    "\nIMPORTANT: August is NOT used anywhere in this step."
)


# ============================================================
# 1. Load complete Jan-Jul empirical failure library
# ============================================================

patterns = pd.read_csv(
    PATTERN_PATH
)

patterns["Date"] = pd.to_datetime(
    patterns["Date"]
).dt.normalize()

patterns = patterns[
    (
        patterns["Date"]
        >= pd.Timestamp("2026-01-01")
    )
    &
    (
        patterns["Date"]
        <= pd.Timestamp("2026-07-31")
    )
].copy()

patterns[
    "Source_DayOfWeek"
] = patterns[
    "Date"
].dt.dayofweek

patterns[
    "Source_Month"
] = patterns[
    "Date"
].dt.to_period(
    "M"
).astype(str)

patterns = patterns.reset_index(
    drop=True
)

patterns[
    "Final_Augmentation_Event_ID"
] = np.arange(
    1,
    len(patterns) + 1
)

print("\n" + "=" * 140)
print("1. JAN–JUL FAILURE PATTERN LIBRARY")
print("=" * 140)

print(
    f"Failure patterns : "
    f"{len(patterns):,}"
)

print(
    f"Unique detectors : "
    f"{patterns['Detector_Code'].nunique():,}"
)

print(
    f"Coverage         : "
    f"{patterns['Date'].min().date()} "
    f"→ {patterns['Date'].max().date()}"
)

monthly_patterns = (
    patterns[
        "Source_Month"
    ]
    .value_counts()
    .sort_index()
)

print("\nPatterns by month:")

print(
    monthly_patterns.to_string()
)


# ============================================================
# 2. Build strict-clean detector-day inventory for Jan-Jul
#
# Requirements:
# - exactly 96 rows
# - 96 valid targets
# - all five lag predictors available on all 96 rows
# ============================================================

print("\n" + "=" * 140)
print("2. BUILDING STRICT-CLEAN JAN–JUL DETECTOR-DAY INVENTORY")
print("=" * 140)

inventory_parts = []

for month in MONTHS:

    path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    df = pd.read_parquet(
        path,
        columns=[
            "Timestamp",
            "Detector_Code",
            "Target_Volume",
            "Lag_1",
            "Lag_2",
            "Lag_4",
            "Lag_8",
            "Lag_96"
        ]
    )

    df[
        "Timestamp"
    ] = pd.to_datetime(
        df["Timestamp"]
    )

    df[
        "Date"
    ] = df[
        "Timestamp"
    ].dt.normalize()

    df[
        "Valid_Target"
    ] = (
        df[
            "Target_Volume"
        ].notna()
        &
        np.isfinite(
            df[
                "Target_Volume"
            ]
        )
        &
        (
            df[
                "Target_Volume"
            ] >= 0
        )
    )

    df[
        "All_Lags_Valid"
    ] = (
        df[
            [
                "Lag_1",
                "Lag_2",
                "Lag_4",
                "Lag_8",
                "Lag_96"
            ]
        ]
        .notna()
        .all(axis=1)
    )

    daily = (
        df
        .groupby(
            [
                "Detector_Code",
                "Date"
            ],
            as_index=False
        )
        .agg(
            Rows=(
                "Timestamp",
                "size"
            ),

            Valid_Targets=(
                "Valid_Target",
                "sum"
            ),

            Valid_Lag_Rows=(
                "All_Lags_Valid",
                "sum"
            )
        )
    )

    strict = daily[
        (
            daily[
                "Rows"
            ] == 96
        )
        &
        (
            daily[
                "Valid_Targets"
            ] == 96
        )
        &
        (
            daily[
                "Valid_Lag_Rows"
            ] == 96
        )
    ].copy()

    strict[
        "DayOfWeek"
    ] = strict[
        "Date"
    ].dt.dayofweek

    strict[
        "Month"
    ] = strict[
        "Date"
    ].dt.to_period(
        "M"
    ).astype(str)

    print(
        f"{month}: "
        f"{len(strict):,} strict-clean detector-days"
    )

    inventory_parts.append(
        strict[
            [
                "Detector_Code",
                "Date",
                "DayOfWeek",
                "Month"
            ]
        ]
    )

    del df, daily, strict
    gc.collect()


strict_days = pd.concat(
    inventory_parts,
    ignore_index=True
)

del inventory_parts
gc.collect()

print(
    f"\nTotal strict-clean detector-days: "
    f"{len(strict_days):,}"
)

print(
    f"Unique detectors represented    : "
    f"{strict_days['Detector_Code'].nunique():,}"
)


# ============================================================
# 3. Require clean injection day AND following calendar day
#
# This guarantees a clean baseline for all effects through
# Lag_96 because each source failure run is confined to its
# detector-day.
# ============================================================

strict_lookup = set(
    zip(
        strict_days[
            "Detector_Code"
        ].astype(int),
        strict_days[
            "Date"
        ]
    )
)

clean48_records = []

for row in strict_days.itertuples(
    index=False
):

    next_date = (
        pd.Timestamp(
            row.Date
        )
        +
        pd.Timedelta(
            days=1
        )
    )

    if (
        next_date
        >
        pd.Timestamp(
            "2026-07-31"
        )
    ):
        continue

    if (
        int(
            row.Detector_Code
        ),
        next_date
    ) in strict_lookup:

        clean48_records.append({
            "Detector_Code":
                int(
                    row.Detector_Code
                ),

            "Date":
                pd.Timestamp(
                    row.Date
                ),

            "DayOfWeek":
                int(
                    row.DayOfWeek
                ),

            "Month":
                row.Month,

            "Next_Date":
                next_date
        })


clean48 = pd.DataFrame(
    clean48_records
)

print("\n" + "=" * 140)
print("3. STRICT-CLEAN 48-HOUR WINDOWS")
print("=" * 140)

print(
    f"Eligible windows : "
    f"{len(clean48):,}"
)

print(
    f"Unique detectors : "
    f"{clean48['Detector_Code'].nunique():,}"
)


# ============================================================
# 4. Candidate-window feasibility for every failure pattern
# ============================================================

candidate_counts = []

for row in patterns.itertuples(
    index=False
):

    pool = clean48[
        (
            clean48[
                "Detector_Code"
            ] == int(
                row.Detector_Code
            )
        )
        &
        (
            clean48[
                "DayOfWeek"
            ] == int(
                row.Source_DayOfWeek
            )
        )
    ]

    same_month = int(
        (
            pool[
                "Month"
            ]
            ==
            row.Source_Month
        ).sum()
    )

    candidate_counts.append({
        "Final_Augmentation_Event_ID":
            int(
                row.Final_Augmentation_Event_ID
            ),

        "SameMonth_SameDOW_Candidates":
            same_month,

        "SameDOW_AllMonths_Candidates":
            int(
                len(pool)
            )
    })


candidate_counts = pd.DataFrame(
    candidate_counts
)

patterns = patterns.merge(
    candidate_counts,
    on="Final_Augmentation_Event_ID",
    how="left",
    validate="one_to_one"
)

feasible_patterns = int(
    (
        patterns[
            "SameDOW_AllMonths_Candidates"
        ] > 0
    ).sum()
)

print("\n" + "=" * 140)
print("4. PATTERN-LEVEL FINAL AUGMENTATION FEASIBILITY")
print("=" * 140)

print(
    f"Patterns with >=1 same-detector + same-DOW candidate:"
)

print(
    f"{feasible_patterns:,} / "
    f"{len(patterns):,} "
    f"({100*feasible_patterns/len(patterns):.2f}%)"
)

print(
    "\nCandidate windows per pattern:"
)

print(
    patterns[
        "SameDOW_AllMonths_Candidates"
    ]
    .describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .to_string()
)


# ============================================================
# 5. Helper to calculate exact affected target footprint
# ============================================================

def affected_footprint(
    injection_date,
    start_interval,
    run_length
):

    failure_start = (
        pd.Timestamp(
            injection_date
        )
        +
        pd.to_timedelta(
            int(
                start_interval
            ) * 15,
            unit="m"
        )
    )

    failure_times = pd.date_range(
        start=failure_start,
        periods=int(
            run_length
        ),
        freq="15min"
    )

    affected = set()

    for offset in (
        LAG_OFFSETS_MINUTES.values()
    ):

        shifted = (
            failure_times
            +
            pd.Timedelta(
                minutes=int(
                    offset
                )
            )
        )

        affected.update(
            shifted.to_pydatetime()
        )

    return (
        failure_start,
        max(
            failure_times
        ),
        max(
            affected
        ),
        affected
    )


# ============================================================
# 6. Conflict-free mapping
#
# Rules:
# - same detector
# - same day-of-week
# - same month preferred
# - strict clean 48h baseline
# - one injection start-day per detector-event map
# - no overlapping affected target footprint for same detector
# - deterministic random seed
# ============================================================

print("\n" + "=" * 140)
print("5. BUILDING CONFLICT-FREE FINAL AUGMENTATION MAP")
print("=" * 140)

rng_global = np.random.default_rng(
    RANDOM_SEED
)

all_assignments = []
detector_audit = []

detectors = sorted(
    patterns[
        "Detector_Code"
    ]
    .unique()
)

for detector_index, detector in enumerate(
    detectors,
    start=1
):

    det_patterns = patterns[
        patterns[
            "Detector_Code"
        ] == detector
    ].copy()

    # Hardest cases first:
    # fewest candidates, then longest failures first.
    det_patterns = (
        det_patterns
        .sort_values(
            [
                "SameDOW_AllMonths_Candidates",
                "Run_Length",
                "Final_Augmentation_Event_ID"
            ],
            ascending=[
                True,
                False,
                True
            ]
        )
        .reset_index(
            drop=True
        )
    )

    # We retry with different deterministic randomizations
    # only if the first greedy assignment is incomplete.
    best_assignment = None
    best_count = -1
    successful_attempt = None

    for attempt in range(
        1,
        101
    ):

        rng = np.random.default_rng(
            RANDOM_SEED
            +
            int(detector) * 1000
            +
            attempt
        )

        occupied_targets = set()
        used_start_dates = set()
        current_assignment = []

        if attempt == 1:

            event_order = det_patterns.copy()

        else:

            # Preserve approximate difficulty ordering,
            # while varying ties to escape rare conflicts.
            temp = det_patterns.copy()

            temp[
                "_Random"
            ] = rng.random(
                len(temp)
            )

            event_order = (
                temp
                .sort_values(
                    [
                        "SameDOW_AllMonths_Candidates",
                        "_Random"
                    ],
                    ascending=[
                        True,
                        True
                    ]
                )
                .drop(
                    columns="_Random"
                )
                .reset_index(
                    drop=True
                )
            )

        for event in event_order.itertuples(
            index=False
        ):

            pool = clean48[
                (
                    clean48[
                        "Detector_Code"
                    ] == int(
                        detector
                    )
                )
                &
                (
                    clean48[
                        "DayOfWeek"
                    ] == int(
                        event.Source_DayOfWeek
                    )
                )
            ].copy()

            # Remove already-used start days for this detector
            pool = pool[
                ~pool[
                    "Date"
                ].isin(
                    used_start_dates
                )
            ].copy()

            if len(
                pool
            ) == 0:

                continue

            pool[
                "Month_Priority"
            ] = np.where(
                pool[
                    "Month"
                ]
                ==
                event.Source_Month,
                0,
                1
            )

            pool[
                "_Random"
            ] = rng.random(
                len(pool)
            )

            pool = (
                pool
                .sort_values(
                    [
                        "Month_Priority",
                        "_Random"
                    ]
                )
                .reset_index(
                    drop=True
                )
            )

            assigned = False

            for candidate in pool.itertuples(
                index=False
            ):

                (
                    failure_start,
                    failure_end,
                    evaluation_end,
                    footprint
                ) = affected_footprint(
                    injection_date=
                        candidate.Date,

                    start_interval=
                        event.Start_Interval_Index,

                    run_length=
                        event.Run_Length
                )

                # Must remain strictly inside Jan-Jul
                if (
                    evaluation_end
                    >
                    pd.Timestamp(
                        "2026-07-31 23:45:00"
                    )
                ):
                    continue

                # Exact affected targets may not overlap
                # any earlier mapped event for this detector.
                if not occupied_targets.isdisjoint(
                    footprint
                ):
                    continue

                match_type = (
                    "SameMonth_SameDOW"
                    if (
                        candidate.Month
                        ==
                        event.Source_Month
                    )
                    else
                    "SameDOW_Fallback"
                )

                current_assignment.append({
                    "Final_Augmentation_Event_ID":
                        int(
                            event.Final_Augmentation_Event_ID
                        ),

                    "Detector_Code":
                        int(
                            detector
                        ),

                    "Source_Failure_Date":
                        pd.Timestamp(
                            event.Date
                        ),

                    "Source_Month":
                        event.Source_Month,

                    "Source_DayOfWeek":
                        int(
                            event.Source_DayOfWeek
                        ),

                    "Target_Injection_Date":
                        pd.Timestamp(
                            candidate.Date
                        ),

                    "Target_Month":
                        candidate.Month,

                    "Target_DayOfWeek":
                        int(
                            candidate.DayOfWeek
                        ),

                    "Start_Interval_Index":
                        int(
                            event.Start_Interval_Index
                        ),

                    "End_Interval_Index":
                        int(
                            event.End_Interval_Index
                        ),

                    "Run_Length":
                        int(
                            event.Run_Length
                        ),

                    "Failure_Duration_Minutes":
                        event.Failure_Duration_Minutes,

                    "Failure_Duration_Hours":
                        event.Failure_Duration_Hours,

                    "Failure_Duration_Group":
                        event.Failure_Duration_Group,

                    "Failure_Start":
                        failure_start,

                    "Failure_End":
                        failure_end,

                    "Evaluation_End":
                        evaluation_end,

                    "Match_Type":
                        match_type,

                    "Mapping_Attempt":
                        attempt
                })

                occupied_targets.update(
                    footprint
                )

                used_start_dates.add(
                    pd.Timestamp(
                        candidate.Date
                    )
                )

                assigned = True

                break

        if len(
            current_assignment
        ) > best_count:

            best_count = len(
                current_assignment
            )

            best_assignment = (
                current_assignment
            )

        if len(
            current_assignment
        ) == len(
            det_patterns
        ):

            successful_attempt = (
                attempt
            )

            best_assignment = (
                current_assignment
            )

            break

    all_assignments.extend(
        best_assignment
    )

    detector_audit.append({
        "Detector_Code":
            int(
                detector
            ),

        "Source_Events":
            int(
                len(
                    det_patterns
                )
            ),

        "Mapped_Events":
            int(
                len(
                    best_assignment
                )
            ),

        "Complete":
            bool(
                len(
                    best_assignment
                )
                ==
                len(
                    det_patterns
                )
            ),

        "Successful_Attempt":
            successful_attempt
    })

    if (
        detector_index % 50 == 0
        or detector_index == len(
            detectors
        )
    ):

        print(
            f"[MAPPING] Detectors processed: "
            f"{detector_index:,} / "
            f"{len(detectors):,}",
            flush=True
        )


final_map = pd.DataFrame(
    all_assignments
)

detector_audit = pd.DataFrame(
    detector_audit
)


# ============================================================
# 7. Mapping audit
# ============================================================

print("\n" + "=" * 140)
print("6. FINAL MAPPING RESULTS")
print("=" * 140)

print(
    f"Source patterns : "
    f"{len(patterns):,}"
)

print(
    f"Mapped events   : "
    f"{len(final_map):,}"
)

print(
    f"Unmapped events : "
    f"{len(patterns)-len(final_map):,}"
)

print(
    f"Source detectors: "
    f"{patterns['Detector_Code'].nunique():,}"
)

print(
    f"Fully mapped detectors: "
    f"{detector_audit['Complete'].sum():,} / "
    f"{len(detector_audit):,}"
)

print("\nMatch types:")

print(
    final_map[
        "Match_Type"
    ]
    .value_counts()
    .to_string()
)


same_dow_pass = bool(
    (
        final_map[
            "Source_DayOfWeek"
        ]
        ==
        final_map[
            "Target_DayOfWeek"
        ]
    ).all()
)

all_inside_period = bool(
    (
        final_map[
            "Evaluation_End"
        ]
        <=
        pd.Timestamp(
            "2026-07-31 23:45:00"
        )
    ).all()
)

reused_start_days = int(
    final_map
    .duplicated(
        subset=[
            "Detector_Code",
            "Target_Injection_Date"
        ]
    )
    .sum()
)

print(
    f"\nDay-of-week preserved          : "
    f"{same_dow_pass}"
)

print(
    f"All dependency effects Jan-Jul : "
    f"{all_inside_period}"
)

print(
    f"Reused detector start-days     : "
    f"{reused_start_days:,}"
)


# ============================================================
# 8. Reconstruct exact affected footprints
#    and verify no detector-target collisions
# ============================================================

verification_records = []

for row in final_map.itertuples(
    index=False
):

    failure_times = pd.date_range(
        start=row.Failure_Start,
        periods=int(
            row.Run_Length
        ),
        freq="15min"
    )

    for lag_name, offset in (
        LAG_OFFSETS_MINUTES.items()
    ):

        target_times = (
            failure_times
            +
            pd.Timedelta(
                minutes=int(
                    offset
                )
            )
        )

        for ts in target_times:

            verification_records.append({
                "Final_Augmentation_Event_ID":
                    int(
                        row.Final_Augmentation_Event_ID
                    ),

                "Detector_Code":
                    int(
                        row.Detector_Code
                    ),

                "Timestamp":
                    pd.Timestamp(
                        ts
                    ),

                "Missing_Lag":
                    lag_name
            })


verification_long = pd.DataFrame(
    verification_records
)

unique_targets = (
    verification_long[
        [
            "Final_Augmentation_Event_ID",
            "Detector_Code",
            "Timestamp"
        ]
    ]
    .drop_duplicates()
)

collision_count = int(
    unique_targets
    .duplicated(
        subset=[
            "Detector_Code",
            "Timestamp"
        ],
        keep=False
    )
    .sum()
)

print("\n" + "=" * 140)
print("7. EXACT TARGET-FOOTPRINT VERIFICATION")
print("=" * 140)

print(
    f"Raw lag-impact records       : "
    f"{len(verification_long):,}"
)

print(
    f"Unique event-target rows     : "
    f"{len(unique_targets):,}"
)

print(
    f"Detector-target collisions   : "
    f"{collision_count:,}"
)


# ============================================================
# 9. Final integrity decision
# ============================================================

integrity_pass = bool(
    len(final_map)
    ==
    len(patterns)

    and
    detector_audit[
        "Complete"
    ].all()

    and
    same_dow_pass

    and
    all_inside_period

    and
    reused_start_days == 0

    and
    collision_count == 0
)

print("\n" + "=" * 140)
print("8. STEP 21 INTEGRITY DECISION")
print("=" * 140)

print(
    f"All 2,223 source patterns mapped : "
    f"{len(final_map) == len(patterns)}"
)

print(
    f"All detectors completely mapped  : "
    f"{detector_audit['Complete'].all()}"
)

print(
    f"Day-of-week preserved             : "
    f"{same_dow_pass}"
)

print(
    f"No reused detector start-days    : "
    f"{reused_start_days == 0}"
)

print(
    f"No affected-target collisions    : "
    f"{collision_count == 0}"
)

print(
    f"All effects remain in Jan-Jul     : "
    f"{all_inside_period}"
)

print(
    f"\nFINAL STEP 21 INTEGRITY PASS: "
    f"{integrity_pass}"
)


# ============================================================
# 10. Save outputs
# ============================================================

strict_path = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Strict_Clean_48h_Windows.parquet"
)

map_path = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Augmentation_Event_Map_ConflictFree.csv"
)

audit_path = os.path.join(
    NB5,
    "Tables",
    "Table_Final_JanJul_Augmentation_Mapping_Audit.csv"
)

feasibility_path = os.path.join(
    NB5,
    "Tables",
    "Table_Final_JanJul_Augmentation_Feasibility.csv"
)

verification_path = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Augmentation_Target_Verification.parquet"
)

summary_path = os.path.join(
    NB5,
    "Results",
    "Final_JanJul_Augmentation_Mapping_Summary.json"
)

clean48.to_parquet(
    strict_path,
    index=False
)

final_map.to_csv(
    map_path,
    index=False
)

detector_audit.to_csv(
    audit_path,
    index=False
)

patterns[
    [
        "Final_Augmentation_Event_ID",
        "Detector_Code",
        "Date",
        "Source_Month",
        "Source_DayOfWeek",
        "Run_Length",
        "Failure_Duration_Group",
        "SameMonth_SameDOW_Candidates",
        "SameDOW_AllMonths_Candidates"
    ]
].to_csv(
    feasibility_path,
    index=False
)

verification_long.to_parquet(
    verification_path,
    index=False
)

summary = {
    "training_period":
        "2026-01-01 to 2026-07-31",

    "final_test_period":
        "August 2026",

    "august_used":
        False,

    "source_failure_patterns":
        int(
            len(patterns)
        ),

    "source_detectors":
        int(
            patterns[
                "Detector_Code"
            ].nunique()
        ),

    "mapped_events":
        int(
            len(final_map)
        ),

    "unique_event_target_rows":
        int(
            len(
                unique_targets
            )
        ),

    "detector_target_collisions":
        int(
            collision_count
        ),

    "same_day_of_week_preserved":
        same_dow_pass,

    "reused_detector_start_days":
        int(
            reused_start_days
        ),

    "integrity_pass":
        integrity_pass
}

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )


print("\n" + "=" * 140)
print("9. SAVED OUTPUTS")
print("=" * 140)

print(f"✅ {strict_path}")
print(f"✅ {map_path}")
print(f"✅ {audit_path}")
print(f"✅ {feasibility_path}")
print(f"✅ {verification_path}")
print(f"✅ {summary_path}")

print("\n" + "=" * 140)
print("STEP 21 COMPLETED")
print("=" * 140)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 22: Materialize FINAL Jan-Jul failure-aware
#          training components
# ============================================================

import os
import gc
import json
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

MAP_PATH = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Augmentation_Event_Map_ConflictFree.csv"
)

VERIFICATION_PATH = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Augmentation_Target_Verification.parquet"
)

MONTHS = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06",
    "2026-07"
]

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

NONLAG_FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend"
]

LAG_COLS = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

print("=" * 140)
print("NOTEBOOK 5 — FINAL JAN–JUL TRAINING DATA MATERIALIZATION")
print("=" * 140)

print(
    "\nIMPORTANT: August is NOT used anywhere in this step."
)


# ============================================================
# 1. Load final augmentation event map
# ============================================================

event_map = pd.read_csv(
    MAP_PATH
)

date_cols = [
    "Source_Failure_Date",
    "Target_Injection_Date",
    "Failure_Start",
    "Failure_End",
    "Evaluation_End"
]

for col in date_cols:
    event_map[col] = pd.to_datetime(
        event_map[col]
    )

print("\n" + "=" * 140)
print("1. FINAL AUGMENTATION EVENT MAP")
print("=" * 140)

print(
    f"Events           : "
    f"{len(event_map):,}"
)

print(
    f"Unique detectors : "
    f"{event_map['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Load exact lag-impact verification
# ============================================================

impact_long = pd.read_parquet(
    VERIFICATION_PATH
)

impact_long[
    "Timestamp"
] = pd.to_datetime(
    impact_long[
        "Timestamp"
    ]
)

print("\n" + "=" * 140)
print("2. EXACT LAG-IMPACT RECORDS")
print("=" * 140)

print(
    f"Raw lag-impact rows : "
    f"{len(impact_long):,}"
)

print(
    f"Events represented  : "
    f"{impact_long['Final_Augmentation_Event_ID'].nunique():,}"
)


# ============================================================
# 3. Collapse impacts to unique affected target rows
# ============================================================

target_map = (
    impact_long[
        [
            "Final_Augmentation_Event_ID",
            "Detector_Code",
            "Timestamp"
        ]
    ]
    .drop_duplicates()
    .copy()
)

for lag in LAG_COLS:

    flag = (
        impact_long[
            impact_long[
                "Missing_Lag"
            ] == lag
        ][
            [
                "Final_Augmentation_Event_ID",
                "Detector_Code",
                "Timestamp"
            ]
        ]
        .drop_duplicates()
        .copy()
    )

    flag[
        f"Missing_{lag}"
    ] = np.int8(1)

    target_map = target_map.merge(
        flag,
        on=[
            "Final_Augmentation_Event_ID",
            "Detector_Code",
            "Timestamp"
        ],
        how="left"
    )


for lag in LAG_COLS:

    target_map[
        f"Missing_{lag}"
    ] = (
        target_map[
            f"Missing_{lag}"
        ]
        .fillna(0)
        .astype("int8")
    )


target_map[
    "Missing_Lag_Count"
] = (
    target_map[
        [
            f"Missing_{lag}"
            for lag in LAG_COLS
        ]
    ]
    .sum(axis=1)
    .astype("int8")
)


def build_pattern(row):

    missing = [
        lag
        for lag in LAG_COLS
        if row[
            f"Missing_{lag}"
        ] == 1
    ]

    return "+".join(
        missing
    )


target_map[
    "Missing_Lag_Pattern"
] = target_map.apply(
    build_pattern,
    axis=1
)


# Add event metadata
event_metadata = event_map[
    [
        "Final_Augmentation_Event_ID",
        "Source_Failure_Date",
        "Target_Injection_Date",
        "Run_Length",
        "Failure_Duration_Hours",
        "Failure_Duration_Group"
    ]
].copy()

target_map = target_map.merge(
    event_metadata,
    on="Final_Augmentation_Event_ID",
    how="left",
    validate="many_to_one"
)


print("\n" + "=" * 140)
print("3. UNIQUE FINAL AUGMENTED TARGETS")
print("=" * 140)

print(
    f"Unique affected targets : "
    f"{len(target_map):,}"
)

print(
    f"Events represented       : "
    f"{target_map['Final_Augmentation_Event_ID'].nunique():,}"
)

print(
    f"Duplicate detector-target timestamps: "
    f"{target_map.duplicated(['Detector_Code','Timestamp']).sum():,}"
)


# ============================================================
# 4. Process Jan-Jul month by month
#
# Simultaneously:
#   A) audit clean vs natural-missing original rows
#   B) save natural failure-aware rows
#   C) extract clean baselines for empirical augmentation
# ============================================================

print("\n" + "=" * 140)
print("4. PROCESSING ORIGINAL JAN–JUL DATA")
print("=" * 140)

natural_parts = []
augmented_baseline_parts = []
monthly_audit = []

for i, month in enumerate(
    MONTHS,
    start=1
):

    print(
        f"\n[PROCESSING] Month {i} / {len(MONTHS)}: {month}",
        flush=True
    )

    path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    df = pd.read_parquet(
        path,
        columns=[
            "Timestamp"
        ]
        +
        FEATURES
        +
        [
            TARGET
        ]
    )

    df[
        "Timestamp"
    ] = pd.to_datetime(
        df[
            "Timestamp"
        ]
    )

    # --------------------------------------------------------
    # Target validity
    # --------------------------------------------------------

    valid_target = (
        df[
            TARGET
        ].notna()
        &
        np.isfinite(
            df[
                TARGET
            ]
        )
        &
        (
            df[
                TARGET
            ] >= 0
        )
    )

    nonlag_valid = (
        df[
            NONLAG_FEATURES
        ]
        .notna()
        .all(axis=1)
    )

    lag_missing = (
        df[
            LAG_COLS
        ]
        .isna()
        .any(axis=1)
    )

    all_lags_valid = (
        ~lag_missing
    )

    clean_mask = (
        valid_target
        &
        nonlag_valid
        &
        all_lags_valid
    )

    natural_mask = (
        valid_target
        &
        nonlag_valid
        &
        lag_missing
    )

    clean_rows = int(
        clean_mask.sum()
    )

    natural_rows = int(
        natural_mask.sum()
    )

    invalid_target_rows = int(
        (
            ~valid_target
        ).sum()
    )

    nonlag_invalid_rows = int(
        (
            valid_target
            &
            ~nonlag_valid
        ).sum()
    )

    print(
        f"  Total rows           : {len(df):,}"
    )

    print(
        f"  Clean model-ready    : {clean_rows:,}"
    )

    print(
        f"  Natural missing-lag  : {natural_rows:,}"
    )

    print(
        f"  Invalid target       : {invalid_target_rows:,}"
    )

    print(
        f"  Invalid non-lag feat : {nonlag_invalid_rows:,}"
    )

    # --------------------------------------------------------
    # Save natural missing-history rows
    # --------------------------------------------------------

    natural = df.loc[
        natural_mask,
        [
            "Timestamp"
        ]
        +
        FEATURES
        +
        [
            TARGET
        ]
    ].copy()

    natural_parts.append(
        natural
    )

    # --------------------------------------------------------
    # Extract empirical augmentation baseline targets
    # --------------------------------------------------------

    month_target_map = target_map[
        target_map[
            "Timestamp"
        ]
        .dt.to_period(
            "M"
        )
        .astype(str)
        ==
        month
    ].copy()

    baseline = df.merge(
        month_target_map,
        on=[
            "Detector_Code",
            "Timestamp"
        ],
        how="inner",
        validate="one_to_one"
    )

    expected_augmented = len(
        month_target_map
    )

    found_augmented = len(
        baseline
    )

    baseline_missing_lags = int(
        baseline[
            LAG_COLS
        ]
        .isna()
        .any(axis=1)
        .sum()
    )

    baseline_invalid_targets = int(
        (
            baseline[
                TARGET
            ].isna()
            |
            ~np.isfinite(
                baseline[
                    TARGET
                ]
            )
            |
            (
                baseline[
                    TARGET
                ] < 0
            )
        ).sum()
    )

    print(
        f"  Aug targets expected : {expected_augmented:,}"
    )

    print(
        f"  Aug targets found    : {found_augmented:,}"
    )

    print(
        f"  Baseline missing lag : {baseline_missing_lags:,}"
    )

    augmented_baseline_parts.append(
        baseline
    )

    monthly_audit.append({
        "Month":
            month,

        "Original_Rows":
            int(
                len(df)
            ),

        "Clean_Model_Ready_Rows":
            clean_rows,

        "Natural_Missing_Lag_Rows":
            natural_rows,

        "Invalid_Target_Rows":
            invalid_target_rows,

        "Invalid_NonLag_Feature_Rows":
            nonlag_invalid_rows,

        "Augmentation_Targets_Expected":
            int(
                expected_augmented
            ),

        "Augmentation_Targets_Found":
            int(
                found_augmented
            ),

        "Augmentation_Baseline_Missing_Lags":
            baseline_missing_lags,

        "Augmentation_Invalid_Targets":
            baseline_invalid_targets
    })

    del df
    del natural
    del baseline
    del month_target_map

    gc.collect()


monthly_audit = pd.DataFrame(
    monthly_audit
)


# ============================================================
# 5. Combine natural missing-history rows
# ============================================================

natural_training = pd.concat(
    natural_parts,
    ignore_index=True
)

del natural_parts
gc.collect()


# Missing-pattern metadata for audit
for lag in LAG_COLS:

    natural_training[
        f"Missing_{lag}"
    ] = (
        natural_training[
            lag
        ]
        .isna()
        .astype("int8")
    )


natural_training[
    "Missing_Lag_Count"
] = (
    natural_training[
        [
            f"Missing_{lag}"
            for lag in LAG_COLS
        ]
    ]
    .sum(axis=1)
    .astype("int8")
)


natural_training[
    "Missing_Lag_Pattern"
] = natural_training.apply(
    build_pattern,
    axis=1
)


print("\n" + "=" * 140)
print("5. FINAL JAN–JUL NATURAL MISSING-HISTORY DATA")
print("=" * 140)

print(
    f"Natural missing rows : "
    f"{len(natural_training):,}"
)

print(
    f"Missing targets      : "
    f"{natural_training[TARGET].isna().sum():,}"
)


natural_pattern_summary = (
    natural_training[
        "Missing_Lag_Pattern"
    ]
    .value_counts()
    .rename_axis(
        "Missing_Lag_Pattern"
    )
    .reset_index(
        name="Rows"
    )
)

natural_pattern_summary[
    "Percent"
] = (
    100
    *
    natural_pattern_summary[
        "Rows"
    ]
    /
    len(
        natural_training
    )
)

print("\nNatural missing-lag patterns:")

print(
    natural_pattern_summary
    .head(20)
    .to_string(
        index=False
    )
)


# ============================================================
# 6. Combine empirical augmentation baseline rows
# ============================================================

augmented = pd.concat(
    augmented_baseline_parts,
    ignore_index=True
)

del augmented_baseline_parts
gc.collect()


print("\n" + "=" * 140)
print("6. EMPIRICAL AUGMENTATION BASELINE AUDIT")
print("=" * 140)

print(
    f"Expected affected targets : "
    f"{len(target_map):,}"
)

print(
    f"Extracted baseline rows   : "
    f"{len(augmented):,}"
)

baseline_missing_total = int(
    augmented[
        LAG_COLS
    ]
    .isna()
    .any(axis=1)
    .sum()
)

baseline_invalid_target_total = int(
    (
        augmented[
            TARGET
        ].isna()
        |
        ~np.isfinite(
            augmented[
                TARGET
            ]
        )
        |
        (
            augmented[
                TARGET
            ] < 0
        )
    ).sum()
)

duplicate_augmented_targets = int(
    augmented
    .duplicated(
        subset=[
            "Detector_Code",
            "Timestamp"
        ]
    )
    .sum()
)

print(
    f"Baseline rows with missing lag : "
    f"{baseline_missing_total:,}"
)

print(
    f"Invalid targets                 : "
    f"{baseline_invalid_target_total:,}"
)

print(
    f"Duplicate detector-target rows  : "
    f"{duplicate_augmented_targets:,}"
)


# ============================================================
# 7. Inject exact missing lags
# ============================================================

for lag in LAG_COLS:

    injection_mask = (
        augmented[
            f"Missing_{lag}"
        ] == 1
    )

    augmented.loc[
        injection_mask,
        lag
    ] = np.nan


# ============================================================
# 8. Validate exact injection counts
# ============================================================

injection_audit = []

for lag in LAG_COLS:

    intended = int(
        augmented[
            f"Missing_{lag}"
        ].sum()
    )

    actual = int(
        augmented[
            lag
        ].isna()
        .sum()
    )

    unexpected = int(
        (
            augmented[
                lag
            ].isna()
            &
            (
                augmented[
                    f"Missing_{lag}"
                ] == 0
            )
        ).sum()
    )

    injection_audit.append({
        "Lag":
            lag,

        "Intended_Missing":
            intended,

        "Actual_Missing":
            actual,

        "Unexpected_Missing":
            unexpected,

        "Exact_Match":
            bool(
                intended == actual
                and
                unexpected == 0
            )
    })


injection_audit = pd.DataFrame(
    injection_audit
)

all_injections_exact = bool(
    injection_audit[
        "Exact_Match"
    ].all()
)

print("\n" + "=" * 140)
print("7. FINAL EMPIRICAL INJECTION VALIDATION")
print("=" * 140)

print(
    injection_audit.to_string(
        index=False
    )
)

print(
    f"\nAll injections exact: "
    f"{all_injections_exact}"
)


# ============================================================
# 9. Augmented missing-pattern distribution
# ============================================================

augmented_pattern_summary = (
    augmented[
        "Missing_Lag_Pattern"
    ]
    .value_counts()
    .rename_axis(
        "Missing_Lag_Pattern"
    )
    .reset_index(
        name="Rows"
    )
)

augmented_pattern_summary[
    "Percent"
] = (
    100
    *
    augmented_pattern_summary[
        "Rows"
    ]
    /
    len(
        augmented
    )
)


print("\n" + "=" * 140)
print("8. FINAL EMPIRICAL AUGMENTED MISSING-LAG PATTERNS")
print("=" * 140)

print(
    augmented_pattern_summary
    .to_string(
        index=False
    )
)


# ============================================================
# 10. Final Jan-Jul training composition
# ============================================================

total_original_rows = int(
    monthly_audit[
        "Original_Rows"
    ].sum()
)

total_clean_rows = int(
    monthly_audit[
        "Clean_Model_Ready_Rows"
    ].sum()
)

total_natural_rows = int(
    monthly_audit[
        "Natural_Missing_Lag_Rows"
    ].sum()
)

total_invalid_targets = int(
    monthly_audit[
        "Invalid_Target_Rows"
    ].sum()
)

total_invalid_nonlag = int(
    monthly_audit[
        "Invalid_NonLag_Feature_Rows"
    ].sum()
)

total_augmented_rows = int(
    len(
        augmented
    )
)

final_training_rows = (
    total_clean_rows
    +
    total_natural_rows
    +
    total_augmented_rows
)


composition = pd.DataFrame([
    {
        "Training_Source":
            "Clean_Model_Ready_JanJul",

        "Rows":
            total_clean_rows
    },

    {
        "Training_Source":
            "Natural_Missing_History_JanJul",

        "Rows":
            total_natural_rows
    },

    {
        "Training_Source":
            "Empirical_Failure_Augmentation_JanJul",

        "Rows":
            total_augmented_rows
    }
])

composition[
    "Percent_of_Final_Training"
] = (
    100
    *
    composition[
        "Rows"
    ]
    /
    final_training_rows
)


print("\n" + "=" * 140)
print("9. FINAL JAN–JUL TRAINING COMPOSITION")
print("=" * 140)

print(
    f"Original Jan-Jul rows        : "
    f"{total_original_rows:,}"
)

print(
    f"Clean model-ready rows       : "
    f"{total_clean_rows:,}"
)

print(
    f"Natural missing-history rows : "
    f"{total_natural_rows:,}"
)

print(
    f"Empirical augmented rows     : "
    f"{total_augmented_rows:,}"
)

print(
    f"Invalid target rows          : "
    f"{total_invalid_targets:,}"
)

print(
    f"Invalid non-lag feature rows : "
    f"{total_invalid_nonlag:,}"
)

print(
    f"\nFINAL R1 TRAINING ROWS       : "
    f"{final_training_rows:,}"
)

print("\nComposition:")

print(
    composition.to_string(
        index=False
    )
)


# ============================================================
# 11. Final integrity checks
# ============================================================

all_original_rows_accounted = bool(
    (
        total_clean_rows
        +
        total_natural_rows
        +
        total_invalid_targets
        +
        total_invalid_nonlag
    )
    ==
    total_original_rows
)

all_augmented_recovered = bool(
    len(
        augmented
    )
    ==
    len(
        target_map
    )
)

all_events_represented = bool(
    augmented[
        "Final_Augmentation_Event_ID"
    ].nunique()
    ==
    event_map[
        "Final_Augmentation_Event_ID"
    ].nunique()
)

integrity_pass = bool(
    all_original_rows_accounted
    and
    total_invalid_targets == 0
    and
    total_invalid_nonlag == 0
    and
    all_augmented_recovered
    and
    baseline_missing_total == 0
    and
    baseline_invalid_target_total == 0
    and
    duplicate_augmented_targets == 0
    and
    all_injections_exact
    and
    all_events_represented
)


print("\n" + "=" * 140)
print("10. STEP 22 INTEGRITY DECISION")
print("=" * 140)

print(
    f"All original rows accounted for : "
    f"{all_original_rows_accounted}"
)

print(
    f"No invalid targets              : "
    f"{total_invalid_targets == 0}"
)

print(
    f"No invalid non-lag features     : "
    f"{total_invalid_nonlag == 0}"
)

print(
    f"All augmented targets recovered : "
    f"{all_augmented_recovered}"
)

print(
    f"All augmentation baselines clean: "
    f"{baseline_missing_total == 0}"
)

print(
    f"No duplicate augmented targets  : "
    f"{duplicate_augmented_targets == 0}"
)

print(
    f"All injections exact            : "
    f"{all_injections_exact}"
)

print(
    f"All 2,223 events represented    : "
    f"{all_events_represented}"
)

print(
    f"\nFINAL STEP 22 INTEGRITY PASS: "
    f"{integrity_pass}"
)


# ============================================================
# 12. Save final training components
# ============================================================

natural_output = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Natural_Missing_Training_Rows.parquet"
)

augmented_output = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Empirical_Augmented_Training_Rows.parquet"
)

monthly_audit_output = os.path.join(
    NB5,
    "Tables",
    "Table_Final_JanJul_Training_Monthly_Audit.csv"
)

composition_output = os.path.join(
    NB5,
    "Tables",
    "Table_Final_JanJul_Training_Composition.csv"
)

injection_output = os.path.join(
    NB5,
    "Tables",
    "Table_Final_JanJul_Augmentation_Injection_Validation.csv"
)

aug_pattern_output = os.path.join(
    NB5,
    "Tables",
    "Table_Final_JanJul_Augmented_Missing_Patterns.csv"
)

natural_pattern_output = os.path.join(
    NB5,
    "Tables",
    "Table_Final_JanJul_Natural_Missing_Patterns.csv"
)

metadata_output = os.path.join(
    NB5,
    "Results",
    "Final_JanJul_Training_Data_Summary.json"
)


natural_training.to_parquet(
    natural_output,
    index=False
)

augmented.to_parquet(
    augmented_output,
    index=False
)

monthly_audit.to_csv(
    monthly_audit_output,
    index=False
)

composition.to_csv(
    composition_output,
    index=False
)

injection_audit.to_csv(
    injection_output,
    index=False
)

augmented_pattern_summary.to_csv(
    aug_pattern_output,
    index=False
)

natural_pattern_summary.to_csv(
    natural_pattern_output,
    index=False
)


metadata = {
    "training_period":
        "2026-01-01 to 2026-07-31",

    "final_test_period":
        "August 2026",

    "august_used":
        False,

    "original_rows":
        total_original_rows,

    "clean_model_ready_rows":
        total_clean_rows,

    "natural_missing_history_rows":
        total_natural_rows,

    "empirical_augmented_rows":
        total_augmented_rows,

    "final_training_rows":
        int(
            final_training_rows
        ),

    "source_failure_events":
        int(
            len(
                event_map
            )
        ),

    "augmented_target_rows":
        int(
            len(
                augmented
            )
        ),

    "integrity_pass":
        integrity_pass
}


with open(
    metadata_output,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )


print("\n" + "=" * 140)
print("11. SAVED OUTPUTS")
print("=" * 140)

print(f"✅ {natural_output}")
print(f"✅ {augmented_output}")
print(f"✅ {monthly_audit_output}")
print(f"✅ {composition_output}")
print(f"✅ {injection_output}")
print(f"✅ {aug_pattern_output}")
print(f"✅ {natural_pattern_output}")
print(f"✅ {metadata_output}")

print("\n" + "=" * 140)
print("STEP 22 COMPLETED")
print("=" * 140)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 23: FINAL R1 TRAINING — JAN–JUL
# ============================================================

import os
import gc
import json
import time
import numpy as np
import pandas as pd
import lightgbm as lgb

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(
    BASE_PATH,
    "Notebook_3_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

FORECASTING_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly"
)

NATURAL_PATH = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Natural_Missing_Training_Rows.parquet"
)

AUGMENTED_PATH = os.path.join(
    NB5,
    "Data",
    "Final_JanJul_Empirical_Augmented_Training_Rows.parquet"
)

SUMMARY_PATH = os.path.join(
    NB5,
    "Results",
    "Final_JanJul_Training_Data_Summary.json"
)

MODEL_OUTPUT = os.path.join(
    NB5,
    "Models",
    "FINAL_R1_FailureAware_LightGBM_JanJul_1200.txt"
)

FEATURE_IMPORTANCE_OUTPUT = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_R1_Feature_Importance.csv"
)

METADATA_OUTPUT = os.path.join(
    NB5,
    "Results",
    "FINAL_R1_JanJul_Training_Metadata.json"
)


FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

LAG_COLS = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

MONTHS = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06",
    "2026-07"
]

print("=" * 140)
print("NOTEBOOK 5 — FINAL R1 JAN–JUL TRAINING")
print("=" * 140)

print(
    "\nIMPORTANT: August is NOT used for training."
)


# ============================================================
# 1. Load expected Step 22 counts
# ============================================================

with open(
    SUMMARY_PATH,
    "r",
    encoding="utf-8"
) as f:

    expected_summary = json.load(
        f
    )


expected_clean = int(
    expected_summary[
        "clean_model_ready_rows"
    ]
)

expected_natural = int(
    expected_summary[
        "natural_missing_history_rows"
    ]
)

expected_augmented = int(
    expected_summary[
        "empirical_augmented_rows"
    ]
)

expected_total = int(
    expected_summary[
        "final_training_rows"
    ]
)

print("\n" + "=" * 140)
print("1. EXPECTED FINAL TRAINING COMPOSITION")
print("=" * 140)

print(
    f"Clean rows      : {expected_clean:,}"
)

print(
    f"Natural missing : {expected_natural:,}"
)

print(
    f"Augmented rows  : {expected_augmented:,}"
)

print(
    f"Expected total  : {expected_total:,}"
)


# ============================================================
# 2. Helper: memory-efficient dtype conversion
# ============================================================

def downcast_training_frame(df):

    df["Detector_Code"] = pd.to_numeric(
        df["Detector_Code"],
        downcast="integer"
    )

    df["Interval_Index"] = pd.to_numeric(
        df["Interval_Index"],
        downcast="integer"
    )

    df["DayOfWeek"] = pd.to_numeric(
        df["DayOfWeek"],
        downcast="integer"
    )

    df["IsWeekend"] = pd.to_numeric(
        df["IsWeekend"],
        downcast="integer"
    )

    for lag in LAG_COLS:

        df[lag] = df[lag].astype(
            "float32"
        )

    df[TARGET] = df[TARGET].astype(
        "float32"
    )

    return df


# ============================================================
# 3. Load Jan-Jul CLEAN model-ready rows
# ============================================================

print("\n" + "=" * 140)
print("2. LOADING CLEAN JAN–JUL ROWS")
print("=" * 140)

clean_parts = []
monthly_counts = []

for i, month in enumerate(
    MONTHS,
    start=1
):

    path = os.path.join(
        FORECASTING_PATH,
        f"FORECASTING_{month}.parquet"
    )

    df = pd.read_parquet(
        path,
        columns=FEATURES + [TARGET]
    )

    valid_target = (
        df[TARGET].notna()
        &
        np.isfinite(
            df[TARGET]
        )
        &
        (
            df[TARGET] >= 0
        )
    )

    all_features_valid = (
        df[FEATURES]
        .notna()
        .all(axis=1)
    )

    clean = df.loc[
        valid_target
        &
        all_features_valid,
        FEATURES + [TARGET]
    ].copy()

    clean = downcast_training_frame(
        clean
    )

    clean_parts.append(
        clean
    )

    monthly_counts.append({
        "Month":
            month,

        "Clean_Rows":
            int(
                len(clean)
            )
    })

    print(
        f"[LOAD] Month {i} / {len(MONTHS)} "
        f"({month}) → {len(clean):,} clean rows",
        flush=True
    )

    del df, clean
    gc.collect()


clean_training = pd.concat(
    clean_parts,
    ignore_index=True
)

del clean_parts
gc.collect()

actual_clean = len(
    clean_training
)

print(
    f"\nTotal clean rows loaded: "
    f"{actual_clean:,}"
)


# ============================================================
# 4. Load final NATURAL missing-history rows
# ============================================================

print("\n" + "=" * 140)
print("3. LOADING FINAL NATURAL MISSING-HISTORY ROWS")
print("=" * 140)

natural_training = pd.read_parquet(
    NATURAL_PATH,
    columns=FEATURES + [TARGET]
)

natural_training = downcast_training_frame(
    natural_training
)

actual_natural = len(
    natural_training
)

print(
    f"Natural rows loaded: "
    f"{actual_natural:,}"
)

print(
    f"Rows with missing predictors: "
    f"{natural_training[FEATURES].isna().any(axis=1).sum():,}"
)


# ============================================================
# 5. Load final EMPIRICAL AUGMENTED rows
# ============================================================

print("\n" + "=" * 140)
print("4. LOADING FINAL EMPIRICAL AUGMENTED ROWS")
print("=" * 140)

augmented_training = pd.read_parquet(
    AUGMENTED_PATH,
    columns=FEATURES + [TARGET]
)

augmented_training = downcast_training_frame(
    augmented_training
)

actual_augmented = len(
    augmented_training
)

print(
    f"Augmented rows loaded: "
    f"{actual_augmented:,}"
)

print(
    f"Rows with missing predictors: "
    f"{augmented_training[FEATURES].isna().any(axis=1).sum():,}"
)


# ============================================================
# 6. Verify exact Step 22 composition
# ============================================================

print("\n" + "=" * 140)
print("5. PRE-TRAINING COMPOSITION VERIFICATION")
print("=" * 140)

print(
    f"Clean count exact      : "
    f"{actual_clean == expected_clean}"
)

print(
    f"Natural count exact    : "
    f"{actual_natural == expected_natural}"
)

print(
    f"Augmented count exact  : "
    f"{actual_augmented == expected_augmented}"
)

if not (
    actual_clean == expected_clean
    and
    actual_natural == expected_natural
    and
    actual_augmented == expected_augmented
):

    raise ValueError(
        "Training composition does not match Step 22."
    )


# ============================================================
# 7. Assemble FINAL R1 training data
# ============================================================

print("\n" + "=" * 140)
print("6. ASSEMBLING FINAL R1 TRAINING DATA")
print("=" * 140)

final_training = pd.concat(
    [
        clean_training,
        natural_training,
        augmented_training
    ],
    ignore_index=True
)

actual_total = len(
    final_training
)

missing_predictor_rows = int(
    final_training[
        FEATURES
    ]
    .isna()
    .any(axis=1)
    .sum()
)

invalid_targets = int(
    (
        final_training[
            TARGET
        ].isna()
        |
        ~np.isfinite(
            final_training[
                TARGET
            ]
        )
        |
        (
            final_training[
                TARGET
            ] < 0
        )
    ).sum()
)

print(
    f"Clean rows                : "
    f"{actual_clean:,}"
)

print(
    f"Natural missing rows      : "
    f"{actual_natural:,}"
)

print(
    f"Empirical augmented rows  : "
    f"{actual_augmented:,}"
)

print(
    f"--------------------------------"
)

print(
    f"FINAL TRAINING ROWS       : "
    f"{actual_total:,}"
)

print(
    f"Expected rows             : "
    f"{expected_total:,}"
)

print(
    f"Rows with missing predictor: "
    f"{missing_predictor_rows:,}"
)

print(
    f"Missing-input prevalence   : "
    f"{100*missing_predictor_rows/actual_total:.4f}%"
)

print(
    f"Invalid targets            : "
    f"{invalid_targets:,}"
)


if actual_total != expected_total:

    raise ValueError(
        "Final row count does not match Step 22."
    )

if invalid_targets != 0:

    raise ValueError(
        "Invalid targets found."
    )


# ============================================================
# 8. Preserve tiny sample for post-save verification
# ============================================================

verification_sample = (
    final_training[
        FEATURES
    ]
    .head(10000)
    .copy()
)


# ============================================================
# 9. Release source frames before LightGBM conversion
# ============================================================

del clean_training
del natural_training
del augmented_training

gc.collect()


# ============================================================
# 10. Prepare LightGBM dataset
# ============================================================

print("\n" + "=" * 140)
print("7. PREPARING LIGHTGBM DATASET")
print("=" * 140)

X_train = final_training[
    FEATURES
]

y_train = final_training[
    TARGET
]

print(
    f"X_train shape: "
    f"{X_train.shape}"
)

print(
    f"y_train shape: "
    f"{y_train.shape}"
)

train_dataset = lgb.Dataset(
    X_train,
    label=y_train,
    feature_name=FEATURES,
    free_raw_data=False
)


# ============================================================
# 11. EXACT selected R1 configuration
# ============================================================

FINAL_PARAMS = {
    "objective":
        "regression",

    "metric":
        [
            "l1",
            "rmse",
            "l2"
        ],

    "learning_rate":
        0.05,

    "num_leaves":
        31,

    "max_depth":
        -1,

    "min_data_in_leaf":
        100,

    "feature_fraction":
        0.9,

    "bagging_fraction":
        0.8,

    "bagging_freq":
        1,

    "lambda_l1":
        0,

    "lambda_l2":
        1,

    "seed":
        42,

    "feature_fraction_seed":
        42,

    "bagging_seed":
        42,

    "data_random_seed":
        42,

    "verbosity":
        -1,

    "num_threads":
        -1
}

NUM_BOOST_ROUND = 1200


print("\n" + "=" * 140)
print("8. FINAL FIXED MODEL CONFIGURATION")
print("=" * 140)

print(
    f"Features        : "
    f"{len(FEATURES)}"
)

print(
    f"Boosting rounds : "
    f"{NUM_BOOST_ROUND:,}"
)

print(
    "Selected regime: R1_FailureAware"
)


# ============================================================
# 12. Visible heartbeat
# ============================================================

training_start = time.time()


def training_progress(
    period=50
):

    def _callback(env):

        iteration = (
            env.iteration + 1
        )

        if (
            iteration == 1
            or
            iteration % period == 0
            or
            iteration == env.end_iteration
        ):

            elapsed = (
                time.time()
                -
                training_start
            )

            print(
                f"[FINAL R1 TRAINING] "
                f"Iteration {iteration:,} / "
                f"{env.end_iteration:,} "
                f"| Elapsed: {elapsed:.1f} sec",
                flush=True
            )

    _callback.order = 10

    return _callback


# ============================================================
# 13. FINAL TRAINING
# ============================================================

print("\n" + "=" * 140)
print("9. FINAL R1 TRAINING — JAN–JUL")
print("=" * 140)

print(
    "Progress will be displayed every 50 iterations.",
    flush=True
)

final_r1 = lgb.train(
    params=FINAL_PARAMS,
    train_set=train_dataset,
    num_boost_round=NUM_BOOST_ROUND,
    callbacks=[
        training_progress(
            period=50
        )
    ]
)

training_seconds = (
    time.time()
    -
    training_start
)

print(
    "\n✅ FINAL R1 training completed"
)

print(
    f"Training time : "
    f"{training_seconds:.2f} seconds"
)

print(
    f"Trees         : "
    f"{final_r1.num_trees():,}"
)

print(
    f"Features      : "
    f"{final_r1.num_feature():,}"
)


# ============================================================
# 14. Save final model immediately
# ============================================================

final_r1.save_model(
    MODEL_OUTPUT
)

print(
    f"\n✅ FINAL MODEL SAVED:\n"
    f"{MODEL_OUTPUT}"
)


# ============================================================
# 15. Verify saved model can be reloaded exactly
# ============================================================

reloaded_model = lgb.Booster(
    model_file=MODEL_OUTPUT
)

pred_original = final_r1.predict(
    verification_sample
)

pred_reloaded = reloaded_model.predict(
    verification_sample
)

max_reload_difference = float(
    np.max(
        np.abs(
            pred_original
            -
            pred_reloaded
        )
    )
)

reload_pass = bool(
    max_reload_difference
    <
    1e-10
)

print("\n" + "=" * 140)
print("10. SAVED MODEL VERIFICATION")
print("=" * 140)

print(
    f"Reloaded trees    : "
    f"{reloaded_model.num_trees():,}"
)

print(
    f"Reloaded features : "
    f"{reloaded_model.num_feature():,}"
)

print(
    f"Maximum prediction difference: "
    f"{max_reload_difference:.12f}"
)

print(
    f"Reload verification pass      : "
    f"{reload_pass}"
)


# ============================================================
# 16. Feature importance
# ============================================================

feature_importance = pd.DataFrame({
    "Feature":
        FEATURES,

    "Split_Importance":
        final_r1.feature_importance(
            importance_type="split"
        ),

    "Gain_Importance":
        final_r1.feature_importance(
            importance_type="gain"
        )
})

feature_importance[
    "Gain_Percent"
] = (
    100
    *
    feature_importance[
        "Gain_Importance"
    ]
    /
    feature_importance[
        "Gain_Importance"
    ].sum()
)

feature_importance = (
    feature_importance
    .sort_values(
        "Gain_Importance",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

print("\n" + "=" * 140)
print("11. FINAL MODEL FEATURE IMPORTANCE")
print("=" * 140)

print(
    feature_importance
    .to_string(
        index=False
    )
)


# ============================================================
# 17. Save feature importance + metadata
# ============================================================

feature_importance.to_csv(
    FEATURE_IMPORTANCE_OUTPUT,
    index=False
)

metadata = {
    "selected_regime":
        "R1_FailureAware",

    "training_period":
        "2026-01-01 to 2026-07-31",

    "final_test_period":
        "August 2026",

    "august_used_for_training":
        False,

    "feature_count":
        len(
            FEATURES
        ),

    "features":
        FEATURES,

    "clean_rows":
        int(
            actual_clean
        ),

    "natural_missing_rows":
        int(
            actual_natural
        ),

    "empirical_augmented_rows":
        int(
            actual_augmented
        ),

    "total_training_rows":
        int(
            actual_total
        ),

    "rows_with_missing_predictors":
        int(
            missing_predictor_rows
        ),

    "num_boost_round":
        NUM_BOOST_ROUND,

    "parameters":
        FINAL_PARAMS,

    "training_seconds":
        float(
            training_seconds
        ),

    "model_path":
        MODEL_OUTPUT,

    "reload_verification_pass":
        reload_pass,

    "maximum_reload_prediction_difference":
        max_reload_difference
}

with open(
    METADATA_OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )


# ============================================================
# 18. Final integrity decision
# ============================================================

integrity_pass = bool(
    actual_clean == expected_clean
    and
    actual_natural == expected_natural
    and
    actual_augmented == expected_augmented
    and
    actual_total == expected_total
    and
    invalid_targets == 0
    and
    final_r1.num_trees() == 1200
    and
    final_r1.num_feature() == 9
    and
    reload_pass
)

print("\n" + "=" * 140)
print("12. STEP 23 FINAL TRAINING INTEGRITY")
print("=" * 140)

print(
    f"Training composition exact : "
    f"{actual_total == expected_total}"
)

print(
    f"No invalid targets         : "
    f"{invalid_targets == 0}"
)

print(
    f"Exactly 1,200 trees        : "
    f"{final_r1.num_trees() == 1200}"
)

print(
    f"Exactly 9 features         : "
    f"{final_r1.num_feature() == 9}"
)

print(
    f"Saved model reloads exactly: "
    f"{reload_pass}"
)

print(
    f"\nFINAL STEP 23 INTEGRITY PASS: "
    f"{integrity_pass}"
)


print("\n" + "=" * 140)
print("13. SAVED OUTPUTS")
print("=" * 140)

print(f"✅ {MODEL_OUTPUT}")
print(f"✅ {FEATURE_IMPORTANCE_OUTPUT}")
print(f"✅ {METADATA_OUTPUT}")

print("\n" + "=" * 140)
print("STEP 23 COMPLETED — FINAL MODEL TRAINING FINISHED")
print("=" * 140)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 24: FINAL UNTOUCHED AUGUST TEST
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import lightgbm as lgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB3 = os.path.join(BASE_PATH, "Notebook_3_Output")
NB4 = os.path.join(BASE_PATH, "Notebook_4_Output")
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

AUGUST_PATH = os.path.join(
    NB3,
    "Data",
    "Forecasting_Monthly",
    "FORECASTING_2026-08.parquet"
)

R0_MODEL_PATH = os.path.join(
    NB3,
    "Models",
    "LightGBM_FINAL_JanJul_1200.txt"
)

R1_MODEL_PATH = os.path.join(
    NB5,
    "Models",
    "FINAL_R1_FailureAware_LightGBM_JanJul_1200.txt"
)

AUGUST_EVENT_PATH = os.path.join(
    NB4,
    "Data",
    "Controlled_August_Failure_Injection_Events.csv"
)

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

LAG_COLS = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

LAG_OFFSETS = {
    "Lag_1": 15,
    "Lag_2": 30,
    "Lag_4": 60,
    "Lag_8": 120,
    "Lag_96": 1440
}

TARGET = "Target_Volume"

print("=" * 145)
print("NOTEBOOK 5 — FINAL UNTOUCHED AUGUST TEST")
print("=" * 145)

print("\nIMPORTANT:")
print(
    "No training, tuning, model selection, or parameter "
    "changes are performed in this step."
)


# ============================================================
# 1. Load frozen models
# ============================================================

r0 = lgb.Booster(
    model_file=R0_MODEL_PATH
)

r1 = lgb.Booster(
    model_file=R1_MODEL_PATH
)

print("\n" + "=" * 145)
print("1. FROZEN FINAL MODELS")
print("=" * 145)

print(
    f"R0 trees/features : "
    f"{r0.num_trees():,} / {r0.num_feature()}"
)

print(
    f"R1 trees/features : "
    f"{r1.num_trees():,} / {r1.num_feature()}"
)

print(
    f"R0 feature match  : "
    f"{r0.feature_name() == FEATURES}"
)

print(
    f"R1 feature match  : "
    f"{r1.feature_name() == FEATURES}"
)


# ============================================================
# 2. Metric function
# ============================================================

def calculate_metrics(y_true, y_pred):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    denominator = np.sum(
        np.abs(y_true)
    )

    wape = (
        100
        *
        np.sum(
            np.abs(
                y_true - y_pred
            )
        )
        /
        denominator
    )

    bias = np.mean(
        y_pred - y_true
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "WAPE": float(wape),
        "Bias": float(bias)
    }


# ============================================================
# 3. Load full August data
# ============================================================

august = pd.read_parquet(
    AUGUST_PATH,
    columns=[
        "Timestamp"
    ]
    +
    FEATURES
    +
    [
        TARGET
    ]
)

august["Timestamp"] = pd.to_datetime(
    august["Timestamp"]
)

valid_target = (
    august[TARGET].notna()
    &
    np.isfinite(
        august[TARGET]
    )
    &
    (
        august[TARGET] >= 0
    )
)

all_features_valid = (
    august[FEATURES]
    .notna()
    .all(axis=1)
)

clean_mask = (
    valid_target
    &
    all_features_valid
)

august_clean = august.loc[
    clean_mask
].copy()

print("\n" + "=" * 145)
print("2. FULL AUGUST CLEAN TEST DATA")
print("=" * 145)

print(
    f"Full August rows   : "
    f"{len(august):,}"
)

print(
    f"Clean model-ready  : "
    f"{len(august_clean):,}"
)

print(
    f"Excluded rows      : "
    f"{len(august)-len(august_clean):,}"
)

print(
    f"Model-ready rate   : "
    f"{100*len(august_clean)/len(august):.4f}%"
)


# ============================================================
# 4. Full clean August predictions
# ============================================================

X_august_clean = august_clean[
    FEATURES
]

y_august_clean = august_clean[
    TARGET
].to_numpy(
    dtype=np.float64
)

r0_full_clean_pred = np.maximum(
    r0.predict(
        X_august_clean
    ),
    0
)

r1_full_clean_pred = np.maximum(
    r1.predict(
        X_august_clean
    ),
    0
)

r0_full_clean_metrics = calculate_metrics(
    y_august_clean,
    r0_full_clean_pred
)

r1_full_clean_metrics = calculate_metrics(
    y_august_clean,
    r1_full_clean_pred
)

print("\n" + "=" * 145)
print("3. FULL AUGUST CLEAN PERFORMANCE")
print("=" * 145)

print("\nR0 — ORIGINAL CLEAN-ONLY")
print("-" * 60)

for k, v in r0_full_clean_metrics.items():
    print(
        f"{k:<10}: {v:.6f}"
    )

print("\nFINAL R1 — FAILURE-AWARE")
print("-" * 60)

for k, v in r1_full_clean_metrics.items():
    print(
        f"{k:<10}: {v:.6f}"
    )


# ============================================================
# 5. Load controlled August events
# ============================================================

events = pd.read_csv(
    AUGUST_EVENT_PATH
)

if "Injection_Event_ID" in events.columns:
    EVENT_ID_COL = "Injection_Event_ID"

elif "Event_ID" in events.columns:
    EVENT_ID_COL = "Event_ID"

else:
    event_candidates = [
        c
        for c in events.columns
        if (
            "event" in c.lower()
            and
            "id" in c.lower()
        )
    ]

    if len(event_candidates) == 0:
        raise ValueError(
            "Could not identify August event ID column."
        )

    EVENT_ID_COL = event_candidates[0]


for col in [
    "Failure_Start",
    "Failure_End",
    "Evaluation_End",
    "Target_Injection_Date"
]:

    if col in events.columns:
        events[col] = pd.to_datetime(
            events[col]
        )


print("\n" + "=" * 145)
print("4. CONTROLLED AUGUST FAILURE EVENTS")
print("=" * 145)

print(
    f"Events           : "
    f"{len(events):,}"
)

print(
    f"Unique detectors : "
    f"{events['Detector_Code'].nunique():,}"
)

print(
    f"Event ID column  : "
    f"{EVENT_ID_COL}"
)


# ============================================================
# 6. Build exact affected target map
# ============================================================

impact_records = []

for row in events.itertuples(
    index=False
):

    event_id = getattr(
        row,
        EVENT_ID_COL
    )

    failure_start = pd.Timestamp(
        row.Failure_Start
    )

    run_length = int(
        row.Run_Length
    )

    failure_times = pd.date_range(
        start=failure_start,
        periods=run_length,
        freq="15min"
    )

    for lag_name, offset in LAG_OFFSETS.items():

        target_times = (
            failure_times
            +
            pd.Timedelta(
                minutes=offset
            )
        )

        for target_ts in target_times:

            impact_records.append({
                "Injection_Event_ID":
                    int(event_id),

                "Detector_Code":
                    int(
                        row.Detector_Code
                    ),

                "Timestamp":
                    pd.Timestamp(
                        target_ts
                    ),

                "Missing_Lag":
                    lag_name
            })


impact_long = pd.DataFrame(
    impact_records
).drop_duplicates()


target_map = (
    impact_long[
        [
            "Injection_Event_ID",
            "Detector_Code",
            "Timestamp"
        ]
    ]
    .drop_duplicates()
    .copy()
)


for lag in LAG_COLS:

    flag = (
        impact_long[
            impact_long[
                "Missing_Lag"
            ] == lag
        ][
            [
                "Injection_Event_ID",
                "Detector_Code",
                "Timestamp"
            ]
        ]
        .drop_duplicates()
        .copy()
    )

    flag[
        f"Missing_{lag}"
    ] = np.int8(1)

    target_map = target_map.merge(
        flag,
        on=[
            "Injection_Event_ID",
            "Detector_Code",
            "Timestamp"
        ],
        how="left"
    )


for lag in LAG_COLS:

    target_map[
        f"Missing_{lag}"
    ] = (
        target_map[
            f"Missing_{lag}"
        ]
        .fillna(0)
        .astype("int8")
    )


target_map[
    "Missing_Lag_Count"
] = (
    target_map[
        [
            f"Missing_{lag}"
            for lag in LAG_COLS
        ]
    ]
    .sum(axis=1)
    .astype("int8")
)


def build_pattern(row):

    missing = [
        lag
        for lag in LAG_COLS
        if row[
            f"Missing_{lag}"
        ] == 1
    ]

    return "+".join(
        missing
    )


target_map[
    "Missing_Lag_Pattern"
] = target_map.apply(
    build_pattern,
    axis=1
)


print("\n" + "=" * 145)
print("5. CONTROLLED AUGUST AFFECTED TARGETS")
print("=" * 145)

print(
    f"Raw lag impacts  : "
    f"{len(impact_long):,}"
)

print(
    f"Affected targets : "
    f"{len(target_map):,}"
)

print(
    f"Events represented: "
    f"{target_map['Injection_Event_ID'].nunique():,}"
)

print(
    f"Duplicate detector-targets: "
    f"{target_map.duplicated(['Detector_Code','Timestamp']).sum():,}"
)


# ============================================================
# 7. Extract clean controlled baselines
# ============================================================

controlled = august.merge(
    target_map,
    on=[
        "Detector_Code",
        "Timestamp"
    ],
    how="inner",
    validate="one_to_one"
)

baseline_missing = int(
    controlled[
        LAG_COLS
    ]
    .isna()
    .any(axis=1)
    .sum()
)

invalid_controlled_targets = int(
    (
        controlled[
            TARGET
        ].isna()
        |
        ~np.isfinite(
            controlled[
                TARGET
            ]
        )
        |
        (
            controlled[
                TARGET
            ] < 0
        )
    ).sum()
)


print("\n" + "=" * 145)
print("6. CONTROLLED AUGUST BASELINE INTEGRITY")
print("=" * 145)

print(
    f"Expected affected rows : "
    f"{len(target_map):,}"
)

print(
    f"Extracted rows         : "
    f"{len(controlled):,}"
)

print(
    f"Baseline missing lags  : "
    f"{baseline_missing:,}"
)

print(
    f"Invalid targets        : "
    f"{invalid_controlled_targets:,}"
)


# ============================================================
# 8. Inject missing predictors
# ============================================================

controlled_injected = (
    controlled[
        FEATURES
    ].copy()
)

for lag in LAG_COLS:

    mask = (
        controlled[
            f"Missing_{lag}"
        ] == 1
    )

    controlled_injected.loc[
        mask,
        lag
    ] = np.nan


injection_checks = []

for lag in LAG_COLS:

    intended = int(
        controlled[
            f"Missing_{lag}"
        ].sum()
    )

    actual = int(
        controlled_injected[
            lag
        ].isna()
        .sum()
    )

    injection_checks.append({
        "Lag":
            lag,

        "Intended_Missing":
            intended,

        "Actual_Missing":
            actual,

        "Exact_Match":
            bool(
                intended == actual
            )
    })


injection_checks = pd.DataFrame(
    injection_checks
)

all_injections_exact = bool(
    injection_checks[
        "Exact_Match"
    ].all()
)


print("\nInjection validation:")

print(
    injection_checks.to_string(
        index=False
    )
)

print(
    f"\nAll injections exact: "
    f"{all_injections_exact}"
)


# ============================================================
# 9. Predictions
# ============================================================

y_controlled = controlled[
    TARGET
].to_numpy(
    dtype=np.float64
)

X_controlled_clean = controlled[
    FEATURES
]

r0_controlled_clean_pred = np.maximum(
    r0.predict(
        X_controlled_clean
    ),
    0
)

r0_controlled_injected_pred = np.maximum(
    r0.predict(
        controlled_injected
    ),
    0
)

r1_controlled_clean_pred = np.maximum(
    r1.predict(
        X_controlled_clean
    ),
    0
)

r1_controlled_injected_pred = np.maximum(
    r1.predict(
        controlled_injected
    ),
    0
)


r0_controlled_clean_metrics = calculate_metrics(
    y_controlled,
    r0_controlled_clean_pred
)

r0_injected_metrics = calculate_metrics(
    y_controlled,
    r0_controlled_injected_pred
)

r1_controlled_clean_metrics = calculate_metrics(
    y_controlled,
    r1_controlled_clean_pred
)

r1_injected_metrics = calculate_metrics(
    y_controlled,
    r1_controlled_injected_pred
)


# ============================================================
# 10. Final results
# ============================================================

print("\n" + "=" * 145)
print("7. FINAL AUGUST CONTROLLED TEST RESULTS")
print("=" * 145)

print("\nR0 — CONTROLLED CLEAN")
print("-" * 60)

for k, v in r0_controlled_clean_metrics.items():
    print(
        f"{k:<10}: {v:.6f}"
    )

print("\nR0 — FAILURE-INJECTED")
print("-" * 60)

for k, v in r0_injected_metrics.items():
    print(
        f"{k:<10}: {v:.6f}"
    )

print("\nFINAL R1 — CONTROLLED CLEAN")
print("-" * 60)

for k, v in r1_controlled_clean_metrics.items():
    print(
        f"{k:<10}: {v:.6f}"
    )

print("\nFINAL R1 — FAILURE-INJECTED")
print("-" * 60)

for k, v in r1_injected_metrics.items():
    print(
        f"{k:<10}: {v:.6f}"
    )


# ============================================================
# 11. Mitigation effect
# ============================================================

clean_wape_delta = (
    r1_full_clean_metrics["WAPE"]
    -
    r0_full_clean_metrics["WAPE"]
)

injected_wape_delta = (
    r1_injected_metrics["WAPE"]
    -
    r0_injected_metrics["WAPE"]
)

injected_wape_improvement = (
    100
    *
    (
        r0_injected_metrics["WAPE"]
        -
        r1_injected_metrics["WAPE"]
    )
    /
    r0_injected_metrics["WAPE"]
)

injected_mae_improvement = (
    100
    *
    (
        r0_injected_metrics["MAE"]
        -
        r1_injected_metrics["MAE"]
    )
    /
    r0_injected_metrics["MAE"]
)

injected_rmse_improvement = (
    100
    *
    (
        r0_injected_metrics["RMSE"]
        -
        r1_injected_metrics["RMSE"]
    )
    /
    r0_injected_metrics["RMSE"]
)


print("\n" + "=" * 145)
print("8. FINAL MITIGATION EFFECT — AUGUST")
print("=" * 145)

print(
    f"Clean WAPE — R0              : "
    f"{r0_full_clean_metrics['WAPE']:.6f}%"
)

print(
    f"Clean WAPE — Final R1        : "
    f"{r1_full_clean_metrics['WAPE']:.6f}%"
)

print(
    f"Clean WAPE change (R1-R0)    : "
    f"{clean_wape_delta:+.6f} pp"
)

print()

print(
    f"Injected WAPE — R0           : "
    f"{r0_injected_metrics['WAPE']:.6f}%"
)

print(
    f"Injected WAPE — Final R1     : "
    f"{r1_injected_metrics['WAPE']:.6f}%"
)

print(
    f"Injected WAPE change         : "
    f"{injected_wape_delta:+.6f} pp"
)

print(
    f"WAPE improvement             : "
    f"{injected_wape_improvement:.2f}%"
)

print(
    f"MAE improvement              : "
    f"{injected_mae_improvement:.2f}%"
)

print(
    f"RMSE improvement             : "
    f"{injected_rmse_improvement:.2f}%"
)

print(
    f"R0 injected bias             : "
    f"{r0_injected_metrics['Bias']:.6f}"
)

print(
    f"R1 injected bias             : "
    f"{r1_injected_metrics['Bias']:.6f}"
)


# ============================================================
# 12. Notebook 4 reproduction check
# ============================================================

REFERENCE_R0_AUGUST_WAPE = 42.8338

reproduction_difference = (
    r0_injected_metrics["WAPE"]
    -
    REFERENCE_R0_AUGUST_WAPE
)

print("\n" + "=" * 145)
print("9. NOTEBOOK 4 R0 REPRODUCTION CHECK")
print("=" * 145)

print(
    f"Notebook 4 reported R0 injected WAPE : "
    f"{REFERENCE_R0_AUGUST_WAPE:.4f}%"
)

print(
    f"Step 24 reproduced R0 injected WAPE  : "
    f"{r0_injected_metrics['WAPE']:.4f}%"
)

print(
    f"Difference                             : "
    f"{reproduction_difference:+.6f} pp"
)


# ============================================================
# 13. Save row-level predictions
# ============================================================

prediction_table = pd.DataFrame({
    "Injection_Event_ID":
        controlled[
            "Injection_Event_ID"
        ].values,

    "Detector_Code":
        controlled[
            "Detector_Code"
        ].values,

    "Timestamp":
        controlled[
            "Timestamp"
        ].values,

    "Actual":
        y_controlled,

    "Missing_Lag_Count":
        controlled[
            "Missing_Lag_Count"
        ].values,

    "Missing_Lag_Pattern":
        controlled[
            "Missing_Lag_Pattern"
        ].values,

    "R0_Clean_Prediction":
        r0_controlled_clean_pred,

    "R0_Injected_Prediction":
        r0_controlled_injected_pred,

    "Final_R1_Clean_Prediction":
        r1_controlled_clean_pred,

    "Final_R1_Injected_Prediction":
        r1_controlled_injected_pred
})


# ============================================================
# 14. Save summary tables
# ============================================================

performance_table = pd.DataFrame([
    {
        "Model":
            "R0_CleanOnly",

        "Condition":
            "Full_August_Clean",

        **r0_full_clean_metrics
    },

    {
        "Model":
            "Final_R1_FailureAware",

        "Condition":
            "Full_August_Clean",

        **r1_full_clean_metrics
    },

    {
        "Model":
            "R0_CleanOnly",

        "Condition":
            "Controlled_August_Clean",

        **r0_controlled_clean_metrics
    },

    {
        "Model":
            "Final_R1_FailureAware",

        "Condition":
            "Controlled_August_Clean",

        **r1_controlled_clean_metrics
    },

    {
        "Model":
            "R0_CleanOnly",

        "Condition":
            "Controlled_August_Injected",

        **r0_injected_metrics
    },

    {
        "Model":
            "Final_R1_FailureAware",

        "Condition":
            "Controlled_August_Injected",

        **r1_injected_metrics
    }
])


comparison_table = pd.DataFrame([
    {
        "Metric": "Full_Clean_WAPE",
        "R0": r0_full_clean_metrics["WAPE"],
        "Final_R1": r1_full_clean_metrics["WAPE"]
    },

    {
        "Metric": "Injected_WAPE",
        "R0": r0_injected_metrics["WAPE"],
        "Final_R1": r1_injected_metrics["WAPE"]
    },

    {
        "Metric": "Injected_MAE",
        "R0": r0_injected_metrics["MAE"],
        "Final_R1": r1_injected_metrics["MAE"]
    },

    {
        "Metric": "Injected_RMSE",
        "R0": r0_injected_metrics["RMSE"],
        "Final_R1": r1_injected_metrics["RMSE"]
    },

    {
        "Metric": "Injected_Bias",
        "R0": r0_injected_metrics["Bias"],
        "Final_R1": r1_injected_metrics["Bias"]
    }
])

comparison_table[
    "Final_R1_minus_R0"
] = (
    comparison_table[
        "Final_R1"
    ]
    -
    comparison_table[
        "R0"
    ]
)


# ============================================================
# 15. Integrity
# ============================================================

integrity_pass = bool(
    len(events) == 604
    and
    len(target_map) == 12662
    and
    len(controlled) == 12662
    and
    baseline_missing == 0
    and
    invalid_controlled_targets == 0
    and
    all_injections_exact
    and
    r0.num_trees() == 1200
    and
    r1.num_trees() == 1200
)


print("\n" + "=" * 145)
print("10. STEP 24 FINAL TEST INTEGRITY")
print("=" * 145)

print(
    f"604 controlled events recovered : "
    f"{len(events) == 604}"
)

print(
    f"12,662 affected rows recovered  : "
    f"{len(target_map) == 12662 and len(controlled) == 12662}"
)

print(
    f"Controlled baselines clean      : "
    f"{baseline_missing == 0}"
)

print(
    f"Targets valid                   : "
    f"{invalid_controlled_targets == 0}"
)

print(
    f"All missing-lag injections exact: "
    f"{all_injections_exact}"
)

print(
    f"\nFINAL STEP 24 INTEGRITY PASS: "
    f"{integrity_pass}"
)


# ============================================================
# 16. Save outputs
# ============================================================

performance_path = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_August_Performance.csv"
)

comparison_path = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_August_R0_vs_R1_Comparison.csv"
)

prediction_path = os.path.join(
    NB5,
    "Data",
    "FINAL_August_Controlled_R0_R1_Predictions.parquet"
)

injection_path = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_August_Injection_Validation.csv"
)

summary_path = os.path.join(
    NB5,
    "Results",
    "FINAL_August_Untouched_Test_Summary.json"
)

performance_table.to_csv(
    performance_path,
    index=False
)

comparison_table.to_csv(
    comparison_path,
    index=False
)

prediction_table.to_parquet(
    prediction_path,
    index=False
)

injection_checks.to_csv(
    injection_path,
    index=False
)


summary = {
    "test_period":
        "August 2026",

    "test_role":
        "Final untouched holdout",

    "model_selection_completed_before_august":
        True,

    "training_or_tuning_performed_on_august":
        False,

    "controlled_events":
        int(len(events)),

    "controlled_affected_rows":
        int(len(controlled)),

    "R0_full_clean_metrics":
        r0_full_clean_metrics,

    "Final_R1_full_clean_metrics":
        r1_full_clean_metrics,

    "R0_failure_injected_metrics":
        r0_injected_metrics,

    "Final_R1_failure_injected_metrics":
        r1_injected_metrics,

    "clean_WAPE_change_pp":
        float(clean_wape_delta),

    "failure_WAPE_improvement_percent":
        float(
            injected_wape_improvement
        ),

    "failure_MAE_improvement_percent":
        float(
            injected_mae_improvement
        ),

    "failure_RMSE_improvement_percent":
        float(
            injected_rmse_improvement
        ),

    "integrity_pass":
        integrity_pass
}


with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )


print("\n" + "=" * 145)
print("11. SAVED OUTPUTS")
print("=" * 145)

print(f"✅ {performance_path}")
print(f"✅ {comparison_path}")
print(f"✅ {prediction_path}")
print(f"✅ {injection_path}")
print(f"✅ {summary_path}")

print("\n" + "=" * 145)
print("STEP 24 COMPLETED — FINAL AUGUST TEST FINISHED")
print("=" * 145)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 25: FINAL EVENT-LEVEL & STATISTICAL ANALYSIS
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from scipy.stats import wilcoxon

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)

NB4 = os.path.join(
    BASE_PATH,
    "Notebook_4_Output"
)

NB5 = os.path.join(
    BASE_PATH,
    "Notebook_5_Output"
)

PRED_PATH = os.path.join(
    NB5,
    "Data",
    "FINAL_August_Controlled_R0_R1_Predictions.parquet"
)

EVENT_PATH = os.path.join(
    NB4,
    "Data",
    "Controlled_August_Failure_Injection_Events.csv"
)

print("=" * 145)
print("NOTEBOOK 5 — FINAL AUGUST EVENT-LEVEL & STATISTICAL ANALYSIS")
print("=" * 145)


# ============================================================
# 1. Load final August predictions
# ============================================================

df = pd.read_parquet(
    PRED_PATH
)

df["Timestamp"] = pd.to_datetime(
    df["Timestamp"]
)

print("\n" + "=" * 145)
print("1. FINAL CONTROLLED AUGUST DATA")
print("=" * 145)

print(
    f"Rows   : {len(df):,}"
)

print(
    f"Events : "
    f"{df['Injection_Event_ID'].nunique():,}"
)

print(
    f"Detectors: "
    f"{df['Detector_Code'].nunique():,}"
)


# ============================================================
# 2. Calculate row-level absolute errors
# ============================================================

df[
    "R0_Clean_AE"
] = np.abs(
    df["Actual"]
    -
    df["R0_Clean_Prediction"]
)

df[
    "R0_Injected_AE"
] = np.abs(
    df["Actual"]
    -
    df["R0_Injected_Prediction"]
)

df[
    "R1_Clean_AE"
] = np.abs(
    df["Actual"]
    -
    df["Final_R1_Clean_Prediction"]
)

df[
    "R1_Injected_AE"
] = np.abs(
    df["Actual"]
    -
    df["Final_R1_Injected_Prediction"]
)


# ============================================================
# 3. Event-level MAE
# ============================================================

event_level = (
    df
    .groupby(
        "Injection_Event_ID",
        as_index=False
    )
    .agg(
        Detector_Code=(
            "Detector_Code",
            "first"
        ),

        Affected_Rows=(
            "Timestamp",
            "size"
        ),

        R0_Clean_MAE=(
            "R0_Clean_AE",
            "mean"
        ),

        R0_Injected_MAE=(
            "R0_Injected_AE",
            "mean"
        ),

        R1_Clean_MAE=(
            "R1_Clean_AE",
            "mean"
        ),

        R1_Injected_MAE=(
            "R1_Injected_AE",
            "mean"
        )
    )
)

event_level[
    "R0_Degradation"
] = (
    event_level[
        "R0_Injected_MAE"
    ]
    -
    event_level[
        "R0_Clean_MAE"
    ]
)

event_level[
    "R1_Degradation"
] = (
    event_level[
        "R1_Injected_MAE"
    ]
    -
    event_level[
        "R1_Clean_MAE"
    ]
)

# Positive value = R1 reduced failure error
event_level[
    "Mitigation_Benefit_MAE"
] = (
    event_level[
        "R0_Injected_MAE"
    ]
    -
    event_level[
        "R1_Injected_MAE"
    ]
)

event_level[
    "R1_Better"
] = (
    event_level[
        "Mitigation_Benefit_MAE"
    ] > 0
)


# ============================================================
# 4. Add empirical failure-event metadata
# ============================================================

events = pd.read_csv(
    EVENT_PATH
)

metadata_cols = [
    "Injection_Event_ID",
    "Run_Length",
    "Failure_Duration_Hours",
    "Failure_Duration_Group"
]

available_metadata = [
    c
    for c in metadata_cols
    if c in events.columns
]

event_level = event_level.merge(
    events[
        available_metadata
    ],
    on="Injection_Event_ID",
    how="left",
    validate="one_to_one"
)


print("\n" + "=" * 145)
print("2. EVENT-LEVEL MITIGATION RESULTS")
print("=" * 145)

n_events = len(
    event_level
)

n_improved = int(
    event_level[
        "R1_Better"
    ].sum()
)

n_worse = int(
    (
        event_level[
            "Mitigation_Benefit_MAE"
        ] < 0
    ).sum()
)

n_ties = (
    n_events
    -
    n_improved
    -
    n_worse
)

improved_percent = (
    100
    *
    n_improved
    /
    n_events
)

print(
    f"Events improved by Final R1 : "
    f"{n_improved:,} / {n_events:,} "
    f"({improved_percent:.2f}%)"
)

print(
    f"Events worse               : "
    f"{n_worse:,}"
)

print(
    f"Ties                       : "
    f"{n_ties:,}"
)

print(
    f"\nMean R0 injected event MAE : "
    f"{event_level['R0_Injected_MAE'].mean():.6f}"
)

print(
    f"Mean R1 injected event MAE : "
    f"{event_level['R1_Injected_MAE'].mean():.6f}"
)

print(
    f"Mean mitigation benefit    : "
    f"{event_level['Mitigation_Benefit_MAE'].mean():.6f}"
)

print(
    f"Median mitigation benefit  : "
    f"{event_level['Mitigation_Benefit_MAE'].median():.6f}"
)


# ============================================================
# 5. Bootstrap confidence interval
# ============================================================

RANDOM_SEED = 42
N_BOOT = 10000

rng = np.random.default_rng(
    RANDOM_SEED
)

benefits = event_level[
    "Mitigation_Benefit_MAE"
].to_numpy(
    dtype=np.float64
)

bootstrap_means = np.empty(
    N_BOOT,
    dtype=np.float64
)

for i in range(
    N_BOOT
):

    sample = rng.choice(
        benefits,
        size=len(benefits),
        replace=True
    )

    bootstrap_means[i] = (
        sample.mean()
    )


mean_benefit = float(
    benefits.mean()
)

ci_low = float(
    np.percentile(
        bootstrap_means,
        2.5
    )
)

ci_high = float(
    np.percentile(
        bootstrap_means,
        97.5
    )
)


print("\n" + "=" * 145)
print("3. BOOTSTRAP CONFIDENCE INTERVAL")
print("=" * 145)

print(
    f"Mean event-level MAE reduction : "
    f"{mean_benefit:.6f}"
)

print(
    f"95% bootstrap CI               : "
    f"[{ci_low:.6f}, {ci_high:.6f}]"
)


# ============================================================
# 6. Paired Wilcoxon tests
# ============================================================

r0_event_error = event_level[
    "R0_Injected_MAE"
].to_numpy()

r1_event_error = event_level[
    "R1_Injected_MAE"
].to_numpy()


wilcox_two_sided = wilcoxon(
    r0_event_error,
    r1_event_error,
    alternative="two-sided"
)

wilcox_one_sided = wilcoxon(
    r0_event_error,
    r1_event_error,
    alternative="greater"
)


print("\n" + "=" * 145)
print("4. PAIRED WILCOXON TEST")
print("=" * 145)

print(
    f"Two-sided statistic : "
    f"{wilcox_two_sided.statistic}"
)

print(
    f"Two-sided p-value   : "
    f"{wilcox_two_sided.pvalue:.12g}"
)

print(
    f"\nOne-sided hypothesis:"
)

print(
    "R0 failure error > Final R1 failure error"
)

print(
    f"One-sided p-value   : "
    f"{wilcox_one_sided.pvalue:.12g}"
)


# ============================================================
# 7. Compare degradation under failures
# ============================================================

print("\n" + "=" * 145)
print("5. EVENT-LEVEL FAILURE DEGRADATION")
print("=" * 145)

print(
    f"R0 mean degradation : "
    f"{event_level['R0_Degradation'].mean():.6f}"
)

print(
    f"R1 mean degradation : "
    f"{event_level['R1_Degradation'].mean():.6f}"
)

degradation_reduction = (
    100
    *
    (
        event_level[
            "R0_Degradation"
        ].mean()
        -
        event_level[
            "R1_Degradation"
        ].mean()
    )
    /
    event_level[
        "R0_Degradation"
    ].mean()
)

print(
    f"Mean degradation reduction: "
    f"{degradation_reduction:.2f}%"
)


# ============================================================
# 8. Failure-duration analysis
# ============================================================

if (
    "Failure_Duration_Group"
    in event_level.columns
):

    duration_summary = (
        event_level
        .groupby(
            "Failure_Duration_Group",
            as_index=False
        )
        .agg(
            Events=(
                "Injection_Event_ID",
                "size"
            ),

            R0_Injected_MAE=(
                "R0_Injected_MAE",
                "mean"
            ),

            R1_Injected_MAE=(
                "R1_Injected_MAE",
                "mean"
            ),

            Mean_Mitigation_Benefit=(
                "Mitigation_Benefit_MAE",
                "mean"
            ),

            Median_Mitigation_Benefit=(
                "Mitigation_Benefit_MAE",
                "median"
            ),

            Improved_Events=(
                "R1_Better",
                "sum"
            )
        )
    )

    duration_summary[
        "Improved_Percent"
    ] = (
        100
        *
        duration_summary[
            "Improved_Events"
        ]
        /
        duration_summary[
            "Events"
        ]
    )

    print("\n" + "=" * 145)
    print("6. PERFORMANCE BY FAILURE DURATION")
    print("=" * 145)

    print(
        duration_summary.to_string(
            index=False
        )
    )

else:

    duration_summary = pd.DataFrame()


# ============================================================
# 9. Performance by missing-lag count
# ============================================================

def grouped_wape(
    group,
    pred_col
):

    denominator = np.sum(
        np.abs(
            group[
                "Actual"
            ]
        )
    )

    return (
        100
        *
        np.sum(
            np.abs(
                group[
                    "Actual"
                ]
                -
                group[
                    pred_col
                ]
            )
        )
        /
        denominator
    )


missing_count_records = []

for count, group in df.groupby(
    "Missing_Lag_Count"
):

    missing_count_records.append({
        "Missing_Lag_Count":
            int(count),

        "Rows":
            int(
                len(group)
            ),

        "R0_Injected_WAPE":
            float(
                grouped_wape(
                    group,
                    "R0_Injected_Prediction"
                )
            ),

        "R1_Injected_WAPE":
            float(
                grouped_wape(
                    group,
                    "Final_R1_Injected_Prediction"
                )
            )
    })


missing_count_summary = pd.DataFrame(
    missing_count_records
).sort_values(
    "Missing_Lag_Count"
)

missing_count_summary[
    "WAPE_Reduction_pp"
] = (
    missing_count_summary[
        "R0_Injected_WAPE"
    ]
    -
    missing_count_summary[
        "R1_Injected_WAPE"
    ]
)

missing_count_summary[
    "WAPE_Improvement_Percent"
] = (
    100
    *
    missing_count_summary[
        "WAPE_Reduction_pp"
    ]
    /
    missing_count_summary[
        "R0_Injected_WAPE"
    ]
)


print("\n" + "=" * 145)
print("7. PERFORMANCE BY NUMBER OF MISSING LAGS")
print("=" * 145)

print(
    missing_count_summary.to_string(
        index=False
    )
)


# ============================================================
# 10. Performance by exact missing-lag pattern
# ============================================================

pattern_records = []

for pattern, group in df.groupby(
    "Missing_Lag_Pattern"
):

    pattern_records.append({
        "Missing_Lag_Pattern":
            pattern,

        "Rows":
            int(
                len(group)
            ),

        "R0_Injected_WAPE":
            float(
                grouped_wape(
                    group,
                    "R0_Injected_Prediction"
                )
            ),

        "R1_Injected_WAPE":
            float(
                grouped_wape(
                    group,
                    "Final_R1_Injected_Prediction"
                )
            )
    })


pattern_summary = pd.DataFrame(
    pattern_records
)

pattern_summary[
    "WAPE_Reduction_pp"
] = (
    pattern_summary[
        "R0_Injected_WAPE"
    ]
    -
    pattern_summary[
        "R1_Injected_WAPE"
    ]
)

pattern_summary[
    "WAPE_Improvement_Percent"
] = (
    100
    *
    pattern_summary[
        "WAPE_Reduction_pp"
    ]
    /
    pattern_summary[
        "R0_Injected_WAPE"
    ]
)

pattern_summary = (
    pattern_summary
    .sort_values(
        "Rows",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 145)
print("8. PERFORMANCE BY MISSING-LAG PATTERN")
print("=" * 145)

print(
    pattern_summary.to_string(
        index=False
    )
)


# ============================================================
# 11. Final statistical summary
# ============================================================

statistical_summary = {
    "test_period":
        "August 2026",

    "controlled_events":
        int(
            n_events
        ),

    "events_improved":
        int(
            n_improved
        ),

    "events_improved_percent":
        float(
            improved_percent
        ),

    "mean_event_MAE_reduction":
        float(
            mean_benefit
        ),

    "median_event_MAE_reduction":
        float(
            np.median(
                benefits
            )
        ),

    "bootstrap_95_CI_mean_MAE_reduction":
        [
            ci_low,
            ci_high
        ],

    "wilcoxon_two_sided_p":
        float(
            wilcox_two_sided.pvalue
        ),

    "wilcoxon_one_sided_p":
        float(
            wilcox_one_sided.pvalue
        ),

    "R0_mean_event_degradation":
        float(
            event_level[
                "R0_Degradation"
            ].mean()
        ),

    "R1_mean_event_degradation":
        float(
            event_level[
                "R1_Degradation"
            ].mean()
        ),

    "degradation_reduction_percent":
        float(
            degradation_reduction
        )
}


# ============================================================
# 12. Save outputs
# ============================================================

event_path = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_August_Event_Level_Analysis.csv"
)

duration_path = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_August_Performance_by_Failure_Duration.csv"
)

missing_count_path = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_August_Performance_by_Missing_Count.csv"
)

pattern_path = os.path.join(
    NB5,
    "Tables",
    "Table_FINAL_August_Performance_by_Missing_Pattern.csv"
)

summary_path = os.path.join(
    NB5,
    "Results",
    "FINAL_August_Statistical_Analysis.json"
)


event_level.to_csv(
    event_path,
    index=False
)

if len(
    duration_summary
) > 0:

    duration_summary.to_csv(
        duration_path,
        index=False
    )

missing_count_summary.to_csv(
    missing_count_path,
    index=False
)

pattern_summary.to_csv(
    pattern_path,
    index=False
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        statistical_summary,
        f,
        indent=2
    )


print("\n" + "=" * 145)
print("9. SAVED OUTPUTS")
print("=" * 145)

print(f"✅ {event_path}")

if len(
    duration_summary
) > 0:
    print(f"✅ {duration_path}")

print(f"✅ {missing_count_path}")
print(f"✅ {pattern_path}")
print(f"✅ {summary_path}")

print("\n" + "=" * 145)
print("STEP 25 COMPLETED")
print("=" * 145)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 26: PAPER-READY FIGURES & MAIN RESULT TABLES
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

TABLES = os.path.join(NB5, "Tables")
RESULTS = os.path.join(NB5, "Results")
FIGURES = os.path.join(NB5, "Figures")

os.makedirs(FIGURES, exist_ok=True)

# ------------------------------------------------------------
# Input files
# ------------------------------------------------------------

AUG_PERF_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Performance.csv"
)

EVENT_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Event_Level_Analysis.csv"
)

DURATION_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Performance_by_Failure_Duration.csv"
)

MISSING_COUNT_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Performance_by_Missing_Count.csv"
)

FEATURE_IMPORTANCE_PATH = os.path.join(
    TABLES,
    "Table_FINAL_R1_Feature_Importance.csv"
)

JULY_SELECTION_PATH = os.path.join(
    TABLES,
    "Table_R0_R1_R2_July_Model_Selection.csv"
)

AUG_SUMMARY_PATH = os.path.join(
    RESULTS,
    "FINAL_August_Untouched_Test_Summary.json"
)

STAT_SUMMARY_PATH = os.path.join(
    RESULTS,
    "FINAL_August_Statistical_Analysis.json"
)

print("=" * 135)
print("NOTEBOOK 5 — PAPER-READY FIGURES & TABLES")
print("=" * 135)


# ============================================================
# 1. Load data
# ============================================================

aug_perf = pd.read_csv(AUG_PERF_PATH)
event_df = pd.read_csv(EVENT_PATH)
duration_df = pd.read_csv(DURATION_PATH)
missing_df = pd.read_csv(MISSING_COUNT_PATH)
importance_df = pd.read_csv(FEATURE_IMPORTANCE_PATH)
july_df = pd.read_csv(JULY_SELECTION_PATH)

with open(AUG_SUMMARY_PATH, "r") as f:
    aug_summary = json.load(f)

with open(STAT_SUMMARY_PATH, "r") as f:
    stat_summary = json.load(f)

print("\n✅ All required result files loaded.")


# ============================================================
# Helper functions
# ============================================================

def save_figure(fig, filename):

    png_path = os.path.join(
        FIGURES,
        filename + ".png"
    )

    pdf_path = os.path.join(
        FIGURES,
        filename + ".pdf"
    )

    fig.savefig(
        png_path,
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        pdf_path,
        bbox_inches="tight"
    )

    plt.close(fig)

    print(f"✅ {png_path}")
    print(f"✅ {pdf_path}")


def annotate_bars(ax, fmt="{:.2f}"):

    for container in ax.containers:

        ax.bar_label(
            container,
            fmt=fmt,
            padding=3,
            fontsize=9
        )


# ============================================================
# 2. FIGURE 1
# Final August clean vs failure-injected WAPE
# ============================================================

print("\n" + "=" * 135)
print("FIGURE 1 — FINAL AUGUST WAPE COMPARISON")
print("=" * 135)

r0_clean = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "R0_CleanOnly"
        )
        &
        (
            aug_perf["Condition"] == "Full_August_Clean"
        ),
        "WAPE"
    ].iloc[0]
)

r1_clean = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "Final_R1_FailureAware"
        )
        &
        (
            aug_perf["Condition"] == "Full_August_Clean"
        ),
        "WAPE"
    ].iloc[0]
)

r0_failure = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "R0_CleanOnly"
        )
        &
        (
            aug_perf["Condition"] == "Controlled_August_Injected"
        ),
        "WAPE"
    ].iloc[0]
)

r1_failure = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "Final_R1_FailureAware"
        )
        &
        (
            aug_perf["Condition"] == "Controlled_August_Injected"
        ),
        "WAPE"
    ].iloc[0]
)

conditions = [
    "Clean Test",
    "Failure-Injected Test"
]

r0_values = [
    r0_clean,
    r0_failure
]

r1_values = [
    r1_clean,
    r1_failure
]

x = np.arange(
    len(conditions)
)

width = 0.36

fig, ax = plt.subplots(
    figsize=(8.2, 5.5)
)

ax.bar(
    x - width / 2,
    r0_values,
    width,
    label="R0: Clean-Only"
)

ax.bar(
    x + width / 2,
    r1_values,
    width,
    label="Final R1: Failure-Aware"
)

ax.set_ylabel(
    "WAPE (%)",
    fontsize=11
)

ax.set_xticks(x)
ax.set_xticklabels(
    conditions
)

ax.set_title(
    "Final August Holdout Performance",
    fontsize=13
)

ax.legend(
    frameon=False
)

ax.grid(
    axis="y",
    alpha=0.25
)

annotate_bars(
    ax,
    fmt="%.2f"
)

ax.text(
    1,
    max(r0_failure, r1_failure) * 0.92,
    "57.22% reduction",
    ha="center",
    fontsize=10
)

fig.tight_layout()

save_figure(
    fig,
    "Figure_Candidate_01_Final_August_WAPE"
)


# ============================================================
# 3. FIGURE 2
# Event-level R0 vs R1 failure MAE scatter
# ============================================================

print("\n" + "=" * 135)
print("FIGURE 2 — EVENT-LEVEL MITIGATION")
print("=" * 135)

x_event = event_df[
    "R0_Injected_MAE"
].to_numpy()

y_event = event_df[
    "R1_Injected_MAE"
].to_numpy()

max_value = float(
    max(
        x_event.max(),
        y_event.max()
    )
)

fig, ax = plt.subplots(
    figsize=(6.8, 6.2)
)

ax.scatter(
    x_event,
    y_event,
    alpha=0.55,
    s=24
)

ax.plot(
    [0, max_value],
    [0, max_value],
    linestyle="--",
    linewidth=1.3,
    label="No improvement"
)

ax.set_xlim(
    0,
    max_value * 1.03
)

ax.set_ylim(
    0,
    max_value * 1.03
)

ax.set_xlabel(
    "R0 Failure-Injected Event MAE",
    fontsize=11
)

ax.set_ylabel(
    "Final R1 Failure-Injected Event MAE",
    fontsize=11
)

ax.set_title(
    "Event-Level Robustness on August Holdout",
    fontsize=13
)

ax.grid(
    alpha=0.20
)

ax.legend(
    frameon=False
)

ax.text(
    0.04,
    0.95,
    "R1 better in 367/604 events (60.76%)\n"
    "Mean MAE reduction = 4.92\n"
    "95% CI = [4.10, 5.81]",
    transform=ax.transAxes,
    va="top",
    fontsize=9.5
)

fig.tight_layout()

save_figure(
    fig,
    "Figure_Candidate_02_Event_Level_Robustness"
)


# ============================================================
# 4. FIGURE 3
# Robustness by number of missing lag predictors
# ============================================================

print("\n" + "=" * 135)
print("FIGURE 3 — PERFORMANCE BY MISSING-LAG COUNT")
print("=" * 135)

missing_df = missing_df.sort_values(
    "Missing_Lag_Count"
)

x = np.arange(
    len(missing_df)
)

width = 0.36

fig, ax = plt.subplots(
    figsize=(8.4, 5.5)
)

ax.bar(
    x - width / 2,
    missing_df[
        "R0_Injected_WAPE"
    ],
    width,
    label="R0: Clean-Only"
)

ax.bar(
    x + width / 2,
    missing_df[
        "R1_Injected_WAPE"
    ],
    width,
    label="Final R1: Failure-Aware"
)

ax.set_xticks(x)

ax.set_xticklabels(
    missing_df[
        "Missing_Lag_Count"
    ].astype(str)
)

ax.set_xlabel(
    "Number of Missing Lag Predictors",
    fontsize=11
)

ax.set_ylabel(
    "WAPE (%)",
    fontsize=11
)

ax.set_title(
    "Robustness as Missing Information Increases",
    fontsize=13
)

ax.legend(
    frameon=False
)

ax.grid(
    axis="y",
    alpha=0.25
)

fig.tight_layout()

save_figure(
    fig,
    "Figure_Candidate_03_WAPE_by_Missing_Lag_Count"
)


# ============================================================
# 5. FIGURE 4
# Performance by empirical failure duration
# ============================================================

print("\n" + "=" * 135)
print("FIGURE 4 — PERFORMANCE BY FAILURE DURATION")
print("=" * 135)

duration_order = [
    "15 min",
    "30–60 min",
    "1.25–2 h",
    "2.25–6 h",
    "6.25–12 h",
    "12.25–23.75 h",
    "24 h"
]

duration_df[
    "Failure_Duration_Group"
] = pd.Categorical(
    duration_df[
        "Failure_Duration_Group"
    ],
    categories=duration_order,
    ordered=True
)

duration_df = (
    duration_df
    .sort_values(
        "Failure_Duration_Group"
    )
)

x = np.arange(
    len(duration_df)
)

width = 0.36

fig, ax = plt.subplots(
    figsize=(10.2, 5.7)
)

ax.bar(
    x - width / 2,
    duration_df[
        "R0_Injected_MAE"
    ],
    width,
    label="R0: Clean-Only"
)

ax.bar(
    x + width / 2,
    duration_df[
        "R1_Injected_MAE"
    ],
    width,
    label="Final R1: Failure-Aware"
)

ax.set_xticks(x)

ax.set_xticklabels(
    duration_df[
        "Failure_Duration_Group"
    ],
    rotation=25,
    ha="right"
)

ax.set_ylabel(
    "Mean Event MAE",
    fontsize=11
)

ax.set_xlabel(
    "Failure Duration",
    fontsize=11
)

ax.set_title(
    "Forecasting Error Across Failure Durations",
    fontsize=13
)

ax.legend(
    frameon=False
)

ax.grid(
    axis="y",
    alpha=0.25
)

fig.tight_layout()

save_figure(
    fig,
    "Figure_Candidate_04_MAE_by_Failure_Duration"
)


# ============================================================
# 6. FIGURE 5
# July validation vs August holdout robustness
# ============================================================

print("\n" + "=" * 135)
print("FIGURE 5 — JULY vs AUGUST GENERALIZATION")
print("=" * 135)

july_r0 = float(
    july_df.loc[
        july_df[
            "Model"
        ] == "R0_CleanOnly",
        "Controlled_Injected_WAPE"
    ].iloc[0]
)

july_r1 = float(
    july_df.loc[
        july_df[
            "Model"
        ] == "R1_FailureAware",
        "Controlled_Injected_WAPE"
    ].iloc[0]
)

periods = [
    "July Validation",
    "August Holdout"
]

r0_values = [
    july_r0,
    r0_failure
]

r1_values = [
    july_r1,
    r1_failure
]

x = np.arange(
    len(periods)
)

width = 0.36

fig, ax = plt.subplots(
    figsize=(8.2, 5.5)
)

ax.bar(
    x - width / 2,
    r0_values,
    width,
    label="R0: Clean-Only"
)

ax.bar(
    x + width / 2,
    r1_values,
    width,
    label="Failure-Aware R1"
)

ax.set_xticks(x)

ax.set_xticklabels(
    periods
)

ax.set_ylabel(
    "Failure-Injected WAPE (%)",
    fontsize=11
)

ax.set_title(
    "Robustness Generalization from Validation to Holdout",
    fontsize=13
)

ax.legend(
    frameon=False
)

ax.grid(
    axis="y",
    alpha=0.25
)

annotate_bars(
    ax,
    fmt="%.2f"
)

fig.tight_layout()

save_figure(
    fig,
    "Figure_Candidate_05_July_August_Generalization"
)


# ============================================================
# 7. FIGURE 6
# Final model feature importance
# ============================================================

print("\n" + "=" * 135)
print("FIGURE 6 — FEATURE IMPORTANCE")
print("=" * 135)

importance_plot = (
    importance_df
    .sort_values(
        "Gain_Percent",
        ascending=True
    )
)

fig, ax = plt.subplots(
    figsize=(8.0, 5.8)
)

ax.barh(
    importance_plot[
        "Feature"
    ],
    importance_plot[
        "Gain_Percent"
    ]
)

ax.set_xlabel(
    "Gain Importance (%)",
    fontsize=11
)

ax.set_ylabel(
    "Feature",
    fontsize=11
)

ax.set_title(
    "Final Failure-Aware LightGBM Feature Importance",
    fontsize=13
)

ax.grid(
    axis="x",
    alpha=0.25
)

fig.tight_layout()

save_figure(
    fig,
    "Figure_Candidate_06_Feature_Importance"
)


# ============================================================
# 8. Main paper result table
# ============================================================

main_results = pd.DataFrame([
    {
        "Evaluation":
            "August Full Clean",

        "Metric":
            "WAPE (%)",

        "R0_CleanOnly":
            r0_clean,

        "Final_R1_FailureAware":
            r1_clean,

        "Improvement":
            r0_clean - r1_clean
    },

    {
        "Evaluation":
            "August Failure-Injected",

        "Metric":
            "WAPE (%)",

        "R0_CleanOnly":
            r0_failure,

        "Final_R1_FailureAware":
            r1_failure,

        "Improvement":
            r0_failure - r1_failure
    },

    {
        "Evaluation":
            "August Failure-Injected",

        "Metric":
            "MAE",

        "R0_CleanOnly":
            aug_summary[
                "R0_failure_injected_metrics"
            ][
                "MAE"
            ],

        "Final_R1_FailureAware":
            aug_summary[
                "Final_R1_failure_injected_metrics"
            ][
                "MAE"
            ],

        "Improvement":
            (
                aug_summary[
                    "R0_failure_injected_metrics"
                ][
                    "MAE"
                ]
                -
                aug_summary[
                    "Final_R1_failure_injected_metrics"
                ][
                    "MAE"
                ]
            )
    },

    {
        "Evaluation":
            "August Failure-Injected",

        "Metric":
            "RMSE",

        "R0_CleanOnly":
            aug_summary[
                "R0_failure_injected_metrics"
            ][
                "RMSE"
            ],

        "Final_R1_FailureAware":
            aug_summary[
                "Final_R1_failure_injected_metrics"
            ][
                "RMSE"
            ],

        "Improvement":
            (
                aug_summary[
                    "R0_failure_injected_metrics"
                ][
                    "RMSE"
                ]
                -
                aug_summary[
                    "Final_R1_failure_injected_metrics"
                ][
                    "RMSE"
                ]
            )
    }
])

main_table_path = os.path.join(
    TABLES,
    "Table_PAPER_Main_Final_Results.csv"
)

main_results.to_csv(
    main_table_path,
    index=False
)


# ============================================================
# 9. Statistical evidence table
# ============================================================

stat_table = pd.DataFrame([
    {
        "Statistic":
            "Controlled failure events",

        "Value":
            stat_summary[
                "controlled_events"
            ]
    },

    {
        "Statistic":
            "Events improved by Final R1 (%)",

        "Value":
            stat_summary[
                "events_improved_percent"
            ]
    },

    {
        "Statistic":
            "Mean event-level MAE reduction",

        "Value":
            stat_summary[
                "mean_event_MAE_reduction"
            ]
    },

    {
        "Statistic":
            "95% bootstrap CI — lower",

        "Value":
            stat_summary[
                "bootstrap_95_CI_mean_MAE_reduction"
            ][0]
    },

    {
        "Statistic":
            "95% bootstrap CI — upper",

        "Value":
            stat_summary[
                "bootstrap_95_CI_mean_MAE_reduction"
            ][1]
    },

    {
        "Statistic":
            "One-sided Wilcoxon p-value",

        "Value":
            stat_summary[
                "wilcoxon_one_sided_p"
            ]
    },

    {
        "Statistic":
            "Mean failure-degradation reduction (%)",

        "Value":
            stat_summary[
                "degradation_reduction_percent"
            ]
    }
])

stat_table_path = os.path.join(
    TABLES,
    "Table_PAPER_Final_Statistical_Evidence.csv"
)

stat_table.to_csv(
    stat_table_path,
    index=False
)


# ============================================================
# 10. Final summary
# ============================================================

print("\n" + "=" * 135)
print("FINAL PAPER-READY OUTPUTS")
print("=" * 135)

print(
    "\nSix candidate figures saved in PNG (600 dpi) + PDF."
)

print(
    f"\n✅ {main_table_path}"
)

print(
    f"✅ {stat_table_path}"
)

print("\nMain final results:")

print(
    main_results.to_string(
        index=False
    )
)

print("\n" + "=" * 135)
print("STEP 26 COMPLETED")
print("=" * 135)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 27: FINAL PUBLICATION-READY MAIN FIGURES
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

TABLES = os.path.join(NB5, "Tables")
RESULTS = os.path.join(NB5, "Results")
FIGURES = os.path.join(NB5, "Figures")

os.makedirs(FIGURES, exist_ok=True)

AUG_PERF_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Performance.csv"
)

EVENT_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Event_Level_Analysis.csv"
)

MISSING_COUNT_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Performance_by_Missing_Count.csv"
)

JULY_SELECTION_PATH = os.path.join(
    TABLES,
    "Table_R0_R1_R2_July_Model_Selection.csv"
)

STAT_SUMMARY_PATH = os.path.join(
    RESULTS,
    "FINAL_August_Statistical_Analysis.json"
)

# ------------------------------------------------------------
# Load data
# ------------------------------------------------------------

aug_perf = pd.read_csv(AUG_PERF_PATH)
event_df = pd.read_csv(EVENT_PATH)
missing_df = pd.read_csv(MISSING_COUNT_PATH)
july_df = pd.read_csv(JULY_SELECTION_PATH)

with open(
    STAT_SUMMARY_PATH,
    "r",
    encoding="utf-8"
) as f:
    stat_summary = json.load(f)

print("=" * 135)
print("NOTEBOOK 5 — FINAL PUBLICATION-READY MAIN FIGURES")
print("=" * 135)


# ============================================================
# 1. Global publication formatting
# ============================================================

plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10,
    "figure.dpi": 150
})


def save_final_figure(fig, filename):

    png_path = os.path.join(
        FIGURES,
        filename + ".png"
    )

    pdf_path = os.path.join(
        FIGURES,
        filename + ".pdf"
    )

    fig.savefig(
        png_path,
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        pdf_path,
        bbox_inches="tight"
    )

    plt.close(fig)

    print(f"✅ {png_path}")
    print(f"✅ {pdf_path}")


# ============================================================
# 2. Retrieve final overall metrics
# ============================================================

r0_clean = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "R0_CleanOnly"
        )
        &
        (
            aug_perf["Condition"] == "Full_August_Clean"
        ),
        "WAPE"
    ].iloc[0]
)

r1_clean = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "Final_R1_FailureAware"
        )
        &
        (
            aug_perf["Condition"] == "Full_August_Clean"
        ),
        "WAPE"
    ].iloc[0]
)

r0_failure = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "R0_CleanOnly"
        )
        &
        (
            aug_perf["Condition"] == "Controlled_August_Injected"
        ),
        "WAPE"
    ].iloc[0]
)

r1_failure = float(
    aug_perf.loc[
        (
            aug_perf["Model"] == "Final_R1_FailureAware"
        )
        &
        (
            aug_perf["Condition"] == "Controlled_August_Injected"
        ),
        "WAPE"
    ].iloc[0]
)

failure_reduction_percent = (
    100
    *
    (
        r0_failure - r1_failure
    )
    /
    r0_failure
)

clean_delta_pp = (
    r1_clean
    -
    r0_clean
)


# ============================================================
# FIGURE 1
# Final August holdout performance
# ============================================================

print("\n" + "=" * 135)
print("FINAL FIGURE 1 — AUGUST HOLDOUT PERFORMANCE")
print("=" * 135)

conditions = [
    "Clean Test",
    "Failure-Injected Test"
]

r0_values = [
    r0_clean,
    r0_failure
]

r1_values = [
    r1_clean,
    r1_failure
]

x = np.arange(
    len(conditions)
)

width = 0.34

fig, ax = plt.subplots(
    figsize=(8.4, 5.6)
)

bars_r0 = ax.bar(
    x - width / 2,
    r0_values,
    width,
    label="R0: Clean-Only"
)

bars_r1 = ax.bar(
    x + width / 2,
    r1_values,
    width,
    label="Final R1: Failure-Aware"
)

ax.set_ylabel(
    "WAPE (%)"
)

ax.set_xticks(x)
ax.set_xticklabels(
    conditions
)

ax.set_title(
    "Final August Holdout Performance"
)

ax.legend(
    frameon=False,
    loc="upper left"
)

ax.grid(
    axis="y",
    alpha=0.22
)

ax.bar_label(
    bars_r0,
    fmt="%.2f",
    padding=3
)

ax.bar_label(
    bars_r1,
    fmt="%.2f",
    padding=3
)

# Clean-data annotation
ax.annotate(
    f"Clean change: {clean_delta_pp:+.2f} pp",
    xy=(0, max(r0_clean, r1_clean)),
    xytext=(0, max(r0_clean, r1_clean) + 6),
    ha="center",
    arrowprops=dict(
        arrowstyle="-",
        linewidth=1
    )
)

# Failure mitigation annotation
failure_mid_y = (
    r1_failure
    +
    (r0_failure - r1_failure) / 2
)

ax.annotate(
    f"{failure_reduction_percent:.2f}% lower WAPE",
    xy=(1, failure_mid_y),
    xytext=(1.20, failure_mid_y),
    ha="left",
    va="center",
    fontsize=10,
    fontweight="bold",
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.2
    )
)

ax.set_ylim(
    0,
    max(r0_failure, r1_failure) * 1.16
)

fig.tight_layout()

save_final_figure(
    fig,
    "Figure_FINAL_01_August_Holdout_Performance"
)


# ============================================================
# FIGURE 2
# Event-level robustness
# ============================================================

print("\n" + "=" * 135)
print("FINAL FIGURE 2 — EVENT-LEVEL ROBUSTNESS")
print("=" * 135)

x_event = (
    event_df[
        "R0_Injected_MAE"
    ].to_numpy()
)

y_event = (
    event_df[
        "R1_Injected_MAE"
    ].to_numpy()
)

max_value = float(
    max(
        x_event.max(),
        y_event.max()
    )
)

improved_n = int(
    stat_summary[
        "events_improved"
    ]
)

improved_pct = float(
    stat_summary[
        "events_improved_percent"
    ]
)

mean_reduction = float(
    stat_summary[
        "mean_event_MAE_reduction"
    ]
)

ci_low = float(
    stat_summary[
        "bootstrap_95_CI_mean_MAE_reduction"
    ][0]
)

ci_high = float(
    stat_summary[
        "bootstrap_95_CI_mean_MAE_reduction"
    ][1]
)

p_value = float(
    stat_summary[
        "wilcoxon_one_sided_p"
    ]
)

fig, ax = plt.subplots(
    figsize=(7.0, 6.3)
)

ax.scatter(
    x_event,
    y_event,
    alpha=0.48,
    s=24
)

ax.plot(
    [0, max_value],
    [0, max_value],
    linestyle="--",
    linewidth=1.3,
    label="Equal error"
)

ax.set_xlim(
    0,
    max_value * 1.03
)

ax.set_ylim(
    0,
    max_value * 1.03
)

ax.set_xlabel(
    "R0 Failure-Injected Event MAE"
)

ax.set_ylabel(
    "Final R1 Failure-Injected Event MAE"
)

ax.set_title(
    "Event-Level Robustness on the August Holdout"
)

ax.grid(
    alpha=0.18
)

ax.legend(
    frameon=False,
    loc="lower right"
)

annotation_text = (
    f"Improved events: {improved_n}/604 ({improved_pct:.2f}%)\n"
    f"Mean MAE reduction: {mean_reduction:.2f}\n"
    f"95% bootstrap CI: [{ci_low:.2f}, {ci_high:.2f}]\n"
    f"Wilcoxon p = {p_value:.2e}"
)

ax.text(
    0.04,
    0.95,
    annotation_text,
    transform=ax.transAxes,
    va="top",
    ha="left",
    fontsize=9.4,
    bbox=dict(
        boxstyle="round,pad=0.4",
        alpha=0.08
    )
)

fig.tight_layout()

save_final_figure(
    fig,
    "Figure_FINAL_02_Event_Level_Robustness"
)


# ============================================================
# FIGURE 3
# WAPE vs number of missing lag predictors
# with explicit sample size
# ============================================================

print("\n" + "=" * 135)
print("FINAL FIGURE 3 — ROBUSTNESS BY MISSING-LAG COUNT")
print("=" * 135)

missing_df = (
    missing_df
    .sort_values(
        "Missing_Lag_Count"
    )
    .reset_index(
        drop=True
    )
)

x = np.arange(
    len(missing_df)
)

width = 0.34

fig, ax = plt.subplots(
    figsize=(8.8, 5.8)
)

bars_r0 = ax.bar(
    x - width / 2,
    missing_df[
        "R0_Injected_WAPE"
    ],
    width,
    label="R0: Clean-Only"
)

bars_r1 = ax.bar(
    x + width / 2,
    missing_df[
        "R1_Injected_WAPE"
    ],
    width,
    label="Final R1: Failure-Aware"
)

tick_labels = [
    f"{int(row.Missing_Lag_Count)}\n(n={int(row.Rows):,})"
    for row in missing_df.itertuples()
]

ax.set_xticks(x)
ax.set_xticklabels(
    tick_labels
)

ax.set_xlabel(
    "Number of Missing Lag Predictors"
)

ax.set_ylabel(
    "WAPE (%)"
)

ax.set_title(
    "Robustness as Missing Information Increases"
)

ax.legend(
    frameon=False,
    loc="upper left"
)

ax.grid(
    axis="y",
    alpha=0.22
)

ax.bar_label(
    bars_r0,
    fmt="%.1f",
    padding=2,
    fontsize=8.5
)

ax.bar_label(
    bars_r1,
    fmt="%.1f",
    padding=2,
    fontsize=8.5
)

# Explicit warning for tiny n=8 category
five_lag_row = missing_df[
    missing_df[
        "Missing_Lag_Count"
    ] == 5
]

if len(
    five_lag_row
) == 1:

    ax.text(
        0.99,
        0.03,
        "Note: the five-lag category contains only 8 rows.",
        transform=ax.transAxes,
        ha="right",
        va="bottom",
        fontsize=8.5
    )

ax.set_ylim(
    0,
    max(
        missing_df[
            "R0_Injected_WAPE"
        ].max(),
        missing_df[
            "R1_Injected_WAPE"
        ].max()
    ) * 1.12
)

fig.tight_layout()

save_final_figure(
    fig,
    "Figure_FINAL_03_Robustness_by_Missing_Lag_Count"
)


# ============================================================
# FIGURE 4
# July validation -> August holdout generalization
# ============================================================

print("\n" + "=" * 135)
print("FINAL FIGURE 4 — VALIDATION TO HOLDOUT GENERALIZATION")
print("=" * 135)

july_r0 = float(
    july_df.loc[
        july_df[
            "Model"
        ] == "R0_CleanOnly",
        "Controlled_Injected_WAPE"
    ].iloc[0]
)

july_r1 = float(
    july_df.loc[
        july_df[
            "Model"
        ] == "R1_FailureAware",
        "Controlled_Injected_WAPE"
    ].iloc[0]
)

periods = [
    "July Validation",
    "August Holdout"
]

r0_values = [
    july_r0,
    r0_failure
]

r1_values = [
    july_r1,
    r1_failure
]

x = np.arange(
    len(periods)
)

width = 0.34

fig, ax = plt.subplots(
    figsize=(8.4, 5.6)
)

bars_r0 = ax.bar(
    x - width / 2,
    r0_values,
    width,
    label="R0: Clean-Only"
)

bars_r1 = ax.bar(
    x + width / 2,
    r1_values,
    width,
    label="Failure-Aware R1"
)

ax.set_xticks(x)
ax.set_xticklabels(
    periods
)

ax.set_ylabel(
    "Failure-Injected WAPE (%)"
)

ax.set_title(
    "Robustness Generalization from Validation to Holdout"
)

ax.legend(
    frameon=False,
    loc="upper right"
)

ax.grid(
    axis="y",
    alpha=0.22
)

ax.bar_label(
    bars_r0,
    fmt="%.2f",
    padding=3
)

ax.bar_label(
    bars_r1,
    fmt="%.2f",
    padding=3
)

# Show stability of R1 result across months
r1_holdout_change = (
    r1_failure - july_r1
)

ax.text(
    0.50,
    0.12,
    (
        "Failure-aware WAPE changed only "
        f"{r1_holdout_change:+.2f} pp\n"
        "from validation to untouched holdout"
    ),
    transform=ax.transAxes,
    ha="center",
    va="center",
    fontsize=9.5,
    bbox=dict(
        boxstyle="round,pad=0.4",
        alpha=0.08
    )
)

ax.set_ylim(
    0,
    max(r0_values) * 1.13
)

fig.tight_layout()

save_final_figure(
    fig,
    "Figure_FINAL_04_Validation_to_Holdout_Generalization"
)


# ============================================================
# 3. Save figure index / recommended manuscript role
# ============================================================

figure_index = pd.DataFrame([
    {
        "Figure":
            "Figure_FINAL_01_August_Holdout_Performance",

        "Recommended_Role":
            "Main Manuscript",

        "Purpose":
            "Primary overall mitigation effect on the untouched August holdout"
    },

    {
        "Figure":
            "Figure_FINAL_02_Event_Level_Robustness",

        "Recommended_Role":
            "Main Manuscript",

        "Purpose":
            "Event-level evidence and statistical robustness"
    },

    {
        "Figure":
            "Figure_FINAL_03_Robustness_by_Missing_Lag_Count",

        "Recommended_Role":
            "Main Manuscript",

        "Purpose":
            "Robustness under increasing missing-input severity"
    },

    {
        "Figure":
            "Figure_FINAL_04_Validation_to_Holdout_Generalization",

        "Recommended_Role":
            "Main Manuscript",

        "Purpose":
            "Validation-to-holdout generalization evidence"
    },

    {
        "Figure":
            "Figure_Candidate_04_MAE_by_Failure_Duration",

        "Recommended_Role":
            "Supplementary",

        "Purpose":
            "Performance stratified by empirical failure duration"
    },

    {
        "Figure":
            "Figure_Candidate_06_Feature_Importance",

        "Recommended_Role":
            "Supplementary",

        "Purpose":
            "Final model feature importance"
    }
])

figure_index_path = os.path.join(
    TABLES,
    "Table_PAPER_Figure_Index.csv"
)

figure_index.to_csv(
    figure_index_path,
    index=False
)


print("\n" + "=" * 135)
print("FINAL FIGURE SET CREATED")
print("=" * 135)

print(
    "\nMain-manuscript figures:"
)

print(
    "1. Final August Holdout Performance"
)

print(
    "2. Event-Level Robustness"
)

print(
    "3. Robustness by Missing-Lag Count"
)

print(
    "4. Validation-to-Holdout Generalization"
)

print(
    "\nSupplementary candidates:"
)

print(
    "- Failure Duration"
)

print(
    "- Feature Importance"
)

print(
    f"\n✅ Figure index saved:\n{figure_index_path}"
)

print("\n" + "=" * 135)
print("STEP 27 COMPLETED")
print("=" * 135)
```

```python
# ============================================================
# NOTEBOOK 5 — FAILURE-AWARE MITIGATION & ROBUST FORECASTING
# STEP 28: CLEAN OLD FIGURES + FINAL POLISH FOR FIGURES 3 & 4
# ============================================================

import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = str(OUTPUT_ROOT)
NB5 = os.path.join(BASE_PATH, "Notebook_5_Output")

TABLES = os.path.join(NB5, "Tables")
FIGURES = os.path.join(NB5, "Figures")

os.makedirs(FIGURES, exist_ok=True)

print("=" * 135)
print("NOTEBOOK 5 — FIGURE CLEANUP + FINAL POLISH")
print("=" * 135)


# ============================================================
# 1. DELETE OBSOLETE FIGURES
# ============================================================

print("\n" + "=" * 135)
print("1. DELETING OBSOLETE FIGURE VERSIONS")
print("=" * 135)

obsolete_prefixes = [
    "Figure_Candidate_01_Final_August_WAPE",
    "Figure_Candidate_02_Event_Level_Robustness",
    "Figure_Candidate_03_WAPE_by_Missing_Lag_Count",
    "Figure_Candidate_05_July_August_Generalization",

    # Current Final 3 and 4 will be replaced below
    "Figure_FINAL_03_Robustness_by_Missing_Lag_Count",
    "Figure_FINAL_04_Validation_to_Holdout_Generalization"
]

deleted_files = []

for prefix in obsolete_prefixes:

    for ext in [
        "png",
        "pdf"
    ]:

        path = os.path.join(
            FIGURES,
            f"{prefix}.{ext}"
        )

        if os.path.exists(path):

            os.remove(path)

            deleted_files.append(path)

            print(
                f"🗑️ Deleted: {os.path.basename(path)}"
            )


if len(deleted_files) == 0:

    print(
        "No obsolete files were found."
    )

else:

    print(
        f"\nDeleted obsolete files: "
        f"{len(deleted_files)}"
    )


# ============================================================
# 2. LOAD DATA
# ============================================================

MISSING_COUNT_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Performance_by_Missing_Count.csv"
)

JULY_SELECTION_PATH = os.path.join(
    TABLES,
    "Table_R0_R1_R2_July_Model_Selection.csv"
)

AUG_PERF_PATH = os.path.join(
    TABLES,
    "Table_FINAL_August_Performance.csv"
)

missing_df = pd.read_csv(
    MISSING_COUNT_PATH
)

july_df = pd.read_csv(
    JULY_SELECTION_PATH
)

aug_perf = pd.read_csv(
    AUG_PERF_PATH
)

print("\n✅ Required data loaded.")


# ============================================================
# 3. Publication formatting
# ============================================================

plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10,
    "figure.dpi": 150
})


def save_final_figure(
    fig,
    filename
):

    png_path = os.path.join(
        FIGURES,
        filename + ".png"
    )

    pdf_path = os.path.join(
        FIGURES,
        filename + ".pdf"
    )

    fig.savefig(
        png_path,
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        pdf_path,
        bbox_inches="tight"
    )

    plt.close(fig)

    print(f"✅ {png_path}")
    print(f"✅ {pdf_path}")


# ============================================================
# 4. RECREATE FINAL FIGURE 3
#    Missing-lag severity
# ============================================================

print("\n" + "=" * 135)
print("2. RECREATING FINAL FIGURE 3")
print("=" * 135)

missing_df = (
    missing_df
    .sort_values(
        "Missing_Lag_Count"
    )
    .reset_index(
        drop=True
    )
)

x = np.arange(
    len(missing_df)
)

width = 0.34

fig, ax = plt.subplots(
    figsize=(8.8, 6.1)
)

bars_r0 = ax.bar(
    x - width / 2,
    missing_df[
        "R0_Injected_WAPE"
    ],
    width,
    label="R0: Clean-Only"
)

bars_r1 = ax.bar(
    x + width / 2,
    missing_df[
        "R1_Injected_WAPE"
    ],
    width,
    label="Final R1: Failure-Aware"
)

tick_labels = [
    (
        f"{int(row.Missing_Lag_Count)}\n"
        f"(n={int(row.Rows):,})"
    )
    for row
    in missing_df.itertuples()
]

ax.set_xticks(x)

ax.set_xticklabels(
    tick_labels
)

ax.set_xlabel(
    "Number of Missing Lag Predictors"
)

ax.set_ylabel(
    "WAPE (%)"
)

ax.set_title(
    "Robustness as Missing Information Increases"
)

ax.legend(
    frameon=False,
    loc="upper left"
)

ax.grid(
    axis="y",
    alpha=0.22
)

ax.bar_label(
    bars_r0,
    fmt="%.1f",
    padding=2,
    fontsize=8.5
)

ax.bar_label(
    bars_r1,
    fmt="%.1f",
    padding=2,
    fontsize=8.5
)

ax.set_ylim(
    0,
    max(
        missing_df[
            "R0_Injected_WAPE"
        ].max(),
        missing_df[
            "R1_Injected_WAPE"
        ].max()
    ) * 1.12
)

# ------------------------------------------------------------
# Footnote OUTSIDE plotting area
# ------------------------------------------------------------

fig.text(
    0.99,
    0.015,
    "Note: the five-lag category contains only 8 affected rows "
    "and should be interpreted cautiously.",
    ha="right",
    va="bottom",
    fontsize=8.5
)

fig.subplots_adjust(
    bottom=0.17
)

save_final_figure(
    fig,
    "Figure_FINAL_03_Robustness_by_Missing_Lag_Count"
)


# ============================================================
# 5. RECREATE FINAL FIGURE 4
#    Validation-to-holdout generalization
# ============================================================

print("\n" + "=" * 135)
print("3. RECREATING FINAL FIGURE 4")
print("=" * 135)

july_r0 = float(
    july_df.loc[
        july_df[
            "Model"
        ] == "R0_CleanOnly",
        "Controlled_Injected_WAPE"
    ].iloc[0]
)

july_r1 = float(
    july_df.loc[
        july_df[
            "Model"
        ] == "R1_FailureAware",
        "Controlled_Injected_WAPE"
    ].iloc[0]
)

aug_r0 = float(
    aug_perf.loc[
        (
            aug_perf[
                "Model"
            ] == "R0_CleanOnly"
        )
        &
        (
            aug_perf[
                "Condition"
            ] == "Controlled_August_Injected"
        ),
        "WAPE"
    ].iloc[0]
)

aug_r1 = float(
    aug_perf.loc[
        (
            aug_perf[
                "Model"
            ] == "Final_R1_FailureAware"
        )
        &
        (
            aug_perf[
                "Condition"
            ] == "Controlled_August_Injected"
        ),
        "WAPE"
    ].iloc[0]
)

periods = [
    "July Validation",
    "August Holdout"
]

r0_values = [
    july_r0,
    aug_r0
]

r1_values = [
    july_r1,
    aug_r1
]

x = np.arange(
    len(periods)
)

width = 0.34

fig, ax = plt.subplots(
    figsize=(8.4, 5.8)
)

bars_r0 = ax.bar(
    x - width / 2,
    r0_values,
    width,
    label="R0: Clean-Only"
)

bars_r1 = ax.bar(
    x + width / 2,
    r1_values,
    width,
    label="Failure-Aware R1"
)

ax.set_xticks(x)

ax.set_xticklabels(
    periods
)

ax.set_ylabel(
    "Failure-Injected WAPE (%)"
)

ax.set_title(
    "Robustness Generalization from Validation to Holdout"
)

ax.legend(
    frameon=False,
    loc="upper right"
)

ax.grid(
    axis="y",
    alpha=0.22
)

ax.bar_label(
    bars_r0,
    fmt="%.2f",
    padding=3
)

ax.bar_label(
    bars_r1,
    fmt="%.2f",
    padding=3
)

r1_change = (
    aug_r1
    -
    july_r1
)

# ------------------------------------------------------------
# Annotation moved to empty upper-middle area
# ------------------------------------------------------------

ax.text(
    0.50,
    0.73,
    (
        "Failure-aware WAPE remained highly stable\n"
        f"Validation → Holdout change: {r1_change:+.2f} pp"
    ),
    transform=ax.transAxes,
    ha="center",
    va="center",
    fontsize=9.5,
    bbox=dict(
        boxstyle="round,pad=0.45",
        facecolor="white",
        edgecolor="0.75",
        alpha=0.92
    )
)

ax.set_ylim(
    0,
    max(r0_values) * 1.15
)

fig.tight_layout()

save_final_figure(
    fig,
    "Figure_FINAL_04_Validation_to_Holdout_Generalization"
)


# ============================================================
# 6. VERIFY FINAL FIGURE DIRECTORY
# ============================================================

print("\n" + "=" * 135)
print("4. FINAL FIGURE DIRECTORY CHECK")
print("=" * 135)

figure_files = sorted(
    [
        os.path.basename(path)
        for path
        in glob.glob(
            os.path.join(
                FIGURES,
                "*"
            )
        )
        if path.lower().endswith(
            (
                ".png",
                ".pdf"
            )
        )
    ]
)

for f in figure_files:

    print(
        f"✅ {f}"
    )


# ============================================================
# 7. Verify required main figures exist
# ============================================================

required_main_figures = [
    "Figure_FINAL_01_August_Holdout_Performance",
    "Figure_FINAL_02_Event_Level_Robustness",
    "Figure_FINAL_03_Robustness_by_Missing_Lag_Count",
    "Figure_FINAL_04_Validation_to_Holdout_Generalization"
]

all_main_figures_present = True

for fig_name in required_main_figures:

    png_exists = os.path.exists(
        os.path.join(
            FIGURES,
            fig_name + ".png"
        )
    )

    pdf_exists = os.path.exists(
        os.path.join(
            FIGURES,
            fig_name + ".pdf"
        )
    )

    status = (
        png_exists
        and
        pdf_exists
    )

    print(
        f"{'✅' if status else '❌'} "
        f"{fig_name}"
    )

    if not status:

        all_main_figures_present = False


print("\n" + "=" * 135)

print(
    f"FINAL MAIN FIGURE SET COMPLETE: "
    f"{all_main_figures_present}"
)

print("=" * 135)

print(
    "\nSTEP 28 COMPLETED — OLD MAIN FIGURES CLEANED "
    "AND FINAL FIGURES UPDATED"
)
```
