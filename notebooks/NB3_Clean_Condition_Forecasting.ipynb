```python
# ============================================================
# NOTEBOOK 3 — CODE 4
# Structural audit before creating forecasting lags
# ============================================================

import pandas as pd
import numpy as np

print("=" * 85)
print("NOTEBOOK 3 — 15-MINUTE STRUCTURAL AUDIT")
print("=" * 85)

# ------------------------------------------------------------
# 1. BASIC EXPECTED VS ACTUAL COVERAGE
# ------------------------------------------------------------

n_detectors = df15["Detector_Code"].nunique()

start_date = df15["Timestamp"].dt.normalize().min()
end_date   = df15["Timestamp"].dt.normalize().max()

n_days = (end_date - start_date).days + 1

expected_rows = n_detectors * n_days * 96
actual_rows = len(df15)
missing_grid_rows = expected_rows - actual_rows

print("\n1. EXPECTED VS ACTUAL COVERAGE")
print("-" * 85)

print(f"Detectors             : {n_detectors:,}")
print(f"Days                  : {n_days:,}")
print(f"Expected rows         : {expected_rows:,}")
print(f"Actual rows           : {actual_rows:,}")
print(f"Absent grid rows      : {missing_grid_rows:,}")
print(
    f"Observed coverage     : "
    f"{100 * actual_rows / expected_rows:.4f}%"
)

# ------------------------------------------------------------
# 2. DUPLICATE DETECTOR-TIMESTAMP CHECK
# ------------------------------------------------------------

duplicate_count = df15.duplicated(
    subset=["Detector_Code", "Timestamp"],
    keep=False
).sum()

print("\n2. DUPLICATE DETECTOR-TIMESTAMP CHECK")
print("-" * 85)

print(f"Rows involved in duplicates: {duplicate_count:,}")

if duplicate_count == 0:
    print("✅ No duplicate Detector-Timestamp records.")
else:
    print("⚠️ Duplicates detected.")

# ------------------------------------------------------------
# 3. INTERVAL INDEX CONSISTENCY
# ------------------------------------------------------------

expected_interval = (
    df15["Timestamp"].dt.hour * 4
    + df15["Timestamp"].dt.minute // 15
)

interval_mismatch = (
    expected_interval != df15["Interval_Index"]
).sum()

print("\n3. INTERVAL INDEX CONSISTENCY")
print("-" * 85)

print(f"Interval-index mismatches: {interval_mismatch:,}")

if interval_mismatch == 0:
    print("✅ Interval_Index perfectly matches Timestamp.")

# ------------------------------------------------------------
# 4. TARGET VOLUME VALIDITY
# ------------------------------------------------------------

negative_count = (df15["Volume"] < 0).sum()
zero_count     = (df15["Volume"] == 0).sum()
positive_count = (df15["Volume"] > 0).sum()

print("\n4. TRAFFIC VOLUME TARGET CHECK")
print("-" * 85)

print(f"Minimum Volume        : {df15['Volume'].min():,}")
print(f"Maximum Volume        : {df15['Volume'].max():,}")
print(f"Negative observations : {negative_count:,}")
print(f"Zero observations     : {zero_count:,}")
print(f"Positive observations : {positive_count:,}")

print(
    f"Zero share            : "
    f"{100 * zero_count / len(df15):.4f}%"
)

# ------------------------------------------------------------
# 5. RECORDS PER DETECTOR-DAY
# ------------------------------------------------------------

audit = df15.copy()
audit["Date"] = audit["Timestamp"].dt.normalize()

records_per_detector_day = (
    audit.groupby(
        ["Detector_Code", "Date"],
        observed=True
    )
    .size()
)

print("\n5. RECORDS PER DETECTOR-DAY")
print("-" * 85)

distribution = (
    records_per_detector_day
    .value_counts()
    .sort_index()
)

print(distribution.to_string())

print("\nDetector-days summary:")
print(records_per_detector_day.describe().to_string())

# ------------------------------------------------------------
# 6. DETECTOR TIMESTAMP GAPS
# ------------------------------------------------------------

gap_df = (
    df15[["Detector_Code", "Timestamp"]]
    .sort_values(["Detector_Code", "Timestamp"])
    .copy()
)

gap_df["Gap_Minutes"] = (
    gap_df.groupby("Detector_Code")["Timestamp"]
    .diff()
    .dt.total_seconds()
    .div(60)
)

within_month_gaps = gap_df[
    gap_df["Gap_Minutes"].notna()
].copy()

normal_15min = (within_month_gaps["Gap_Minutes"] == 15).sum()
larger_gaps  = (within_month_gaps["Gap_Minutes"] > 15).sum()

print("\n6. TIMESTAMP GAP CHECK")
print("-" * 85)

print(f"Normal 15-min transitions : {normal_15min:,}")
print(f"Transitions > 15 min      : {larger_gaps:,}")

if larger_gaps > 0:
    print("\nMost common gap lengths > 15 minutes:")

    gap_distribution = (
        within_month_gaps.loc[
            within_month_gaps["Gap_Minutes"] > 15,
            "Gap_Minutes"
        ]
        .value_counts()
        .sort_values(ascending=False)
        .head(20)
    )

    print(gap_distribution.to_string())

# ------------------------------------------------------------
# 7. ZERO-VOLUME DAY FLAG
# ------------------------------------------------------------

print("\n7. ZERO-VOLUME-DAY FLAG")
print("-" * 85)

print(
    df15["Is_Zero_Volume_Day"]
    .value_counts(dropna=False)
    .to_string()
)

# ------------------------------------------------------------
# 8. SENSITIVITY COHORT
# ------------------------------------------------------------

print("\n8. SENSITIVITY COHORT")
print("-" * 85)

sensitivity_detector_count = (
    df15.loc[
        df15["Sensitivity_Cohort"],
        "Detector_Code"
    ]
    .nunique()
)

print(
    f"Sensitivity detectors : "
    f"{sensitivity_detector_count:,} / {n_detectors:,}"
)

print(
    df15["Sensitivity_Cohort"]
    .value_counts(dropna=False)
    .to_string()
)

print("\n" + "=" * 85)
print("STRUCTURAL AUDIT COMPLETE")
print("=" * 85)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 5
# Audit all monthly 15-minute modeling files
# ============================================================

import pandas as pd
import os
import glob

MONTHLY_DIR = os.path.join(
    NOTEBOOK2_PATH,
    "Clean_15min_Monthly"
)

monthly_files = sorted(
    glob.glob(
        os.path.join(
            MONTHLY_DIR,
            "CLEAN_15MIN_*.parquet"
        )
    )
)

print("=" * 90)
print("NOTEBOOK 3 — ALL-MONTH 15-MINUTE DATA AUDIT")
print("=" * 90)

print(f"\nMonthly files found: {len(monthly_files)}")

results = []

for file in monthly_files:

    temp = pd.read_parquet(file)

    month_name = os.path.basename(file).replace(
        "CLEAN_15MIN_", ""
    ).replace(".parquet", "")

    # Basic coverage
    min_time = temp["Timestamp"].min()
    max_time = temp["Timestamp"].max()

    n_detectors = temp["Detector_Code"].nunique()
    n_rows = len(temp)

    # Detector-day counts
    temp_dates = temp["Timestamp"].dt.normalize()

    detector_days = (
        temp.assign(Date=temp_dates)
        .groupby(
            ["Detector_Code", "Date"],
            observed=True
        )
        .size()
    )

    complete_detector_days = (detector_days == 96).sum()
    incomplete_detector_days = (detector_days != 96).sum()

    # Calendar days
    n_days = (
        max_time.normalize()
        - min_time.normalize()
    ).days + 1

    expected_detector_days = (
        n_detectors * n_days
    )

    observed_detector_days = len(detector_days)

    absent_detector_days = (
        expected_detector_days
        - observed_detector_days
    )

    expected_rows = (
        expected_detector_days * 96
    )

    coverage_pct = (
        100 * n_rows / expected_rows
    )

    # Volume checks
    negative_volume = (
        temp["Volume"] < 0
    ).sum()

    zero_volume = (
        temp["Volume"] == 0
    ).sum()

    # Sensitivity detectors
    sensitivity_detectors = (
        temp.loc[
            temp["Sensitivity_Cohort"],
            "Detector_Code"
        ]
        .nunique()
    )

    results.append({
        "Month": month_name,
        "Start": min_time,
        "End": max_time,
        "Days": n_days,
        "Detectors": n_detectors,
        "Rows": n_rows,
        "Expected_Rows": expected_rows,
        "Coverage_%": coverage_pct,
        "Observed_Detector_Days": observed_detector_days,
        "Absent_Detector_Days": absent_detector_days,
        "Incomplete_Observed_Days": incomplete_detector_days,
        "Negative_Volumes": negative_volume,
        "Zero_Volume_%": (
            100 * zero_volume / n_rows
        ),
        "Sensitivity_Detectors": sensitivity_detectors
    })

    print(
        f"✅ {month_name}: "
        f"{n_rows:,} rows | "
        f"{coverage_pct:.3f}% coverage | "
        f"{absent_detector_days:,} absent detector-days"
    )

    del temp


# ============================================================
# FINAL SUMMARY TABLE
# ============================================================

monthly_audit = pd.DataFrame(results)

print("\n" + "=" * 90)
print("MONTHLY AUDIT SUMMARY")
print("=" * 90)

display(
    monthly_audit.style.format({
        "Rows": "{:,.0f}",
        "Expected_Rows": "{:,.0f}",
        "Coverage_%": "{:.3f}",
        "Observed_Detector_Days": "{:,.0f}",
        "Absent_Detector_Days": "{:,.0f}",
        "Incomplete_Observed_Days": "{:,.0f}",
        "Negative_Volumes": "{:,.0f}",
        "Zero_Volume_%": "{:.2f}",
        "Sensitivity_Detectors": "{:,.0f}"
    })
)

print("\n" + "=" * 90)
print("OVERALL SUMMARY")
print("=" * 90)

print(
    f"Total rows: "
    f"{monthly_audit['Rows'].sum():,}"
)

print(
    f"Total expected rows: "
    f"{monthly_audit['Expected_Rows'].sum():,}"
)

print(
    f"Overall coverage: "
    f"{100 * monthly_audit['Rows'].sum() / monthly_audit['Expected_Rows'].sum():.4f}%"
)

print(
    f"Total absent detector-days: "
    f"{monthly_audit['Absent_Detector_Days'].sum():,}"
)

print(
    f"Incomplete observed detector-days: "
    f"{monthly_audit['Incomplete_Observed_Days'].sum():,}"
)

print(
    f"Negative traffic-volume observations: "
    f"{monthly_audit['Negative_Volumes'].sum():,}"
)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 6
# Prototype leakage-safe forecasting features — January 2026
# ============================================================

import pandas as pd
import numpy as np

print("=" * 90)
print("NOTEBOOK 3 — FORECAST FEATURE PROTOTYPE")
print("=" * 90)

# ------------------------------------------------------------
# 1. SORT DATA
# ------------------------------------------------------------

jan = df15.sort_values(
    ["Detector_Code", "Timestamp"]
).reset_index(drop=True).copy()

print(f"\nInput rows: {len(jan):,}")

# ------------------------------------------------------------
# 2. TARGET
# ------------------------------------------------------------

# Target volume at prediction timestamp t
jan["Target_Volume"] = jan["Volume"].astype("int16")

# ------------------------------------------------------------
# 3. TIME FEATURES
# These are known in advance and therefore safe predictors
# ------------------------------------------------------------

jan["Hour"] = jan["Timestamp"].dt.hour.astype("uint8")
jan["Minute"] = jan["Timestamp"].dt.minute.astype("uint8")
jan["DayOfWeek"] = jan["Timestamp"].dt.dayofweek.astype("uint8")
jan["IsWeekend"] = (
    jan["Timestamp"].dt.dayofweek >= 5
).astype("uint8")

# Interval_Index already represents 00:00–23:45 as 0–95

# ------------------------------------------------------------
# 4. LEAKAGE-SAFE LAG FEATURES
# ------------------------------------------------------------

lag_definitions = {
    "Lag_1": 1,      # 15 min
    "Lag_2": 2,      # 30 min
    "Lag_4": 4,      # 1 hour
    "Lag_8": 8,      # 2 hours
    "Lag_96": 96     # 24 hours
}

grouped_volume = jan.groupby(
    "Detector_Code",
    sort=False
)["Volume"]

grouped_time = jan.groupby(
    "Detector_Code",
    sort=False
)["Timestamp"]

for lag_name, lag_steps in lag_definitions.items():

    expected_minutes = lag_steps * 15

    shifted_volume = grouped_volume.shift(lag_steps)
    shifted_time = grouped_time.shift(lag_steps)

    actual_difference = (
        jan["Timestamp"] - shifted_time
    ).dt.total_seconds() / 60

    # Only retain lag if it comes from EXACTLY the required time
    valid_time_match = (
        actual_difference == expected_minutes
    )

    jan[lag_name] = shifted_volume.where(
        valid_time_match
    ).astype("float32")


# ------------------------------------------------------------
# 5. VERIFY TEMPORAL SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("LAG AVAILABILITY")
print("=" * 90)

for lag_name in lag_definitions:

    available = jan[lag_name].notna().sum()
    missing = jan[lag_name].isna().sum()

    print(
        f"{lag_name:<8} | "
        f"Available: {available:>10,} "
        f"({100 * available / len(jan):6.2f}%) | "
        f"Unavailable: {missing:>8,}"
    )


# ------------------------------------------------------------
# 6. MODEL-READY FLAG
# ------------------------------------------------------------

required_lags = list(lag_definitions.keys())

jan["All_Lags_Available"] = (
    jan[required_lags]
    .notna()
    .all(axis=1)
)

model_ready_rows = jan["All_Lags_Available"].sum()

print("\n" + "=" * 90)
print("MODEL-READY ROWS")
print("=" * 90)

print(f"Total observations : {len(jan):,}")
print(f"All lags available : {model_ready_rows:,}")

print(
    f"Usable percentage  : "
    f"{100 * model_ready_rows / len(jan):.4f}%"
)


# ------------------------------------------------------------
# 7. IMPORTANT LEAKAGE CHECK
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("DATA LEAKAGE CHECK")
print("=" * 90)

print("Predictors derived only from:")
print("  ✓ Past traffic observations")
print("  ✓ Prediction timestamp")
print("  ✓ Detector identifier")

print("\nExplicitly NOT used as predictors:")
print("  ✗ Is_Zero_Volume_Day")
print("  ✗ Sensitivity_Cohort")
print("  ✗ Any future traffic value")
print("  ✗ Any target-derived daily information")

print("\n✅ No target leakage introduced by this feature construction.")


# ------------------------------------------------------------
# 8. SAMPLE MODELING ROWS
# ------------------------------------------------------------

columns_to_show = [
    "Timestamp",
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "Target_Volume",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96",
    "All_Lags_Available"
]

print("\n" + "=" * 90)
print("SAMPLE FORECASTING DATA")
print("=" * 90)

display(
    jan.loc[
        jan["All_Lags_Available"],
        columns_to_show
    ].head(15)
)


# ------------------------------------------------------------
# 9. FINAL SANITY CHECK
# ------------------------------------------------------------

assert (
    jan.loc[
        jan["Lag_1"].notna(),
        "Lag_1"
    ] >= 0
).all()

assert (
    jan.loc[
        jan["Lag_96"].notna(),
        "Lag_96"
    ] >= 0
).all()

print("\n" + "=" * 90)
print("✅ JANUARY FORECAST FEATURE PROTOTYPE COMPLETED SUCCESSFULLY")
print("=" * 90)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 8
# Build leakage-safe forecasting datasets for all 8 months
# Preserve lags correctly across month boundaries
# ============================================================

import pandas as pd
import numpy as np
import os
import glob
import gc

print("=" * 95)
print("NOTEBOOK 3 — BUILD ALL MONTHLY FORECASTING DATASETS")
print("=" * 95)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

MONTHLY_SOURCE_DIR = os.path.join(
    NOTEBOOK2_PATH,
    "Clean_15min_Monthly"
)

FORECAST_MONTHLY_DIR = os.path.join(
    DATA_DIR,
    "Forecasting_Monthly"
)

os.makedirs(
    FORECAST_MONTHLY_DIR,
    exist_ok=True
)

monthly_files = sorted(
    glob.glob(
        os.path.join(
            MONTHLY_SOURCE_DIR,
            "CLEAN_15MIN_*.parquet"
        )
    )
)

print(f"\nMonthly source files found: {len(monthly_files)}")

# ------------------------------------------------------------
# LAG DEFINITIONS
# ------------------------------------------------------------

lag_definitions = {
    "Lag_1": 1,      # 15 minutes
    "Lag_2": 2,      # 30 minutes
    "Lag_4": 4,      # 60 minutes
    "Lag_8": 8,      # 120 minutes
    "Lag_96": 96     # 24 hours
}

required_lags = list(lag_definitions.keys())

summary_rows = []

# This carries ONLY the last calendar day of the previous month
prev_tail = None


# ============================================================
# PROCESS MONTH BY MONTH
# ============================================================

for file in monthly_files:

    month = (
        os.path.basename(file)
        .replace("CLEAN_15MIN_", "")
        .replace(".parquet", "")
    )

    print("\n" + "-" * 95)
    print(f"Processing {month}")
    print("-" * 95)

    # --------------------------------------------------------
    # LOAD CURRENT MONTH
    # --------------------------------------------------------

    current = pd.read_parquet(file)

    current = current.sort_values(
        ["Detector_Code", "Timestamp"]
    ).reset_index(drop=True)

    original_rows = len(current)

    # --------------------------------------------------------
    # ADD PREVIOUS MONTH TAIL
    # --------------------------------------------------------

    if prev_tail is not None:

        bridge = pd.concat(
            [prev_tail, current],
            ignore_index=True
        )

        bridge = bridge.sort_values(
            ["Detector_Code", "Timestamp"]
        ).reset_index(drop=True)

    else:
        bridge = current.copy()

    # --------------------------------------------------------
    # TARGET
    # --------------------------------------------------------

    bridge["Target_Volume"] = (
        bridge["Volume"]
        .astype("int16")
    )

    # --------------------------------------------------------
    # TIME FEATURES
    # --------------------------------------------------------

    bridge["Hour"] = (
        bridge["Timestamp"]
        .dt.hour
        .astype("uint8")
    )

    bridge["Minute"] = (
        bridge["Timestamp"]
        .dt.minute
        .astype("uint8")
    )

    bridge["DayOfWeek"] = (
        bridge["Timestamp"]
        .dt.dayofweek
        .astype("uint8")
    )

    bridge["IsWeekend"] = (
        bridge["Timestamp"]
        .dt.dayofweek
        .ge(5)
        .astype("uint8")
    )

    # --------------------------------------------------------
    # TIMESTAMP-SAFE LAGS
    # --------------------------------------------------------

    grouped_volume = bridge.groupby(
        "Detector_Code",
        sort=False
    )["Volume"]

    grouped_time = bridge.groupby(
        "Detector_Code",
        sort=False
    )["Timestamp"]

    for lag_name, lag_steps in lag_definitions.items():

        expected_minutes = lag_steps * 15

        shifted_volume = grouped_volume.shift(
            lag_steps
        )

        shifted_time = grouped_time.shift(
            lag_steps
        )

        actual_minutes = (
            bridge["Timestamp"] - shifted_time
        ).dt.total_seconds().div(60)

        exact_time_match = (
            actual_minutes == expected_minutes
        )

        bridge[lag_name] = (
            shifted_volume
            .where(exact_time_match)
            .astype("float32")
        )

    # --------------------------------------------------------
    # KEEP CURRENT MONTH ONLY
    # --------------------------------------------------------

    month_start = current["Timestamp"].min()
    month_end = current["Timestamp"].max()

    forecast = bridge.loc[
        (bridge["Timestamp"] >= month_start)
        &
        (bridge["Timestamp"] <= month_end)
    ].copy()

    forecast = forecast.sort_values(
        ["Detector_Code", "Timestamp"]
    ).reset_index(drop=True)

    # Safety check
    assert len(forecast) == original_rows

    # --------------------------------------------------------
    # MODEL-READY FLAG
    # --------------------------------------------------------

    forecast["All_Lags_Available"] = (
        forecast[required_lags]
        .notna()
        .all(axis=1)
    )

    # --------------------------------------------------------
    # SPLIT LABEL
    # --------------------------------------------------------

    if month <= "2026-06":
        split = "Train"
    elif month == "2026-07":
        split = "Validation"
    else:
        split = "Test"

    forecast["Split"] = split

    # --------------------------------------------------------
    # IMPORTANT:
    # Quality variables remain metadata only.
    # They are NOT predictor variables.
    # --------------------------------------------------------

    final_columns = [
        "Timestamp",
        "Detector_Code",
        "Interval_Index",

        # Target
        "Target_Volume",

        # Safe time predictors
        "Hour",
        "Minute",
        "DayOfWeek",
        "IsWeekend",

        # Past traffic predictors
        "Lag_1",
        "Lag_2",
        "Lag_4",
        "Lag_8",
        "Lag_96",

        # Analysis metadata — NOT predictors
        "Is_Zero_Volume_Day",
        "Sensitivity_Cohort",

        # Dataset control
        "All_Lags_Available",
        "Split"
    ]

    forecast = forecast[final_columns]

    # --------------------------------------------------------
    # CHECK FOR NEGATIVE TARGETS
    # --------------------------------------------------------

    assert (
        forecast["Target_Volume"] >= 0
    ).all()

    # --------------------------------------------------------
    # SAVE MONTHLY FILE
    # --------------------------------------------------------

    output_file = os.path.join(
        FORECAST_MONTHLY_DIR,
        f"FORECASTING_{month}.parquet"
    )

    forecast.to_parquet(
        output_file,
        index=False
    )

    # --------------------------------------------------------
    # MONTHLY SUMMARY
    # --------------------------------------------------------

    ready_rows = int(
        forecast["All_Lags_Available"].sum()
    )

    month_summary = {
        "Month": month,
        "Split": split,
        "Rows": len(forecast),
        "Model_Ready_Rows": ready_rows,
        "Model_Ready_%":
            100 * ready_rows / len(forecast),

        "Lag_1_Available_%":
            100 * forecast["Lag_1"].notna().mean(),

        "Lag_2_Available_%":
            100 * forecast["Lag_2"].notna().mean(),

        "Lag_4_Available_%":
            100 * forecast["Lag_4"].notna().mean(),

        "Lag_8_Available_%":
            100 * forecast["Lag_8"].notna().mean(),

        "Lag_96_Available_%":
            100 * forecast["Lag_96"].notna().mean()
    }

    summary_rows.append(month_summary)

    print(
        f"Rows        : {len(forecast):,}"
    )

    print(
        f"Model-ready : {ready_rows:,} "
        f"({month_summary['Model_Ready_%']:.3f}%)"
    )

    print(
        f"Lag-96 avail: "
        f"{month_summary['Lag_96_Available_%']:.3f}%"
    )

    print(f"✅ Saved: {output_file}")

    # --------------------------------------------------------
    # PREPARE LAST DAY FOR NEXT MONTH
    # --------------------------------------------------------

    last_calendar_day = (
        current["Timestamp"]
        .dt.normalize()
        .max()
    )

    prev_tail = current.loc[
        current["Timestamp"].dt.normalize()
        == last_calendar_day
    ].copy()

    print(
        f"Carry-forward rows for next month: "
        f"{len(prev_tail):,}"
    )

    # --------------------------------------------------------
    # FREE MEMORY
    # --------------------------------------------------------

    del current
    del bridge
    del forecast

    gc.collect()


# ============================================================
# FINAL FEATURE AVAILABILITY SUMMARY
# ============================================================

feature_summary = pd.DataFrame(
    summary_rows
)

print("\n" + "=" * 95)
print("FORECASTING FEATURE AVAILABILITY SUMMARY")
print("=" * 95)

display(
    feature_summary.style.format({
        "Rows": "{:,.0f}",
        "Model_Ready_Rows": "{:,.0f}",
        "Model_Ready_%": "{:.3f}",
        "Lag_1_Available_%": "{:.3f}",
        "Lag_2_Available_%": "{:.3f}",
        "Lag_4_Available_%": "{:.3f}",
        "Lag_8_Available_%": "{:.3f}",
        "Lag_96_Available_%": "{:.3f}"
    })
)


# ============================================================
# SAVE SUMMARY
# ============================================================

summary_output = os.path.join(
    TABLES_DIR,
    "Table_Forecasting_Feature_Availability.csv"
)

feature_summary.to_csv(
    summary_output,
    index=False
)

print("\n✅ Forecasting availability table saved:")
print(summary_output)


# ============================================================
# OVERALL COUNTS
# ============================================================

total_rows = int(
    feature_summary["Rows"].sum()
)

total_ready = int(
    feature_summary["Model_Ready_Rows"].sum()
)

print("\n" + "=" * 95)
print("OVERALL FORECASTING DATASET")
print("=" * 95)

print(f"Total observations : {total_rows:,}")
print(f"Model-ready rows   : {total_ready:,}")

print(
    f"Overall usable %   : "
    f"{100 * total_ready / total_rows:.4f}%"
)

print("\nSplit totals:")

display(
    feature_summary.groupby(
        "Split",
        observed=True
    )[["Rows", "Model_Ready_Rows"]]
    .sum()
)

print("\n" + "=" * 95)
print("✅ ALL MONTHLY FORECASTING DATASETS CREATED AND SAVED")
print("=" * 95)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 9
# Baseline forecasting experiment — Validation set only
# ============================================================

import pandas as pd
import numpy as np
import os
import json
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 90)
print("NOTEBOOK 3 — VALIDATION BASELINE FORECASTING EXPERIMENT")
print("=" * 90)

# ------------------------------------------------------------
# 1. LOAD VALIDATION DATA ONLY
# ------------------------------------------------------------

validation_file = os.path.join(
    FORECAST_MONTHLY_DIR,
    "FORECASTING_2026-07.parquet"
)

val = pd.read_parquet(validation_file)

# Use exactly the rows where ALL model features are available
val = val.loc[
    val["All_Lags_Available"]
].copy()

print(f"\nValidation rows used: {len(val):,}")
print(
    f"Period: {val['Timestamp'].min()} "
    f"→ {val['Timestamp'].max()}"
)

print(
    f"Unique detectors: "
    f"{val['Detector_Code'].nunique():,}"
)

# ------------------------------------------------------------
# 2. ACTUAL TARGET
# ------------------------------------------------------------

y_true = val["Target_Volume"].astype("float64")

# ------------------------------------------------------------
# 3. BASELINE PREDICTIONS
# ------------------------------------------------------------

baseline_predictions = {
    "Persistence_15min": val["Lag_1"].astype("float64"),
    "Seasonal_Naive_24h": val["Lag_96"].astype("float64")
}

# ------------------------------------------------------------
# 4. METRIC FUNCTION
# ------------------------------------------------------------

def calculate_metrics(y_true, y_pred):

    error = y_pred - y_true

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    # WAPE is robust to individual zero-volume observations
    denominator = np.abs(y_true).sum()

    wape = (
        100
        * np.abs(error).sum()
        / denominator
        if denominator > 0
        else np.nan
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    bias = error.mean()

    return {
        "MAE": mae,
        "RMSE": rmse,
        "WAPE_%": wape,
        "R2": r2,
        "Mean_Bias": bias
    }

# ------------------------------------------------------------
# 5. CALCULATE BASELINE METRICS
# ------------------------------------------------------------

baseline_results = []

for model_name, prediction in baseline_predictions.items():

    metrics = calculate_metrics(
        y_true,
        prediction
    )

    baseline_results.append({
        "Model": model_name,
        "N": len(y_true),
        **metrics
    })

baseline_metrics = pd.DataFrame(
    baseline_results
)

# ------------------------------------------------------------
# 6. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("VALIDATION BASELINE PERFORMANCE")
print("=" * 90)

display(
    baseline_metrics.style.format({
        "N": "{:,.0f}",
        "MAE": "{:.4f}",
        "RMSE": "{:.4f}",
        "WAPE_%": "{:.3f}",
        "R2": "{:.5f}",
        "Mean_Bias": "{:.4f}"
    })
)

# ------------------------------------------------------------
# 7. IDENTIFY BEST BASELINE
# ------------------------------------------------------------

best_mae = baseline_metrics.loc[
    baseline_metrics["MAE"].idxmin()
]

best_rmse = baseline_metrics.loc[
    baseline_metrics["RMSE"].idxmin()
]

print("\n" + "=" * 90)
print("BEST BASELINE")
print("=" * 90)

print(
    f"Lowest MAE  : "
    f"{best_mae['Model']} "
    f"({best_mae['MAE']:.4f})"
)

print(
    f"Lowest RMSE : "
    f"{best_rmse['Model']} "
    f"({best_rmse['RMSE']:.4f})"
)

# ------------------------------------------------------------
# 8. SAVE PAPER-RELEVANT TABLE
# ------------------------------------------------------------

baseline_table_path = os.path.join(
    TABLES_DIR,
    "Table_Validation_Baseline_Performance.csv"
)

baseline_metrics.to_csv(
    baseline_table_path,
    index=False
)

print("\n✅ Baseline performance table saved:")
print(baseline_table_path)

# ------------------------------------------------------------
# 9. SAVE RESULTS CONFIGURATION
# ------------------------------------------------------------

baseline_summary = {
    "dataset": "Validation",
    "period": {
        "start": str(val["Timestamp"].min()),
        "end": str(val["Timestamp"].max())
    },
    "rows": int(len(val)),
    "detectors": int(
        val["Detector_Code"].nunique()
    ),
    "baselines": {
        row["Model"]: {
            "MAE": float(row["MAE"]),
            "RMSE": float(row["RMSE"]),
            "WAPE_%": float(row["WAPE_%"]),
            "R2": float(row["R2"]),
            "Mean_Bias": float(row["Mean_Bias"])
        }
        for _, row in baseline_metrics.iterrows()
    }
}

baseline_result_path = os.path.join(
    RESULTS_DIR,
    "Validation_Baseline_Results.json"
)

with open(
    baseline_result_path,
    "w"
) as f:

    json.dump(
        baseline_summary,
        f,
        indent=4
    )

print("\n✅ Detailed baseline results saved:")
print(baseline_result_path)

# ------------------------------------------------------------
# 10. FINAL SAFETY CHECK
# ------------------------------------------------------------

assert (
    val["Split"] == "Validation"
).all()

print("\n" + "=" * 90)
print("✅ VALIDATION BASELINE EXPERIMENT COMPLETED")
print("✅ TEST SET HAS NOT BEEN USED")
print("=" * 90)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 10
# LightGBM clean-condition forecasting model
# Train: Jan–Jun 2026
# Validation: July 2026
# Test: NOT TOUCHED
# ============================================================

import os
import gc
import json
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# LIGHTGBM
# ------------------------------------------------------------

try:
    import lightgbm as lgb
except ImportError:
    !pip -q install lightgbm
    import lightgbm as lgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 95)
print("NOTEBOOK 3 — LIGHTGBM CLEAN-CONDITION FORECASTING MODEL")
print("=" * 95)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

FORECAST_MONTHLY_DIR = str(OUTPUT_ROOT / "Notebook_3_Output" / "Data" / "Forecasting_Monthly")

NOTEBOOK3_OUTPUT = str(OUTPUT_ROOT / "Notebook_3_Output")

TABLES_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Tables")
RESULTS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Results")
MODELS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Models")
DATA_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Data")

# ------------------------------------------------------------
# FEATURES
# ------------------------------------------------------------

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

print("\nPredictor features:")
for f in FEATURES:
    print(f"  • {f}")

print(f"\nTarget: {TARGET}")

# ============================================================
# 1. LOAD TRAINING MONTHS
# ============================================================

train_months = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06"
]

train_parts = []

read_columns = (
    FEATURES
    + [TARGET, "All_Lags_Available"]
)

print("\n" + "=" * 95)
print("LOADING TRAINING DATA")
print("=" * 95)

for month in train_months:

    file = os.path.join(
        FORECAST_MONTHLY_DIR,
        f"FORECASTING_{month}.parquet"
    )

    temp = pd.read_parquet(
        file,
        columns=read_columns
    )

    temp = temp.loc[
        temp["All_Lags_Available"]
    ].drop(
        columns="All_Lags_Available"
    )

    train_parts.append(temp)

    print(
        f"✅ {month}: "
        f"{len(temp):,} model-ready rows"
    )

train_df = pd.concat(
    train_parts,
    ignore_index=True
)

del train_parts
gc.collect()

print(
    f"\nTotal training rows: "
    f"{len(train_df):,}"
)

# ============================================================
# 2. LOAD JULY VALIDATION
# ============================================================

validation_file = os.path.join(
    FORECAST_MONTHLY_DIR,
    "FORECASTING_2026-07.parquet"
)

validation_columns = (
    FEATURES
    + [
        TARGET,
        "All_Lags_Available",
        "Timestamp"
    ]
)

val_df = pd.read_parquet(
    validation_file,
    columns=validation_columns
)

val_df = val_df.loc[
    val_df["All_Lags_Available"]
].drop(
    columns="All_Lags_Available"
).reset_index(drop=True)

print(
    f"Validation rows: "
    f"{len(val_df):,}"
)

# ============================================================
# 3. MEMORY-EFFICIENT DTYPES
# ============================================================

float_features = [
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

for col in float_features:
    train_df[col] = train_df[col].astype("float32")
    val_df[col] = val_df[col].astype("float32")

train_df[TARGET] = train_df[TARGET].astype("float32")
val_df[TARGET] = val_df[TARGET].astype("float32")

# ------------------------------------------------------------
# Split X / y
# ------------------------------------------------------------

X_train = train_df[FEATURES]
y_train = train_df[TARGET]

X_val = val_df[FEATURES]
y_val = val_df[TARGET]

print("\n" + "=" * 95)
print("FINAL MODELING SETS")
print("=" * 95)

print(f"X_train : {X_train.shape}")
print(f"X_val   : {X_val.shape}")

# ============================================================
# 4. LIGHTGBM MODEL
# ============================================================

model = lgb.LGBMRegressor(
    objective="regression",

    n_estimators=1200,
    learning_rate=0.05,

    num_leaves=31,
    max_depth=-1,
    min_child_samples=100,

    max_bin=63,

    subsample=0.80,
    subsample_freq=1,

    colsample_bytree=0.90,

    reg_lambda=1.0,

    random_state=42,
    n_jobs=-1,
    verbosity=-1
)

print("\n" + "=" * 95)
print("TRAINING LIGHTGBM")
print("=" * 95)

model.fit(
    X_train,
    y_train,

    eval_set=[
        (X_val, y_val)
    ],

    eval_metric=[
        "l1",
        "rmse"
    ],

    callbacks=[
        lgb.early_stopping(
            stopping_rounds=50,
            verbose=True
        ),
        lgb.log_evaluation(
            period=50
        )
    ]
)

# ============================================================
# 5. VALIDATION PREDICTIONS
# ============================================================

pred_raw = model.predict(
    X_val,
    num_iteration=model.best_iteration_
)

negative_predictions = int(
    (pred_raw < 0).sum()
)

# Traffic volume cannot be negative
pred = np.clip(
    pred_raw,
    0,
    None
)

print("\nNegative raw predictions clipped to 0:")
print(f"{negative_predictions:,}")

# ============================================================
# 6. METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    error = y_pred - y_true

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    denominator = np.abs(y_true).sum()

    wape = (
        100
        * np.abs(error).sum()
        / denominator
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    bias = error.mean()

    return {
        "MAE": mae,
        "RMSE": rmse,
        "WAPE_%": wape,
        "R2": r2,
        "Mean_Bias": bias
    }


lgb_metrics = calculate_metrics(
    y_val,
    pred
)

print("\n" + "=" * 95)
print("LIGHTGBM — JULY VALIDATION PERFORMANCE")
print("=" * 95)

for key, value in lgb_metrics.items():

    if key == "R2":
        print(f"{key:<12}: {value:.6f}")
    else:
        print(f"{key:<12}: {value:.4f}")

print(
    f"\nBest boosting iteration: "
    f"{model.best_iteration_}"
)

# ============================================================
# 7. COMPARE WITH BASELINES
# ============================================================

baseline_file = os.path.join(
    TABLES_DIR,
    "Table_Validation_Baseline_Performance.csv"
)

comparison = pd.read_csv(
    baseline_file
)

lgb_row = pd.DataFrame([{
    "Model": "LightGBM",
    "N": len(y_val),
    **lgb_metrics
}])

comparison = pd.concat(
    [comparison, lgb_row],
    ignore_index=True
)

comparison = comparison.sort_values(
    "MAE"
).reset_index(drop=True)

print("\n" + "=" * 95)
print("VALIDATION MODEL COMPARISON")
print("=" * 95)

display(
    comparison.style.format({
        "N": "{:,.0f}",
        "MAE": "{:.4f}",
        "RMSE": "{:.4f}",
        "WAPE_%": "{:.3f}",
        "R2": "{:.5f}",
        "Mean_Bias": "{:.4f}"
    })
)

# ============================================================
# 8. IMPROVEMENT OVER BEST BASELINE
# ============================================================

baseline_mae = comparison.loc[
    comparison["Model"] == "Persistence_15min",
    "MAE"
].iloc[0]

baseline_rmse = comparison.loc[
    comparison["Model"] == "Persistence_15min",
    "RMSE"
].iloc[0]

mae_improvement = (
    100
    * (baseline_mae - lgb_metrics["MAE"])
    / baseline_mae
)

rmse_improvement = (
    100
    * (baseline_rmse - lgb_metrics["RMSE"])
    / baseline_rmse
)

print("\n" + "=" * 95)
print("IMPROVEMENT OVER PERSISTENCE BASELINE")
print("=" * 95)

print(
    f"MAE improvement  : "
    f"{mae_improvement:.3f}%"
)

print(
    f"RMSE improvement : "
    f"{rmse_improvement:.3f}%"
)

# ============================================================
# 9. SAVE MODEL
# ============================================================

model_file = os.path.join(
    MODELS_DIR,
    "LightGBM_Clean_15min.txt"
)

model.booster_.save_model(
    model_file
)

print("\n✅ Model saved:")
print(model_file)

# ============================================================
# 10. SAVE COMPARISON TABLE
# ============================================================

comparison_file = os.path.join(
    TABLES_DIR,
    "Table_Validation_Model_Comparison.csv"
)

comparison.to_csv(
    comparison_file,
    index=False
)

print("\n✅ Model comparison table saved:")
print(comparison_file)

# ============================================================
# 11. SAVE VALIDATION PREDICTIONS
# ============================================================

predictions_df = pd.DataFrame({
    "Timestamp":
        val_df["Timestamp"].values,

    "Detector_Code":
        val_df["Detector_Code"].values,

    "Actual":
        y_val.values,

    "LightGBM_Prediction":
        pred.astype("float32"),

    "Persistence_15min":
        val_df["Lag_1"].values,

    "Seasonal_Naive_24h":
        val_df["Lag_96"].values
})

predictions_file = os.path.join(
    DATA_DIR,
    "Validation_LightGBM_Predictions.parquet"
)

predictions_df.to_parquet(
    predictions_file,
    index=False
)

print("\n✅ Validation predictions saved:")
print(predictions_file)

# ============================================================
# 12. SAVE FEATURE IMPORTANCE
# ============================================================

importance_df = pd.DataFrame({
    "Feature": FEATURES,
    "Importance_Gain":
        model.booster_.feature_importance(
            importance_type="gain"
        ),
    "Importance_Split":
        model.booster_.feature_importance(
            importance_type="split"
        )
})

importance_df["Gain_%"] = (
    100
    * importance_df["Importance_Gain"]
    / importance_df["Importance_Gain"].sum()
)

importance_df = importance_df.sort_values(
    "Importance_Gain",
    ascending=False
).reset_index(drop=True)

importance_file = os.path.join(
    TABLES_DIR,
    "Table_LightGBM_Feature_Importance.csv"
)

importance_df.to_csv(
    importance_file,
    index=False
)

print("\n✅ Feature importance saved:")
print(importance_file)

print("\nFeature importance:")
display(
    importance_df.style.format({
        "Importance_Gain": "{:,.2f}",
        "Importance_Split": "{:,.0f}",
        "Gain_%": "{:.2f}"
    })
)

# ============================================================
# 13. SAVE RESULT SUMMARY
# ============================================================

result_summary = {
    "model": "LightGBM",
    "forecast_horizon": "15 minutes",

    "train_period":
        "2026-01-01 to 2026-06-30",

    "validation_period":
        "2026-07-01 to 2026-07-31",

    "train_rows":
        int(len(y_train)),

    "validation_rows":
        int(len(y_val)),

    "best_iteration":
        int(model.best_iteration_),

    "metrics":
        {
            k: float(v)
            for k, v in lgb_metrics.items()
        },

    "improvement_over_persistence": {
        "MAE_%":
            float(mae_improvement),

        "RMSE_%":
            float(rmse_improvement)
    },

    "negative_predictions_clipped":
        negative_predictions,

    "test_set_used":
        False
}

result_file = os.path.join(
    RESULTS_DIR,
    "LightGBM_Validation_Results.json"
)

with open(result_file, "w") as f:
    json.dump(
        result_summary,
        f,
        indent=4
    )

print("\n✅ Detailed results saved:")
print(result_file)

# ============================================================
# 14. FINAL CHECK
# ============================================================

print("\n" + "=" * 95)
print("✅ LIGHTGBM EXPERIMENT COMPLETED")
print("✅ TRAIN = JAN–JUN")
print("✅ VALIDATION = JULY")
print("✅ TEST = AUGUST — STILL UNTOUCHED")
print("=" * 95)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 12
# XGBoost clean-condition forecasting model
# Train: Jan–Jun 2026
# Validation: July 2026
# Test: NOT TOUCHED
# ============================================================

import os
import gc
import json
import numpy as np
import pandas as pd

try:
    import xgboost as xgb
except ImportError:
    !pip -q install xgboost
    import xgboost as xgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 95)
print("NOTEBOOK 3 — XGBOOST CLEAN-CONDITION FORECASTING MODEL")
print("=" * 95)

print(f"\nXGBoost version: {xgb.__version__}")

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

FORECAST_MONTHLY_DIR = str(OUTPUT_ROOT / "Notebook_3_Output" / "Data" / "Forecasting_Monthly")

NOTEBOOK3_OUTPUT = str(OUTPUT_ROOT / "Notebook_3_Output")

TABLES_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Tables")
RESULTS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Results")
MODELS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Models")
DATA_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Data")

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

# ============================================================
# 1. REUSE DATA FROM LIGHTGBM IF AVAILABLE
#    OTHERWISE LOAD IT
# ============================================================

required_objects = [
    "X_train",
    "y_train",
    "X_val",
    "y_val",
    "val_df"
]

reuse_data = all(
    name in globals()
    for name in required_objects
)

if reuse_data:

    print("\n✅ Reusing Train/Validation data already in RAM.")
    print(f"Training rows   : {len(y_train):,}")
    print(f"Validation rows : {len(y_val):,}")

else:

    print("\nLoading Train/Validation data...")

    train_months = [
        "2026-01",
        "2026-02",
        "2026-03",
        "2026-04",
        "2026-05",
        "2026-06"
    ]

    train_parts = []

    read_columns = (
        FEATURES
        + [TARGET, "All_Lags_Available"]
    )

    for month in train_months:

        file = os.path.join(
            FORECAST_MONTHLY_DIR,
            f"FORECASTING_{month}.parquet"
        )

        temp = pd.read_parquet(
            file,
            columns=read_columns
        )

        temp = temp.loc[
            temp["All_Lags_Available"]
        ].drop(
            columns="All_Lags_Available"
        )

        train_parts.append(temp)

        print(
            f"✅ {month}: "
            f"{len(temp):,} rows"
        )

    train_df = pd.concat(
        train_parts,
        ignore_index=True
    )

    del train_parts
    gc.collect()

    validation_file = os.path.join(
        FORECAST_MONTHLY_DIR,
        "FORECASTING_2026-07.parquet"
    )

    val_df = pd.read_parquet(
        validation_file,
        columns=(
            FEATURES
            + [
                TARGET,
                "All_Lags_Available",
                "Timestamp"
            ]
        )
    )

    val_df = val_df.loc[
        val_df["All_Lags_Available"]
    ].drop(
        columns="All_Lags_Available"
    ).reset_index(drop=True)

    X_train = train_df[FEATURES]
    y_train = train_df[TARGET]

    X_val = val_df[FEATURES]
    y_val = val_df[TARGET]

# ------------------------------------------------------------
# Ensure compact numeric types
# ------------------------------------------------------------

X_train = X_train.copy()
X_val = X_val.copy()

for col in FEATURES:
    X_train[col] = X_train[col].astype("float32")
    X_val[col] = X_val[col].astype("float32")

y_train = y_train.astype("float32")
y_val = y_val.astype("float32")

print("\n" + "=" * 95)
print("FINAL MODELING SETS")
print("=" * 95)

print(f"X_train : {X_train.shape}")
print(f"X_val   : {X_val.shape}")

# ============================================================
# 2. XGBOOST MODEL
# ============================================================

xgb_model = xgb.XGBRegressor(
    objective="reg:squarederror",

    n_estimators=1200,
    learning_rate=0.05,

    max_depth=8,
    min_child_weight=20,

    subsample=0.80,
    colsample_bytree=0.90,

    reg_lambda=1.0,

    tree_method="hist",
    max_bin=63,

    eval_metric="rmse",
    early_stopping_rounds=50,

    random_state=42,
    n_jobs=-1
)

print("\n" + "=" * 95)
print("TRAINING XGBOOST")
print("=" * 95)

xgb_model.fit(
    X_train,
    y_train,

    eval_set=[
        (X_val, y_val)
    ],

    verbose=50
)

# ============================================================
# 3. VALIDATION PREDICTIONS
# ============================================================

pred_raw = xgb_model.predict(
    X_val
)

negative_predictions = int(
    (pred_raw < 0).sum()
)

pred = np.clip(
    pred_raw,
    0,
    None
)

print("\nNegative raw predictions clipped to 0:")
print(f"{negative_predictions:,}")

# ============================================================
# 4. METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    error = y_pred - y_true

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    denominator = np.abs(y_true).sum()

    wape = (
        100
        * np.abs(error).sum()
        / denominator
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    bias = error.mean()

    return {
        "MAE": mae,
        "RMSE": rmse,
        "WAPE_%": wape,
        "R2": r2,
        "Mean_Bias": bias
    }


xgb_metrics = calculate_metrics(
    y_val,
    pred
)

print("\n" + "=" * 95)
print("XGBOOST — JULY VALIDATION PERFORMANCE")
print("=" * 95)

for key, value in xgb_metrics.items():

    if key == "R2":
        print(f"{key:<12}: {value:.6f}")
    else:
        print(f"{key:<12}: {value:.4f}")

best_iteration = getattr(
    xgb_model,
    "best_iteration",
    xgb_model.n_estimators - 1
)

print(
    f"\nBest boosting iteration: "
    f"{best_iteration}"
)

# ============================================================
# 5. UPDATE MODEL COMPARISON
# ============================================================

comparison_file = os.path.join(
    TABLES_DIR,
    "Table_Validation_Model_Comparison.csv"
)

comparison = pd.read_csv(
    comparison_file
)

# Avoid duplicate XGBoost row if rerun
comparison = comparison.loc[
    comparison["Model"] != "XGBoost"
].copy()

xgb_row = pd.DataFrame([{
    "Model": "XGBoost",
    "N": len(y_val),
    **xgb_metrics
}])

comparison = pd.concat(
    [comparison, xgb_row],
    ignore_index=True
)

comparison = comparison.sort_values(
    "MAE"
).reset_index(drop=True)

print("\n" + "=" * 95)
print("VALIDATION MODEL COMPARISON")
print("=" * 95)

display(
    comparison.style.format({
        "N": "{:,.0f}",
        "MAE": "{:.4f}",
        "RMSE": "{:.4f}",
        "WAPE_%": "{:.3f}",
        "R2": "{:.5f}",
        "Mean_Bias": "{:.4f}"
    })
)

comparison.to_csv(
    comparison_file,
    index=False
)

# ============================================================
# 6. IMPROVEMENT OVER PERSISTENCE
# ============================================================

baseline_row = comparison.loc[
    comparison["Model"] == "Persistence_15min"
].iloc[0]

mae_improvement = (
    100
    * (
        baseline_row["MAE"]
        - xgb_metrics["MAE"]
    )
    / baseline_row["MAE"]
)

rmse_improvement = (
    100
    * (
        baseline_row["RMSE"]
        - xgb_metrics["RMSE"]
    )
    / baseline_row["RMSE"]
)

print("\n" + "=" * 95)
print("IMPROVEMENT OVER PERSISTENCE BASELINE")
print("=" * 95)

print(
    f"MAE improvement  : "
    f"{mae_improvement:.3f}%"
)

print(
    f"RMSE improvement : "
    f"{rmse_improvement:.3f}%"
)

# ============================================================
# 7. SAVE XGBOOST MODEL
# ============================================================

model_file = os.path.join(
    MODELS_DIR,
    "XGBoost_Clean_15min.json"
)

xgb_model.save_model(
    model_file
)

print("\n✅ XGBoost model saved:")
print(model_file)

# ============================================================
# 8. SAVE FEATURE IMPORTANCE
# ============================================================

booster = xgb_model.get_booster()

gain_scores = booster.get_score(
    importance_type="gain"
)

split_scores = booster.get_score(
    importance_type="weight"
)

importance_rows = []

for feature in FEATURES:

    importance_rows.append({
        "Feature": feature,
        "Importance_Gain":
            float(gain_scores.get(feature, 0)),
        "Importance_Split":
            float(split_scores.get(feature, 0))
    })

xgb_importance = pd.DataFrame(
    importance_rows
)

gain_total = (
    xgb_importance["Importance_Gain"].sum()
)

xgb_importance["Gain_%"] = (
    100
    * xgb_importance["Importance_Gain"]
    / gain_total
)

xgb_importance = xgb_importance.sort_values(
    "Importance_Gain",
    ascending=False
).reset_index(drop=True)

importance_file = os.path.join(
    TABLES_DIR,
    "Table_XGBoost_Feature_Importance.csv"
)

xgb_importance.to_csv(
    importance_file,
    index=False
)

print("\n✅ XGBoost feature importance saved:")
print(importance_file)

display(
    xgb_importance.style.format({
        "Importance_Gain": "{:,.3f}",
        "Importance_Split": "{:,.0f}",
        "Gain_%": "{:.2f}"
    })
)

# ============================================================
# 9. ADD XGBOOST PREDICTIONS TO SAVED VALIDATION FILE
# ============================================================

predictions_file = os.path.join(
    DATA_DIR,
    "Validation_LightGBM_Predictions.parquet"
)

predictions_df = pd.read_parquet(
    predictions_file
)

assert (
    len(predictions_df) == len(pred)
)

predictions_df[
    "XGBoost_Prediction"
] = pred.astype("float32")

final_predictions_file = os.path.join(
    DATA_DIR,
    "Validation_All_Model_Predictions.parquet"
)

predictions_df.to_parquet(
    final_predictions_file,
    index=False
)

print("\n✅ Combined validation predictions saved:")
print(final_predictions_file)

# ============================================================
# 10. SAVE RESULT SUMMARY
# ============================================================

result_summary = {

    "model": "XGBoost",

    "forecast_horizon":
        "15 minutes",

    "train_period":
        "2026-01-01 to 2026-06-30",

    "validation_period":
        "2026-07-01 to 2026-07-31",

    "train_rows":
        int(len(y_train)),

    "validation_rows":
        int(len(y_val)),

    "best_iteration":
        int(best_iteration),

    "metrics": {
        k: float(v)
        for k, v in xgb_metrics.items()
    },

    "improvement_over_persistence": {
        "MAE_%":
            float(mae_improvement),

        "RMSE_%":
            float(rmse_improvement)
    },

    "negative_predictions_clipped":
        negative_predictions,

    "test_set_used":
        False
}

result_file = os.path.join(
    RESULTS_DIR,
    "XGBoost_Validation_Results.json"
)

with open(
    result_file,
    "w"
) as f:

    json.dump(
        result_summary,
        f,
        indent=4
    )

print("\n✅ Detailed XGBoost results saved:")
print(result_file)

print("\n" + "=" * 95)
print("✅ XGBOOST EXPERIMENT COMPLETED")
print("✅ TEST SET (AUGUST) IS STILL UNTOUCHED")
print("=" * 95)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 14
# Controlled LightGBM extension
# Purpose:
# Check whether the previous 1200-tree ceiling limited performance
#
# Train      : Jan–Jun 2026
# Validation : July 2026
# Test       : August — NOT TOUCHED
# Primary selection metric: RMSE
# ============================================================

import os
import gc
import json
import numpy as np
import pandas as pd
import lightgbm as lgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 100)
print("NOTEBOOK 3 — CONTROLLED LIGHTGBM EXTENSION")
print("=" * 100)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

NOTEBOOK3_OUTPUT = str(OUTPUT_ROOT / "Notebook_3_Output")

FORECAST_MONTHLY_DIR = os.path.join(
    NOTEBOOK3_OUTPUT,
    "Data",
    "Forecasting_Monthly"
)

TABLES_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Tables")
RESULTS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Results")
MODELS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Models")
DATA_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Data")

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

# ============================================================
# 1. REUSE DATA IF STILL IN MEMORY
# ============================================================

required_objects = [
    "X_train",
    "y_train",
    "X_val",
    "y_val",
    "val_df"
]

reuse_data = all(
    name in globals()
    for name in required_objects
)

if reuse_data:

    print("\n✅ Reusing Train/Validation data already in RAM.")
    print(f"Training rows   : {len(y_train):,}")
    print(f"Validation rows : {len(y_val):,}")

else:

    print("\n⚠️ Data not found in RAM — reloading from saved monthly files.")

    train_months = [
        "2026-01",
        "2026-02",
        "2026-03",
        "2026-04",
        "2026-05",
        "2026-06"
    ]

    parts = []

    read_cols = FEATURES + [
        TARGET,
        "All_Lags_Available"
    ]

    for month in train_months:

        path = os.path.join(
            FORECAST_MONTHLY_DIR,
            f"FORECASTING_{month}.parquet"
        )

        temp = pd.read_parquet(
            path,
            columns=read_cols
        )

        temp = temp.loc[
            temp["All_Lags_Available"]
        ].drop(
            columns="All_Lags_Available"
        )

        parts.append(temp)

        print(
            f"✅ {month}: "
            f"{len(temp):,} rows"
        )

    train_df = pd.concat(
        parts,
        ignore_index=True
    )

    del parts
    gc.collect()

    val_path = os.path.join(
        FORECAST_MONTHLY_DIR,
        "FORECASTING_2026-07.parquet"
    )

    val_df = pd.read_parquet(
        val_path,
        columns=FEATURES + [
            TARGET,
            "Timestamp",
            "All_Lags_Available"
        ]
    )

    val_df = val_df.loc[
        val_df["All_Lags_Available"]
    ].drop(
        columns="All_Lags_Available"
    ).reset_index(drop=True)

    X_train = train_df[FEATURES]
    y_train = train_df[TARGET]

    X_val = val_df[FEATURES]
    y_val = val_df[TARGET]


# ------------------------------------------------------------
# Compact dtypes
# ------------------------------------------------------------

X_train = X_train.copy()
X_val = X_val.copy()

for col in FEATURES:
    X_train[col] = X_train[col].astype("float32")
    X_val[col] = X_val[col].astype("float32")

y_train = y_train.astype("float32")
y_val = y_val.astype("float32")

print("\n" + "=" * 100)
print("MODELING SETS")
print("=" * 100)

print(f"Train      : {X_train.shape}")
print(f"Validation : {X_val.shape}")

# ============================================================
# 2. CONTROLLED EXTENDED MODEL
# ============================================================

tuned_lgb = lgb.LGBMRegressor(

    objective="regression",

    # Main controlled change:
    n_estimators=2400,

    learning_rate=0.05,

    num_leaves=31,
    max_depth=-1,
    min_child_samples=100,

    max_bin=63,

    subsample=0.80,
    subsample_freq=1,

    colsample_bytree=0.90,

    reg_lambda=1.0,

    random_state=42,
    n_jobs=-1,
    verbosity=-1
)

print("\n" + "=" * 100)
print("TRAINING EXTENDED LIGHTGBM")
print("=" * 100)

print(
    "Only controlled change: "
    "maximum boosting rounds 1200 → 2400"
)

tuned_lgb.fit(

    X_train,
    y_train,

    eval_set=[
        (X_val, y_val)
    ],

    # Primary tuning metric
    eval_metric="rmse",

    callbacks=[
        lgb.early_stopping(
            stopping_rounds=100,
            verbose=True
        ),

        lgb.log_evaluation(
            period=100
        )
    ]
)

# ============================================================
# 3. VALIDATION PREDICTIONS
# ============================================================

pred_raw = tuned_lgb.predict(
    X_val,
    num_iteration=tuned_lgb.best_iteration_
)

negative_predictions = int(
    (pred_raw < 0).sum()
)

pred = np.clip(
    pred_raw,
    0,
    None
)

# ============================================================
# 4. METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    error = y_pred - y_true

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    wape = (
        100
        * np.abs(error).sum()
        / np.abs(y_true).sum()
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    bias = error.mean()

    return {
        "MAE": mae,
        "RMSE": rmse,
        "WAPE_%": wape,
        "R2": r2,
        "Mean_Bias": bias
    }


tuned_metrics = calculate_metrics(
    y_val,
    pred
)

print("\n" + "=" * 100)
print("EXTENDED LIGHTGBM — VALIDATION PERFORMANCE")
print("=" * 100)

for key, value in tuned_metrics.items():

    if key == "R2":
        print(f"{key:<12}: {value:.6f}")
    else:
        print(f"{key:<12}: {value:.4f}")

print(
    f"\nBest iteration: "
    f"{tuned_lgb.best_iteration_}"
)

print(
    f"Negative raw predictions clipped: "
    f"{negative_predictions:,}"
)

# ============================================================
# 5. COMPARE WITH ORIGINAL LIGHTGBM
# ============================================================

comparison_file = os.path.join(
    TABLES_DIR,
    "Table_Validation_Model_Comparison.csv"
)

comparison = pd.read_csv(
    comparison_file
)

original_lgb = comparison.loc[
    comparison["Model"] == "LightGBM"
].iloc[0]

original_rmse = float(
    original_lgb["RMSE"]
)

original_mae = float(
    original_lgb["MAE"]
)

rmse_gain_pct = (
    100
    * (original_rmse - tuned_metrics["RMSE"])
    / original_rmse
)

mae_gain_pct = (
    100
    * (original_mae - tuned_metrics["MAE"])
    / original_mae
)

print("\n" + "=" * 100)
print("ORIGINAL vs EXTENDED LIGHTGBM")
print("=" * 100)

comparison_lgb = pd.DataFrame([
    {
        "Model": "LightGBM_1200",
        "MAE": original_mae,
        "RMSE": original_rmse,
        "WAPE_%": float(
            original_lgb["WAPE_%"]
        ),
        "R2": float(
            original_lgb["R2"]
        )
    },

    {
        "Model": "LightGBM_Extended",
        "MAE": tuned_metrics["MAE"],
        "RMSE": tuned_metrics["RMSE"],
        "WAPE_%": tuned_metrics["WAPE_%"],
        "R2": tuned_metrics["R2"]
    }
])

display(
    comparison_lgb.style.format({
        "MAE": "{:.4f}",
        "RMSE": "{:.4f}",
        "WAPE_%": "{:.3f}",
        "R2": "{:.6f}"
    })
)

print(
    f"\nRMSE improvement over original LightGBM: "
    f"{rmse_gain_pct:.3f}%"
)

print(
    f"MAE improvement over original LightGBM : "
    f"{mae_gain_pct:.3f}%"
)

# ============================================================
# 6. AUTOMATIC STOP / KEEP DECISION
# ============================================================

print("\n" + "=" * 100)
print("TUNING DECISION")
print("=" * 100)

# Predefined practical threshold:
# <0.5% RMSE improvement = marginal
# >=0.5% RMSE improvement = worthwhile

if tuned_metrics["RMSE"] < original_rmse:

    if rmse_gain_pct >= 0.5:

        decision = (
            "KEEP_EXTENDED_MODEL"
        )

        print(
            "✅ Meaningful RMSE improvement (≥0.5%)."
        )

        print(
            "✅ Extended LightGBM should replace "
            "the original model."
        )

    else:

        decision = (
            "MARGINAL_IMPROVEMENT"
        )

        print(
            "⚠️ Extended model improved RMSE, "
            "but by <0.5%."
        )

        print(
            "➡️ Improvement is marginal; "
            "additional tuning is probably unnecessary."
        )

else:

    decision = (
        "KEEP_ORIGINAL_MODEL"
    )

    print(
        "✅ Original LightGBM remains better."
    )

    print(
        "➡️ Stop tuning and retain the original model."
    )

# ============================================================
# 7. SAVE EXTENDED MODEL
# ============================================================

model_path = os.path.join(
    MODELS_DIR,
    "LightGBM_Extended_Validation.txt"
)

tuned_lgb.booster_.save_model(
    model_path
)

print("\n✅ Extended model saved:")
print(model_path)

# ============================================================
# 8. SAVE RESULTS TABLE
# ============================================================

tuning_table_path = os.path.join(
    TABLES_DIR,
    "Table_LightGBM_Controlled_Tuning.csv"
)

comparison_lgb.to_csv(
    tuning_table_path,
    index=False
)

print("\n✅ Tuning comparison table saved:")
print(tuning_table_path)

# ============================================================
# 9. SAVE EXTENDED VALIDATION PREDICTIONS
# ============================================================

prediction_path = os.path.join(
    DATA_DIR,
    "Validation_LightGBM_Extended_Predictions.parquet"
)

pd.DataFrame({
    "Timestamp":
        val_df["Timestamp"].values,

    "Detector_Code":
        X_val["Detector_Code"].values,

    "Actual":
        y_val.values,

    "Prediction":
        pred.astype("float32")
}).to_parquet(
    prediction_path,
    index=False
)

print("\n✅ Extended predictions saved:")
print(prediction_path)

# ============================================================
# 10. SAVE DECISION RECORD
# ============================================================

result = {

    "experiment":
        "Controlled LightGBM Extension",

    "primary_selection_metric":
        "RMSE",

    "original_max_estimators":
        1200,

    "extended_max_estimators":
        2400,

    "extended_best_iteration":
        int(tuned_lgb.best_iteration_),

    "original_RMSE":
        original_rmse,

    "extended_RMSE":
        float(tuned_metrics["RMSE"]),

    "RMSE_improvement_%":
        float(rmse_gain_pct),

    "MAE_improvement_%":
        float(mae_gain_pct),

    "decision":
        decision,

    "test_set_used":
        False
}

result_path = os.path.join(
    RESULTS_DIR,
    "LightGBM_Controlled_Tuning_Decision.json"
)

with open(
    result_path,
    "w"
) as f:

    json.dump(
        result,
        f,
        indent=4
    )

print("\n✅ Tuning decision saved:")
print(result_path)

print("\n" + "=" * 100)
print("✅ CONTROLLED LIGHTGBM TUNING COMPLETED")
print("✅ AUGUST TEST SET REMAINS UNTOUCHED")
print("=" * 100)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 15
# FINAL LightGBM Clean-Condition Model
#
# Final training : Jan–Jul 2026
# Final test     : Aug 2026
#
# Hyperparameters are LOCKED.
# NO tuning / early stopping on August.
# ============================================================

import os
import gc
import json
import numpy as np
import pandas as pd
import lightgbm as lgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 100)
print("NOTEBOOK 3 — FINAL LIGHTGBM TEST EVALUATION")
print("=" * 100)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

NOTEBOOK3_OUTPUT = str(OUTPUT_ROOT / "Notebook_3_Output")

FORECAST_MONTHLY_DIR = os.path.join(
    NOTEBOOK3_OUTPUT,
    "Data",
    "Forecasting_Monthly"
)

TABLES_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Tables")
RESULTS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Results")
MODELS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Models")
DATA_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Data")

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

# ============================================================
# 1. LOAD FINAL TRAINING DATA: JAN–JUL
# ============================================================

final_train_months = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06",
    "2026-07"
]

train_parts = []

read_columns = (
    FEATURES
    + [
        TARGET,
        "All_Lags_Available"
    ]
)

print("\n" + "=" * 100)
print("LOADING FINAL TRAINING DATA — JAN TO JUL")
print("=" * 100)

for month in final_train_months:

    path = os.path.join(
        FORECAST_MONTHLY_DIR,
        f"FORECASTING_{month}.parquet"
    )

    temp = pd.read_parquet(
        path,
        columns=read_columns
    )

    temp = temp.loc[
        temp["All_Lags_Available"]
    ].drop(
        columns="All_Lags_Available"
    )

    # Compact types
    for col in FEATURES:
        temp[col] = temp[col].astype("float32")

    temp[TARGET] = temp[TARGET].astype("float32")

    train_parts.append(temp)

    print(
        f"✅ {month}: "
        f"{len(temp):,} rows"
    )

final_train = pd.concat(
    train_parts,
    ignore_index=True
)

del train_parts
gc.collect()

X_final_train = final_train[FEATURES]
y_final_train = final_train[TARGET]

print(
    f"\nFinal training observations: "
    f"{len(final_train):,}"
)

print(
    f"Training matrix: "
    f"{X_final_train.shape}"
)

# ============================================================
# 2. LOCKED FINAL MODEL
# ============================================================

final_lgb = lgb.LGBMRegressor(

    objective="regression",

    # LOCKED FROM VALIDATION STAGE
    n_estimators=1200,
    learning_rate=0.05,

    num_leaves=31,
    max_depth=-1,
    min_child_samples=100,

    max_bin=63,

    subsample=0.80,
    subsample_freq=1,

    colsample_bytree=0.90,

    reg_lambda=1.0,

    random_state=42,
    n_jobs=-1,
    verbosity=-1
)

print("\n" + "=" * 100)
print("TRAINING FINAL LIGHTGBM ON JAN–JUL")
print("=" * 100)

print("Hyperparameters are LOCKED.")
print("August is NOT used for tuning or early stopping.")

final_lgb.fit(
    X_final_train,
    y_final_train,
    callbacks=[
        lgb.log_evaluation(period=100)
    ]
)

print("\n✅ Final LightGBM training completed.")

# ------------------------------------------------------------
# Free final training matrix before loading test
# ------------------------------------------------------------

del final_train
del X_final_train
del y_final_train

gc.collect()

# ============================================================
# 3. LOAD AUGUST FINAL TEST
# ============================================================

test_path = os.path.join(
    FORECAST_MONTHLY_DIR,
    "FORECASTING_2026-08.parquet"
)

test_columns = (
    FEATURES
    + [
        TARGET,
        "Timestamp",
        "All_Lags_Available"
    ]
)

test_df = pd.read_parquet(
    test_path,
    columns=test_columns
)

test_df = test_df.loc[
    test_df["All_Lags_Available"]
].reset_index(drop=True)

for col in FEATURES:
    test_df[col] = test_df[col].astype("float32")

test_df[TARGET] = test_df[TARGET].astype("float32")

X_test = test_df[FEATURES]
y_test = test_df[TARGET]

print("\n" + "=" * 100)
print("FINAL UNSEEN TEST SET")
print("=" * 100)

print(
    f"Test rows       : "
    f"{len(test_df):,}"
)

print(
    f"Period          : "
    f"{test_df['Timestamp'].min()} "
    f"→ {test_df['Timestamp'].max()}"
)

print(
    f"Unique detectors: "
    f"{test_df['Detector_Code'].nunique():,}"
)

# ============================================================
# 4. FINAL LIGHTGBM PREDICTIONS
# ============================================================

pred_raw = final_lgb.predict(
    X_test
)

negative_predictions = int(
    (pred_raw < 0).sum()
)

final_pred = np.clip(
    pred_raw,
    0,
    None
)

print(
    f"\nNegative raw predictions clipped: "
    f"{negative_predictions:,}"
)

# ============================================================
# 5. METRIC FUNCTION
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    error = y_pred - y_true

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    wape = (
        100
        * np.abs(error).sum()
        / np.abs(y_true).sum()
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    bias = error.mean()

    return {
        "MAE": mae,
        "RMSE": rmse,
        "WAPE_%": wape,
        "R2": r2,
        "Mean_Bias": bias
    }

# ============================================================
# 6. FINAL LIGHTGBM METRICS
# ============================================================

lgb_test_metrics = calculate_metrics(
    y_test,
    final_pred
)

print("\n" + "=" * 100)
print("FINAL LIGHTGBM — AUGUST TEST PERFORMANCE")
print("=" * 100)

for key, value in lgb_test_metrics.items():

    if key == "R2":
        print(f"{key:<12}: {value:.6f}")
    else:
        print(f"{key:<12}: {value:.4f}")

# ============================================================
# 7. AUGUST BASELINES
# ============================================================

persistence_pred = (
    test_df["Lag_1"]
    .astype("float64")
    .values
)

seasonal_pred = (
    test_df["Lag_96"]
    .astype("float64")
    .values
)

persistence_metrics = calculate_metrics(
    y_test,
    persistence_pred
)

seasonal_metrics = calculate_metrics(
    y_test,
    seasonal_pred
)

final_test_comparison = pd.DataFrame([
    {
        "Model": "LightGBM_Final",
        "N": len(y_test),
        **lgb_test_metrics
    },
    {
        "Model": "Persistence_15min",
        "N": len(y_test),
        **persistence_metrics
    },
    {
        "Model": "Seasonal_Naive_24h",
        "N": len(y_test),
        **seasonal_metrics
    }
])

final_test_comparison = (
    final_test_comparison
    .sort_values("RMSE")
    .reset_index(drop=True)
)

print("\n" + "=" * 100)
print("AUGUST FINAL TEST COMPARISON")
print("=" * 100)

display(
    final_test_comparison.style.format({
        "N": "{:,.0f}",
        "MAE": "{:.4f}",
        "RMSE": "{:.4f}",
        "WAPE_%": "{:.3f}",
        "R2": "{:.6f}",
        "Mean_Bias": "{:.4f}"
    })
)

# ============================================================
# 8. IMPROVEMENT OVER PERSISTENCE
# ============================================================

mae_improvement = (
    100
    * (
        persistence_metrics["MAE"]
        - lgb_test_metrics["MAE"]
    )
    / persistence_metrics["MAE"]
)

rmse_improvement = (
    100
    * (
        persistence_metrics["RMSE"]
        - lgb_test_metrics["RMSE"]
    )
    / persistence_metrics["RMSE"]
)

wape_improvement = (
    100
    * (
        persistence_metrics["WAPE_%"]
        - lgb_test_metrics["WAPE_%"]
    )
    / persistence_metrics["WAPE_%"]
)

print("\n" + "=" * 100)
print("FINAL IMPROVEMENT OVER PERSISTENCE BASELINE")
print("=" * 100)

print(
    f"MAE reduction  : "
    f"{mae_improvement:.3f}%"
)

print(
    f"RMSE reduction : "
    f"{rmse_improvement:.3f}%"
)

print(
    f"WAPE reduction : "
    f"{wape_improvement:.3f}%"
)

# ============================================================
# 9. SAVE FINAL MODEL
# ============================================================

final_model_path = os.path.join(
    MODELS_DIR,
    "LightGBM_FINAL_JanJul_1200.txt"
)

final_lgb.booster_.save_model(
    final_model_path
)

print("\n✅ FINAL model saved:")
print(final_model_path)

# ============================================================
# 10. SAVE FINAL TEST TABLE
# ============================================================

test_table_path = os.path.join(
    TABLES_DIR,
    "Table_August_Final_Test_Performance.csv"
)

final_test_comparison.to_csv(
    test_table_path,
    index=False
)

print("\n✅ Final test table saved:")
print(test_table_path)

# ============================================================
# 11. SAVE TEST PREDICTIONS
# ============================================================

test_predictions = pd.DataFrame({

    "Timestamp":
        test_df["Timestamp"].values,

    "Detector_Code":
        test_df["Detector_Code"].values,

    "Actual":
        y_test.values,

    "LightGBM_Prediction":
        final_pred.astype("float32"),

    "Persistence_15min":
        persistence_pred.astype("float32"),

    "Seasonal_Naive_24h":
        seasonal_pred.astype("float32")
})

test_prediction_path = os.path.join(
    DATA_DIR,
    "August_Final_Test_Predictions.parquet"
)

test_predictions.to_parquet(
    test_prediction_path,
    index=False
)

print("\n✅ Final test predictions saved:")
print(test_prediction_path)

# ============================================================
# 12. SAVE PAPER-READY RESULT RECORD
# ============================================================

final_result = {

    "model":
        "LightGBM",

    "forecast_horizon":
        "15 minutes",

    "final_training_period":
        "2026-01-01 to 2026-07-31",

    "final_test_period":
        "2026-08-01 to 2026-08-30",

    "fixed_hyperparameters": {
        "n_estimators": 1200,
        "learning_rate": 0.05,
        "num_leaves": 31,
        "min_child_samples": 100,
        "max_bin": 63,
        "subsample": 0.80,
        "colsample_bytree": 0.90,
        "reg_lambda": 1.0
    },

    "test_rows":
        int(len(y_test)),

    "test_metrics": {
        k: float(v)
        for k, v in lgb_test_metrics.items()
    },

    "improvement_over_persistence": {
        "MAE_reduction_%":
            float(mae_improvement),

        "RMSE_reduction_%":
            float(rmse_improvement),

        "WAPE_reduction_%":
            float(wape_improvement)
    },

    "negative_predictions_clipped":
        negative_predictions,

    "test_used_for_tuning":
        False
}

final_result_path = os.path.join(
    RESULTS_DIR,
    "LightGBM_FINAL_August_Test_Results.json"
)

with open(
    final_result_path,
    "w"
) as f:

    json.dump(
        final_result,
        f,
        indent=4
    )

print("\n✅ Final paper-ready result record saved:")
print(final_result_path)

print("\n" + "=" * 100)
print("✅ FINAL LIGHTGBM CLEAN-CONDITION TEST COMPLETE")
print("✅ AUGUST WAS USED ONLY AS AN UNSEEN TEST SET")
print("✅ NO TEST-SET TUNING WAS PERFORMED")
print("=" * 100)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 16
# FINAL XGBoost Clean-Condition Model
#
# Final training : Jan–Jul 2026
# Final test     : Aug 2026
#
# Hyperparameters LOCKED from July validation.
# NO tuning / early stopping on August.
# ============================================================

import os
import gc
import json
import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 100)
print("NOTEBOOK 3 — FINAL XGBOOST TEST EVALUATION")
print("=" * 100)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

NOTEBOOK3_OUTPUT = str(OUTPUT_ROOT / "Notebook_3_Output")

FORECAST_MONTHLY_DIR = os.path.join(
    NOTEBOOK3_OUTPUT,
    "Data",
    "Forecasting_Monthly"
)

TABLES_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Tables")
RESULTS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Results")
MODELS_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Models")
DATA_DIR = os.path.join(NOTEBOOK3_OUTPUT, "Data")

FEATURES = [
    "Detector_Code",
    "Interval_Index",
    "DayOfWeek",
    "IsWeekend",
    "Lag_1",
    "Lag_2",
    "Lag_4",
    "Lag_8",
    "Lag_96"
]

TARGET = "Target_Volume"

# ============================================================
# 1. LOAD FINAL TRAINING DATA — JAN TO JUL
# ============================================================

final_train_months = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06",
    "2026-07"
]

train_parts = []

read_columns = FEATURES + [
    TARGET,
    "All_Lags_Available"
]

print("\n" + "=" * 100)
print("LOADING FINAL TRAINING DATA — JAN TO JUL")
print("=" * 100)

for month in final_train_months:

    path = os.path.join(
        FORECAST_MONTHLY_DIR,
        f"FORECASTING_{month}.parquet"
    )

    temp = pd.read_parquet(
        path,
        columns=read_columns
    )

    temp = temp.loc[
        temp["All_Lags_Available"]
    ].drop(
        columns="All_Lags_Available"
    )

    for col in FEATURES:
        temp[col] = temp[col].astype("float32")

    temp[TARGET] = temp[TARGET].astype("float32")

    train_parts.append(temp)

    print(
        f"✅ {month}: "
        f"{len(temp):,} rows"
    )

final_train = pd.concat(
    train_parts,
    ignore_index=True
)

del train_parts
gc.collect()

X_final_train = final_train[FEATURES]
y_final_train = final_train[TARGET]

print(
    f"\nFinal training observations: "
    f"{len(final_train):,}"
)

print(
    f"Training matrix: "
    f"{X_final_train.shape}"
)

# ============================================================
# 2. LOCKED FINAL XGBOOST MODEL
# ============================================================

# July validation best_iteration = 623 (zero-based)
# Therefore final fixed number of trees = 624

FINAL_XGB_TREES = 624

final_xgb = xgb.XGBRegressor(

    objective="reg:squarederror",

    n_estimators=FINAL_XGB_TREES,
    learning_rate=0.05,

    max_depth=8,
    min_child_weight=20,

    subsample=0.80,
    colsample_bytree=0.90,

    reg_lambda=1.0,

    tree_method="hist",
    max_bin=63,

    random_state=42,
    n_jobs=-1
)

print("\n" + "=" * 100)
print("TRAINING FINAL XGBOOST ON JAN–JUL")
print("=" * 100)

print("Hyperparameters are LOCKED.")
print(f"Fixed boosting trees: {FINAL_XGB_TREES}")
print("August is NOT used for tuning or early stopping.")

final_xgb.fit(
    X_final_train,
    y_final_train,
    verbose=False
)

print("\n✅ Final XGBoost training completed.")

# Free training data
del final_train
del X_final_train
del y_final_train
gc.collect()

# ============================================================
# 3. LOAD AUGUST TEST
# ============================================================

test_path = os.path.join(
    FORECAST_MONTHLY_DIR,
    "FORECASTING_2026-08.parquet"
)

test_df = pd.read_parquet(
    test_path,
    columns=FEATURES + [
        TARGET,
        "Timestamp",
        "All_Lags_Available"
    ]
)

test_df = test_df.loc[
    test_df["All_Lags_Available"]
].reset_index(drop=True)

for col in FEATURES:
    test_df[col] = test_df[col].astype("float32")

test_df[TARGET] = test_df[TARGET].astype("float32")

X_test = test_df[FEATURES]
y_test = test_df[TARGET]

print("\n" + "=" * 100)
print("FINAL UNSEEN TEST SET")
print("=" * 100)

print(f"Test rows        : {len(test_df):,}")
print(
    f"Period           : "
    f"{test_df['Timestamp'].min()} → "
    f"{test_df['Timestamp'].max()}"
)
print(
    f"Unique detectors : "
    f"{test_df['Detector_Code'].nunique():,}"
)

# ============================================================
# 4. PREDICTIONS
# ============================================================

xgb_pred_raw = final_xgb.predict(X_test)

negative_predictions = int(
    (xgb_pred_raw < 0).sum()
)

xgb_pred = np.clip(
    xgb_pred_raw,
    0,
    None
)

print(
    f"\nNegative raw predictions clipped: "
    f"{negative_predictions:,}"
)

# ============================================================
# 5. METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    error = y_pred - y_true

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    wape = (
        100
        * np.abs(error).sum()
        / np.abs(y_true).sum()
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    bias = error.mean()

    return {
        "MAE": mae,
        "RMSE": rmse,
        "WAPE_%": wape,
        "R2": r2,
        "Mean_Bias": bias
    }


xgb_test_metrics = calculate_metrics(
    y_test,
    xgb_pred
)

print("\n" + "=" * 100)
print("FINAL XGBOOST — AUGUST TEST PERFORMANCE")
print("=" * 100)

for key, value in xgb_test_metrics.items():

    if key == "R2":
        print(f"{key:<12}: {value:.6f}")
    else:
        print(f"{key:<12}: {value:.4f}")

# ============================================================
# 6. LOAD EXISTING LIGHTGBM + BASELINE TEST TABLE
# ============================================================

test_table_path = os.path.join(
    TABLES_DIR,
    "Table_August_Final_Test_Performance.csv"
)

final_comparison = pd.read_csv(
    test_table_path
)

# Prevent duplication if rerun
final_comparison = final_comparison.loc[
    final_comparison["Model"] != "XGBoost_Final"
].copy()

xgb_row = pd.DataFrame([{
    "Model": "XGBoost_Final",
    "N": len(y_test),
    **xgb_test_metrics
}])

final_comparison = pd.concat(
    [final_comparison, xgb_row],
    ignore_index=True
)

final_comparison = (
    final_comparison
    .sort_values("RMSE")
    .reset_index(drop=True)
)

print("\n" + "=" * 100)
print("FINAL AUGUST TEST — ALL MODEL COMPARISON")
print("=" * 100)

display(
    final_comparison.style.format({
        "N": "{:,.0f}",
        "MAE": "{:.4f}",
        "RMSE": "{:.4f}",
        "WAPE_%": "{:.3f}",
        "R2": "{:.6f}",
        "Mean_Bias": "{:.4f}"
    })
)

# Update final table
final_comparison.to_csv(
    test_table_path,
    index=False
)

# ============================================================
# 7. IMPROVEMENT OVER PERSISTENCE
# ============================================================

persistence = final_comparison.loc[
    final_comparison["Model"] == "Persistence_15min"
].iloc[0]

mae_improvement = (
    100
    * (persistence["MAE"] - xgb_test_metrics["MAE"])
    / persistence["MAE"]
)

rmse_improvement = (
    100
    * (persistence["RMSE"] - xgb_test_metrics["RMSE"])
    / persistence["RMSE"]
)

print("\n" + "=" * 100)
print("XGBOOST IMPROVEMENT OVER PERSISTENCE")
print("=" * 100)

print(
    f"MAE reduction  : "
    f"{mae_improvement:.3f}%"
)

print(
    f"RMSE reduction : "
    f"{rmse_improvement:.3f}%"
)

# ============================================================
# 8. SAVE FINAL MODEL
# ============================================================

model_path = os.path.join(
    MODELS_DIR,
    "XGBoost_FINAL_JanJul_624.json"
)

final_xgb.save_model(
    model_path
)

print("\n✅ Final XGBoost model saved:")
print(model_path)

# ============================================================
# 9. ADD XGBOOST TO AUGUST PREDICTIONS
# ============================================================

prediction_path = os.path.join(
    DATA_DIR,
    "August_Final_Test_Predictions.parquet"
)

predictions = pd.read_parquet(
    prediction_path
)

assert len(predictions) == len(xgb_pred)

predictions[
    "XGBoost_Prediction"
] = xgb_pred.astype("float32")

final_prediction_path = os.path.join(
    DATA_DIR,
    "August_Final_All_Model_Predictions.parquet"
)

predictions.to_parquet(
    final_prediction_path,
    index=False
)

print("\n✅ Combined August predictions saved:")
print(final_prediction_path)

# ============================================================
# 10. SAVE FINAL RESULT RECORD
# ============================================================

result = {

    "model": "XGBoost",

    "forecast_horizon":
        "15 minutes",

    "final_training_period":
        "2026-01-01 to 2026-07-31",

    "final_test_period":
        "2026-08-01 to 2026-08-30",

    "fixed_trees":
        FINAL_XGB_TREES,

    "test_rows":
        int(len(y_test)),

    "test_metrics": {
        k: float(v)
        for k, v in xgb_test_metrics.items()
    },

    "improvement_over_persistence": {
        "MAE_reduction_%":
            float(mae_improvement),

        "RMSE_reduction_%":
            float(rmse_improvement)
    },

    "negative_predictions_clipped":
        negative_predictions,

    "test_used_for_tuning":
        False
}

result_path = os.path.join(
    RESULTS_DIR,
    "XGBoost_FINAL_August_Test_Results.json"
)

with open(result_path, "w") as f:
    json.dump(
        result,
        f,
        indent=4
    )

print("\n✅ Final XGBoost result record saved:")
print(result_path)

print("\n" + "=" * 100)
print("✅ FINAL XGBOOST TEST COMPLETE")
print("✅ AUGUST WAS NOT USED FOR TUNING")
print("=" * 100)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 19
# Publication-ready Error by Time of Day
# Final LightGBM — August Test
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 95)
print("NOTEBOOK 3 — LIGHTGBM TEST ERROR BY TIME OF DAY")
print("=" * 95)

# ------------------------------------------------------------
# 1. LOAD AUGUST FINAL PREDICTIONS
# ------------------------------------------------------------

prediction_file = os.path.join(
    DATA_DIR,
    "August_Final_All_Model_Predictions.parquet"
)

error_df = pd.read_parquet(
    prediction_file,
    columns=[
        "Timestamp",
        "Actual",
        "LightGBM_Prediction"
    ]
)

# ------------------------------------------------------------
# 2. CREATE 15-MINUTE INTERVAL INDEX
# ------------------------------------------------------------

error_df["Interval_Index"] = (
    error_df["Timestamp"].dt.hour * 4
    + error_df["Timestamp"].dt.minute // 15
)

error_df["Absolute_Error"] = (
    error_df["LightGBM_Prediction"]
    - error_df["Actual"]
).abs()

error_df["Squared_Error"] = (
    error_df["LightGBM_Prediction"]
    - error_df["Actual"]
) ** 2

# ------------------------------------------------------------
# 3. AGGREGATE BY TIME OF DAY
# ------------------------------------------------------------

time_error = (
    error_df
    .groupby("Interval_Index")
    .agg(
        MAE=("Absolute_Error", "mean"),
        MSE=("Squared_Error", "mean"),
        Mean_Observed_Volume=("Actual", "mean"),
        Observations=("Actual", "size")
    )
    .reset_index()
)

time_error["RMSE"] = np.sqrt(
    time_error["MSE"]
)

# Human-readable time labels
time_error["Time"] = pd.to_datetime(
    time_error["Interval_Index"] * 15,
    unit="m"
).dt.strftime("%H:%M")

print("\nError range across the day:")
print(
    f"Lowest MAE : "
    f"{time_error['MAE'].min():.4f}"
)

print(
    f"Highest MAE: "
    f"{time_error['MAE'].max():.4f}"
)

worst_row = time_error.loc[
    time_error["MAE"].idxmax()
]

best_row = time_error.loc[
    time_error["MAE"].idxmin()
]

print(
    f"\nLowest-error time : "
    f"{best_row['Time']} "
    f"(MAE={best_row['MAE']:.4f})"
)

print(
    f"Highest-error time: "
    f"{worst_row['Time']} "
    f"(MAE={worst_row['MAE']:.4f})"
)

# ------------------------------------------------------------
# 4. PUBLICATION-READY FIGURE
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 5.5)
)

ax.plot(
    time_error["Interval_Index"],
    time_error["MAE"],
    linewidth=2
)

ax.set_xlabel(
    "Time of Day",
    fontsize=12
)

ax.set_ylabel(
    "Mean Absolute Error\n(vehicles per 15-min interval)",
    fontsize=12
)

ax.set_title(
    "Final LightGBM Forecast Error by Time of Day",
    fontsize=13,
    pad=14
)

# Hourly ticks
hourly_positions = np.arange(
    0,
    96,
    4
)

hourly_labels = [
    f"{hour:02d}:00"
    for hour in range(24)
]

ax.set_xticks(
    hourly_positions
)

ax.set_xticklabels(
    hourly_labels,
    rotation=45,
    ha="right",
    fontsize=9
)

ax.tick_params(
    axis="y",
    labelsize=10
)

ax.grid(
    linestyle="--",
    alpha=0.30
)

plt.tight_layout()

# ------------------------------------------------------------
# 5. SAVE FIGURE
# ------------------------------------------------------------

png_path = os.path.join(
    FIGURES_DIR,
    "Figure_August_LightGBM_MAE_by_Time_of_Day.png"
)

pdf_path = os.path.join(
    FIGURES_DIR,
    "Figure_August_LightGBM_MAE_by_Time_of_Day.pdf"
)

plt.savefig(
    png_path,
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    pdf_path,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 6. SAVE TABLE
# ------------------------------------------------------------

table_path = os.path.join(
    TABLES_DIR,
    "Table_August_LightGBM_Error_by_Time_of_Day.csv"
)

time_error[
    [
        "Interval_Index",
        "Time",
        "MAE",
        "RMSE",
        "Mean_Observed_Volume",
        "Observations"
    ]
].to_csv(
    table_path,
    index=False
)

print("\n✅ Time-of-day error figure saved:")
print(png_path)

print("\n✅ Vector PDF saved:")
print(pdf_path)

print("\n✅ Time-of-day error table saved:")
print(table_path)

print("\n" + "=" * 95)
print("✅ TIME-OF-DAY ERROR ANALYSIS COMPLETED")
print("=" * 95)
```

```python
# ============================================================
# NOTEBOOK 3 — CODE 20
# Publication-ready WAPE by Time of Day
# Final LightGBM — August Test
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 95)
print("NOTEBOOK 3 — LIGHTGBM RELATIVE ERROR BY TIME OF DAY")
print("=" * 95)

# ------------------------------------------------------------
# 1. LOAD FINAL AUGUST PREDICTIONS
# ------------------------------------------------------------

prediction_file = os.path.join(
    DATA_DIR,
    "August_Final_All_Model_Predictions.parquet"
)

df_time = pd.read_parquet(
    prediction_file,
    columns=[
        "Timestamp",
        "Actual",
        "LightGBM_Prediction"
    ]
)

# ------------------------------------------------------------
# 2. TIME INDEX + ERRORS
# ------------------------------------------------------------

df_time["Interval_Index"] = (
    df_time["Timestamp"].dt.hour * 4
    + df_time["Timestamp"].dt.minute // 15
)

df_time["Absolute_Error"] = (
    df_time["LightGBM_Prediction"]
    - df_time["Actual"]
).abs()

# ------------------------------------------------------------
# 3. AGGREGATE
# ------------------------------------------------------------

time_wape = (
    df_time
    .groupby("Interval_Index")
    .agg(
        Absolute_Error_Sum=("Absolute_Error", "sum"),
        Actual_Volume_Sum=("Actual", "sum"),
        Mean_Observed_Volume=("Actual", "mean"),
        Observations=("Actual", "size")
    )
    .reset_index()
)

time_wape["WAPE_%"] = (
    100
    * time_wape["Absolute_Error_Sum"]
    / time_wape["Actual_Volume_Sum"]
)

time_wape["Time"] = pd.to_datetime(
    time_wape["Interval_Index"] * 15,
    unit="m"
).dt.strftime("%H:%M")

# ------------------------------------------------------------
# 4. SUMMARY
# ------------------------------------------------------------

best_row = time_wape.loc[
    time_wape["WAPE_%"].idxmin()
]

worst_row = time_wape.loc[
    time_wape["WAPE_%"].idxmax()
]

print(
    f"\nLowest WAPE : "
    f"{best_row['WAPE_%']:.3f}% "
    f"at {best_row['Time']}"
)

print(
    f"Highest WAPE: "
    f"{worst_row['WAPE_%']:.3f}% "
    f"at {worst_row['Time']}"
)

# ------------------------------------------------------------
# 5. FIGURE
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 5.5)
)

ax.plot(
    time_wape["Interval_Index"],
    time_wape["WAPE_%"],
    linewidth=2
)

ax.set_xlabel(
    "Time of Day",
    fontsize=12
)

ax.set_ylabel(
    "WAPE (%)",
    fontsize=12
)

ax.set_title(
    "Final LightGBM Relative Forecast Error by Time of Day",
    fontsize=13,
    pad=14
)

hourly_positions = np.arange(
    0,
    96,
    4
)

hourly_labels = [
    f"{hour:02d}:00"
    for hour in range(24)
]

ax.set_xticks(hourly_positions)

ax.set_xticklabels(
    hourly_labels,
    rotation=45,
    ha="right",
    fontsize=9
)

ax.tick_params(
    axis="y",
    labelsize=10
)

ax.grid(
    linestyle="--",
    alpha=0.30
)

plt.tight_layout()

# ------------------------------------------------------------
# 6. SAVE FIGURE
# ------------------------------------------------------------

png_path = os.path.join(
    FIGURES_DIR,
    "Figure_August_LightGBM_WAPE_by_Time_of_Day.png"
)

pdf_path = os.path.join(
    FIGURES_DIR,
    "Figure_August_LightGBM_WAPE_by_Time_of_Day.pdf"
)

plt.savefig(
    png_path,
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    pdf_path,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 7. SAVE TABLE
# ------------------------------------------------------------

table_path = os.path.join(
    TABLES_DIR,
    "Table_August_LightGBM_WAPE_by_Time_of_Day.csv"
)

time_wape[
    [
        "Interval_Index",
        "Time",
        "WAPE_%",
        "Mean_Observed_Volume",
        "Observations"
    ]
].to_csv(
    table_path,
    index=False
)

print("\n✅ Relative-error figure saved:")
print(png_path)

print("\n✅ Vector PDF saved:")
print(pdf_path)

print("\n✅ Relative-error table saved:")
print(table_path)

print("\n" + "=" * 95)
print("✅ RELATIVE TIME-OF-DAY ERROR ANALYSIS COMPLETED")
print("=" * 95)
```
